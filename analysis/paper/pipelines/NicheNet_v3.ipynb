# Pipeline (not re-run): TIL-zone DE, pathway filters and NicheNet per zone

**Produces** the frozen inputs of the TIL-zone figures, written to EFS `results/visium/til_zones/`:
- `deg_pseudobulk_edger/` (`config.ZONE_DE_DIR`): edgeR TIL-high vs TIL-low DE per TIL zone on Visium
  pseudobulk, and the UP / DOWN gene panels;
- `nichenet_filtering/` (`config.NICHENET_FILTERING_DIR`): ULM pathway scores per zone, snRNA-seq inputs,
  ligand / receptor DE per cell type;
- `nichenet_analysis/<zone>/` (`config.NICHENET_DIR`): `all_axes.csv`, `top_axes.csv`, `axes_catalog.csv`,
  `axes_shortlist.csv` and the hand-picked `axes_selected.csv` drawn in Fig 6a and Supp Fig 6a–b.

Spencer Watson's notebook, from `analysis/survival/` (outputs of the original run). Cell numbers
below count from the cell after this one.
- Cells 162–174 (cell-type correlation modules, Fig 2d–e) are run in `fig2de_celltype_modules.ipynb`.
- The circos cells 80 (TIL+), 116 (BZ1) and 152 (BZ2) are run in `fig6a_suppfig6_NicheNet_circos.ipynb`.

Known inconsistencies of this run:
- `PATHWAY_FILTER` differs between zones: TIL+ 1,934 genes (cell 51: Hallmark sets + a brain ECM list),
  BZ1 1,537 genes in the run behind the BZ1 axes against 1,963 printed by cell 87 as saved (Hallmark +
  KEGG Medicus + BioCarta sets + custom lists), BZ2 1,934 (cell 123). The Methods describe one filter.
- The TIL+ block ran a different enrichment (cell 49: sign-aligned ULM on the receiver-filtered target
  programs) where BZ1 and BZ2 ran the zone-DE ULM of cells 85 / 121.
- The frozen TIL+ `axes_selected.csv` predates the current TIL+ axis catalog: 6 of its 54 axis IDs
  (H03C0E5, H1E8A4B, H238F85, H5281DC, HF15A68, L517C1F) are not in `TILplus/axes_catalog.csv`, so
  cell 77 fails on the current state and Supp Fig 6a cannot be rebuilt from it; the frozen file is the
  input of the circos plot. The BZ1 and BZ2 selections match their catalogs (46 and 44 axes).

### Create new log_norm layer for merged single cell object

import scanpy as sc
import scipy.sparse as sp

COUNT_LAYER = "unnormalized_counts"
OUT_LAYER   = "LogNormalize"

TARGET_SUM     = 1e4
BATCH_KEY      = "center"   # set to None if you do not want batch-aware HVGs
HVG_N_TOP      = 3000
N_PCS          = 50
N_NEIGHBORS    = 15
UMAP_MIN_DIST  = 0.3
UMAP_SPREAD    = 1.0
RANDOM_STATE   = 13

if COUNT_LAYER not in adata_sc.layers:
    raise KeyError(f"adata_sc.layers['{COUNT_LAYER}'] not found. Available: {list(adata_sc.layers.keys())}")

if BATCH_KEY is not None and BATCH_KEY not in adata_sc.obs.columns:
    raise KeyError(f"adata_sc.obs['{BATCH_KEY}'] not found. Set BATCH_KEY=None or choose an existing key.")

adata_sc.layers[OUT_LAYER] = adata_sc.layers[COUNT_LAYER].copy()
sc.pp.normalize_total(adata_sc, layer=OUT_LAYER, target_sum=TARGET_SUM)
sc.pp.log1p(adata_sc, layer=OUT_LAYER)

sc.pp.highly_variable_genes(
    adata_sc,
    n_top_genes=HVG_N_TOP,
    flavor="seurat_v3",
    layer=COUNT_LAYER,
    batch_key=BATCH_KEY,
    subset=False,
)

adata_sc.obsm.pop("X_pca", None)
adata_sc.obsm.pop("X_umap", None)
adata_sc.uns.pop("neighbors", None)
adata_sc.obsp.pop("connectivities", None)
adata_sc.obsp.pop("distances", None)

sc.tl.pca(
    adata_sc,
    n_comps=N_PCS,
    svd_solver="arpack",
    use_highly_variable=True,
)

sc.pp.neighbors(adata_sc, n_neighbors=N_NEIGHBORS, n_pcs=N_PCS)
sc.tl.umap(adata_sc, min_dist=UMAP_MIN_DIST, spread=UMAP_SPREAD, random_state=RANDOM_STATE)

print("done")
print("LogNormalize layer:", OUT_LAYER, adata_sc.layers[OUT_LAYER].shape, type(adata_sc.layers[OUT_LAYER]))
print("HVGs:", int(adata_sc.var["highly_variable"].sum()))
# print("X_pca:", adata_sc.obsm["X_pca"].shape, "X_umap:", adata_sc.obsm["X_umap"].shape)

### Load lognorm and UMAP from file

```python
import os
import numpy as np
import pandas as pd
import scipy.sparse as sp

LOAD_DIR = "/home/sagemaker-user/gbm-collaboration_2/analysis_sc_merged"
LAYER    = "LogNormalize"
UMAP_KEY = "X_umap"

# file checks
npz_path = f"{LOAD_DIR}/{LAYER}.npz"
if not os.path.exists(npz_path):
    raise FileNotFoundError(f"Missing: {npz_path}")

for fn in ("obs_names.csv", "var_names.csv", "umap_coords.csv"):
    p = f"{LOAD_DIR}/{fn}"
    if not os.path.exists(p):
        raise FileNotFoundError(f"Missing: {p}")

# load saved artifacts
obs_saved = pd.read_csv(f"{LOAD_DIR}/obs_names.csv")["obs"].astype(str)
var_saved = pd.read_csv(f"{LOAD_DIR}/var_names.csv")["var"].astype(str)
M_saved   = sp.load_npz(npz_path)

U_saved = pd.read_csv(f"{LOAD_DIR}/umap_coords.csv", index_col=0)
U_saved.index = U_saved.index.astype(str)

# hard shape checks
if M_saved.shape != (len(obs_saved), len(var_saved)):
    raise ValueError(
        f"M_saved shape {M_saved.shape} does not match "
        f"(len(obs_saved), len(var_saved)) = {(len(obs_saved), len(var_saved))}"
    )

if obs_saved.duplicated().any():
    raise ValueError("Duplicate entries found in saved obs_names.csv")
if var_saved.duplicated().any():
    raise ValueError("Duplicate entries found in saved var_names.csv")

if U_saved.shape[1] < 2:
    raise ValueError(f"umap_coords.csv must have at least 2 columns, got {U_saved.shape[1]}")

# current object indices
obs_cur = pd.Index(adata_sc.obs_names.astype(str))
var_cur = pd.Index(adata_sc.var_names.astype(str))

# overlap
obs_common = obs_cur.intersection(pd.Index(obs_saved))
var_common = var_cur.intersection(pd.Index(var_saved))

if len(obs_common) == 0:
    raise ValueError("No overlapping obs_names between current adata_sc and saved obs_names.")
if len(var_common) == 0:
    raise ValueError("No overlapping var_names between current adata_sc and saved var_names.")

# indexers
cur_r = obs_cur.get_indexer(obs_common)
sav_r = pd.Index(obs_saved).get_indexer(obs_common)
cur_c = var_cur.get_indexer(var_common)
sav_c = pd.Index(var_saved).get_indexer(var_common)

# rebuild layer into current shape (only overlapping entries get filled)
sub = M_saved[sav_r, :][:, sav_c].tocoo()
rows = cur_r[sub.row]
cols = cur_c[sub.col]
M_cur = sp.csr_matrix((sub.data, (rows, cols)), shape=(len(obs_cur), len(var_cur)))
adata_sc.layers[LAYER] = M_cur

# align UMAP
U_cur = np.full((len(obs_cur), 2), np.nan, dtype=float)

U_common = U_saved.loc[obs_common]
U_vals = U_common.iloc[:, :2].to_numpy(dtype=float, copy=False)

U_cur[cur_r, :] = U_vals
adata_sc.obsm[UMAP_KEY] = U_cur

print("loaded into adata_sc:")
print("  layer:", LAYER, adata_sc.layers[LAYER].shape, "nnz:", int(adata_sc.layers[LAYER].nnz))
print("  umap:", UMAP_KEY, adata_sc.obsm[UMAP_KEY].shape, "nan_rows:", int(np.isnan(adata_sc.obsm[UMAP_KEY]).any(axis=1).sum()))
print("  obs_overlap:", len(obs_common), "/", len(obs_cur))
print("  var_overlap:", len(var_common), "/", len(var_cur))
```

### Load curated cell annotations from file

```python
import os
import pandas as pd

# user vars
ADATA_OBJ = adata_sc  # set to adata_sc (single cell) or adata (visium), explicitly
KEY_PATH = "/home/sagemaker-user/custom-file-systems/efs/fs-09913c1f7db79b6fd/Analysis_Repo/Branch_infercnv_visium_backup/analysis/SingleCell_Annotation_Key.csv"

LEVEL4_KEY = "celltype_level4_scanvi"
NEW_COLS = [
    "celltype_level4_curated",
    "celltype_level3_curated",
    "celltype_level3_figure",
    "celltype_level2_curated",
    "celltype_level2_figure",
    "celltype_level1_curated",
]

if not os.path.exists(KEY_PATH):
    raise FileNotFoundError(f"Missing KEY_PATH: {KEY_PATH}")

key = pd.read_csv(KEY_PATH)

# checks
if LEVEL4_KEY not in ADATA_OBJ.obs.columns:
    raise KeyError(f"Missing {LEVEL4_KEY} in ADATA_OBJ.obs")
missing_cols = [c for c in [LEVEL4_KEY] + NEW_COLS if c not in key.columns]
if missing_cols:
    raise KeyError(f"Missing column(s) in key file: {missing_cols}")

# normalize + deduplicate key
key = key.copy()
key[LEVEL4_KEY] = key[LEVEL4_KEY].astype(str).str.strip()
for c in NEW_COLS:
    key[c] = key[c].astype(str).where(key[c].notna(), None)

key = key.drop_duplicates(subset=[LEVEL4_KEY], keep="first").copy()

# mapping dicts from level4 -> curated levels
key_map = key.set_index(LEVEL4_KEY)[NEW_COLS].to_dict(orient="series")

# apply mappings to create new obs columns
src = ADATA_OBJ.obs[LEVEL4_KEY].astype(str).str.strip()

for col in NEW_COLS:
    mapped = src.map(key_map[col])
    out = mapped.fillna("Unmapped").astype(str)

    cats = pd.Index(key[col].dropna().astype(str).unique())
    if "Unmapped" in out.values and "Unmapped" not in cats:
        cats = cats.append(pd.Index(["Unmapped"]))

    ADATA_OBJ.obs[col] = pd.Categorical(out, categories=cats)

# quick report
lvl4_all = pd.Index(src.unique())
lvl4_in_key = pd.Index(key[LEVEL4_KEY].astype(str).unique())
missing = lvl4_all.difference(lvl4_in_key)

print(f"Target object: {'adata_sc' if ADATA_OBJ is adata_sc else 'adata'}")
print(f"Added columns: {NEW_COLS}")
print(f"Level4 labels in target: {len(lvl4_all)}")
print(f"Level4 labels covered by key: {len(lvl4_all.intersection(lvl4_in_key))}")
print(f"Level4 labels missing from key: {len(missing)}")
if len(missing):
    print("Unmapped labels:", ", ".join(list(missing)[:20]) + (" ..." if len(missing) > 20 else ""))
```

### Load Cell2Location absolute abundance data

```python
# Load ALL C2L abundance data (CHUV, UKER, Charite) - COMBINED
import gc
from joblib import Parallel, delayed
from joblib.externals.loky import get_reusable_executor

def load_sample(center_name, sample_id, h5ad_file):
    """Load and process a single sample using Cell2LocationLoader approach."""
    try:
        if isinstance(h5ad_file, S3Path) or str(h5ad_file).startswith('s3://'):
            s3 = s3fs.S3FileSystem(anon=False)
            h5ad_path_str = str(h5ad_file).replace('s3://', '')
            with s3.open(h5ad_path_str, 'rb') as f:
                adata = ad.read_h5ad(f)
        else:
            adata = ad.read_h5ad(h5ad_file)
        
        if 'means_cell_abundance_w_sf' not in adata.obsm:
            return None
        
        celltype_names = None
        if 'mod' in adata.uns and 'factor_names' in adata.uns['mod']:
            celltype_names = adata.uns['mod']['factor_names']
        elif 'factor_names' in adata.uns:
            celltype_names = adata.uns['factor_names']
        elif 'cell_type_names' in adata.uns:
            celltype_names = adata.uns['cell_type_names']
        
        if celltype_names is None:
            n_celltypes = adata.obsm['means_cell_abundance_w_sf'].shape[1]
            celltype_names = [f'celltype_{i}' for i in range(n_celltypes)]
        
        celltype_names = [f'meanscell_abundance_w_sf_{name}' for name in celltype_names]
        
        cell_abundance = adata.obsm['means_cell_abundance_w_sf']
        if issparse(cell_abundance):
            cell_abundance = cell_abundance.toarray()
        
        df = pd.DataFrame(cell_abundance, index=adata.obs_names, columns=celltype_names)
        df = df.fillna(0)
        df.index = pd.MultiIndex.from_tuples(
            [(sample_id, idx) for idx in df.index],
            names=['sample', 'barcode']
        )
        df['total_cells'] = df.sum(axis=1)
        
        return df
    except Exception as e:
        return None


def load_center(center_name, cell2loc_path):
    """Load one center with original fast parallel, then kill workers."""
    print(f"\nProcessing {center_name}...")
    base_path = S3Path(str(cell2loc_path))
    sample_dirs = [d.name for d in base_path.iterdir() if d.is_dir()]
    print(f"  Found {len(sample_dirs)} sample dirs for {center_name}")
    
    samples_to_process = [
        (center_name, sample_id, base_path / sample_id / "cohort_level_4.h5ad")
        for sample_id in sample_dirs
    ]
    
    # Original fast parallel call
    center_results = Parallel(n_jobs=-1)(
        delayed(load_sample)(center_name, sample_id, h5ad_file)
        for center_name, sample_id, h5ad_file in tqdm(samples_to_process, desc=f"  {center_name}")
    )
    
    center_results = [df for df in center_results if df is not None]
    
    for df in center_results:
        new_index = pd.MultiIndex.from_tuples(
            [(center_name, *idx) for idx in df.index],
            names=["center", "sample", "barcode"],
        )
        df.index = new_index
        df["center"] = center_name
    
    df_center = pd.concat(center_results, axis=0)
    
    print(f"  Loaded {df_center.index.get_level_values('sample').nunique()} samples from {center_name}")
    print(f"  Total spots: {df_center.shape[0]}")
    
    # FORCE KILL parallel workers to prevent hang on next call
    get_reusable_executor().shutdown(wait=True)
    gc.collect()
    
    return df_center


# Load each center (workers killed between each)
df_chuv = load_center("CHUV", DATA_PATH_CHUV_FILES["Cell2location"])
df_uker = load_center("UKER", DATA_PATH_UKER_FILES["Cell2location"])
df_cr = load_center("CR", DATA_PATH_CR_FILES["Cell2location"])


# Merge all centers
df_all = pd.concat([df_chuv, df_uker, df_cr], axis=0)

print("\n" + "=" * 60)
print("Combined df_all summary:")
print("  Total spots:", df_all.shape[0])
print("  Total samples:", df_all.index.get_level_values("sample").nunique())
print("  Index levels:", df_all.index.names)
print("  Example columns:", df_all.columns[:5].tolist())
print("=" * 60)
```

### Load custom cell annotations from file for Visium

```python
import pandas as pd

# user vars
KEY_PATH = "/home/sagemaker-user/custom-file-systems/efs/fs-09913c1f7db79b6fd/Analysis_Repo/Branch_infercnv_visium_backup/analysis/SingleCell_Annotation_Key.csv"
LEVEL4_KEY = "celltype_level4_scanvi"
NEW_COLS = [
    "celltype_level4_curated",
    "celltype_level3_curated",
    "celltype_level3_figure",
    "celltype_level2_curated",
    "celltype_level2_figure",
    "celltype_level1_curated",
]

# 1) Load key
key_raw = pd.read_csv(KEY_PATH)

# 2) Basic column checks
required_cols = [LEVEL4_KEY] + NEW_COLS
missing_cols = [c for c in required_cols if c not in key_raw.columns]
if missing_cols:
    raise ValueError(f"Missing columns in key file: {missing_cols}")

# 3) Deduplicate on the level4 key and standardize to string
key_unique = key_raw.drop_duplicates(subset=[LEVEL4_KEY]).copy()
key_unique[LEVEL4_KEY] = key_unique[LEVEL4_KEY].astype(str)

# 4) Build a tidy DataFrame indexed by level4
celltype_key = key_unique.set_index(LEVEL4_KEY)[NEW_COLS]
celltype_key = celltype_key.apply(lambda s: s.astype(str))

# 5) Attach as the canonical celltype hierarchy in adata.uns
adata.uns["celltype_annotation_key"] = celltype_key

# 6) Quick coverage checks vs Visium abundance columns
lvl4_in_key = pd.Index(celltype_key.index)

obs_cols = pd.Index(adata.obs.columns)

# deconv_cohort_level_4 columns should use the raw level4 names
deconv_level4_cols = obs_cols.intersection(lvl4_in_key)

# C2L totals use "meanscell_abundance_w_sf_<level4>"
c2l_cols = [c for c in obs_cols if c.startswith("meanscell_abundance_w_sf_")]
c2l_lvl4 = pd.Index([c.replace("meanscell_abundance_w_sf_", "") for c in c2l_cols])
c2l_in_key = c2l_lvl4.intersection(lvl4_in_key)

missing_for_deconv = lvl4_in_key.difference(deconv_level4_cols)
missing_for_c2l = lvl4_in_key.difference(c2l_in_key)

print("Stored celltype_annotation_key in adata.uns['celltype_annotation_key']")
print(f"Total level4 types in key: {len(lvl4_in_key)}")
print(f"Matching deconv level4 columns in adata.obs: {len(deconv_level4_cols)}")
print(f"Matching C2L level4 types (from meanscell_abundance_w_sf_*): {len(c2l_in_key)}")

if len(missing_for_deconv):
    print(f"Level4 types in key not seen as deconv columns: {len(missing_for_deconv)}")
if len(missing_for_c2l):
    print(f"Level4 types in key not seen in C2L columns: {len(missing_for_c2l)}")
```

### Define TIL-low and TIL-high patients

```python
import pandas as pd
import numpy as np

VISIUM_SAMPLE_COL = "sample"
SC_SAMPLE_COL     = "orig.ident"
TIL_FLAG_COL      = "TIL_positive"

QUANTILE_CUTOFF = 0.75
MIN_TIL_SPOTS_FOR_HIGH = 0
MIN_TOTAL_SPOTS = 0
MIN_TIL_FRACTION_FOR_HIGH = None

KEY_LEN = 9

if "adata" not in globals():
    raise ValueError("adata (Visium) not found in the session.")
if "adata_sc" not in globals():
    raise ValueError("adata_sc (single-cell) not found in the session.")

for col in [VISIUM_SAMPLE_COL, TIL_FLAG_COL]:
    if col not in adata.obs.columns:
        raise KeyError(f"Missing '{col}' in adata.obs")

if SC_SAMPLE_COL not in adata_sc.obs.columns:
    raise KeyError(f"Missing '{SC_SAMPLE_COL}' in adata_sc.obs")

visium_id_full = adata.obs[VISIUM_SAMPLE_COL].astype(str).str.replace(r"_vis$", "", regex=True)
visium_key = visium_id_full.str[:KEY_LEN]

df = (
    pd.DataFrame(
        {
            "join_key": visium_key.values,
            "visium_id_example": visium_id_full.values,
            "til_flag": adata.obs[TIL_FLAG_COL].astype(int).values,
        },
        index=adata.obs_names,
    )
    .groupby("join_key", as_index=False)
    .agg(
        TIL_spots=("til_flag", "sum"),
        Total_Spots=("til_flag", "size"),
        visium_id_example=("visium_id_example", lambda s: s.iloc[0]),
    )
)

df["TIL_fraction"] = df["TIL_spots"] / df["Total_Spots"]
df["TIL_percent"] = 100.0 * df["TIL_fraction"]

if MIN_TOTAL_SPOTS > 0:
    df = df[df["Total_Spots"] >= int(MIN_TOTAL_SPOTS)].copy()

if df.empty:
    raise ValueError("No groups remain after applying MIN_TOTAL_SPOTS filter.")

cutoff = float(df["TIL_fraction"].quantile(float(QUANTILE_CUTOFF)))
if MIN_TIL_FRACTION_FOR_HIGH is not None:
    cutoff = max(cutoff, float(MIN_TIL_FRACTION_FOR_HIGH))

df["TIL_group"] = "TIL_low"
high_mask = df["TIL_fraction"] >= cutoff
if MIN_TIL_SPOTS_FOR_HIGH > 0:
    high_mask &= df["TIL_spots"] >= int(MIN_TIL_SPOTS_FOR_HIGH)
df.loc[high_mask, "TIL_group"] = "TIL_high"

key_to_label = df.set_index("join_key")["TIL_group"].to_dict()

adata.obs["TIL_group"] = visium_key.map(key_to_label)
adata.obs["TIL_group"] = pd.Categorical(
    adata.obs["TIL_group"],
    categories=["TIL_high", "TIL_low"],
    ordered=True,
)

sc_id_full = adata_sc.obs[SC_SAMPLE_COL].astype(str)
sc_key = sc_id_full.str[:KEY_LEN]

adata_sc.obs["sample_id"] = sc_key.astype(str)

sc_labels = sc_key.map(key_to_label).fillna("Unmapped")
adata_sc.obs["TIL_group"] = pd.Categorical(
    sc_labels,
    categories=["TIL_high", "TIL_low", "Unmapped"],
    ordered=False,
)

adata.uns["TIL_group_stats"] = df.set_index("join_key")

print(f"Join key: first {KEY_LEN} chars of Visium sample (after stripping '_vis') and of single-cell orig.ident")
print(f"TIL_fraction cutoff for TIL_high (Q={QUANTILE_CUTOFF:.2f}): {cutoff:.6f}")

print("\nVisium group counts (unique join keys):")
print(df["TIL_group"].value_counts().to_string())

print("\nSingle-cell TIL_group value counts:")
print(adata_sc.obs["TIL_group"].value_counts(dropna=False).to_string())

n_sc_mapped = int((adata_sc.obs["TIL_group"].astype(str) != "Unmapped").sum())
print(f"\nSingle-cell mapped (not Unmapped): {n_sc_mapped} / {adata_sc.n_obs}")

vis_keys = set(pd.unique(visium_key))
sc_keys = set(pd.unique(sc_key))
only_in_sc = sorted(sc_keys - vis_keys)
only_in_vis = sorted(vis_keys - sc_keys)

print(f"\nKey coverage (unique join keys):")
print(f"  keys in Visium: {len(vis_keys)}")
print(f"  keys in scRNA : {len(sc_keys)}")
print(f"  keys only in scRNA (unmappable from Visium): {len(only_in_sc)}")
print(f"  keys only in Visium (no scRNA): {len(only_in_vis)}")
print(f"  examples only_in_sc (up to 20): {only_in_sc[:20]}")
print(f"  examples only_in_vis (up to 20): {only_in_vis[:20]}")
```

### Pseudobulk TIL zones per patient

```python
import numpy as np
import pandas as pd
import scipy.sparse as sp

ZONE_KEY = "TIL_zone"
GROUP_KEY = "TIL_group"
SAMPLE_KEY = "sample"
RAW_COUNT_LAYER = "raw_count"

ZONES = ["TIL+", "BZ1", "BZ2", "TIL-"]
GROUPS = ["TIL_low", "TIL_high"]  # low is reference

MIN_SPOTS_PER_SAMPLE_IN_ZONE = 1
MIN_SAMPLES_PER_GROUP = 3

for k in [ZONE_KEY, GROUP_KEY, SAMPLE_KEY]:
    if k not in adata_deg.obs.columns:
        raise KeyError(f"adata_deg.obs['{k}'] not found")
if RAW_COUNT_LAYER not in adata_deg.layers:
    raise KeyError(f"adata_deg.layers['{RAW_COUNT_LAYER}'] not found")

def pseudobulk_counts_for_zone(adata_in, zone: str):
    obs = adata_in.obs
    mask = (obs[ZONE_KEY].astype(str) == str(zone)) & (obs[GROUP_KEY].astype(str).isin(GROUPS))
    idx = np.flatnonzero(mask.to_numpy())
    if idx.size == 0:
        return None

    sub_obs = obs.iloc[idx][[SAMPLE_KEY, GROUP_KEY]].copy()
    sub_obs[SAMPLE_KEY] = sub_obs[SAMPLE_KEY].astype(str)
    sub_obs[GROUP_KEY] = sub_obs[GROUP_KEY].astype(str)

    spot_counts = sub_obs[SAMPLE_KEY].value_counts()
    keep_samples = spot_counts[spot_counts >= int(MIN_SPOTS_PER_SAMPLE_IN_ZONE)].index
    keep_mask = sub_obs[SAMPLE_KEY].isin(keep_samples).to_numpy()

    idx = idx[keep_mask]
    sub_obs = sub_obs.iloc[keep_mask].copy()
    if idx.size == 0:
        return None

    nuniq = sub_obs.groupby(SAMPLE_KEY)[GROUP_KEY].nunique()
    bad = nuniq[nuniq > 1]
    if len(bad):
        raise ValueError(f"{len(bad)} samples map to >1 group in zone={zone}. Examples: {bad.index[:5].tolist()}")

    sample_group = sub_obs.groupby(SAMPLE_KEY)[GROUP_KEY].first()

    X = adata_in.layers[RAW_COUNT_LAYER]
    samples = sub_obs[SAMPLE_KEY].to_numpy()
    uniq_samples, inv = np.unique(samples, return_inverse=True)

    if sp.issparse(X):
        Xz = X[idx, :].tocsr()
        G = sp.csr_matrix(
            (np.ones(inv.size, dtype=np.float32), (inv, np.arange(inv.size))),
            shape=(len(uniq_samples), inv.size),
        )
        Xpb = (G @ Xz).toarray()
    else:
        Xz = np.asarray(X)[idx, :]
        Xpb = np.zeros((len(uniq_samples), adata_in.n_vars), dtype=np.float64)
        np.add.at(Xpb, inv, Xz)

    Xpb = np.rint(Xpb).astype(np.int64)
    if (Xpb < 0).any():
        raise ValueError(f"Negative counts after pseudobulk in zone={zone}")

    gene_mask = Xpb.sum(axis=0) > 0
    genes = pd.Index(adata_in.var_names.astype(str))[gene_mask]
    Xpb = Xpb[:, gene_mask]

    meta = pd.DataFrame(index=pd.Index(uniq_samples, name="sample"))
    meta["TIL_group"] = sample_group.reindex(uniq_samples).astype(str).values
    meta["n_spots_in_zone"] = spot_counts.reindex(uniq_samples).fillna(0).astype(int).values
    meta["zone"] = str(zone)

    counts_df = pd.DataFrame(Xpb.T, index=genes, columns=uniq_samples)
    return counts_df, meta

counts_by_zone = {}
meta_by_zone = {}
summary_rows = []

for zone in ZONES:
    out = pseudobulk_counts_for_zone(adata_deg, zone)
    if out is None:
        summary_rows.append({"zone": zone, "status": "skip", "reason": "no spots after filtering"})
        continue

    counts_df, meta = out
    meta["TIL_group"] = pd.Categorical(meta["TIL_group"].astype(str), categories=GROUPS, ordered=True)

    n_by_group = meta["TIL_group"].value_counts().to_dict()
    ok = all(int(n_by_group.get(g, 0)) >= int(MIN_SAMPLES_PER_GROUP) for g in GROUPS)

    summary_rows.append({
        "zone": zone,
        "status": "ok" if ok else "skip",
        "reason": "" if ok else f"too few samples per group: {n_by_group}",
        "n_samples_total": int(meta.shape[0]),
        "n_TIL_low": int(n_by_group.get("TIL_low", 0)),
        "n_TIL_high": int(n_by_group.get("TIL_high", 0)),
        "n_spots_total_in_zone": int(meta["n_spots_in_zone"].sum()),
        "median_spots_per_sample": float(np.median(meta["n_spots_in_zone"].to_numpy())),
        "n_genes_nonzero": int(counts_df.shape[0]),
    })

    if ok:
        counts_by_zone[zone] = counts_df
        meta_by_zone[zone] = meta

df_pb_summary = pd.DataFrame(summary_rows)
display(df_pb_summary)
print("Zones kept for DE:", list(counts_by_zone.keys()))
```

### DE analysis of each TIL zone

```python
import pandas as pd
import numpy as np
from pathlib import Path

import rpy2.robjects as ro
from rpy2.robjects import pandas2ri
from rpy2.robjects.conversion import localconverter

GROUPS = ["TIL_low", "TIL_high"]  # low is reference

OUT_DIR = Path("/home/sagemaker-user/custom-file-systems/efs/fs-09913c1f7db79b6fd/results/visium/til_zones/deg_pseudobulk_edger")
OUT_DIR.mkdir(parents=True, exist_ok=True)

ro.r("suppressPackageStartupMessages(library(edgeR))")

r_run_edger = ro.r(r"""
function(counts_mat, group_vec) {
  group_vec <- factor(group_vec, levels=c("TIL_low","TIL_high"))

  counts <- as.matrix(counts_mat)
  storage.mode(counts) <- "integer"

  y <- DGEList(counts=counts)
  design <- model.matrix(~group_vec)

  keep <- filterByExpr(y, design=design)
  y <- y[keep, , keep.lib.sizes=FALSE]
  y <- calcNormFactors(y)

  y <- estimateDisp(y, design, robust=TRUE)
  fit <- glmQLFit(y, design, robust=TRUE)
  qlf <- glmQLFTest(fit, coef=2)

  tt <- topTags(qlf, n=Inf)$table
  tt$gene <- rownames(tt)
  tt <- tt[, c("gene","logFC","logCPM","F","PValue","FDR")]
  tt
}
""")

de_tables = {}
summary = []

for zone in counts_by_zone.keys():
    counts_df = counts_by_zone[zone]
    meta = meta_by_zone[zone]

    group_vec = meta.loc[counts_df.columns, "TIL_group"].astype(str).tolist()

    with localconverter(ro.default_converter + pandas2ri.converter):
        r_counts = ro.conversion.py2rpy(counts_df)

    res_r = r_run_edger(r_counts, ro.StrVector(group_vec))

    with localconverter(ro.default_converter + pandas2ri.converter):
        df_de = ro.conversion.rpy2py(res_r)

    n_by_group = meta["TIL_group"].astype(str).value_counts().to_dict()
    df_de["zone"] = zone
    df_de["contrast"] = "TIL_high_vs_TIL_low"
    df_de["n_samples_total"] = int(counts_df.shape[1])
    df_de["n_TIL_low"] = int(n_by_group.get("TIL_low", 0))
    df_de["n_TIL_high"] = int(n_by_group.get("TIL_high", 0))

    out_de = OUT_DIR / f"DE_edger_zone_{zone}_TIL_high_vs_TIL_low.csv"
    out_meta = OUT_DIR / f"DE_edger_zone_{zone}_sample_metadata.csv"

    df_de.to_csv(out_de, index=False)
    meta.reset_index().to_csv(out_meta, index=False)

    de_tables[zone] = df_de

    summary.append({
        "zone": zone,
        "n_genes_tested": int(df_de.shape[0]),
        "n_samples_total": int(counts_df.shape[1]),
        "n_TIL_low": int(n_by_group.get("TIL_low", 0)),
        "n_TIL_high": int(n_by_group.get("TIL_high", 0)),
        "de_csv": str(out_de),
        "meta_csv": str(out_meta),
    })

df_de_summary = pd.DataFrame(summary)
summary_csv = OUT_DIR / "DE_edger_summary.csv"
df_de_summary.to_csv(summary_csv, index=False)

print("Wrote:", summary_csv)
display(df_de_summary)
```

### Create UP and DOWN gene panels

```python
import pandas as pd
import numpy as np
from pathlib import Path

FDR_MAX = 0.05
LOGFC_MIN = 0.25          # set to 0.0 if you want "all significant" regardless of effect size
MIN_LOGCPM = None         # e.g. 0.0 or 1.0, or keep as None to skip

if "df_de_summary" not in globals():
    raise NameError("df_de_summary is not defined. Run the edgeR summary cell that creates it.")

out_dir = Path(df_de_summary["de_csv"].iloc[0]).parent
out_dir.mkdir(parents=True, exist_ok=True)

de_tables = {}
panel_rows = []
panel_counts = []

for _, row in df_de_summary.iterrows():
    zone = str(row["zone"])
    de_path = Path(row["de_csv"])
    if not de_path.exists():
        raise FileNotFoundError(f"Missing DE file for zone={zone}: {de_path}")

    df = pd.read_csv(de_path)

    required = ["gene", "logFC", "logCPM", "PValue", "FDR"]
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(f"DE table for zone={zone} is missing columns: {missing}")

    df = df.dropna(subset=["gene", "logFC", "FDR"]).copy()
    df["gene"] = df["gene"].astype(str)

    if MIN_LOGCPM is not None:
        df = df[df["logCPM"] >= float(MIN_LOGCPM)].copy()

    sig = df[df["FDR"] <= float(FDR_MAX)].copy()

    up = sig[sig["logFC"] >= float(LOGFC_MIN)].sort_values(["FDR", "PValue"], ascending=[True, True]).copy()
    down = sig[sig["logFC"] <= -float(LOGFC_MIN)].sort_values(["FDR", "PValue"], ascending=[True, True]).copy()

    up_genes = up["gene"].tolist()
    down_genes = down["gene"].tolist()

    (out_dir / "gene_panels").mkdir(parents=True, exist_ok=True)
    pd.Series(up_genes, name="gene").to_csv(out_dir / "gene_panels" / f"{zone}_UP_FDR{FDR_MAX}_logFC{LOGFC_MIN}.txt",
                                            index=False, header=False)
    pd.Series(down_genes, name="gene").to_csv(out_dir / "gene_panels" / f"{zone}_DOWN_FDR{FDR_MAX}_logFC{LOGFC_MIN}.txt",
                                              index=False, header=False)

    up_min = up[["gene", "logFC", "logCPM", "PValue", "FDR"]].copy()
    up_min["zone"] = zone
    up_min["direction"] = "UP"

    down_min = down[["gene", "logFC", "logCPM", "PValue", "FDR"]].copy()
    down_min["zone"] = zone
    down_min["direction"] = "DOWN"

    panel_rows.append(up_min)
    panel_rows.append(down_min)

    panel_counts.append({
        "zone": zone,
        "n_sig_total": int(sig.shape[0]),
        "n_up": int(len(up_genes)),
        "n_down": int(len(down_genes)),
        "FDR_MAX": float(FDR_MAX),
        "LOGFC_MIN": float(LOGFC_MIN),
        "MIN_LOGCPM": (None if MIN_LOGCPM is None else float(MIN_LOGCPM)),
        "de_csv": str(de_path),
    })

    de_tables[zone] = df

df_panels = pd.concat(panel_rows, ignore_index=True)
df_panel_counts = pd.DataFrame(panel_counts).sort_values("zone")

panels_csv = out_dir / f"DE_gene_panels_by_zone_FDR{FDR_MAX}_logFC{LOGFC_MIN}.csv"
counts_csv = out_dir / f"DE_gene_panel_counts_FDR{FDR_MAX}_logFC{LOGFC_MIN}.csv"

df_panels.to_csv(panels_csv, index=False)
df_panel_counts.to_csv(counts_csv, index=False)

print("Wrote:", panels_csv)
print("Wrote:", counts_csv)
display(df_panel_counts)
```

### Cell type association with TIL zones

```python
import numpy as np
import pandas as pd
from pathlib import Path
from scipy.stats import wilcoxon

ZONE_KEY = "TIL_zone"
GROUP_KEY = "TIL_group"
SAMPLE_KEY = "sample"

DECONV_PREFIX = "meanscell_abundance_w_sf_"
KEY_PATH = "/home/sagemaker-user/custom-file-systems/efs/fs-09913c1f7db79b6fd/Analysis_Repo/Branch_infercnv_visium_backup/analysis/SingleCell_Annotation_Key.csv"
SCANVI_KEY = "celltype_level4_scanvi"
CUR_KEY = "celltype_level4_curated"

ZONES = ["TIL+", "BZ1", "BZ2", "TIL-"]
GROUPS = ["TIL_low", "TIL_high"]

OUT_DIR = Path("/home/sagemaker-user/custom-file-systems/efs/fs-09913c1f7db79b6fd/results/visium/til_zones/celltype_zone_specificity_curated")
OUT_DIR.mkdir(parents=True, exist_ok=True)

def bh_fdr(pvals: np.ndarray) -> np.ndarray:
    p = np.asarray(pvals, dtype=float)
    out = np.full_like(p, np.nan, dtype=float)

    ok = np.isfinite(p)
    if ok.sum() == 0:
        return out

    p_ok = p[ok]
    m = p_ok.size
    order = np.argsort(p_ok)
    ranked = p_ok[order]
    q = ranked * (m / (np.arange(1, m + 1)))
    q = np.minimum.accumulate(q[::-1])[::-1]
    q = np.clip(q, 0.0, 1.0)

    q_full = np.empty_like(p_ok)
    q_full[order] = q
    out[ok] = q_full
    return out

for k in [ZONE_KEY, GROUP_KEY, SAMPLE_KEY]:
    if k not in adata_deg.obs.columns:
        raise KeyError(f"adata_deg.obs['{k}'] not found")

key_path = Path(KEY_PATH)
if not key_path.exists():
    raise FileNotFoundError(f"Missing KEY_PATH: {KEY_PATH}")

key = pd.read_csv(key_path)
for c in [SCANVI_KEY, CUR_KEY]:
    if c not in key.columns:
        raise KeyError(f"Missing column '{c}' in key file")

key = key.copy()
key[SCANVI_KEY] = key[SCANVI_KEY].astype(str).str.strip()
key[CUR_KEY] = key[CUR_KEY].astype(str).str.strip()
scanvi_to_cur = dict(zip(key[SCANVI_KEY].tolist(), key[CUR_KEY].tolist()))

pref_cols = [c for c in adata_deg.obs.columns if c.startswith(DECONV_PREFIX)]
if len(pref_cols) == 0:
    raise ValueError(f"No deconvolution columns found with prefix: {DECONV_PREFIX}")

base_labels = [c[len(DECONV_PREFIX):] for c in pref_cols]
missing_in_key = sorted(set(base_labels).difference(set(scanvi_to_cur.keys())))
if len(missing_in_key) > 0:
    raise ValueError(
        f"{len(missing_in_key)} deconv base labels are missing from the key mapping. "
        f"Examples: {missing_in_key[:20]}"
    )

col_to_cur = {c: scanvi_to_cur[c[len(DECONV_PREFIX):]] for c in pref_cols}

obs = adata_deg.obs
df = obs[[SAMPLE_KEY, GROUP_KEY, ZONE_KEY] + pref_cols].copy()
df[SAMPLE_KEY] = df[SAMPLE_KEY].astype(str)
df[GROUP_KEY] = df[GROUP_KEY].astype(str)
df[ZONE_KEY] = df[ZONE_KEY].astype(str)

df = df[df[GROUP_KEY].isin(GROUPS)].copy()
df = df[df[ZONE_KEY].isin(ZONES)].copy()
if df.empty:
    raise ValueError("No rows remain after filtering to GROUPS and ZONES")

df_means = (
    df.groupby([SAMPLE_KEY, GROUP_KEY, ZONE_KEY], observed=True)[pref_cols]
      .mean()
      .rename(columns=col_to_cur)
      .reset_index()
)

celltypes_cur = [c for c in df_means.columns if c not in [SAMPLE_KEY, GROUP_KEY, ZONE_KEY]]
if len(celltypes_cur) == 0:
    raise ValueError("No curated celltype columns present after renaming")

coverage = (
    df_means[[SAMPLE_KEY, GROUP_KEY, ZONE_KEY]]
    .drop_duplicates()
    .groupby([GROUP_KEY, ZONE_KEY], observed=True)[SAMPLE_KEY]
    .nunique()
    .unstack(ZONE_KEY)
    .reindex(index=GROUPS, columns=ZONES)
)
print("n_samples with >=1 spot in zone (by group):")
display(coverage)

eps = 1e-9
rows = []

for g in GROUPS:
    df_g = df_means[df_means[GROUP_KEY] == g].copy()
    if df_g.empty:
        continue

    for ct in celltypes_cur:
        mat = df_g.pivot_table(index=SAMPLE_KEY, columns=ZONE_KEY, values=ct, aggfunc="mean")

        present_zones = [z for z in ZONES if z in mat.columns]
        if len(present_zones) < 2:
            continue

        for z in present_zones:
            other = [zz for zz in present_zones if zz != z]
            v_zone = mat[z]
            v_rest = mat[other].mean(axis=1, skipna=True)

            m = v_zone.notna() & v_rest.notna()
            n = int(m.sum())
            if n == 0:
                continue

            l2 = np.log2((v_zone[m].to_numpy(dtype=float) + eps) / (v_rest[m].to_numpy(dtype=float) + eps))

            mean_zone = float(np.mean(v_zone[m].to_numpy(dtype=float)))
            mean_rest = float(np.mean(v_rest[m].to_numpy(dtype=float)))
            mean_l2 = float(np.mean(l2))
            med_l2 = float(np.median(l2))

            p = np.nan
            if n >= 5:
                try:
                    p = float(wilcoxon(l2, zero_method="wilcox", alternative="two-sided").pvalue)
                except Exception:
                    p = np.nan

            rows.append({
                "group": g,
                "zone": z,
                "celltype_level4_curated": ct,
                "n_samples_used": n,
                "mean_zone": mean_zone,
                "mean_rest_other_zones": mean_rest,
                "delta_zone_minus_rest": mean_zone - mean_rest,
                "mean_log2FC_zone_vs_rest": mean_l2,
                "median_log2FC_zone_vs_rest": med_l2,
                "pval_wilcoxon_paired": p,
            })

res = pd.DataFrame(rows)
if res.empty:
    raise ValueError("No results generated. This usually means too few zones per sample within group after aggregation.")

res["fdr_bh_within_group_zone"] = np.nan
for (g, z), sub_idx in res.groupby(["group", "zone"]).groups.items():
    res.loc[sub_idx, "fdr_bh_within_group_zone"] = bh_fdr(res.loc[sub_idx, "pval_wilcoxon_paired"].to_numpy())

out_csv = OUT_DIR / "celltype_zone_specificity_level4_curated.csv"
res.to_csv(out_csv, index=False)
print("Wrote:", str(out_csv))

for g in GROUPS:
    for z in ZONES:
        sub = res[(res["group"] == g) & (res["zone"] == z)].copy()
        if sub.empty:
            continue
        sub = sub.sort_values("mean_log2FC_zone_vs_rest", key=lambda s: np.abs(s), ascending=False)

        print(f"\n(group={g}) zone={z}: top 20 by abs(mean_log2FC_zone_vs_rest)")
        print(sub[[
            "celltype_level4_curated",
            "n_samples_used",
            "mean_zone",
            "mean_rest_other_zones",
            "mean_log2FC_zone_vs_rest",
            "pval_wilcoxon_paired",
            "fdr_bh_within_group_zone",
        ]].head(20).to_string(index=False))
```

### Load and filter NicheNet priors based on Omnipath allinfo consensus_score (secreted-intracelullar)

```python
%%R -o high_conf_ligands -o high_conf_receptors -o lr_network -o ligand_target_matrix
options(timeout = 3600)

suppressPackageStartupMessages({
  library(dplyr)
  library(readr)
})

# vars
# Consensus cutoffs: you can use absolute thresholds, percentile thresholds, or both
USE_PERCENTILES <- TRUE
CONS_PCTL_LIGAND   <- 0.50
CONS_PCTL_RECEPTOR <- 0.50

# Absolute floors (still enforced even if USE_PERCENTILES=TRUE)
MIN_CONS_LIGAND   <- 3
MIN_CONS_RECEPTOR <- 3

# Penalize intracellular annotations
# net_score = max_role_consensus - max_intracellular_consensus
MIN_NET_LIGAND   <- 2
MIN_NET_RECEPTOR <- 1

# Role and topology constraints
REQUIRE_TRANSMITTER <- TRUE
REQUIRE_RECEIVER    <- TRUE
REQUIRE_LIGAND_TOPOLOGY_SEC_OR_PMP <- TRUE     # secreted or plasma membrane peripheral
REQUIRE_RECEPTOR_PMTM <- TRUE                  # plasma membrane transmembrane

# Cache and load NicheNet priors
cache_dir <- Sys.getenv("NICHENET_CACHE", unset = file.path(Sys.getenv("HOME"), ".cache", "nichenetr"))
dir.create(cache_dir, recursive = TRUE, showWarnings = FALSE)
cat("Cache directory:", cache_dir, "\n")

read_and_cache_rds <- function(name, url) {
  out <- file.path(cache_dir, paste0(name, ".rds"))
  if (file.exists(out)) {
    cat("Loading cached:", name, "\n")
    return(readRDS(out))
  }
  cat("Downloading:", name, "\n")
  obj <- readRDS(url(url))
  saveRDS(obj, out)
  obj
}

base_url <- "https://zenodo.org/records/7074291/files/"

lr_network_raw <- read_and_cache_rds(
  "lr_network",
  paste0(base_url, "lr_network_human_21122021.rds?download=1")
)

ligand_target_matrix <- read_and_cache_rds(
  "ligand_target_matrix",
  paste0(base_url, "ligand_target_matrix_nsga2r_final.rds?download=1")
)

weighted_networks <- read_and_cache_rds(
  "weighted_networks",
  paste0(base_url, "weighted_networks_nsga2r_final.rds?download=1")
)

stopifnot(is.data.frame(lr_network_raw))
stopifnot(all(c("from", "to") %in% colnames(lr_network_raw)))
stopifnot(is.matrix(ligand_target_matrix) || is.array(ligand_target_matrix))

# Orient ligand_target_matrix (rows=ligands, cols=targets)
canon_ligs <- c("TGFB1", "TNF", "IFNG", "CXCL12", "VEGFA", "CCL2", "IL6", "PDGFB")
canon_tgts <- c("COL1A1", "COL3A1", "DCN", "LUM", "CXCL10", "ICAM1", "VCAM1", "STAT1")

score_orientation <- function(mat, ligs, tgts) {
  rn <- rownames(mat)
  cn <- colnames(mat)
  pct_lig_rows <- sum(ligs %in% rn) / max(1L, length(ligs))
  pct_lig_cols <- sum(ligs %in% cn) / max(1L, length(ligs))
  pct_tgt_rows <- sum(tgts %in% rn) / max(1L, length(tgts))
  pct_tgt_cols <- sum(tgts %in% cn) / max(1L, length(tgts))
  (pct_lig_rows + pct_tgt_cols) - (pct_lig_cols + pct_tgt_rows)
}

if (score_orientation(t(ligand_target_matrix), canon_ligs, canon_tgts) >
    score_orientation(ligand_target_matrix, canon_ligs, canon_tgts)) {
  cat("Transposing ligand_target_matrix to orient rows=ligands, cols=targets\n")
  ligand_target_matrix <- t(ligand_target_matrix)
  saveRDS(ligand_target_matrix, file.path(cache_dir, "ligand_target_matrix.rds"))
}

cat("\nNicheNet priors loaded:\n")
cat("  lr_network:", nrow(lr_network_raw), "interactions\n")
cat("  ligand_target_matrix:", nrow(ligand_target_matrix), "ligands x", ncol(ligand_target_matrix), "targets\n")


# Fetch OmniPath intercell annotations from web service
cat("\nFetching OmniPath intercell annotations...\n")

# Ask for the columns we actually use, to avoid relying on defaults
intercell_url <- paste0(
  "https://omnipathdb.org/intercell?format=tsv&entity_types=protein&scope=generic",
  "&fields=genesymbol,category,parent,consensus_score,transmitter,receiver,secreted,plasma_membrane_peripheral,plasma_membrane_transmembrane"
)

intercell_cache <- file.path(cache_dir, "omnipath_intercell_strict_fields.tsv")
if (file.exists(intercell_cache)) {
  cat("Loading cached intercell annotations\n")
  intercell <- read_tsv(intercell_cache, show_col_types = FALSE)
} else {
  cat("Downloading intercell annotations...\n")
  intercell <- read_tsv(intercell_url, show_col_types = FALSE)
  write_tsv(intercell, intercell_cache)
}

cat("Intercell annotations:", nrow(intercell), "rows\n")

need_cols <- c(
  "genesymbol","category","parent","consensus_score",
  "transmitter","receiver",
  "secreted","plasma_membrane_peripheral","plasma_membrane_transmembrane"
)
missing_cols <- setdiff(need_cols, colnames(intercell))
if (length(missing_cols) > 0) {
  stop("Missing required intercell columns: ", paste(missing_cols, collapse = ", "))
}


# Build per-gene max scores, topology flags, and net scores
ligand_categories <- c("ligand","cell_surface_ligand")
receptor_categories <- c("receptor")

intercell2 <- intercell %>%
  mutate(
    genesymbol = as.character(genesymbol),
    category = as.character(category),
    parent = as.character(parent),
    consensus_score = suppressWarnings(as.numeric(consensus_score)),
    transmitter = as.logical(transmitter),
    receiver = as.logical(receiver),
    secreted = as.logical(secreted),
    plasma_membrane_peripheral = as.logical(plasma_membrane_peripheral),
    plasma_membrane_transmembrane = as.logical(plasma_membrane_transmembrane),

    is_intracellular = (parent == "intracellular") | (category == "intracellular"),
    is_liglike = (category %in% ligand_categories) | (parent %in% ligand_categories),
    is_reclike = (category %in% receptor_categories) | (parent %in% receptor_categories)
  )

summ <- intercell2 %>%
  group_by(genesymbol) %>%
  summarise(
    max_lig = suppressWarnings(max(consensus_score[is_liglike], na.rm = TRUE)),
    max_rec = suppressWarnings(max(consensus_score[is_reclike], na.rm = TRUE)),
    max_intra = suppressWarnings(max(consensus_score[is_intracellular], na.rm = TRUE)),

    net_lig = max_lig - max_intra,
    net_rec = max_rec - max_intra,

    transmitter_any = any(transmitter %in% TRUE, na.rm = TRUE),
    receiver_any    = any(receiver %in% TRUE, na.rm = TRUE),

    secreted_any = any(secreted %in% TRUE, na.rm = TRUE),
    pmp_any      = any(plasma_membrane_peripheral %in% TRUE, na.rm = TRUE),
    pmtm_any     = any(plasma_membrane_transmembrane %in% TRUE, na.rm = TRUE),
    .groups = "drop"
  )

# Percentile thresholds (computed on finite max scores)
thr_lig <- if (USE_PERCENTILES) {
  quantile(summ$max_lig[is.finite(summ$max_lig)], probs = CONS_PCTL_LIGAND, na.rm = TRUE, names = FALSE)
} else {
  -Inf
}

thr_rec <- if (USE_PERCENTILES) {
  quantile(summ$max_rec[is.finite(summ$max_rec)], probs = CONS_PCTL_RECEPTOR, na.rm = TRUE, names = FALSE)
} else {
  -Inf
}

thr_lig <- max(thr_lig, MIN_CONS_LIGAND)
thr_rec <- max(thr_rec, MIN_CONS_RECEPTOR)

cat("\nThresholds used:\n")
cat("  ligand max_consensus >=", thr_lig, "\n")
cat("  receptor max_consensus >=", thr_rec, "\n")
cat("  MIN_NET_LIGAND:", MIN_NET_LIGAND, "\n")
cat("  MIN_NET_RECEPTOR:", MIN_NET_RECEPTOR, "\n")

# Ligand filter (no blacklists)
lig_filt <- summ %>%
  filter(is.finite(max_lig), max_lig >= thr_lig, is.finite(net_lig), net_lig >= MIN_NET_LIGAND)

if (REQUIRE_TRANSMITTER) {
  lig_filt <- lig_filt %>% filter(transmitter_any)
}
if (REQUIRE_LIGAND_TOPOLOGY_SEC_OR_PMP) {
  lig_filt <- lig_filt %>% filter(secreted_any | pmp_any)
}

# Receptor filter (no blacklists)
rec_filt <- summ %>%
  filter(is.finite(max_rec), max_rec >= thr_rec, is.finite(net_rec), net_rec >= MIN_NET_RECEPTOR)

if (REQUIRE_RECEIVER) {
  rec_filt <- rec_filt %>% filter(receiver_any)
}
if (REQUIRE_RECEPTOR_PMTM) {
  rec_filt <- rec_filt %>% filter(pmtm_any)
}

high_conf_ligands   <- unique(lig_filt$genesymbol)
high_conf_receptors <- unique(rec_filt$genesymbol)

cat("\nHigh-confidence genes:\n")
cat("  Ligands:", length(high_conf_ligands), "\n")
cat("  Receptors:", length(high_conf_receptors), "\n")


# Cross-reference with NicheNet priors and filter LR network
nn_ligands   <- unique(as.character(lr_network_raw$from))
nn_receptors <- unique(as.character(lr_network_raw$to))
ltm_ligands  <- rownames(ligand_target_matrix)

cat("\nOverlap with NicheNet:\n")
cat("  LR ligands in filtered set:", length(intersect(high_conf_ligands, nn_ligands)), "/", length(nn_ligands), "\n")
cat("  LR receptors in filtered set:", length(intersect(high_conf_receptors, nn_receptors)), "/", length(nn_receptors), "\n")
cat("  LTM ligands in filtered set:", length(intersect(high_conf_ligands, ltm_ligands)), "/", length(ltm_ligands), "\n")

lr_network <- lr_network_raw %>%
  filter(from %in% high_conf_ligands, to %in% high_conf_receptors)

cat("\nFiltered lr_network:", nrow(lr_network), "interactions (was", nrow(lr_network_raw), ")\n")

# Diagnostics (print-only)
cat("\nDiagnostic - selected genes:\n")
check_genes <- c("COPA", "CD74", "B2M", "CALM3", "ARF1", "TGFB1", "TNF", "IFNG", "CXCL12", "VEGFA")
for (g in check_genes) {
  in_lig <- g %in% high_conf_ligands
  in_rec <- g %in% high_conf_receptors
  status <- if (in_lig && in_rec) {
    "PASS (ligand + receptor)"
  } else if (in_lig) {
    "PASS (ligand)"
  } else if (in_rec) {
    "PASS (receptor)"
  } else {
    "FILTERED OUT"
  }
  cat("  ", g, ": ", status, "\n", sep = "")
}

cat("\nExporting to Python: high_conf_ligands, high_conf_receptors, lr_network, ligand_target_matrix\n")
```

### Pass NicheNet priors to python

```python
# ### Convert NicheNet priors to Python
# Converts R objects from the Omnipath filtering cell into Python DataFrames and sets.

import numpy as np
import pandas as pd

required_exports = ["high_conf_ligands", "high_conf_receptors", "lr_network", "ligand_target_matrix"]
missing = [x for x in required_exports if x not in globals()]
if missing:
    raise NameError(f"Missing R exports: {missing}. Run the Omnipath filtering cell first.")

# Convert high_conf_ligands and high_conf_receptors to Python sets
if hasattr(high_conf_ligands, '__iter__') and not isinstance(high_conf_ligands, str):
    high_conf_ligands_set = set(str(x) for x in high_conf_ligands)
else:
    high_conf_ligands_set = {str(high_conf_ligands)}

if hasattr(high_conf_receptors, '__iter__') and not isinstance(high_conf_receptors, str):
    high_conf_receptors_set = set(str(x) for x in high_conf_receptors)
else:
    high_conf_receptors_set = {str(high_conf_receptors)}

# Convert lr_network to pandas DataFrame
if not isinstance(lr_network, pd.DataFrame):
    lr_network = pd.DataFrame(lr_network)

if "from" in lr_network.columns and "to" in lr_network.columns:
    lr_network["from"] = lr_network["from"].astype(str)
    lr_network["to"] = lr_network["to"].astype(str)
else:
    raise KeyError(f"lr_network missing 'from'/'to' columns. Found: {list(lr_network.columns)}")

# Convert ligand_target_matrix to pandas DataFrame
if not isinstance(ligand_target_matrix, pd.DataFrame):
    import rpy2.robjects as ro
    rownames_r = ro.r["rownames"]
    colnames_r = ro.r["colnames"]
    
    ltm_array = np.asarray(ligand_target_matrix)
    ltm_rownames = [str(x) for x in list(rownames_r(ro.globalenv["ligand_target_matrix"]))]
    ltm_colnames = [str(x) for x in list(colnames_r(ro.globalenv["ligand_target_matrix"]))]
    
    ligand_target_matrix = pd.DataFrame(
        ltm_array,
        index=pd.Index(ltm_rownames, name="ligand"),
        columns=pd.Index(ltm_colnames, name="target")
    )
else:
    ligand_target_matrix.index = ligand_target_matrix.index.astype(str)
    ligand_target_matrix.columns = ligand_target_matrix.columns.astype(str)

# Validate LTM orientation (rows=ligands, cols=targets)
canon_ligs = {"TGFB1", "TNF", "IFNG", "CXCL12", "VEGFA", "CCL2", "IL6", "PDGFB"}
canon_tgts = {"COL1A1", "COL3A1", "DCN", "LUM", "CXCL10", "ICAM1", "VCAM1", "STAT1"}

ligs_in_rows = len(canon_ligs & set(ligand_target_matrix.index))
ligs_in_cols = len(canon_ligs & set(ligand_target_matrix.columns))
tgts_in_rows = len(canon_tgts & set(ligand_target_matrix.index))
tgts_in_cols = len(canon_tgts & set(ligand_target_matrix.columns))

if (ligs_in_cols + tgts_in_rows) > (ligs_in_rows + tgts_in_cols):
    print("LTM appears transposed, fixing...")
    ligand_target_matrix = ligand_target_matrix.T

# Create derived sets
ltm_ligands = set(ligand_target_matrix.index.astype(str))
ltm_targets = set(ligand_target_matrix.columns.astype(str))
lr_ligands = set(lr_network["from"].unique())
lr_receptors = set(lr_network["to"].unique())

# Store in dict for organized access
nichenet_priors = {
    "lr_network": lr_network,
    "ligand_target_matrix": ligand_target_matrix,
    "high_conf_ligands": high_conf_ligands_set,
    "high_conf_receptors": high_conf_receptors_set,
    "ltm_ligands": ltm_ligands,
    "ltm_targets": ltm_targets,
    "lr_ligands": lr_ligands,
    "lr_receptors": lr_receptors,
}

print("NicheNet priors loaded into Python:")
print(f"  high_conf_ligands: {len(high_conf_ligands_set)}")
print(f"  high_conf_receptors: {len(high_conf_receptors_set)}")
print(f"  lr_network: {len(lr_network)} interactions")
print(f"  ligand_target_matrix: {ligand_target_matrix.shape[0]} x {ligand_target_matrix.shape[1]}")
```

# TIL+ zone NicheNet

### Define TIL+ senders and receivers

### GSEA of HIGH and LOW TIL+ programs

```python
import numpy as np
import pandas as pd
from pathlib import Path
import requests
import decoupler as dc

# Paths

cache_dir = Path.home() / ".cache" / "msigdb"
cache_dir.mkdir(parents=True, exist_ok=True)

EFS_ROOT = Path("/mnt/custom-file-systems/efs/fs-09913c1f7db79b6fd")
FILTER_DIR = EFS_ROOT / "results/visium/til_zones/nichenet_filtering"

high_targets_file = FILTER_DIR / "TILplus_targets_HIGH_visiumUP_receiverFiltered_LTMsupported.csv"
low_targets_file  = FILTER_DIR / "TILplus_targets_LOW_visiumDOWN_receiverFiltered_LTMsupported.csv"
visium_de_file    = EFS_ROOT / "results/visium/til_zones/deg_pseudobulk_edger/DE_edger_zone_TIL+_TIL_high_vs_TIL_low.csv"

for p in [high_targets_file, low_targets_file, visium_de_file]:
    if not p.exists():
        raise FileNotFoundError(f"Missing: {p}")

# MSigDB GMT download + parse

urls = {
    "hallmark": "https://data.broadinstitute.org/gsea-msigdb/msigdb/release/2023.2.Hs/h.all.v2023.2.Hs.symbols.gmt",
    "kegg":     "https://data.broadinstitute.org/gsea-msigdb/msigdb/release/2023.2.Hs/c2.cp.kegg_medicus.v2023.2.Hs.symbols.gmt",
    "biocarta": "https://data.broadinstitute.org/gsea-msigdb/msigdb/release/2023.2.Hs/c2.cp.biocarta.v2023.2.Hs.symbols.gmt",
}
paths = {k: cache_dir / Path(v).name for k, v in urls.items()}

def download_text(url: str, out_path: Path, timeout: int = 120) -> None:
    if out_path.exists() and out_path.stat().st_size > 0:
        return
    r = requests.get(url, timeout=timeout)
    r.raise_for_status()
    out_path.write_text(r.text)

def parse_gmt(gmt_path: Path) -> dict:
    gene_sets = {}
    with gmt_path.open("r") as f:
        for line in f:
            parts = line.rstrip("\n").split("\t")
            if len(parts) >= 3:
                gene_sets[parts[0]] = [g for g in parts[2:] if g]
    return gene_sets

for k, url in urls.items():
    download_text(url, paths[k])

hallmark_sets = parse_gmt(paths["hallmark"])
kegg_sets = parse_gmt(paths["kegg"])
biocarta_sets = parse_gmt(paths["biocarta"])

def gmt_to_net(gene_sets: dict) -> pd.DataFrame:
    rows = [(gs, g, 1.0) for gs, genes in gene_sets.items() for g in genes]
    return pd.DataFrame(rows, columns=["source", "target", "weight"])

net = pd.concat(
    [gmt_to_net(hallmark_sets), gmt_to_net(kegg_sets), gmt_to_net(biocarta_sets)],
    ignore_index=True
)

# Inputs: target gene lists + Visium DE

def load_gene_list(path: Path) -> set:
    df = pd.read_csv(path)
    if "gene" in df.columns:
        return set(df["gene"].astype(str))
    if df.shape[1] == 1:
        return set(df.iloc[:, 0].astype(str))
    raise ValueError(f"Could not find a gene column in {path.name}. Columns: {df.columns.tolist()}")

high_genes = load_gene_list(high_targets_file)
low_genes  = load_gene_list(low_targets_file)

visium_de = pd.read_csv(visium_de_file)
for c in ["gene", "logFC"]:
    if c not in visium_de.columns:
        raise ValueError(f"Missing column '{c}' in {visium_de_file.name}. Columns: {visium_de.columns.tolist()}")

visium_de["gene"] = visium_de["gene"].astype(str)
visium_de["logFC"] = pd.to_numeric(visium_de["logFC"], errors="coerce")

# Quick orientation sanity check:
# For TIL_high_vs_low, HIGH genes should skew positive, LOW genes should skew negative.
m_high = np.nanmedian(visium_de.loc[visium_de["gene"].isin(high_genes), "logFC"].to_numpy())
m_low  = np.nanmedian(visium_de.loc[visium_de["gene"].isin(low_genes),  "logFC"].to_numpy())
print(f"Median logFC among HIGH genes: {m_high:.4f}")
print(f"Median logFC among LOW genes : {m_low:.4f}")
if (np.isfinite(m_high) and m_high < 0) or (np.isfinite(m_low) and m_low > 0):
    print("WARNING: logFC sign looks inconsistent with expected TIL_high_vs_low orientation.")


# ULM scoring (flip LOW sign so 'TIL_low program' is positive)

def bh_fdr(pvals: np.ndarray) -> np.ndarray:
    p = np.asarray(pvals, dtype=float)
    q = np.full_like(p, np.nan, dtype=float)
    ok = np.isfinite(p)
    if ok.sum() == 0:
        return q
    p0 = p[ok]
    n = p0.size
    order = np.argsort(p0)
    ranked = p0[order]
    adj = ranked * n / (np.arange(1, n + 1))
    adj = np.minimum.accumulate(adj[::-1])[::-1]
    out = np.empty_like(p0)
    out[order] = np.clip(adj, 0, 1)
    q[ok] = out
    return q

def run_ulm(gene_set: set, sign: float, label: str) -> tuple[pd.DataFrame, int]:
    df = visium_de.loc[visium_de["gene"].isin(gene_set), ["gene", "logFC"]].dropna().copy()
    if df.empty:
        return pd.DataFrame(), 0

    df = df.drop_duplicates(subset=["gene"], keep="first")
    w = (sign * df.set_index("gene")["logFC"]).astype(float)

    mat = w.to_frame().T
    mat.index = [label]

    est, pvs = dc.run_ulm(
        mat=mat,
        net=net,
        source="source",
        target="target",
        weight="weight",
        min_n=5,
        verbose=False,
    )

    res = pd.DataFrame({
        "gene_set": est.columns.astype(str),
        "score": est.iloc[0].to_numpy(dtype=float),
        "pval":  pvs.iloc[0].to_numpy(dtype=float),
    })
    res["fdr"] = bh_fdr(res["pval"].to_numpy())
    res["collection"] = np.select(
        [
            res["gene_set"].str.startswith("HALLMARK_"),
            res["gene_set"].str.startswith("KEGG_"),
            res["gene_set"].str.startswith("BIOCARTA_"),
        ],
        ["Hallmark", "KEGG", "BioCarta"],
        default="Other",
    )
    return res, mat.shape[1]

ulm_high, n_high = run_ulm(high_genes, sign=+1.0, label="TIL_high_program")
ulm_low,  n_low  = run_ulm(low_genes,  sign=-1.0, label="TIL_low_program")  # SIGN FLIP HERE

out_high = FILTER_DIR / "TILplus_ULM_high_program_all_sets_signAligned.csv"
out_low  = FILTER_DIR / "TILplus_ULM_low_program_all_sets_signAligned.csv"

ulm_high.sort_values("score", ascending=False).to_csv(out_high, index=False)
ulm_low.sort_values("score", ascending=False).to_csv(out_low, index=False)

print(f"\nGene sets loaded: Hallmark={len(hallmark_sets)}, KEGG={len(kegg_sets)}, BioCarta={len(biocarta_sets)}")
print(f"Network rows: {len(net)} (sets={net['source'].nunique()}, genes={net['target'].nunique()})")
print(f"HIGH genes used: {n_high}")
print(f"LOW genes used:  {n_low}")
print(f"\nWrote:\n  {out_high}\n  {out_low}")

def show_top(res: pd.DataFrame, title: str, n: int = 25) -> None:
    if res.empty:
        print(f"\n{title}: no results")
        return
    top = res.sort_values("score", ascending=False).head(n).copy()
    top["score"] = top["score"].round(3)
    top["pval"] = top["pval"].map(lambda x: f"{x:.2e}" if np.isfinite(x) else "nan")
    top["fdr"] = top["fdr"].map(lambda x: f"{x:.2e}" if np.isfinite(x) else "nan")
    print(f"\n{title}: top {n} (score desc, sign-aligned)")
    print(top[["gene_set", "collection", "score", "pval", "fdr"]].to_string(index=False))

show_top(ulm_high, "HIGH program (TIL-high UP)")
show_top(ulm_low,  "LOW program (TIL-low UP)")
```

### Zone background gene universe

```python
import numpy as np
import pandas as pd
import scipy.sparse as sp

# Defines these as sorted Python lists of gene symbols for downstream %%R cells:
#   background_high, background_low
#   ligands_high, ligands_low
#   receptors_high, receptors_low
#
# Also keeps a structured dict for inspection:
#   universes_panel

LAYER = "LogNormalize"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
MIN_PCT_EXPRESSED = 0.10

required = ["adata_sc", "nichenet_priors", "SENDERS", "RECEIVERS"]
missing = [x for x in required if x not in globals()]
if missing:
    raise NameError(f"Missing required variables: {missing}")

if LAYER not in adata_sc.layers:
    raise KeyError(
        f"adata_sc.layers['{LAYER}'] not found. Available layers: {list(adata_sc.layers.keys())}"
    )
for k in [CT_KEY, GROUP_KEY]:
    if k not in adata_sc.obs.columns:
        raise KeyError(
            f"adata_sc.obs['{k}'] not found. Available obs columns (first 60): {list(adata_sc.obs.columns)[:60]}"
        )

for k in ["ltm_targets", "ltm_ligands", "lr_ligands", "lr_receptors"]:
    if k not in nichenet_priors:
        raise KeyError(f"nichenet_priors['{k}'] not found. Available keys: {list(nichenet_priors.keys())}")

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

senders_std = set(std_ct(x) for x in SENDERS)
receivers_std = set(std_ct(x) for x in RECEIVERS)

ct = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)

expected_groups = {"TIL_high", "TIL_low"}
groups_present = set(grp.unique())
if not expected_groups.issubset(groups_present):
    raise ValueError(f"Expected {expected_groups} in {GROUP_KEY}, found: {sorted(groups_present)}")

var_names = pd.Index(adata_sc.var_names.astype(str))
var_set = set(var_names)

X = adata_sc.layers[LAYER]
if sp.issparse(X):
    X = X.tocsr()
else:
    X = sp.csr_matrix(X)

ltm_targets = set(map(str, nichenet_priors["ltm_targets"]))
ltm_ligands = set(map(str, nichenet_priors["ltm_ligands"]))
lr_ligands = set(map(str, nichenet_priors["lr_ligands"]))
lr_receptors = set(map(str, nichenet_priors["lr_receptors"]))

candidate_background = ltm_targets
candidate_ligands = ltm_ligands.intersection(lr_ligands)
candidate_receptors = lr_receptors

def expressed_genes(mask_cells: np.ndarray, candidate_genes: set, min_pct: float) -> set:
    idx = np.flatnonzero(mask_cells)
    if idx.size == 0:
        return set()

    genes_present = sorted(candidate_genes.intersection(var_set))
    if not genes_present:
        return set()

    gene_idx = var_names.get_indexer(genes_present)
    Xsub = X[idx, :][:, gene_idx]

    nnz = np.asarray(Xsub.getnnz(axis=0)).ravel()
    pct = nnz / float(idx.size)

    keep = pct >= float(min_pct)
    return set(np.asarray(genes_present, dtype=object)[keep])

universes_panel = {}

for g in ["TIL_high", "TIL_low"]:
    sender_mask = (grp == g).to_numpy() & ct.isin(senders_std).to_numpy()
    receiver_mask = (grp == g).to_numpy() & ct.isin(receivers_std).to_numpy()

    bg = expressed_genes(receiver_mask, candidate_background, MIN_PCT_EXPRESSED)
    lig = expressed_genes(sender_mask, candidate_ligands, MIN_PCT_EXPRESSED)
    rec = expressed_genes(receiver_mask, candidate_receptors, MIN_PCT_EXPRESSED)

    universes_panel[g] = {
        "background": bg,
        "ligands": lig,
        "receptors": rec,
        "n_sender_cells": int(sender_mask.sum()),
        "n_receiver_cells": int(receiver_mask.sum()),
    }

background_high = sorted(map(str, universes_panel["TIL_high"]["background"]))
background_low = sorted(map(str, universes_panel["TIL_low"]["background"]))
ligands_high = sorted(map(str, universes_panel["TIL_high"]["ligands"]))
ligands_low = sorted(map(str, universes_panel["TIL_low"]["ligands"]))
receptors_high = sorted(map(str, universes_panel["TIL_high"]["receptors"]))
receptors_low = sorted(map(str, universes_panel["TIL_low"]["receptors"]))

summary_df = pd.DataFrame(
    [
        {
            "group": g,
            "n_sender_cells": universes_panel[g]["n_sender_cells"],
            "n_receiver_cells": universes_panel[g]["n_receiver_cells"],
            "n_background": len(universes_panel[g]["background"]),
            "n_ligands": len(universes_panel[g]["ligands"]),
            "n_receptors": len(universes_panel[g]["receptors"]),
        }
        for g in ["TIL_high", "TIL_low"]
    ]
)

print(summary_df.to_string(index=False))

print("\nUniverse integrity checks:")
print("  candidate_background (ltm_targets):", len(candidate_background))
print("  candidate_ligands (ltm_ligands ∩ lr_ligands):", len(candidate_ligands))
print("  candidate_receptors (lr_receptors):", len(candidate_receptors))
print("  var genes:", len(var_names))
print("  overlap candidate_ligands with var:", len(candidate_ligands.intersection(var_set)))
print("  overlap candidate_receptors with var:", len(candidate_receptors.intersection(var_set)))

if len(background_high) == 0 or len(background_low) == 0:
    raise ValueError("Background is empty for at least one group. Check receiver panel, group labels, and MIN_PCT_EXPRESSED.")
if len(ligands_high) == 0 or len(ligands_low) == 0:
    raise ValueError("Ligands are empty for at least one group. Check sender panel, group labels, and MIN_PCT_EXPRESSED.")
if len(receptors_high) == 0 or len(receptors_low) == 0:
    raise ValueError("Receptors are empty for at least one group. Check receiver panel, group labels, and MIN_PCT_EXPRESSED.")

print("\nExported lists ready for R:")
print("  background_high:", len(background_high), "background_low:", len(background_low))
print("  ligands_high   :", len(ligands_high), "ligands_low   :", len(ligands_low))
print("  receptors_high :", len(receptors_high), "receptors_low :", len(receptors_low))
```

### Target gene universe

```python
import numpy as np
import pandas as pd
import scipy.sparse as sp

# Builds canonical objects for this session:
#   targets_high, targets_low
#   universes (dict with targets/ligands/background/receptors per direction)
# Keeps de_df for downstream axis construction (Cell 5 expects de_df).

LAYER = "LogNormalize"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
ZONE_KEY = ZONE  # CHANGED: was "TIL+"

MIN_PCT_EXPRESSED = 0.10

required = [
    "adata_sc", "nichenet_priors", "SENDERS", "RECEIVERS",
    "de_tables", "PATHWAY_FILTER", "FDR_CUTOFF", "LOGFC_CUTOFF"
]
missing = [x for x in required if x not in globals()]
if missing:
    raise NameError(f"Missing required objects/params (run upstream cells): {missing}")

if LAYER not in adata_sc.layers:
    raise KeyError(f"adata_sc.layers['{LAYER}'] not found. Available: {list(adata_sc.layers.keys())}")

for k in [CT_KEY, GROUP_KEY]:
    if k not in adata_sc.obs.columns:
        raise KeyError(f"adata_sc.obs['{k}'] not found. Available (first 60): {list(adata_sc.obs.columns)[:60]}")

if ZONE_KEY not in de_tables:
    raise KeyError(f"de_tables['{ZONE_KEY}'] not found. Available keys: {list(de_tables.keys())}")

for k in ["ltm_targets", "ltm_ligands", "lr_ligands", "lr_receptors"]:
    if k not in nichenet_priors:
        raise KeyError(f"nichenet_priors['{k}'] not found. Available keys: {list(nichenet_priors.keys())}")

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

senders_std = set(std_ct(x) for x in SENDERS)
receivers_std = set(std_ct(x) for x in RECEIVERS)

ct = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)

expected_groups = {"TIL_high", "TIL_low"}
groups_present = set(grp.unique())
if not expected_groups.issubset(groups_present):
    raise ValueError(f"Expected {expected_groups} in {GROUP_KEY}, found: {sorted(groups_present)}")

var_names = pd.Index(adata_sc.var_names.astype(str))
var_set = set(var_names)

X = adata_sc.layers[LAYER]
if sp.issparse(X):
    X = X.tocsr()
else:
    X = sp.csr_matrix(X)

ltm_targets = set(map(str, nichenet_priors["ltm_targets"]))
ltm_ligands = set(map(str, nichenet_priors["ltm_ligands"]))
lr_ligands = set(map(str, nichenet_priors["lr_ligands"]))
lr_receptors = set(map(str, nichenet_priors["lr_receptors"]))

candidate_background = ltm_targets
candidate_ligands = ltm_ligands.intersection(lr_ligands)
candidate_receptors = lr_receptors

def expressed_genes(mask_cells: np.ndarray, candidate_genes: set, min_pct: float) -> set:
    idx = np.flatnonzero(mask_cells)
    if idx.size == 0:
        return set()

    genes_present = sorted(candidate_genes.intersection(var_set))
    if not genes_present:
        return set()

    gene_idx = var_names.get_indexer(genes_present)
    Xsub = X[idx, :][:, gene_idx]

    nnz = np.asarray(Xsub.getnnz(axis=0)).ravel()
    pct = nnz / float(idx.size)

    keep = pct >= float(min_pct)
    return set(np.asarray(genes_present, dtype=object)[keep])

panel = {}
for g in ["TIL_high", "TIL_low"]:
    sender_mask = (grp == g).to_numpy() & ct.isin(senders_std).to_numpy()
    receiver_mask = (grp == g).to_numpy() & ct.isin(receivers_std).to_numpy()

    panel[g] = {
        "n_sender_cells": int(sender_mask.sum()),
        "n_receiver_cells": int(receiver_mask.sum()),
        "background": expressed_genes(receiver_mask, candidate_background, MIN_PCT_EXPRESSED),
        "ligands": expressed_genes(sender_mask, candidate_ligands, MIN_PCT_EXPRESSED),
        "receptors": expressed_genes(receiver_mask, candidate_receptors, MIN_PCT_EXPRESSED),
    }

de_df = de_tables[ZONE_KEY].copy()

req = {"gene", "FDR", "logFC"}
if not req.issubset(set(de_df.columns)):
    raise KeyError(f"de_tables['{ZONE_KEY}'] must contain {sorted(req)}. Found: {list(de_df.columns)}")

de_df["gene"] = de_df["gene"].astype(str)
de_df["FDR"] = pd.to_numeric(de_df["FDR"], errors="coerce")
de_df["logFC"] = pd.to_numeric(de_df["logFC"], errors="coerce")

# DE-defined targets by direction (sign anchored to TIL_high_vs_low)
de_high = set(
    de_df.loc[
        (de_df["FDR"] <= float(FDR_CUTOFF)) & (de_df["logFC"] >= float(LOGFC_CUTOFF)),
        "gene"
    ].dropna()
)
de_low = set(
    de_df.loc[
        (de_df["FDR"] <= float(FDR_CUTOFF)) & (de_df["logFC"] <= -float(LOGFC_CUTOFF)),
        "gene"
    ].dropna()
)

# Gate to receiver-expressed background per group
targets_high = de_high.intersection(panel["TIL_high"]["background"])
targets_low = de_low.intersection(panel["TIL_low"]["background"])

targets_high_pre = set(targets_high)
targets_low_pre = set(targets_low)

# Optional pathway filter (apply after background gating)
if PATHWAY_FILTER is not None:
    pf = set(map(str, PATHWAY_FILTER))
    targets_high = targets_high.intersection(pf)
    targets_low = targets_low.intersection(pf)

universes = {
    "targets_high": sorted(map(str, targets_high)),
    "targets_low": sorted(map(str, targets_low)),
    "ligands_high": sorted(map(str, panel["TIL_high"]["ligands"])),
    "ligands_low": sorted(map(str, panel["TIL_low"]["ligands"])),
    "background_high": sorted(map(str, panel["TIL_high"]["background"])),
    "background_low": sorted(map(str, panel["TIL_low"]["background"])),
    "receptors_high": sorted(map(str, panel["TIL_high"]["receptors"])),
    "receptors_low": sorted(map(str, panel["TIL_low"]["receptors"])),
}

summary_df = pd.DataFrame(
    [
        {
            "group": g,
            "n_sender_cells": panel[g]["n_sender_cells"],
            "n_receiver_cells": panel[g]["n_receiver_cells"],
            "n_background": len(panel[g]["background"]),
            "n_ligands": len(panel[g]["ligands"]),
            "n_receptors": len(panel[g]["receptors"]),
        }
        for g in ["TIL_high", "TIL_low"]
    ]
)

print(summary_df.to_string(index=False))

print("\nTargets:")
print("  targets_high (DE∩background) pre pathway:", len(targets_high_pre), "post:", len(targets_high))
print("  targets_low  (DE∩background) pre pathway:", len(targets_low_pre),  "post:", len(targets_low))
print("  pathway filter enabled:", PATHWAY_FILTER is not None)

if PATHWAY_FILTER is not None:
    pf_set = set(map(str, PATHWAY_FILTER))
    print("  targets_high removed by PATHWAY_FILTER:", len(targets_high_pre - pf_set))
    print("  targets_low  removed by PATHWAY_FILTER:", len(targets_low_pre - pf_set))

if len(universes["targets_high"]) == 0 or len(universes["targets_low"]) == 0:
    raise ValueError(
        "Targets empty after DE/background gating and (optional) pathway filter. "
        "Check thresholds, DE orientation, and PATHWAY_FILTER."
    )
```

### Map targets to receiver cell types by expression fraction

```python
# %% [markdown]
# ### Map targets to receiver cell types by expression fraction

# %%
import numpy as np
import pandas as pd
import scipy.sparse as sp

# === CLEAR STALE VARIABLES ===
for _v in ["target_receiver_map_high", "target_receiver_map_low", "target_panel_coverage",
           "global_best_high", "global_best_low", "all_celltypes", "valid_celltypes", 
           "excluded_celltypes", "ct", "grp", "X", "var_names", "var_set"]:
    if _v in globals():
        del globals()[_v]

required = ["adata_sc", "targets_high", "targets_low", "RECEIVERS", "FRACTION_CUTOFF"]
missing = [x for x in required if x not in globals()]
if missing:
    raise NameError(f"Missing required objects/params: {missing}")

# === VERIFY INPUTS ===
print(f"INPUT CHECK:")
print(f"  targets_high: {len(targets_high)} genes")
print(f"  targets_low: {len(targets_low)} genes")
print(f"  RECEIVERS: {len(RECEIVERS)} cell types")
print(f"  FRACTION_CUTOFF: {FRACTION_CUTOFF}")

LAYER = "LogNormalize"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
SAMPLE_KEY = "sample_id"

MIN_SAMPLES_EXPRESSING = 3
PANEL_COVERAGE_THRESHOLD = 0.50

# Minimum cells required for a cell type to be considered in global best computation
# This prevents rare cell types (e.g., 1 cell) from dominating due to noise
MIN_CELLS_FOR_GLOBAL_BEST = 50

if LAYER not in adata_sc.layers:
    raise KeyError(f"adata_sc.layers['{LAYER}'] not found.")
for k in [CT_KEY, GROUP_KEY, SAMPLE_KEY]:
    if k not in adata_sc.obs.columns:
        raise KeyError(f"adata_sc.obs['{k}'] not found.")

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

def is_garbage_ct(ct_name: str) -> bool:
    """Exclude Doublet, Other, Unknown, Unassigned cell types."""
    s = str(ct_name)
    for bad in ["Doublet", "doublet", "Other", "other", "Unknown", "Unassigned"]:
        if bad in s:
            return True
    return False

receivers_std = sorted(set(std_ct(x) for x in RECEIVERS))

# Build fresh from adata_sc
ct = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)
sample_ids = adata_sc.obs[SAMPLE_KEY].astype(str)

X = adata_sc.layers[LAYER]
if sp.issparse(X):
    X = X.tocsr()
else:
    X = sp.csr_matrix(X)

var_names = pd.Index(adata_sc.var_names.astype(str))
var_set = set(var_names)

# Get valid cell types (excluding garbage)
all_celltypes_raw = sorted(ct.unique())
valid_celltypes = [c for c in all_celltypes_raw if not is_garbage_ct(c)]
excluded_celltypes = [c for c in all_celltypes_raw if is_garbage_ct(c)]

print(f"\nCELL TYPE FILTERING:")
print(f"  Raw cell types: {len(all_celltypes_raw)}")
print(f"  Valid cell types (no Doublet/Other): {len(valid_celltypes)}")
print(f"  Excluded: {len(excluded_celltypes)}")

# Compute cell counts per cell type per group
celltype_counts = {}
for group_label in ["TIL_high", "TIL_low"]:
    celltype_counts[group_label] = {}
    for celltype in valid_celltypes:
        m = (grp == group_label).to_numpy() & (ct == celltype).to_numpy()
        n_cells = m.sum()
        celltype_counts[group_label][celltype] = n_cells

# Filter valid cell types for global best by minimum cell count
valid_celltypes_for_global = {}
for group_label in ["TIL_high", "TIL_low"]:
    valid_celltypes_for_global[group_label] = [
        c for c in valid_celltypes 
        if celltype_counts[group_label][c] >= MIN_CELLS_FOR_GLOBAL_BEST
    ]
    n_excluded_by_count = len(valid_celltypes) - len(valid_celltypes_for_global[group_label])
    print(f"  {group_label}: {len(valid_celltypes_for_global[group_label])} cell types with >= {MIN_CELLS_FOR_GLOBAL_BEST} cells ({n_excluded_by_count} excluded)")

def build_trm_for_group(group_label: str, targets_panel) -> pd.DataFrame:
    """Build target-receiver map for panel receivers only."""
    targets_panel = [str(x) for x in list(targets_panel)]
    targets_present = [t for t in targets_panel if t in var_set]
    if len(targets_present) == 0:
        raise ValueError(f"No targets found in var_names for {group_label}.")

    print(f"  {group_label}: {len(targets_present)} targets in var_names")
    
    gene_idx = var_names.get_indexer(targets_present)
    rows = []

    for rct in receivers_std:
        m = (grp == group_label).to_numpy() & (ct == rct).to_numpy()
        idx = np.flatnonzero(m)
        if idx.size == 0:
            continue

        sample_sub = sample_ids.iloc[idx].values
        n_samples = len(np.unique(sample_sub))

        Xsub = X[idx, :][:, gene_idx]
        nnz = np.asarray(Xsub.getnnz(axis=0)).ravel()
        pct = nnz / float(idx.size)
        mean_expr = np.asarray(Xsub.mean(axis=0)).ravel()
        Xsub_dense = Xsub.toarray()

        for i, target in enumerate(targets_present):
            gene_expr = Xsub_dense[:, i]
            n_samples_expressing = len(set(sample_sub[gene_expr > 0]))
            sample_frac = n_samples_expressing / n_samples if n_samples > 0 else 0.0

            if pct[i] >= float(FRACTION_CUTOFF) and n_samples_expressing >= MIN_SAMPLES_EXPRESSING:
                rows.append({
                    "target": target,
                    "receiver_celltype": rct,
                    "target_pct_expr": float(pct[i]),
                    "target_mean_expr": float(mean_expr[i]),
                    "n_samples": n_samples,
                    "n_samples_expressing": n_samples_expressing,
                    "sample_frac": float(sample_frac),
                })

    if not rows:
        return pd.DataFrame(columns=[
            "target", "receiver_celltype", "target_pct_expr", "target_mean_expr",
            "n_samples", "n_samples_expressing", "sample_frac", "specificity", "receiver_rank_score"
        ])

    out = pd.DataFrame(rows)

    # Specificity within panel
    target_mean_all = out.groupby("target")["target_mean_expr"].transform("mean").replace(0, np.nan)
    out["specificity"] = (out["target_mean_expr"] / target_mean_all).fillna(0)

    # Receiver rank score
    out["_mr"] = out.groupby("target")["target_mean_expr"].rank(ascending=False)
    out["_pr"] = out.groupby("target")["target_pct_expr"].rank(ascending=False)
    out["_sfr"] = out.groupby("target")["sample_frac"].rank(ascending=False)
    out["_spr"] = out.groupby("target")["specificity"].rank(ascending=False)
    out["receiver_rank_score"] = 0.35/out["_mr"] + 0.25/out["_pr"] + 0.20/out["_sfr"] + 0.20/out["_spr"]
    out = out.drop(columns=["_mr", "_pr", "_sfr", "_spr"])

    return out

def compute_global_best(group_label: str, targets_list) -> pd.DataFrame:
    """Find global best cell type per target.
    
    Only considers cell types with >= MIN_CELLS_FOR_GLOBAL_BEST cells
    to avoid noise from rare populations.
    """
    targets_list = [str(x) for x in targets_list if str(x) in var_set]
    if not targets_list:
        return pd.DataFrame(columns=["target", "global_best_celltype", "global_best_mean_expr", "global_best_n_cells"])

    gene_idx = var_names.get_indexer(targets_list)
    best = {t: (None, 0.0, 0) for t in targets_list}  # (celltype, mean_expr, n_cells)

    # Only use cell types with sufficient cells
    celltypes_to_check = valid_celltypes_for_global[group_label]
    
    for celltype in celltypes_to_check:
        m = (grp == group_label).to_numpy() & (ct == celltype).to_numpy()
        idx = np.flatnonzero(m)
        if idx.size == 0:
            continue
        n_cells = idx.size
        Xsub = X[idx, :][:, gene_idx]
        mean_expr = np.asarray(Xsub.mean(axis=0)).ravel()
        for i, target in enumerate(targets_list):
            if mean_expr[i] > best[target][1]:
                best[target] = (celltype, float(mean_expr[i]), n_cells)

    results = [{"target": t, "global_best_celltype": v[0], "global_best_mean_expr": v[1], "global_best_n_cells": v[2]} 
               for t, v in best.items() if v[0] is not None]
    return pd.DataFrame(results)

def add_coverage(trm: pd.DataFrame, global_best: pd.DataFrame) -> pd.DataFrame:
    if trm.empty:
        return trm
    trm = trm.merge(global_best, on="target", how="left")
    panel_best = trm.groupby("target")["target_mean_expr"].transform("max")
    trm["panel_coverage_ratio"] = np.where(
        trm["global_best_mean_expr"] > 0,
        panel_best / trm["global_best_mean_expr"],
        1.0
    )
    trm["panel_is_adequate"] = trm["panel_coverage_ratio"] >= PANEL_COVERAGE_THRESHOLD
    trm["coverage_flag"] = np.where(trm["panel_is_adequate"], "OK", "LOW_COVERAGE")
    return trm

# === BUILD TRMs ===
print("\nBUILDING TARGET-RECEIVER MAPS:")
target_receiver_map_high = build_trm_for_group("TIL_high", targets_high)
target_receiver_map_low = build_trm_for_group("TIL_low", targets_low)

# === COMPUTE GLOBAL BEST ===
print(f"\nCOMPUTING GLOBAL BEST (min {MIN_CELLS_FOR_GLOBAL_BEST} cells, excluding Doublets/Other):")
global_best_high = compute_global_best("TIL_high", targets_high)
global_best_low = compute_global_best("TIL_low", targets_low)

# Verify no garbage in global_best
for name, gb in [("high", global_best_high), ("low", global_best_low)]:
    if not gb.empty:
        garbage = [ct for ct in gb["global_best_celltype"].dropna().unique() if is_garbage_ct(ct)]
        if garbage:
            raise RuntimeError(f"BUG: Garbage cell types in global_best_{name}: {garbage}")
        print(f"  global_best_{name}: {len(gb)} targets, no garbage cell types ✓")

# === ADD COVERAGE ===
target_receiver_map_high = add_coverage(target_receiver_map_high, global_best_high)
target_receiver_map_low = add_coverage(target_receiver_map_low, global_best_low)

# === SUMMARY ===
print("\nTARGET-RECEIVER MAP SUMMARY:")
for name, trm in [("TIL_high", target_receiver_map_high), ("TIL_low", target_receiver_map_low)]:
    print(f"\n  {name}:")
    print(f"    rows: {len(trm)}")
    print(f"    targets: {trm['target'].nunique()}")
    print(f"    receivers: {trm['receiver_celltype'].nunique()}")
    if "coverage_flag" in trm.columns:
        n_ok = (trm["coverage_flag"] == "OK").sum()
        n_low = (trm["coverage_flag"] == "LOW_COVERAGE").sum()
        print(f"    coverage: OK={n_ok}, LOW_COVERAGE={n_low}")

# Show LOW_COVERAGE targets
print("\nLOW_COVERAGE TARGETS:")
for name, trm in [("TIL_high", target_receiver_map_high), ("TIL_low", target_receiver_map_low)]:
    low = trm[trm["coverage_flag"] == "LOW_COVERAGE"][["target", "global_best_celltype", "global_best_n_cells", "panel_coverage_ratio"]].drop_duplicates()
    if low.empty:
        print(f"  {name}: None")
    else:
        print(f"  {name}:")
        for _, r in low.iterrows():
            print(f"    {r['target']}: global_best={r['global_best_celltype']} (n={r['global_best_n_cells']}), coverage={r['panel_coverage_ratio']:.1%}")

if target_receiver_map_high.empty or target_receiver_map_low.empty:
    raise ValueError("One or both target_receiver_maps are empty!")

# Store for downstream
target_panel_coverage = pd.concat([
    global_best_high.assign(direction="TIL_high"),
    global_best_low.assign(direction="TIL_low")
], ignore_index=True)

print("\n✓ Created: target_receiver_map_high, target_receiver_map_low, target_panel_coverage")
```

### Build SCE (python)

```python
from pathlib import Path

import numpy as np
import pandas as pd
import scipy.sparse as sp
from scipy.io import mmwrite

if "adata_sc" not in globals():
    raise NameError("adata_sc not found")
if "SENDERS" not in globals() or "RECEIVERS" not in globals():
    raise NameError("SENDERS and/or RECEIVERS not found")

COUNTS_LAYER = "unnormalized_counts"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
SAMPLE_KEY = "sample_id"
BATCH_KEY = "orig.ident"

if COUNTS_LAYER not in adata_sc.layers:
    raise KeyError(f"adata_sc.layers['{COUNTS_LAYER}'] not found. Available: {list(adata_sc.layers.keys())}")

for k in [CT_KEY, GROUP_KEY, SAMPLE_KEY, BATCH_KEY]:
    if k not in adata_sc.obs.columns:
        raise KeyError(f"adata_sc.obs['{k}'] not found. Available obs cols (first 50): {list(adata_sc.obs.columns)[:50]}")

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

def drop_bad_ct(ct_list):
    out = []
    for ct in ct_list:
        s = str(ct)
        if s.startswith("Doublet"):
            continue
        if s.startswith("Other"):
            continue
        out.append(s)
    return sorted(set(out))

senders_py = drop_bad_ct(list(SENDERS))
receivers_py = drop_bad_ct(list(RECEIVERS))

senders_std = [std_ct(x) for x in senders_py]
receivers_std = [std_ct(x) for x in receivers_py]
panel_ct_std = sorted(set(senders_std) | set(receivers_std))

ct_std = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)

m = grp.isin(["TIL_low", "TIL_high"]) & ct_std.isin(panel_ct_std)
n_keep = int(m.sum())
if n_keep == 0:
    raise ValueError("No cells left after filtering to (TIL_low/TIL_high) and sender/receiver panel celltypes.")

adata_sub = adata_sc[m].copy()

X = adata_sub.layers[COUNTS_LAYER]
if not sp.issparse(X):
    X = sp.csr_matrix(X)
else:
    X = X.tocsr()

EFS_ROOT = Path("/mnt/custom-file-systems/efs/fs-09913c1f7db79b6fd")
zone_dir = ZONE.replace("+", "plus").replace("-", "minus")  # CHANGED: derive from ZONE
out_dir = EFS_ROOT / "results" / "visium" / "til_zones" / "nichenet_filtering" / f"mnn_{zone_dir}_input"  # CHANGED: use zone_dir
out_dir.mkdir(parents=True, exist_ok=True)

for fname in ["counts.mtx", "features.tsv", "barcodes.tsv", "metadata.csv", "senders.txt", "receivers.txt"]:
    fpath = out_dir / fname
    if fpath.exists():
        fpath.unlink()

counts_gene_cell = X.T.tocsc()
mmwrite(str(out_dir / "counts.mtx"), counts_gene_cell)

genes = pd.Series(adata_sub.var_names.astype(str))
genes.to_csv(out_dir / "features.tsv", sep="\t", index=False, header=False)

cells = pd.Series(adata_sub.obs_names.astype(str))
cells.to_csv(out_dir / "barcodes.tsv", sep="\t", index=False, header=False)

ct_sub_std = adata_sub.obs[CT_KEY].astype(str).map(std_ct).values

meta = pd.DataFrame(
    {
        "cell": cells.values,
        "celltype": ct_sub_std,
        "sample": adata_sub.obs[SAMPLE_KEY].astype(str).values,
        "group": adata_sub.obs[GROUP_KEY].astype(str).values,
        "batch": adata_sub.obs[BATCH_KEY].astype(str).values,
    }
)
meta.to_csv(out_dir / "metadata.csv", index=False)

pd.Series(senders_std).to_csv(out_dir / "senders.txt", index=False, header=False)
pd.Series(receivers_std).to_csv(out_dir / "receivers.txt", index=False, header=False)

# ADDED: Export for R cell
SCE_INPUT_DIR = f"mnn_{zone_dir}_input"

print("Exported for SCE:")
print("  out_dir:", str(out_dir))
print("  genes x cells:", counts_gene_cell.shape[0], "x", counts_gene_cell.shape[1])
print("  nnz:", int(counts_gene_cell.nnz))
print("  groups:", dict(pd.Series(meta["group"]).value_counts()))
print("  celltypes in panel (senders, receivers):", len(senders_std), ",", len(receivers_std))
```

### Build SCE (R)

### Compute ligand activity per direction

```python
# Make R-safe copies (rpy2 cannot convert Python sets)
targets_high_r = sorted(list(targets_high))
targets_low_r  = sorted(list(targets_low))

background_high_r = list(background_high)
background_low_r  = list(background_low)
ligands_high_r    = list(ligands_high)
ligands_low_r     = list(ligands_low)

print("R-ready objects:")
print("  targets_high_r:", type(targets_high_r), len(targets_high_r))
print("  targets_low_r :", type(targets_low_r),  len(targets_low_r))
print("  background_high_r:", type(background_high_r), len(background_high_r))
print("  background_low_r :", type(background_low_r),  len(background_low_r))
print("  ligands_high_r:", type(ligands_high_r), len(ligands_high_r))
print("  ligands_low_r :", type(ligands_low_r),  len(ligands_low_r))
```

```python
%%R -i targets_high_r -i targets_low_r -i background_high_r -i background_low_r -i ligands_high_r -i ligands_low_r
suppressPackageStartupMessages({
  library(dplyr)
  library(tibble)
  library(nichenetr)
})

stopifnot(exists("ligand_target_matrix"))

to_chr_unique <- function(x) {
  unique(as.character(unlist(x, use.names = FALSE)))
}

targets_high_chr    <- to_chr_unique(targets_high_r)
targets_low_chr     <- to_chr_unique(targets_low_r)
background_high_chr <- to_chr_unique(background_high_r)
background_low_chr  <- to_chr_unique(background_low_r)
ligands_high_chr    <- to_chr_unique(ligands_high_r)
ligands_low_chr     <- to_chr_unique(ligands_low_r)

ltm <- ligand_target_matrix
ltm_ligands <- rownames(ltm)
ltm_targets <- colnames(ltm)

targets_high_use <- intersect(targets_high_chr, ltm_targets)
targets_low_use  <- intersect(targets_low_chr,  ltm_targets)

bg_high_use <- intersect(background_high_chr, ltm_targets)
bg_low_use  <- intersect(background_low_chr,  ltm_targets)

lig_high_use <- intersect(ligands_high_chr, ltm_ligands)
lig_low_use  <- intersect(ligands_low_chr,  ltm_ligands)

get_top_ltm_ligands <- function(targets_use, ltm, k_per_target) {
  out <- character()
  if (length(targets_use) == 0) return(out)

  for (t in targets_use) {
    w <- ltm[, t, drop = TRUE]
    w <- w[w > 0]
    if (length(w) == 0) next
    top <- sort(w, decreasing = TRUE)
    if (length(top) > k_per_target) top <- top[seq_len(k_per_target)]
    out <- c(out, names(top))
  }
  unique(out)
}

LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY <- 10

top_ltm_high <- get_top_ltm_ligands(targets_high_use, ltm, LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY)
top_ltm_low  <- get_top_ltm_ligands(targets_low_use,  ltm, LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY)

potential_high <- union(lig_high_use, top_ltm_high)
potential_low  <- union(lig_low_use,  top_ltm_low)

cat("Ligand activity inputs after LTM intersection + LTM augmentation:\n")
cat("  LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY=", LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY, "\n", sep = "")
cat("  TIL_high: targets=", length(targets_high_use),
    " background=", length(bg_high_use),
    " sender_panel_ligs=", length(lig_high_use),
    " ltm_top_ligs=", length(top_ltm_high),
    " potential_ligs=", length(potential_high), "\n", sep = "")
cat("  TIL_low : targets=", length(targets_low_use),
    " background=", length(bg_low_use),
    " sender_panel_ligs=", length(lig_low_use),
    " ltm_top_ligs=", length(top_ltm_low),
    " potential_ligs=", length(potential_low), "\n\n", sep = "")

if (length(targets_high_use) < 5) stop("Too few TIL_high targets after intersecting with ligand_target_matrix columns.")
if (length(targets_low_use)  < 5) stop("Too few TIL_low targets after intersecting with ligand_target_matrix columns.")
if (length(bg_high_use) < 50) stop("Too few TIL_high background genes after intersecting with ligand_target_matrix columns.")
if (length(bg_low_use)  < 50) stop("Too few TIL_low background genes after intersecting with ligand_target_matrix columns.")
if (length(potential_high) < 20) stop("Too few TIL_high potential ligands after augmentation.")
if (length(potential_low)  < 20) stop("Too few TIL_low potential ligands after augmentation.")

ligand_activities_high <- predict_ligand_activities(
  geneset = targets_high_use,
  background_expressed_genes = bg_high_use,
  potential_ligands = potential_high,
  ligand_target_matrix = ltm
) %>%
  as_tibble() %>%
  mutate(
    direction = "TIL_high",
    test_ligand = as.character(test_ligand),
    aupr_corrected = as.numeric(aupr_corrected),
    in_sender_panel = test_ligand %in% lig_high_use,
    in_ltm_top = test_ligand %in% top_ltm_high
  ) %>%
  arrange(desc(aupr_corrected), test_ligand) %>%
  mutate(rank = dplyr::row_number()) %>%
  select(test_ligand, direction, aupr_corrected, rank, in_sender_panel, in_ltm_top, everything())

ligand_activities_low <- predict_ligand_activities(
  geneset = targets_low_use,
  background_expressed_genes = bg_low_use,
  potential_ligands = potential_low,
  ligand_target_matrix = ltm
) %>%
  as_tibble() %>%
  mutate(
    direction = "TIL_low",
    test_ligand = as.character(test_ligand),
    aupr_corrected = as.numeric(aupr_corrected),
    in_sender_panel = test_ligand %in% lig_low_use,
    in_ltm_top = test_ligand %in% top_ltm_low
  ) %>%
  arrange(desc(aupr_corrected), test_ligand) %>%
  mutate(rank = dplyr::row_number()) %>%
  select(test_ligand, direction, aupr_corrected, rank, in_sender_panel, in_ltm_top, everything())

ligand_activities <- bind_rows(ligand_activities_high, ligand_activities_low)

cat("Ligand activities computed:\n")
cat("  rows:", nrow(ligand_activities), "\n")
cat("  TIL_high:", nrow(ligand_activities_high), "\n")
cat("  TIL_low :", nrow(ligand_activities_low), "\n\n")

cat("Top 10 (TIL_high):\n")
print(ligand_activities_high %>% slice_head(n = 10) %>% select(test_ligand, aupr_corrected, rank, in_sender_panel, in_ltm_top))
cat("\nTop 10 (TIL_low):\n")
print(ligand_activities_low %>% slice_head(n = 10) %>% select(test_ligand, aupr_corrected, rank, in_sender_panel, in_ltm_top))
```

### Ligand-receptor DE per cell type

```python
# Pass zone prefix to R for file naming
zone_prefix = ZONE.replace("+", "plus").replace("-", "minus")
print(f"Zone prefix for R: {zone_prefix}")
```

```python
%%R -i zone_prefix
# LR-DE: edgeR pseudobulk per celltype, TIL_high vs TIL_low
suppressPackageStartupMessages({
  library(Matrix)
  library(SingleCellExperiment)
  library(SummarizedExperiment)
  library(S4Vectors)
  library(dplyr)
  library(tibble)
  library(readr)
  library(edgeR)
})

stopifnot(exists("sce"), exists("senders_oi"), exists("receivers_oi"))
stopifnot(exists("lr_network"), exists("ligand_target_matrix"))
stopifnot(exists("ligand_activities"))
stopifnot("counts" %in% SummarizedExperiment::assayNames(sce))

EFS_ROOT  <- "/mnt/custom-file-systems/efs/fs-09913c1f7db79b6fd"
FILTER_DIR <- file.path(EFS_ROOT, "results", "visium", "til_zones", "nichenet_filtering")
dir.create(FILTER_DIR, recursive = TRUE, showWarnings = FALSE)

GROUPS <- c("TIL_low","TIL_high")

MIN_CELLS_PER_SAMPLE  <- 5
MIN_SAMPLES_PER_GROUP <- 3

CT_COL     <- "celltype"
GROUP_COL  <- "group"
SAMPLE_COL <- "sample_id"

cd0 <- as.data.frame(colData(sce))
stopifnot(CT_COL %in% colnames(cd0), GROUP_COL %in% colnames(cd0), SAMPLE_COL %in% colnames(cd0))

cd0[[CT_COL]]     <- as.character(cd0[[CT_COL]])
cd0[[GROUP_COL]]  <- as.character(cd0[[GROUP_COL]])
cd0[[SAMPLE_COL]] <- as.character(cd0[[SAMPLE_COL]])

bad_sample <- is.na(cd0[[SAMPLE_COL]]) | cd0[[SAMPLE_COL]] == ""
if (any(bad_sample)) {
  ex <- unique(cd0[[SAMPLE_COL]][bad_sample])[1:min(10, sum(bad_sample))]
  stop(paste0("sample_id contains NA/empty values (n=", sum(bad_sample), "). Examples: ", paste(ex, collapse=", ")))
}

bad_group <- is.na(cd0[[GROUP_COL]]) | !(cd0[[GROUP_COL]] %in% GROUPS)
if (any(bad_group)) {
  ex <- unique(cd0[[GROUP_COL]][bad_group])[1:min(10, sum(bad_group))]
  stop(paste0("group has NA or unexpected labels (n=", sum(bad_group), "). Examples: ", paste(ex, collapse=", ")))
}

cat("Using columns:\n")
cat("  celltype:", CT_COL, "\n")
cat("  group   :", GROUP_COL, "\n")
cat("  sample  :", SAMPLE_COL, "\n")
cat("Cells:", nrow(cd0), "\n\n")

cat("Group counts in sce:\n")
print(table(cd0[[GROUP_COL]], useNA = "ifany"))
cat("\n")

assays_avail <- assayNames(sce)
if (!("counts" %in% assays_avail)) {
  stop(paste0("sce is missing assay 'counts'. Available assays: ", paste(assays_avail, collapse=", ")))
}
counts_all <- assay(sce, "counts")
if (!is(counts_all, "dgCMatrix")) counts_all <- as(counts_all, "CsparseMatrix")
counts_all <- as(counts_all, "dgCMatrix")

cat("Counts matrix:\n")
cat("  class:", class(counts_all)[1], "\n")
cat("  dim  :", nrow(counts_all), "x", ncol(counts_all), "\n\n")

cat("Ligand activities directions:\n")
print(table(as.character(ligand_activities$direction), useNA="ifany"))
cat("\n")

# Harmonize lr_network columns
lr_cols <- colnames(lr_network)
lig_col <- if ("from" %in% lr_cols) "from" else if ("ligand" %in% lr_cols) "ligand" else stop("lr_network missing ligand column (from/ligand).")
rec_col <- if ("to" %in% lr_cols) "to" else if ("receptor" %in% lr_cols) "receptor" else stop("lr_network missing receptor column (to/receptor).")

lr_tbl <- lr_network %>%
  dplyr::distinct(.data[[lig_col]], .data[[rec_col]])
colnames(lr_tbl) <- c("from","to")

# Feasible ligands: use NicheNet activity list, constrained to LTM rows + sce genes
ltm <- ligand_target_matrix
ltm_ligands <- rownames(ltm)
ltm_targets <- colnames(ltm)

feasible_ligands <- intersect(
  unique(as.character(ligand_activities$test_ligand)),
  intersect(ltm_ligands, rownames(counts_all))
)

# Feasible receptors: receptors paired to feasible ligands, constrained to sce genes
feasible_receptors <- intersect(
  unique(lr_tbl$to[lr_tbl$from %in% feasible_ligands]),
  rownames(counts_all)
)

cat("DE universes:\n")
cat("  feasible ligands   :", length(feasible_ligands), "\n")
cat("  feasible receptors :", length(feasible_receptors), "\n\n")

if (length(feasible_ligands) < 20) stop("Too few feasible ligands after intersecting activity ligands with LTM rows and sce genes.")
if (length(feasible_receptors) < 20) stop("Too few feasible receptors after intersecting LR receptors with sce genes.")

# Fast pseudobulk using sparse model matrix
pseudobulk_by_sample <- function(counts_gene_cell, sample_vec) {
  sample_f <- factor(sample_vec)
  M <- Matrix::sparse.model.matrix(~0 + sample_f)
  colnames(M) <- levels(sample_f)
  pb <- counts_gene_cell %*% M
  as(pb, "dgCMatrix")
}

run_edger_for_celltype <- function(celltype_name, genes_keep, mode_label) {
  tryCatch({

    idx_ct <- which(cd0[[CT_COL]] == celltype_name & cd0[[GROUP_COL]] %in% GROUPS)
    if (length(idx_ct) == 0) {
      return(list(status="skip", reason="no cells after filtering", de=NULL, meta=NULL))
    }

    meta_ct <- cd0[idx_ct, , drop=FALSE]

    cells_per_sample <- meta_ct %>%
      dplyr::count(.data[[SAMPLE_COL]], name="n_cells")

    keep_samples <- cells_per_sample %>%
      dplyr::filter(n_cells >= MIN_CELLS_PER_SAMPLE) %>%
      dplyr::pull(.data[[SAMPLE_COL]])

    if (length(keep_samples) == 0) {
      return(list(status="skip", reason=paste0("no samples with >= ", MIN_CELLS_PER_SAMPLE, " cells"), de=NULL, meta=cells_per_sample))
    }

    keep_mask <- meta_ct[[SAMPLE_COL]] %in% keep_samples
    meta_ct <- meta_ct[keep_mask, , drop=FALSE]
    idx_ct  <- idx_ct[keep_mask]

    sample_group_tbl <- meta_ct %>%
      dplyr::group_by(.data[[SAMPLE_COL]]) %>%
      dplyr::summarise(
        n_cells = dplyr::n(),
        n_groups = dplyr::n_distinct(.data[[GROUP_COL]]),
        group = dplyr::first(.data[[GROUP_COL]]),
        .groups="drop"
      )

    bad <- sample_group_tbl %>% dplyr::filter(n_groups > 1)
    if (nrow(bad) > 0) {
      return(list(status="skip", reason="sample maps to >1 group within this celltype", de=NULL, meta=sample_group_tbl))
    }

    n_by_group <- sample_group_tbl %>% dplyr::count(group, name="n_samples")
    n_low  <- n_by_group %>% dplyr::filter(group=="TIL_low")  %>% dplyr::pull(n_samples)
    n_high <- n_by_group %>% dplyr::filter(group=="TIL_high") %>% dplyr::pull(n_samples)
    n_low  <- ifelse(length(n_low)==0, 0, n_low)
    n_high <- ifelse(length(n_high)==0, 0, n_high)

    if (n_low < MIN_SAMPLES_PER_GROUP || n_high < MIN_SAMPLES_PER_GROUP) {
      return(list(
        status="skip",
        reason=paste0("too few samples per group (low=", n_low, ", high=", n_high, ")"),
        de=NULL,
        meta=sample_group_tbl
      ))
    }

    genes_keep <- intersect(as.character(genes_keep), rownames(counts_all))
    if (length(genes_keep) < 10) {
      return(list(status="skip", reason="too few genes in universe after intersecting with sce genes", de=NULL, meta=sample_group_tbl))
    }

    counts_sub <- counts_all[genes_keep, idx_ct, drop=FALSE]
    pb <- pseudobulk_by_sample(counts_sub, meta_ct[[SAMPLE_COL]])

    group_vec <- sample_group_tbl$group
    names(group_vec) <- sample_group_tbl[[SAMPLE_COL]]
    group_vec <- group_vec[colnames(pb)]

    if (any(is.na(group_vec))) {
      na_samples <- colnames(pb)[is.na(group_vec)]
      return(list(
        status="error",
        reason=paste0("group_vec has NA after aligning to pseudobulk columns. Examples: ", paste(head(na_samples, 10), collapse=", ")),
        de=NULL,
        meta=sample_group_tbl
      ))
    }

    group_vec <- factor(group_vec, levels=GROUPS)

    y <- DGEList(counts = as.matrix(pb))
    design <- model.matrix(~ group_vec)

    keep <- filterByExpr(y, design=design)
    y <- y[keep, , keep.lib.sizes=FALSE]
    y <- calcNormFactors(y)

    y <- estimateDisp(y, design, robust=TRUE)
    fit <- glmQLFit(y, design, robust=TRUE)
    qlf <- glmQLFTest(fit, coef=2)

    tt <- topTags(qlf, n=Inf)$table %>%
      tibble::rownames_to_column("gene") %>%
      dplyr::as_tibble() %>%
      dplyr::transmute(
        gene,
        logFC,
        logCPM,
        F,
        PValue,
        FDR,
        celltype = celltype_name,
        mode = mode_label,
        n_samples_total = ncol(pb),
        n_TIL_low = sum(group_vec=="TIL_low"),
        n_TIL_high = sum(group_vec=="TIL_high")
      )

    meta_out <- sample_group_tbl %>%
      dplyr::rename(sample = .data[[SAMPLE_COL]]) %>%
      dplyr::mutate(celltype = celltype_name, mode = mode_label)

    list(status="ok", reason="", de=tt, meta=meta_out)

  }, error = function(e) {
    list(status="error", reason=paste0("edgeR failed: ", conditionMessage(e)), de=NULL, meta=NULL)
  })
}

run_panel <- function(celltypes_vec, genes_keep, mode_label) {
  de_list <- list()
  meta_list <- list()
  status_rows <- list()

  for (ct in celltypes_vec) {
    res <- run_edger_for_celltype(ct, genes_keep, mode_label)
    status_rows[[ct]] <- tibble(
      mode = mode_label,
      celltype = ct,
      status = res$status,
      reason = res$reason
    )
    if (identical(res$status, "ok")) {
      de_list[[ct]] <- res$de
      meta_list[[ct]] <- res$meta
    } else {
      if (!is.null(res$meta)) meta_list[[ct]] <- res$meta
    }
  }

  list(
    status = dplyr::bind_rows(status_rows),
    de = if (length(de_list)) dplyr::bind_rows(de_list) else tibble(),
    meta = if (length(meta_list)) dplyr::bind_rows(meta_list) else tibble()
  )
}

# Run
sender_res   <- run_panel(senders_oi,   feasible_ligands,   "sender_ligand")
receiver_res <- run_panel(receivers_oi, feasible_receptors, "receiver_receptor")

# CHANGED: Use zone_prefix for output file names
out_sender_de     <- file.path(FILTER_DIR, paste0(zone_prefix, "_sender_ligand_DE_edger_by_celltype_TILhigh_vs_TILlow_activityLigs_LTMrows.csv"))
out_receiver_de   <- file.path(FILTER_DIR, paste0(zone_prefix, "_receiver_receptor_DE_edger_by_celltype_TILhigh_vs_TILlow_LRreceptorsOfActivityLigs.csv"))
out_sender_meta   <- file.path(FILTER_DIR, paste0(zone_prefix, "_sender_pseudobulk_samples_by_celltype.csv"))
out_receiver_meta <- file.path(FILTER_DIR, paste0(zone_prefix, "_receiver_pseudobulk_samples_by_celltype.csv"))
out_status        <- file.path(FILTER_DIR, paste0(zone_prefix, "_LR_DE_status_by_celltype.csv"))

readr::write_csv(sender_res$de, out_sender_de)
readr::write_csv(receiver_res$de, out_receiver_de)
readr::write_csv(sender_res$meta, out_sender_meta)
readr::write_csv(receiver_res$meta, out_receiver_meta)
readr::write_csv(dplyr::bind_rows(sender_res$status, receiver_res$status), out_status)

cat("Wrote:\n")
cat(" ", out_sender_de, "\n")
cat(" ", out_receiver_de, "\n")
cat(" ", out_sender_meta, "\n")
cat(" ", out_receiver_meta, "\n")
cat(" ", out_status, "\n\n")

cat("Celltypes with OK DE (senders):", sum(sender_res$status$status=="ok"), "/", length(senders_oi), "\n")
cat("Celltypes with OK DE (receivers):", sum(receiver_res$status$status=="ok"), "/", length(receivers_oi), "\n\n")

if (any(sender_res$status$status=="error") || any(receiver_res$status$status=="error")) {
  cat("Errors detected in some celltypes:\n")
  print(dplyr::bind_rows(sender_res$status, receiver_res$status) %>% dplyr::filter(status=="error"))
  cat("\n")
}

if (nrow(sender_res$de) > 0) {
  cat("Top 10 ligand DE rows overall (by FDR):\n")
  print(sender_res$de %>% dplyr::arrange(FDR) %>% dplyr::slice_head(n=10))
  cat("\n")
}
if (nrow(receiver_res$de) > 0) {
  cat("Top 10 receptor DE rows overall (by FDR):\n")
  print(receiver_res$de %>% dplyr::arrange(FDR) %>% dplyr::slice_head(n=10))
  cat("\n")
}
```

### Construct axis scores

```python
# %% [markdown]
# ### Construct axis scores

# %%
import numpy as np
import pandas as pd
import scipy.sparse as sp
from pathlib import Path

# === CLEAR STALE VARIABLES ===
for _v in ["all_axes", "top_axes", "sender_candidates", "ligand_sender_gating", 
           "base", "axes", "trm_high", "trm_low", "ltm_df", "act_by_dir",
           "allowed_ligands_by_dir", "lr_use_by_dir", "receptors_by_dir"]:
    if _v in globals():
        del globals()[_v]

# === VERIFY INPUTS ===
required = ["target_receiver_map_high", "target_receiver_map_low", "SENDERS", "RECEIVERS",
            "ligand_activities", "sender_de", "receiver_de", "lr_network", 
            "ligand_target_matrix", "de_df", "OUT_DIR",
            "TOP_LIGANDS_PER_DIRECTION", "LTM_TOPK_PER_TARGET", "TOP_AXES_PER_TARGET",
            "TOP_LIGANDS_PER_TARGET", "TOP_AXES_PER_LIGAND_PER_TARGET", "SENDER_TOPK_PER_LIGAND"]
missing = [x for x in required if x not in globals()]
if missing:
    raise NameError(f"Missing: {missing}")

print("INPUT CHECK:")
print(f"  target_receiver_map_high: {len(target_receiver_map_high)} rows, {target_receiver_map_high['target'].nunique()} targets")
print(f"  target_receiver_map_low: {len(target_receiver_map_low)} rows, {target_receiver_map_low['target'].nunique()} targets")

LAYER = "LogNormalize"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
SAMPLE_KEY = "sample_id"

MIN_PCT_EXPRESSED_LIGAND = 0.10
MIN_SAMPLES_EXPRESSING_LIGAND = 3

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

senders_std = sorted({std_ct(x) for x in SENDERS})
receivers_std = sorted({std_ct(x) for x in RECEIVERS})

TOP_LIGANDS_PER_DIRECTION = int(TOP_LIGANDS_PER_DIRECTION)
LTM_TOPK_PER_TARGET = int(LTM_TOPK_PER_TARGET)
TOP_AXES_PER_TARGET = int(TOP_AXES_PER_TARGET)
TOP_LIGANDS_PER_TARGET = int(TOP_LIGANDS_PER_TARGET)
TOP_AXES_PER_LIGAND_PER_TARGET = int(TOP_AXES_PER_LIGAND_PER_TARGET)
SENDER_TOPK_PER_LIGAND = int(SENDER_TOPK_PER_LIGAND) if SENDER_TOPK_PER_LIGAND is not None else None

LIGANDS_FROM_LTM_PER_TARGET = 10
KEEP_LIGANDS = []

OUT_DIR = Path(OUT_DIR)
OUT_DIR.mkdir(parents=True, exist_ok=True)

print(f"\nSELECTION CONTROLS:")
print(f"  TOP_LIGANDS_PER_DIRECTION: {TOP_LIGANDS_PER_DIRECTION}")
print(f"  LTM_TOPK_PER_TARGET: {LTM_TOPK_PER_TARGET}")
print(f"  TOP_AXES_PER_TARGET: {TOP_AXES_PER_TARGET}")
print(f"  TOP_LIGANDS_PER_TARGET: {TOP_LIGANDS_PER_TARGET}")
print(f"  TOP_AXES_PER_LIGAND_PER_TARGET: {TOP_AXES_PER_LIGAND_PER_TARGET}")
print(f"  SENDER_TOPK_PER_LIGAND: {SENDER_TOPK_PER_LIGAND}")

# === COPY INPUTS ===
ligand_activities = pd.DataFrame(ligand_activities).copy()
sender_de = pd.DataFrame(sender_de).copy()
receiver_de = pd.DataFrame(receiver_de).copy()

ligand_activities["test_ligand"] = ligand_activities["test_ligand"].astype(str)
ligand_activities["direction"] = ligand_activities["direction"].astype(str)
ligand_activities["rank"] = pd.to_numeric(ligand_activities["rank"], errors="coerce")
ligand_activities["aupr_corrected"] = pd.to_numeric(ligand_activities["aupr_corrected"], errors="coerce")

for df_ in (sender_de, receiver_de):
    df_["gene"] = df_["gene"].astype(str)
    df_["celltype"] = df_["celltype"].astype(str)
    df_["logFC"] = pd.to_numeric(df_["logFC"], errors="coerce")
    df_["FDR"] = pd.to_numeric(df_["FDR"], errors="coerce")

lr_df0 = pd.DataFrame(lr_network).copy()
cols = set(lr_df0.columns.astype(str))
if {"from", "to"}.issubset(cols):
    lr_df = lr_df0.rename(columns={"from": "ligand", "to": "receptor"})[["ligand", "receptor"]].copy()
elif {"ligand", "receptor"}.issubset(cols):
    lr_df = lr_df0[["ligand", "receptor"]].copy()
else:
    raise KeyError(f"lr_network columns not recognized: {list(lr_df0.columns)}")
lr_df["ligand"] = lr_df["ligand"].astype(str)
lr_df["receptor"] = lr_df["receptor"].astype(str)
lr_df = lr_df.drop_duplicates()

# Fresh copy of TRMs
trm_high = pd.DataFrame(target_receiver_map_high).copy()
trm_low = pd.DataFrame(target_receiver_map_low).copy()

for trm in (trm_high, trm_low):
    trm["target"] = trm["target"].astype(str)
    trm["receiver_celltype"] = trm["receiver_celltype"].astype(str)
    for col in ["target_pct_expr", "target_mean_expr", "receiver_rank_score", "panel_coverage_ratio"]:
        if col in trm.columns:
            trm[col] = pd.to_numeric(trm[col], errors="coerce")

ltm_df_full = pd.DataFrame(
    ligand_target_matrix.values,
    index=pd.Index(ligand_target_matrix.index.astype(str), name="ligand"),
    columns=pd.Index(ligand_target_matrix.columns.astype(str), name="target"),
)

def get_ltm_ligands_for_targets(targets, ltm_df, k):
    targets = [t for t in targets if t in ltm_df.columns]
    ligs = set()
    for t in targets:
        w = ltm_df[t]
        w = w[w > 0].nlargest(k)
        ligs.update(w.index.astype(str))
    return ligs

# === ACTIVITY MAPS ===
act_by_dir = {}
for direction in ["TIL_high", "TIL_low"]:
    act = ligand_activities[ligand_activities["direction"] == direction].dropna(subset=["rank"])
    act = act.sort_values("rank")
    act_by_dir[direction] = {
        "top_ligs": act.nsmallest(TOP_LIGANDS_PER_DIRECTION, "rank")["test_ligand"].tolist(),
        "activity_map": act.set_index("test_ligand")["aupr_corrected"].to_dict(),
        "rank_map": act.set_index("test_ligand")["rank"].to_dict(),
    }

# === LIGAND UNIVERSE ===
allowed_ligands_by_dir = {}
for direction, trm in [("TIL_high", trm_high), ("TIL_low", trm_low)]:
    top_act = set(act_by_dir[direction]["top_ligs"])
    dir_targets = sorted(set(trm["target"]) & set(ltm_df_full.columns))
    top_ltm = get_ltm_ligands_for_targets(dir_targets, ltm_df_full, LIGANDS_FROM_LTM_PER_TARGET)
    allowed_ligands_by_dir[direction] = top_act | top_ltm | set(KEEP_LIGANDS)
    print(f"{direction} ligand universe: {len(allowed_ligands_by_dir[direction])}")

need_ligs = sorted(allowed_ligands_by_dir["TIL_high"] | allowed_ligands_by_dir["TIL_low"])
need_tgts = sorted(set(trm_high["target"]) | set(trm_low["target"]))
use_ligs = sorted(set(ltm_df_full.index) & set(need_ligs))
use_tgts = sorted(set(ltm_df_full.columns) & set(need_tgts))
ltm_df = ltm_df_full.loc[use_ligs, use_tgts].copy()

print(f"LTM subset: {len(use_ligs)} ligands x {len(use_tgts)} targets")

# === SENDER GATING ===
ct_std = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)
sample_ids = adata_sc.obs[SAMPLE_KEY].astype(str)

X = adata_sc.layers[LAYER]
if sp.issparse(X):
    X = X.tocsr()
else:
    X = sp.csr_matrix(X)

var_names = pd.Index(adata_sc.var_names.astype(str))
var_set = set(var_names)
ligs_in_var = sorted(set(use_ligs) & var_set)

sender_rows = []
for direction in ["TIL_high", "TIL_low"]:
    dir_ligs = sorted(allowed_ligands_by_dir[direction] & set(ligs_in_var))
    if not dir_ligs:
        continue
    lig_idx = var_names.get_indexer(dir_ligs)

    for sct in senders_std:
        m = (grp == direction).to_numpy() & (ct_std == sct).to_numpy()
        idx = np.flatnonzero(m)
        if idx.size == 0:
            continue

        sample_sub = sample_ids.iloc[idx].values
        n_samples = len(np.unique(sample_sub))
        Xsub = X[idx, :][:, lig_idx]
        pct = np.asarray(Xsub.getnnz(axis=0)).ravel() / idx.size
        mean_expr = np.asarray(Xsub.mean(axis=0)).ravel()
        Xsub_dense = Xsub.toarray()

        for i, lig in enumerate(dir_ligs):
            n_samp_expr = len(set(sample_sub[Xsub_dense[:, i] > 0]))
            if pct[i] >= MIN_PCT_EXPRESSED_LIGAND and n_samp_expr >= MIN_SAMPLES_EXPRESSING_LIGAND:
                sender_rows.append({
                    "direction": direction, "ligand": lig, "sender_celltype": sct,
                    "pct_expr": float(pct[i]), "mean_expr": float(mean_expr[i]),
                    "n_samples_expressing": n_samp_expr,
                    "sample_frac": n_samp_expr / n_samples if n_samples > 0 else 0,
                })

sender_gating = pd.DataFrame(sender_rows)

# Sender specificity and rank score
for direction in ["TIL_high", "TIL_low"]:
    mask = sender_gating["direction"] == direction
    if mask.sum() == 0:
        continue
    lig_mean = sender_gating.loc[mask].groupby("ligand")["mean_expr"].transform("mean").replace(0, np.nan)
    sender_gating.loc[mask, "specificity"] = (sender_gating.loc[mask, "mean_expr"] / lig_mean).fillna(0)

sender_gating["_mr"] = sender_gating.groupby(["direction", "ligand"])["mean_expr"].rank(ascending=False)
sender_gating["_pr"] = sender_gating.groupby(["direction", "ligand"])["pct_expr"].rank(ascending=False)
sender_gating["_sfr"] = sender_gating.groupby(["direction", "ligand"])["sample_frac"].rank(ascending=False)
sender_gating["_spr"] = sender_gating.groupby(["direction", "ligand"])["specificity"].rank(ascending=False)
sender_gating["sender_rank_score"] = 0.35/sender_gating["_mr"] + 0.25/sender_gating["_pr"] + 0.20/sender_gating["_sfr"] + 0.20/sender_gating["_spr"]
sender_gating = sender_gating.drop(columns=["_mr", "_pr", "_sfr", "_spr"])

# Merge sender DE
sender_de_use = sender_de.rename(columns={"gene": "ligand", "celltype": "sender_celltype", "logFC": "ligand_logFC", "FDR": "ligand_FDR"})
sender_candidates = sender_gating.merge(sender_de_use[["ligand", "sender_celltype", "ligand_logFC", "ligand_FDR"]], on=["ligand", "sender_celltype"], how="left")

# Concordance bonus
exp_sign = np.where(sender_candidates["direction"] == "TIL_high", 1, -1)
conc = sender_candidates["ligand_logFC"].notna() & ((exp_sign * sender_candidates["ligand_logFC"]) > 0)
sender_candidates.loc[conc, "sender_rank_score"] += 0.05

sender_candidates = sender_candidates.sort_values(["direction", "ligand", "sender_rank_score"], ascending=[True, True, False])
if SENDER_TOPK_PER_LIGAND is not None:
    sender_candidates = sender_candidates.groupby(["direction", "ligand"], as_index=False).head(SENDER_TOPK_PER_LIGAND)

print(f"Sender candidates: {len(sender_candidates)} rows")

# === LR BY DIRECTION ===
lr_by_dir = {}
for d in ["TIL_high", "TIL_low"]:
    lr_by_dir[d] = lr_df[lr_df["ligand"].isin(allowed_ligands_by_dir[d])].copy()
    lr_by_dir[d]["direction"] = d

# === LTM TOP-K ===
def ltm_topk(direction, trm):
    tgts = sorted(set(trm["target"]) & set(ltm_df.columns))
    elig_ligs = sorted(set(lr_by_dir[direction]["ligand"]) & set(ltm_df.index))
    if not tgts or not elig_ligs:
        return pd.DataFrame(columns=["direction", "target", "ligand", "ltm_weight"])
    sub = ltm_df.loc[elig_ligs, tgts]
    rows = []
    for t in tgts:
        w = sub[t]
        w = w[w > 0].nlargest(LTM_TOPK_PER_TARGET)
        for lig, wt in w.items():
            rows.append({"direction": direction, "target": t, "ligand": str(lig), "ltm_weight": float(wt)})
    return pd.DataFrame(rows)

topk_high = ltm_topk("TIL_high", trm_high)
topk_low = ltm_topk("TIL_low", trm_low)

# === MERGE TRM ===
trm_cols = ["target", "receiver_celltype", "target_pct_expr", "target_mean_expr", "receiver_rank_score",
            "global_best_celltype", "panel_coverage_ratio", "coverage_flag"]
trm_cols_high = [c for c in trm_cols if c in trm_high.columns]
trm_cols_low = [c for c in trm_cols if c in trm_low.columns]

base_high = topk_high.merge(trm_high[trm_cols_high], on="target", how="inner")
base_low = topk_low.merge(trm_low[trm_cols_low], on="target", how="inner")
base = pd.concat([base_high, base_low], ignore_index=True)

print(f"Base: {len(base)} rows (high={len(base_high)}, low={len(base_low)})")

# Attach activity
base["ligand_activity"] = np.nan
base["ligand_activity_rank"] = np.nan
for d in ["TIL_high", "TIL_low"]:
    m = base["direction"] == d
    base.loc[m, "ligand_activity"] = base.loc[m, "ligand"].map(act_by_dir[d]["activity_map"])
    base.loc[m, "ligand_activity_rank"] = base.loc[m, "ligand"].map(act_by_dir[d]["rank_map"])

# Attach sender
base = base.merge(sender_candidates, on=["direction", "ligand"], how="inner")
print(f"After sender merge: {len(base)} rows")

# Attach receptor
lr_dir = pd.concat([lr_by_dir["TIL_high"], lr_by_dir["TIL_low"]], ignore_index=True)
axes = base.merge(lr_dir, on=["direction", "ligand"], how="inner")

receiver_tbl = receiver_de.rename(columns={"celltype": "receiver_celltype", "gene": "receptor", "logFC": "receptor_logFC", "FDR": "receptor_FDR"})
axes = axes.merge(receiver_tbl[["receiver_celltype", "receptor", "receptor_logFC", "receptor_FDR"]], on=["receiver_celltype", "receptor"], how="left")

# Attach target DE
de_use = de_df[["gene", "logFC", "FDR"]].copy()
de_use.columns = ["target", "target_logFC", "target_FDR"]
axes = axes.merge(de_use, on="target", how="left")

axes = axes.dropna(subset=["receptor_logFC"])
print(f"After receptor filter: {len(axes)} axes")

# === SCORE ===
def score_axes(df):
    df = df.copy()
    for d in ["TIL_high", "TIL_low"]:
        m = df["direction"] == d
        if m.sum() == 0:
            continue
        ranks = df.loc[m, "ligand_activity_rank"]
        max_r = ranks.max()
        df.loc[m, "activity_pctl"] = 1 - (ranks - 1) / max_r if max_r > 1 else 1.0
        df.loc[m, "ltm_pctl"] = df.loc[m, "ltm_weight"].rank(pct=True)
        if "receiver_rank_score" in df.columns:
            df.loc[m, "receiver_rank_pctl"] = df.loc[m, "receiver_rank_score"].rank(pct=True)
        else:
            df.loc[m, "receiver_rank_pctl"] = 0.5

    exp_sign = np.where(df["direction"] == "TIL_high", 1, -1)
    df["ligand_concordant"] = (exp_sign * df["ligand_logFC"]) > 0
    df["receptor_concordant"] = (exp_sign * df["receptor_logFC"]) > 0

    df["axis_score"] = (
        0.35 * df["activity_pctl"].fillna(0) +
        0.25 * df["ltm_pctl"].fillna(0) +
        0.15 * df["receiver_rank_pctl"].fillna(0) +
        0.10 * df["ligand_concordant"].astype(float) +
        0.15 * df["receptor_concordant"].astype(float)
    )
    return df

all_axes = score_axes(axes)
all_axes = all_axes.sort_values(["direction", "target", "axis_score"], ascending=[True, True, False])
all_axes["axis_rank"] = all_axes.groupby(["direction", "target"]).cumcount() + 1
all_axes = all_axes.drop_duplicates(subset=["direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype"])

# === TOP AXES SELECTION ===
lig_best = all_axes.groupby(["direction", "target", "ligand"]).agg(
    best_score=("axis_score", "max"), best_ltm=("ltm_weight", "max"), best_act=("ligand_activity", "max"), n=("axis_score", "size")
).reset_index()
lig_best = lig_best.sort_values(["direction", "target", "best_score", "best_ltm", "best_act"], ascending=[True, True, False, False, False])
lig_best["lig_rank"] = lig_best.groupby(["direction", "target"]).cumcount() + 1
keep_ligs = lig_best[lig_best["lig_rank"] <= TOP_LIGANDS_PER_TARGET][["direction", "target", "ligand"]]

sel = all_axes.merge(keep_ligs, on=["direction", "target", "ligand"], how="inner")
sel = sel.sort_values(["direction", "target", "ligand", "axis_score"], ascending=[True, True, True, False])
sel["ax_in_lig"] = sel.groupby(["direction", "target", "ligand"]).cumcount() + 1
sel = sel[sel["ax_in_lig"] <= TOP_AXES_PER_LIGAND_PER_TARGET]
sel = sel.sort_values(["direction", "target", "axis_score"], ascending=[True, True, False])
sel["ax_final"] = sel.groupby(["direction", "target"]).cumcount() + 1
top_axes = sel[sel["ax_final"] <= TOP_AXES_PER_TARGET].copy()
top_axes["axis_rank"] = top_axes.groupby(["direction", "target"]).cumcount() + 1

# === SUMMARY ===
print(f"\nAXES BUILT:")
print(f"  all_axes: {len(all_axes)} rows, {all_axes['target'].nunique()} targets")
print(f"  top_axes: {len(top_axes)} rows, {top_axes['target'].nunique()} targets")

for d in ["TIL_high", "TIL_low"]:
    sub = all_axes[all_axes["direction"] == d]
    print(f"\n  {d}: {len(sub)} axes, {sub['target'].nunique()} targets, {sub['ligand'].nunique()} ligands")

# === SAVE ===
all_axes.to_csv(OUT_DIR / "all_axes.csv", index=False)
top_axes.to_csv(OUT_DIR / "top_axes.csv", index=False)
print(f"\n✓ Saved to {OUT_DIR}")
```

### Summarize axis scores for plot selection

```python
# %% [markdown]
# ### Summarize axis scores for plot selection

# %%
import pandas as pd
import numpy as np
from pathlib import Path
import hashlib

# === CLEAR STALE VARIABLES ===
for _v in ["df", "df_all", "df_top", "axes_catalog", "axes_shortlist", "short", "short2"]:
    if _v in globals():
        del globals()[_v]

# === VERIFY INPUTS ===
if "OUT_DIR" not in globals():
    raise NameError("OUT_DIR not found.")
if "top_axes" not in globals():
    raise NameError("top_axes not found. Run Cell 5 first.")

OUT_DIR = Path(OUT_DIR)

print("INPUT CHECK:")
print(f"  top_axes: {len(top_axes)} rows, {top_axes['target'].nunique()} targets")
if "all_axes" in globals():
    print(f"  all_axes: {len(all_axes)} rows (for reference, not used in catalog)")

# Use top_axes (filtered) for catalog, NOT all_axes
df = pd.DataFrame(top_axes).copy()

req_cols = {"direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype",
            "target_logFC", "target_FDR", "ligand_activity", "ltm_weight",
            "ligand_concordant", "receptor_concordant", "axis_score", "axis_rank"}
missing = req_cols - set(df.columns)
if missing:
    raise KeyError(f"top_axes missing columns: {sorted(missing)}")

# Normalize dtypes
for c in ["direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype"]:
    df[c] = df[c].astype(str)
for c in ["axis_score", "axis_rank", "target_logFC", "target_FDR", "ligand_activity", "ltm_weight"]:
    df[c] = pd.to_numeric(df[c], errors="coerce")
df["ligand_concordant"] = df["ligand_concordant"].astype(bool)
df["receptor_concordant"] = df["receptor_concordant"].astype(bool)

if "receiver_rank_score" in df.columns:
    df["receiver_rank_score"] = pd.to_numeric(df["receiver_rank_score"], errors="coerce")
if "panel_coverage_ratio" in df.columns:
    df["panel_coverage_ratio"] = pd.to_numeric(df["panel_coverage_ratio"], errors="coerce")

# === FILTER 1: positive activity ===
n_before = len(df)
df = df[df["ligand_activity"] > 0].copy()
print(f"\nFilter: ligand_activity > 0: {n_before} -> {len(df)} axes")

# === ABUNDANCE ANNOTATIONS ===
def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

abundance_available = False
if "EFS_ROOT" in globals() and "ZONE" in globals():
    ab_path = Path(EFS_ROOT) / "results/visium/til_zones/celltype_zone_specificity_curated/celltype_zone_specificity_level4_curated.csv"

    if ab_path.exists():
        ab = pd.read_csv(ab_path)
        ab["group"] = ab["group"].astype(str)
        ab["zone"] = ab["zone"].astype(str)
        ab["celltype_std"] = ab["celltype_level4_curated"].astype(str).map(std_ct)
        ab_zone = ab[ab["zone"] == str(ZONE)].copy()

        delta_tbl = ab_zone.pivot_table(index="celltype_std", columns="group", values="mean_zone", aggfunc="first").reset_index()
        if "TIL_high" in delta_tbl.columns and "TIL_low" in delta_tbl.columns:
            delta_tbl["mean_zone_high_minus_low"] = pd.to_numeric(delta_tbl["TIL_high"], errors="coerce") - pd.to_numeric(delta_tbl["TIL_low"], errors="coerce")
            delta_tbl = delta_tbl[["celltype_std", "mean_zone_high_minus_low"]]
        else:
            delta_tbl = pd.DataFrame({"celltype_std": ab_zone["celltype_std"].unique(), "mean_zone_high_minus_low": np.nan})

        df["sender_celltype_std"] = df["sender_celltype"].map(std_ct)
        df["receiver_celltype_std"] = df["receiver_celltype"].map(std_ct)

        # Sender abundance
        sender_ab = ab_zone.rename(columns={"celltype_std": "sender_celltype_std", "mean_zone": "sender_mean_zone"})
        sender_ab = sender_ab[["group", "sender_celltype_std", "sender_mean_zone"]].copy()
        df = df.merge(sender_ab, left_on=["direction", "sender_celltype_std"], right_on=["group", "sender_celltype_std"], how="left")
        df = df.drop(columns=["group"], errors="ignore")

        # Receiver abundance
        receiver_ab = ab_zone.rename(columns={"celltype_std": "receiver_celltype_std", "mean_zone": "receiver_mean_zone"})
        receiver_ab = receiver_ab[["group", "receiver_celltype_std", "receiver_mean_zone"]].copy()
        df = df.merge(receiver_ab, left_on=["direction", "receiver_celltype_std"], right_on=["group", "receiver_celltype_std"], how="left")
        df = df.drop(columns=["group"], errors="ignore")

        # Delta
        delta_sender = delta_tbl.rename(columns={"celltype_std": "sender_celltype_std", "mean_zone_high_minus_low": "sender_mean_zone_high_minus_low"})
        delta_receiver = delta_tbl.rename(columns={"celltype_std": "receiver_celltype_std", "mean_zone_high_minus_low": "receiver_mean_zone_high_minus_low"})
        df = df.merge(delta_sender, on="sender_celltype_std", how="left")
        df = df.merge(delta_receiver, on="receiver_celltype_std", how="left")

        print(f"Abundance annotations loaded from: {ab_path}")
        abundance_available = True
    else:
        print(f"Abundance CSV not found: {ab_path}")
else:
    print("EFS_ROOT or ZONE not set, skipping abundance annotations")

# === FILTER 2: concordance/abundance ===
ABUNDANCE_THRESHOLD = 0.05
n_before = len(df)

if abundance_available:
    high_mask = df["direction"] == "TIL_high"
    low_mask = df["direction"] == "TIL_low"
    high_keep = df["ligand_concordant"] | (df["sender_mean_zone_high_minus_low"] > ABUNDANCE_THRESHOLD)
    low_keep = df["ligand_concordant"] | (df["sender_mean_zone_high_minus_low"] < -ABUNDANCE_THRESHOLD)
    df = df[(high_mask & high_keep) | (low_mask & low_keep)].copy()
else:
    df = df[df["ligand_concordant"]].copy()

print(f"Filter: concordance/abundance: {n_before} -> {len(df)} axes")

# === AXIS ID ===
key_cols = ["direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype"]
df = df.drop_duplicates(subset=key_cols).copy()

def axis_hash_id(row):
    s = "|".join(str(row[c]) for c in key_cols)
    h = hashlib.md5(s.encode()).hexdigest()[:6].upper()
    return ("H" if row["direction"] == "TIL_high" else "L") + h

df["axis_id"] = df.apply(axis_hash_id, axis=1)

# === RANK SCORE ===
df["de_strength"] = -np.log10(df["target_FDR"].clip(lower=1e-300))
df["rank_score"] = df["axis_score"] * df["de_strength"] * df["target_logFC"].abs()

# === VERIFY AXES PER TARGET ===
axes_per_target = df.groupby(["direction", "target"]).size()
print(f"\nAxes per target: min={axes_per_target.min()}, max={axes_per_target.max()}, mean={axes_per_target.mean():.1f}")

# === CATALOG COLUMNS ===
catalog_cols = ["axis_id", "direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype",
                "axis_score", "axis_rank", "rank_score", "target_logFC", "target_FDR",
                "ligand_activity", "ltm_weight", "ligand_concordant", "receptor_concordant"]

# Add optional columns if present
for c in ["receiver_rank_score", "global_best_celltype", "global_best_n_cells", "panel_coverage_ratio", "coverage_flag",
          "sender_mean_zone", "receiver_mean_zone", "sender_mean_zone_high_minus_low", "receiver_mean_zone_high_minus_low"]:
    if c in df.columns:
        catalog_cols.append(c)

axes_catalog = df[[c for c in catalog_cols if c in df.columns]].copy()
axes_catalog.to_csv(OUT_DIR / "axes_catalog.csv", index=False)

# === SHORTLIST (further filtered subset) ===
AXIS_SCORE_MIN = 0.5
MAX_AXIS_RANK_PER_TARGET = 3
TARGET_FDR_MAX = 0.05
TARGET_LOGFC_MIN = 0.25
AXES_PER_TARGET_DE_BAL = 2

short = df[(df["axis_score"] > AXIS_SCORE_MIN) & (df["axis_rank"] <= MAX_AXIS_RANK_PER_TARGET)].copy()
short = short.dropna(subset=["target_FDR", "target_logFC", "rank_score"])
short = short[short["target_FDR"] <= TARGET_FDR_MAX]

high = short[(short["direction"] == "TIL_high") & (short["target_logFC"] >= TARGET_LOGFC_MIN)]
low = short[(short["direction"] == "TIL_low") & (short["target_logFC"] <= -TARGET_LOGFC_MIN)]
short = pd.concat([high, low])
short = short.sort_values(["direction", "target", "rank_score"], ascending=[True, True, False])
axes_shortlist = short.groupby(["direction", "target"], as_index=False).head(AXES_PER_TARGET_DE_BAL)
axes_shortlist = axes_shortlist[[c for c in catalog_cols if c in axes_shortlist.columns]]
axes_shortlist.to_csv(OUT_DIR / "axes_shortlist.csv", index=False)

# === SUMMARY ===
print(f"\nCATALOG SUMMARY:")
for d in ["TIL_high", "TIL_low"]:
    sub = df[df["direction"] == d]
    print(f"  {d}: {len(sub)} axes, {sub['target'].nunique()} targets, {sub['ligand'].nunique()} ligands")

if "coverage_flag" in df.columns:
    print(f"\nPANEL COVERAGE:")
    for d in ["TIL_high", "TIL_low"]:
        sub = df[df["direction"] == d]
        if sub.empty:
            continue
        n_ok = (sub["coverage_flag"] == "OK").sum()
        n_low = (sub["coverage_flag"] == "LOW_COVERAGE").sum()
        print(f"  {d}: OK={n_ok}, LOW_COVERAGE={n_low}")

print(f"\nSHORTLIST: {len(axes_shortlist)} axes, {axes_shortlist['target'].nunique()} targets")

print(f"\nWROTE:")
print(f"  {OUT_DIR / 'axes_catalog.csv'}")
print(f"  {OUT_DIR / 'axes_shortlist.csv'}")

# === DISPLAY ===
display_cols = ["axis_id", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype",
                "axis_score", "rank_score", "target_logFC"]
if "coverage_flag" in axes_catalog.columns:
    display_cols.append("coverage_flag")
display_cols = [c for c in display_cols if c in axes_catalog.columns]

print(f"\nTOP 15 TIL_high:")
display(axes_catalog[axes_catalog["direction"] == "TIL_high"].nlargest(15, "rank_score")[display_cols])

print(f"\nTOP 15 TIL_low:")
display(axes_catalog[axes_catalog["direction"] == "TIL_low"].nlargest(15, "rank_score")[display_cols])

# === WARNINGS ===
if "coverage_flag" in axes_shortlist.columns:
    low_cov = axes_shortlist[axes_shortlist["coverage_flag"] == "LOW_COVERAGE"]
    if len(low_cov) > 0:
        print(f"\n⚠️ WARNING: {len(low_cov)} shortlist axes have LOW_COVERAGE:")
        for _, r in low_cov.iterrows():
            gb_ct = r.get('global_best_celltype', 'N/A')
            gb_n = r.get('global_best_n_cells', 'N/A')
            print(f"  {r['axis_id']}: {r['target']} (global_best={gb_ct}, n={gb_n})")

print("\n✓ Done")
```

### Circos plotting

```python
# Pass ZONE to R for circos plotting
ZONE_FOR_R = str(ZONE)
print(f"Passing ZONE to R: {ZONE_FOR_R}")
```

```python
%%R -i ZONE_FOR_R
suppressPackageStartupMessages({
  library(dplyr)
  library(readr)
  library(circlize)
})

options(bitmapType = "cairo")

EFS_ROOT <- "/home/sagemaker-user/custom-file-systems/efs/fs-09913c1f7db79b6fd"
ZONE <- ZONE_FOR_R
zone_dir <- gsub("\\+", "plus", gsub("-", "minus", ZONE))
OUT_DIR <- file.path(EFS_ROOT, "results", "visium", "til_zones", "nichenet_analysis", zone_dir)
stopifnot(dir.exists(OUT_DIR))

AXES_SELECTED_CSV <- file.path(OUT_DIR, "axes_selected.csv")
stopifnot(file.exists(AXES_SELECTED_CSV))

MAX_EDGES <- Inf

LINK_ALPHA <- 0.5

BORDER_COL <- "#777777"
BORDER_LWD <- 0.25

TITLE_CEX <- 1.00

START_DEGREE <- -8
SMALL_GAP <- 1
BIG_GAP <- 5

WIDTH_MIN <- 0.10
WIDTH_MAX <- 2.50

COLOR_CLIP_LO <- 0.02
COLOR_CLIP_HI <- 0.98

CANVAS_PAD <- 0.40

LABEL_CEX <- 0.46
LABEL_RADIUS <- 1.0
LABEL_NICEFACING <- TRUE
LABEL_MIN_DEG <- 0

OUT_PNG <- file.path(OUT_DIR, "circos_selected_axes_senders_receivers_only.png")
PNG_SIZE <- 2600
PNG_RES <- 240

axes <- readr::read_csv(AXES_SELECTED_CSV, show_col_types = FALSE)

required_cols <- c(
  "axis_id","direction",
  "sender_celltype","receiver_celltype",
  "ligand","target",
  "rank_score","ligand_activity"
)
missing <- setdiff(required_cols, colnames(axes))
if (length(missing) > 0) stop(paste0("axes_selected.csv missing columns: ", paste(missing, collapse = ", ")))

axes <- axes %>%
  transmute(
    axis_id = as.character(axis_id),
    direction = as.character(direction),
    sender = as.character(sender_celltype),
    receiver = as.character(receiver_celltype),
    ligand = as.character(ligand),
    target = as.character(target),
    rank_score = as.numeric(rank_score),
    ligand_activity = as.numeric(ligand_activity)
  ) %>%
  filter(
    direction %in% c("TIL_high","TIL_low"),
    is.finite(rank_score),
    is.finite(ligand_activity)
  )

if (nrow(axes) == 0) stop("No rows left after filtering. Check direction/rank_score/ligand_activity.")

if (is.finite(MAX_EDGES) && nrow(axes) > MAX_EDGES) {
  axes <- axes %>% slice_head(n = MAX_EDGES)
}

rescale01 <- function(x) {
  x <- as.numeric(x)
  ok <- is.finite(x)
  if (sum(ok) == 0) return(rep(0.5, length(x)))
  lo <- min(x[ok]); hi <- max(x[ok])
  if (!is.finite(lo) || !is.finite(hi) || lo == hi) return(rep(0.5, length(x)))
  y <- (x - lo) / (hi - lo)
  y[!is.finite(y)] <- 0.5
  pmin(pmax(y, 0), 1)
}

clip01_quantile <- function(x, qlo = 0.02, qhi = 0.98) {
  x <- as.numeric(x)
  ok <- is.finite(x)
  if (sum(ok) == 0) return(rep(0.5, length(x)))
  lo <- suppressWarnings(as.numeric(stats::quantile(x[ok], probs = qlo, na.rm = TRUE, names = FALSE)))
  hi <- suppressWarnings(as.numeric(stats::quantile(x[ok], probs = qhi, na.rm = TRUE, names = FALSE)))
  if (!is.finite(lo) || !is.finite(hi) || lo == hi) return(rescale01(x))
  y <- (x - lo) / (hi - lo)
  y[!is.finite(y)] <- 0.5
  pmin(pmax(y, 0), 1)
}

axes <- axes %>%
  mutate(
    width_norm = rescale01(ligand_activity),
    w = WIDTH_MIN + width_norm * (WIDTH_MAX - WIDTH_MIN)
  )

rank_norm <- clip01_quantile(axes$rank_score, COLOR_CLIP_LO, COLOR_CLIP_HI)

col_high <- circlize::colorRamp2(c(0, 1), c("#FEE6CE", "#D55E00"))
col_low  <- circlize::colorRamp2(c(0, 1), c("#DEEBF7", "#0072B2"))

axes$link_col <- vapply(seq_len(nrow(axes)), function(i) {
  if (axes$direction[i] == "TIL_high") col_high(rank_norm[i]) else col_low(rank_norm[i])
}, character(1))
axes$link_col <- adjustcolor(axes$link_col, alpha.f = LINK_ALPHA)

S_PREFIX <- "S|"
R_PREFIX <- "R|"
clean_label <- function(x) sub("^[SR]\\|", "", x)

senders <- axes %>%
  group_by(sender) %>%
  summarise(total = sum(w), .groups = "drop") %>%
  arrange(desc(total)) %>%
  pull(sender)

receivers <- axes %>%
  group_by(receiver) %>%
  summarise(total = sum(w), .groups = "drop") %>%
  arrange(desc(total)) %>%
  pull(receiver)

sender_sectors <- paste0(S_PREFIX, senders)
receiver_sectors <- paste0(R_PREFIX, receivers)
sectors <- c(sender_sectors, receiver_sectors)

ct_names <- sort(unique(c(senders, receivers)))
ct_cols <- setNames(grDevices::hcl.colors(length(ct_names), palette = "Dark 3"), ct_names)

grid_cols <- c(
  setNames(ct_cols[clean_label(sender_sectors)], sender_sectors),
  setNames(ct_cols[clean_label(receiver_sectors)], receiver_sectors)
)

axes <- axes %>%
  mutate(
    s_sector = paste0(S_PREFIX, sender),
    r_sector = paste0(R_PREFIX, receiver),
    row_id = seq_len(n())
  )

sender_totals <- tapply(axes$w, axes$s_sector, sum)
receiver_totals <- tapply(axes$w, axes$r_sector, sum)

totals <- setNames(rep(1, length(sectors)), sectors)
if (length(sender_totals) > 0) totals[names(sender_totals)] <- sender_totals
if (length(receiver_totals) > 0) totals[names(receiver_totals)] <- receiver_totals
totals[!is.finite(totals) | totals <= 0] <- 1

xlim_mat <- cbind(rep(0, length(sectors)), as.numeric(totals[sectors]))
rownames(xlim_mat) <- sectors

# Collapse ligands per sender:
# Allocate one contiguous sender-side block per (sender, ligand), then subdivide within block for links.
lig_groups <- axes %>%
  group_by(s_sector, ligand) %>%
  summarise(
    group_w = sum(w),
    group_order = min(row_id),
    .groups = "drop"
  ) %>%
  arrange(s_sector, group_order)

sender_cursor <- setNames(rep(0, length(sender_sectors)), sender_sectors)
lig_groups$lig_start <- NA_real_
lig_groups$lig_end <- NA_real_

for (k in seq_len(nrow(lig_groups))) {
  ss <- lig_groups$s_sector[k]
  ww <- lig_groups$group_w[k]
  st <- sender_cursor[ss]
  en <- st + ww
  lig_groups$lig_start[k] <- st
  lig_groups$lig_end[k] <- en
  sender_cursor[ss] <- en
}

axes <- axes %>%
  left_join(lig_groups %>% select(s_sector, ligand, lig_start, lig_end), by = c("s_sector", "ligand")) %>%
  group_by(s_sector, ligand) %>%
  arrange(row_id, .by_group = TRUE) %>%
  mutate(
    s_x1 = lig_start + dplyr::lag(cumsum(w), default = 0),
    s_x2 = lig_start + cumsum(w)
  ) %>%
  ungroup()

# Receiver allocation stays per-edge (unchanged)
receiver_cursor <- setNames(rep(0, length(receiver_sectors)), receiver_sectors)
r_x1 <- numeric(nrow(axes)); r_x2 <- numeric(nrow(axes))

for (i in seq_len(nrow(axes))) {
  rr <- axes$r_sector[i]
  ww <- axes$w[i]
  r_x1[i] <- receiver_cursor[rr]
  r_x2[i] <- receiver_cursor[rr] + ww
  receiver_cursor[rr] <- r_x2[i]
}

axes$r_x1 <- r_x1; axes$r_x2 <- r_x2
axes$s_mid <- (axes$s_x1 + axes$s_x2) / 2
axes$r_mid <- (axes$r_x1 + axes$r_x2) / 2

gap_after <- c(
  rep(SMALL_GAP, max(length(sender_sectors) - 1, 0)), BIG_GAP,
  rep(SMALL_GAP, max(length(receiver_sectors) - 1, 0)), BIG_GAP
)

thin_by_min_deg <- function(theta, min_deg) {
  if (!is.finite(min_deg) || min_deg <= 0) return(rep(TRUE, length(theta)))
  ord <- order(theta)
  keep <- rep(FALSE, length(theta))
  last <- NA_real_
  for (j in ord) {
    if (!is.finite(theta[j])) next
    if (is.na(last) || abs(theta[j] - last) >= min_deg) {
      keep[j] <- TRUE
      last <- theta[j]
    }
  }
  keep
}

plot_one <- function() {
  circos.clear()
  circos.par(
    start.degree = START_DEGREE,
    gap.after = gap_after,
    cell.padding = c(0, 0, 0, 0),
    track.margin = c(0.01, 0.01),
    canvas.xlim = c(-(1 + CANVAS_PAD), 1 + CANVAS_PAD),
    canvas.ylim = c(-(1 + CANVAS_PAD), 1 + CANVAS_PAD)
  )

  circos.initialize(factors = sectors, xlim = xlim_mat)

  circos.trackPlotRegion(
    factors = sectors,
    ylim = c(0, 1),
    track.height = 0.08,
    bg.border = NA,
    panel.fun = function(x, y) {
      si <- get.cell.meta.data("sector.index")
      xl <- get.cell.meta.data("xlim")
      circos.rect(xl[1], 0, xl[2], 1, col = grid_cols[si], border = NA)
    }
  )

  for (i in seq_len(nrow(axes))) {
    circos.link(
      sector.index1 = axes$s_sector[i],
      point1 = c(axes$s_x1[i], axes$s_x2[i]),
      sector.index2 = axes$r_sector[i],
      point2 = c(axes$r_x1[i], axes$r_x2[i]),
      col = axes$link_col[i],
      border = BORDER_COL,
      lwd = BORDER_LWD,
      directional = 1,
      arr.type = "big.arrow",
      arr.length = 0.04
    )
  }

  # Sender-side ligand labels: one per (sender, ligand), positioned at the center of the ligand block
  ligand_labels <- lig_groups %>%
    mutate(lig_mid = (lig_start + lig_end) / 2)

  theta_lig <- rep(NA_real_, nrow(ligand_labels))
  for (k in seq_len(nrow(ligand_labels))) {
    ss <- ligand_labels$s_sector[k]
    m <- circlize(ligand_labels$lig_mid[k], 0.5, sector.index = ss, track.index = 1)
    theta_lig[k] <- m[1, 1]
  }

  keep_lig <- rep(TRUE, nrow(ligand_labels))
  if (LABEL_MIN_DEG > 0) {
    for (ss in unique(ligand_labels$s_sector)) {
      idx <- which(ligand_labels$s_sector == ss)
      keep_lig[idx] <- thin_by_min_deg(theta_lig[idx], LABEL_MIN_DEG)
    }
  }

  # Receiver-side target labels remain per-edge (unchanged)
  theta_r <- rep(NA_real_, nrow(axes))
  for (rr in unique(axes$r_sector)) {
    idx <- which(is.finite(axes$r_mid) & axes$r_sector == rr)
    if (length(idx) == 0) next
    m <- circlize(axes$r_mid[idx], rep(0.5, length(idx)), sector.index = rr, track.index = 1)
    theta_r[idx] <- m[, 1]
  }

  keep_r <- rep(TRUE, nrow(axes))
  if (LABEL_MIN_DEG > 0) {
    for (rr in unique(axes$r_sector)) {
      idx <- which(axes$r_sector == rr)
      keep_r[idx] <- thin_by_min_deg(theta_r[idx], LABEL_MIN_DEG)
    }
  }

  op <- par(xpd = NA)

  for (k in seq_len(nrow(ligand_labels))) {
    if (!keep_lig[k] || !is.finite(theta_lig[k])) next

    ang <- theta_lig[k] * pi / 180
    x <- LABEL_RADIUS * cos(ang)
    y <- LABEL_RADIUS * sin(ang)

    srt <- theta_lig[k]
    adjx <- 0
    if (LABEL_NICEFACING && theta_lig[k] > 90 && theta_lig[k] < 270) {
      srt <- theta_lig[k] + 180
      adjx <- 1
    }

    text(
      x = x, y = y,
      labels = ligand_labels$ligand[k],
      srt = srt,
      adj = c(adjx, 0.5),
      cex = LABEL_CEX
    )
  }

  for (i in seq_len(nrow(axes))) {
    if (keep_r[i] && is.finite(theta_r[i])) {
      ang <- theta_r[i] * pi / 180
      x <- LABEL_RADIUS * cos(ang)
      y <- LABEL_RADIUS * sin(ang)

      srt <- theta_r[i]
      adjx <- 0
      if (LABEL_NICEFACING && theta_r[i] > 90 && theta_r[i] < 270) {
        srt <- theta_r[i] + 180
        adjx <- 1
      }

      text(x = x, y = y, labels = axes$target[i], srt = srt, adj = c(adjx, 0.5), cex = LABEL_CEX)
    }
  }

  par(op)

  legend(
    "topleft",
    legend = ct_names,
    fill = ct_cols[ct_names],
    border = NA,
    cex = 0.75,
    bty = "n",
    title = "Cell types"
  )

  legend(
    "bottomleft",
    legend = c("TIL_high axes", "TIL_low axes"),
    fill = c("#D55E00", "#0072B2"),
    border = NA,
    cex = 0.95,
    bty = "n",
    title = "Axis direction"
  )

  title(
    main = paste0(
      "Selected axes circos | ", zone_dir,
      " | edges=", nrow(axes),
      " | width ~ ligand_activity, color ~ rank_score"
    ),
    cex.main = TITLE_CEX
  )

  circos.clear()
}

# Write PNG once
png(OUT_PNG, width = PNG_SIZE, height = PNG_SIZE, res = PNG_RES, type = "cairo")
plot_one()
dev.off()

OUT_PDF <- sub("\\.png$", ".pdf", OUT_PNG)

pdf(OUT_PDF, width = PNG_SIZE / PNG_RES, height = PNG_SIZE / PNG_RES, useDingbats = FALSE)
plot_one()
dev.off()

cat("Wrote:", OUT_PNG, "\n")
cat("Wrote:", OUT_PDF, "\n")
cat("Edges:", nrow(axes), "\n")
cat("ligand_activity range:", min(axes$ligand_activity), "to", max(axes$ligand_activity), "\n")
cat("rank_score range:", min(axes$rank_score), "to", max(axes$rank_score), "\n")
cat("rank_score clip quantiles:", COLOR_CLIP_LO, "to", COLOR_CLIP_HI, "\n")
cat("Label radius:", LABEL_RADIUS, " Label cex:", LABEL_CEX, " Label min deg:", LABEL_MIN_DEG, "\n")
```

### GSEA of HIGH and LOW BZ1 programs

```python
# %% [markdown]
# ### ULM pathway enrichment scoring

# %%
import numpy as np
import pandas as pd
from pathlib import Path
import requests
import decoupler as dc

# User config
ZONE = "BZ1"  # Options: "TIL+", "BZ1", "BZ2"

# Paths
cache_dir = Path.home() / ".cache" / "msigdb"
cache_dir.mkdir(parents=True, exist_ok=True)

EFS_ROOT = Path("/mnt/custom-file-systems/efs/fs-09913c1f7db79b6fd")
DE_DIR = EFS_ROOT / "results/visium/til_zones/deg_pseudobulk_edger"
OUT_DIR = EFS_ROOT / "results/visium/til_zones/nichenet_filtering"
OUT_DIR.mkdir(parents=True, exist_ok=True)

# Zone-specific DE file
visium_de_file = DE_DIR / f"DE_edger_zone_{ZONE}_TIL_high_vs_TIL_low.csv"

print(f"ZONE: {ZONE}")
print(f"DE file: {visium_de_file.name}")

if not visium_de_file.exists():
    raise FileNotFoundError(f"Missing: {visium_de_file}")

print("Input file found")

# MSigDB GMT download and parse
urls = {
    "hallmark": "https://data.broadinstitute.org/gsea-msigdb/msigdb/release/2023.2.Hs/h.all.v2023.2.Hs.symbols.gmt",
    "kegg":     "https://data.broadinstitute.org/gsea-msigdb/msigdb/release/2023.2.Hs/c2.cp.kegg_medicus.v2023.2.Hs.symbols.gmt",
    "biocarta": "https://data.broadinstitute.org/gsea-msigdb/msigdb/release/2023.2.Hs/c2.cp.biocarta.v2023.2.Hs.symbols.gmt",
}
paths = {k: cache_dir / Path(v).name for k, v in urls.items()}

def download_text(url: str, out_path: Path, timeout: int = 120) -> None:
    if out_path.exists() and out_path.stat().st_size > 0:
        return
    r = requests.get(url, timeout=timeout)
    r.raise_for_status()
    out_path.write_text(r.text)

def parse_gmt(gmt_path: Path) -> dict:
    gene_sets = {}
    with gmt_path.open("r") as f:
        for line in f:
            parts = line.rstrip("\n").split("\t")
            if len(parts) >= 3:
                gene_sets[parts[0]] = [g for g in parts[2:] if g]
    return gene_sets

for k, url in urls.items():
    download_text(url, paths[k])

hallmark_sets = parse_gmt(paths["hallmark"])
kegg_sets = parse_gmt(paths["kegg"])
biocarta_sets = parse_gmt(paths["biocarta"])

def gmt_to_net(gene_sets: dict) -> pd.DataFrame:
    rows = [(gs, g, 1.0) for gs, genes in gene_sets.items() for g in genes]
    return pd.DataFrame(rows, columns=["source", "target", "weight"])

net = pd.concat(
    [gmt_to_net(hallmark_sets), gmt_to_net(kegg_sets), gmt_to_net(biocarta_sets)],
    ignore_index=True
)

# Load full DE results (all genes)
visium_de = pd.read_csv(visium_de_file)
for c in ["gene", "logFC"]:
    if c not in visium_de.columns:
        raise ValueError(f"Missing column '{c}' in {visium_de_file.name}")

visium_de["gene"] = visium_de["gene"].astype(str)
visium_de["logFC"] = pd.to_numeric(visium_de["logFC"], errors="coerce")
visium_de = visium_de.dropna(subset=["logFC"])
visium_de = visium_de.drop_duplicates(subset=["gene"], keep="first")

print(f"Genes in DE results: {len(visium_de)}")

# BH FDR correction
def bh_fdr(pvals: np.ndarray) -> np.ndarray:
    p = np.asarray(pvals, dtype=float)
    q = np.full_like(p, np.nan, dtype=float)
    ok = np.isfinite(p)
    if ok.sum() == 0:
        return q
    p0 = p[ok]
    n = p0.size
    order = np.argsort(p0)
    ranked = p0[order]
    adj = ranked * n / (np.arange(1, n + 1))
    adj = np.minimum.accumulate(adj[::-1])[::-1]
    out = np.empty_like(p0)
    out[order] = np.clip(adj, 0, 1)
    q[ok] = out
    return q

# Run ULM on all genes
# Positive logFC = upregulated in TIL_high
# Negative logFC = upregulated in TIL_low
def run_ulm_all_genes(de_df: pd.DataFrame, label: str) -> pd.DataFrame:
    w = de_df.set_index("gene")["logFC"].astype(float)
    mat = w.to_frame().T
    mat.index = [label]

    est, pvs = dc.run_ulm(
        mat=mat,
        net=net,
        source="source",
        target="target",
        weight="weight",
        min_n=5,
        verbose=False,
    )

    res = pd.DataFrame({
        "gene_set": est.columns.astype(str),
        "score": est.iloc[0].to_numpy(dtype=float),
        "pval": pvs.iloc[0].to_numpy(dtype=float),
    })
    res["fdr"] = bh_fdr(res["pval"].to_numpy())
    res["collection"] = np.select(
        [
            res["gene_set"].str.startswith("HALLMARK_"),
            res["gene_set"].str.startswith("KEGG_"),
            res["gene_set"].str.startswith("BIOCARTA_"),
        ],
        ["Hallmark", "KEGG", "BioCarta"],
        default="Other",
    )
    return res

ulm_results = run_ulm_all_genes(visium_de, label=f"{ZONE}_TIL_high_vs_low")

# Positive score = enriched in TIL_high direction
# Negative score = enriched in TIL_low direction
ulm_high = ulm_results[ulm_results["score"] > 0].sort_values("score", ascending=False).copy()
ulm_low = ulm_results[ulm_results["score"] < 0].sort_values("score", ascending=True).copy()

# Save outputs
zone_file_prefix = ZONE.replace("+", "plus").replace("-", "minus")
out_all = OUT_DIR / f"{zone_file_prefix}_ULM_all_pathways.csv"
out_high = OUT_DIR / f"{zone_file_prefix}_ULM_TIL_high_enriched.csv"
out_low = OUT_DIR / f"{zone_file_prefix}_ULM_TIL_low_enriched.csv"

ulm_results.sort_values("score", ascending=False).to_csv(out_all, index=False)
ulm_high.to_csv(out_high, index=False)
ulm_low.to_csv(out_low, index=False)

print(f"\nGene sets: Hallmark={len(hallmark_sets)}, KEGG={len(kegg_sets)}, BioCarta={len(biocarta_sets)}")
print(f"Network: {net['source'].nunique()} sets, {net['target'].nunique()} genes")
print(f"\nWrote:")
print(f"  {out_all}")
print(f"  {out_high}")
print(f"  {out_low}")

# Display top results
def show_top(res: pd.DataFrame, title: str, n: int = 20) -> None:
    if res.empty:
        print(f"\n{title}: no results")
        return
    top = res.head(n).copy()
    top["score"] = top["score"].round(3)
    top["pval"] = top["pval"].map(lambda x: f"{x:.2e}" if np.isfinite(x) else "nan")
    top["fdr"] = top["fdr"].map(lambda x: f"{x:.2e}" if np.isfinite(x) else "nan")
    print(f"\n{title}")
    print(top[["gene_set", "collection", "score", "pval", "fdr"]].to_string(index=False))

show_top(ulm_high, f"Top pathways enriched in TIL_high ({ZONE})")
show_top(ulm_low, f"Top pathways enriched in TIL_low ({ZONE})")
```

### Define pathway filters

### Zone background gene universe

```python
import numpy as np
import pandas as pd
import scipy.sparse as sp

# Defines these as sorted Python lists of gene symbols for downstream %%R cells:
#   background_high, background_low
#   ligands_high, ligands_low
#   receptors_high, receptors_low
#
# Also keeps a structured dict for inspection:
#   universes_panel

LAYER = "LogNormalize"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
MIN_PCT_EXPRESSED = 0.10

required = ["adata_sc", "nichenet_priors", "SENDERS", "RECEIVERS"]
missing = [x for x in required if x not in globals()]
if missing:
    raise NameError(f"Missing required variables: {missing}")

if LAYER not in adata_sc.layers:
    raise KeyError(
        f"adata_sc.layers['{LAYER}'] not found. Available layers: {list(adata_sc.layers.keys())}"
    )
for k in [CT_KEY, GROUP_KEY]:
    if k not in adata_sc.obs.columns:
        raise KeyError(
            f"adata_sc.obs['{k}'] not found. Available obs columns (first 60): {list(adata_sc.obs.columns)[:60]}"
        )

for k in ["ltm_targets", "ltm_ligands", "lr_ligands", "lr_receptors"]:
    if k not in nichenet_priors:
        raise KeyError(f"nichenet_priors['{k}'] not found. Available keys: {list(nichenet_priors.keys())}")

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

senders_std = set(std_ct(x) for x in SENDERS)
receivers_std = set(std_ct(x) for x in RECEIVERS)

ct = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)

expected_groups = {"TIL_high", "TIL_low"}
groups_present = set(grp.unique())
if not expected_groups.issubset(groups_present):
    raise ValueError(f"Expected {expected_groups} in {GROUP_KEY}, found: {sorted(groups_present)}")

var_names = pd.Index(adata_sc.var_names.astype(str))
var_set = set(var_names)

X = adata_sc.layers[LAYER]
if sp.issparse(X):
    X = X.tocsr()
else:
    X = sp.csr_matrix(X)

ltm_targets = set(map(str, nichenet_priors["ltm_targets"]))
ltm_ligands = set(map(str, nichenet_priors["ltm_ligands"]))
lr_ligands = set(map(str, nichenet_priors["lr_ligands"]))
lr_receptors = set(map(str, nichenet_priors["lr_receptors"]))

candidate_background = ltm_targets
candidate_ligands = ltm_ligands.intersection(lr_ligands)
candidate_receptors = lr_receptors

def expressed_genes(mask_cells: np.ndarray, candidate_genes: set, min_pct: float) -> set:
    idx = np.flatnonzero(mask_cells)
    if idx.size == 0:
        return set()

    genes_present = sorted(candidate_genes.intersection(var_set))
    if not genes_present:
        return set()

    gene_idx = var_names.get_indexer(genes_present)
    Xsub = X[idx, :][:, gene_idx]

    nnz = np.asarray(Xsub.getnnz(axis=0)).ravel()
    pct = nnz / float(idx.size)

    keep = pct >= float(min_pct)
    return set(np.asarray(genes_present, dtype=object)[keep])

universes_panel = {}

for g in ["TIL_high", "TIL_low"]:
    sender_mask = (grp == g).to_numpy() & ct.isin(senders_std).to_numpy()
    receiver_mask = (grp == g).to_numpy() & ct.isin(receivers_std).to_numpy()

    bg = expressed_genes(receiver_mask, candidate_background, MIN_PCT_EXPRESSED)
    lig = expressed_genes(sender_mask, candidate_ligands, MIN_PCT_EXPRESSED)
    rec = expressed_genes(receiver_mask, candidate_receptors, MIN_PCT_EXPRESSED)

    universes_panel[g] = {
        "background": bg,
        "ligands": lig,
        "receptors": rec,
        "n_sender_cells": int(sender_mask.sum()),
        "n_receiver_cells": int(receiver_mask.sum()),
    }

background_high = sorted(map(str, universes_panel["TIL_high"]["background"]))
background_low = sorted(map(str, universes_panel["TIL_low"]["background"]))
ligands_high = sorted(map(str, universes_panel["TIL_high"]["ligands"]))
ligands_low = sorted(map(str, universes_panel["TIL_low"]["ligands"]))
receptors_high = sorted(map(str, universes_panel["TIL_high"]["receptors"]))
receptors_low = sorted(map(str, universes_panel["TIL_low"]["receptors"]))

summary_df = pd.DataFrame(
    [
        {
            "group": g,
            "n_sender_cells": universes_panel[g]["n_sender_cells"],
            "n_receiver_cells": universes_panel[g]["n_receiver_cells"],
            "n_background": len(universes_panel[g]["background"]),
            "n_ligands": len(universes_panel[g]["ligands"]),
            "n_receptors": len(universes_panel[g]["receptors"]),
        }
        for g in ["TIL_high", "TIL_low"]
    ]
)

print(summary_df.to_string(index=False))

print("\nUniverse integrity checks:")
print("  candidate_background (ltm_targets):", len(candidate_background))
print("  candidate_ligands (ltm_ligands ∩ lr_ligands):", len(candidate_ligands))
print("  candidate_receptors (lr_receptors):", len(candidate_receptors))
print("  var genes:", len(var_names))
print("  overlap candidate_ligands with var:", len(candidate_ligands.intersection(var_set)))
print("  overlap candidate_receptors with var:", len(candidate_receptors.intersection(var_set)))

if len(background_high) == 0 or len(background_low) == 0:
    raise ValueError("Background is empty for at least one group. Check receiver panel, group labels, and MIN_PCT_EXPRESSED.")
if len(ligands_high) == 0 or len(ligands_low) == 0:
    raise ValueError("Ligands are empty for at least one group. Check sender panel, group labels, and MIN_PCT_EXPRESSED.")
if len(receptors_high) == 0 or len(receptors_low) == 0:
    raise ValueError("Receptors are empty for at least one group. Check receiver panel, group labels, and MIN_PCT_EXPRESSED.")

print("\nExported lists ready for R:")
print("  background_high:", len(background_high), "background_low:", len(background_low))
print("  ligands_high   :", len(ligands_high), "ligands_low   :", len(ligands_low))
print("  receptors_high :", len(receptors_high), "receptors_low :", len(receptors_low))
```

### Target gene universe

```python
import numpy as np
import pandas as pd
import scipy.sparse as sp

# Builds canonical objects for this session:
#   targets_high, targets_low
#   universes (dict with targets/ligands/background/receptors per direction)
# Keeps de_df for downstream axis construction (Cell 5 expects de_df).

LAYER = "LogNormalize"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
ZONE_KEY = ZONE  # CHANGED: was "TIL+"

MIN_PCT_EXPRESSED = 0.10

required = [
    "adata_sc", "nichenet_priors", "SENDERS", "RECEIVERS",
    "de_tables", "PATHWAY_FILTER", "FDR_CUTOFF", "LOGFC_CUTOFF"
]
missing = [x for x in required if x not in globals()]
if missing:
    raise NameError(f"Missing required objects/params (run upstream cells): {missing}")

if LAYER not in adata_sc.layers:
    raise KeyError(f"adata_sc.layers['{LAYER}'] not found. Available: {list(adata_sc.layers.keys())}")

for k in [CT_KEY, GROUP_KEY]:
    if k not in adata_sc.obs.columns:
        raise KeyError(f"adata_sc.obs['{k}'] not found. Available (first 60): {list(adata_sc.obs.columns)[:60]}")

if ZONE_KEY not in de_tables:
    raise KeyError(f"de_tables['{ZONE_KEY}'] not found. Available keys: {list(de_tables.keys())}")

for k in ["ltm_targets", "ltm_ligands", "lr_ligands", "lr_receptors"]:
    if k not in nichenet_priors:
        raise KeyError(f"nichenet_priors['{k}'] not found. Available keys: {list(nichenet_priors.keys())}")

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

senders_std = set(std_ct(x) for x in SENDERS)
receivers_std = set(std_ct(x) for x in RECEIVERS)

ct = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)

expected_groups = {"TIL_high", "TIL_low"}
groups_present = set(grp.unique())
if not expected_groups.issubset(groups_present):
    raise ValueError(f"Expected {expected_groups} in {GROUP_KEY}, found: {sorted(groups_present)}")

var_names = pd.Index(adata_sc.var_names.astype(str))
var_set = set(var_names)

X = adata_sc.layers[LAYER]
if sp.issparse(X):
    X = X.tocsr()
else:
    X = sp.csr_matrix(X)

ltm_targets = set(map(str, nichenet_priors["ltm_targets"]))
ltm_ligands = set(map(str, nichenet_priors["ltm_ligands"]))
lr_ligands = set(map(str, nichenet_priors["lr_ligands"]))
lr_receptors = set(map(str, nichenet_priors["lr_receptors"]))

candidate_background = ltm_targets
candidate_ligands = ltm_ligands.intersection(lr_ligands)
candidate_receptors = lr_receptors

def expressed_genes(mask_cells: np.ndarray, candidate_genes: set, min_pct: float) -> set:
    idx = np.flatnonzero(mask_cells)
    if idx.size == 0:
        return set()

    genes_present = sorted(candidate_genes.intersection(var_set))
    if not genes_present:
        return set()

    gene_idx = var_names.get_indexer(genes_present)
    Xsub = X[idx, :][:, gene_idx]

    nnz = np.asarray(Xsub.getnnz(axis=0)).ravel()
    pct = nnz / float(idx.size)

    keep = pct >= float(min_pct)
    return set(np.asarray(genes_present, dtype=object)[keep])

panel = {}
for g in ["TIL_high", "TIL_low"]:
    sender_mask = (grp == g).to_numpy() & ct.isin(senders_std).to_numpy()
    receiver_mask = (grp == g).to_numpy() & ct.isin(receivers_std).to_numpy()

    panel[g] = {
        "n_sender_cells": int(sender_mask.sum()),
        "n_receiver_cells": int(receiver_mask.sum()),
        "background": expressed_genes(receiver_mask, candidate_background, MIN_PCT_EXPRESSED),
        "ligands": expressed_genes(sender_mask, candidate_ligands, MIN_PCT_EXPRESSED),
        "receptors": expressed_genes(receiver_mask, candidate_receptors, MIN_PCT_EXPRESSED),
    }

de_df = de_tables[ZONE_KEY].copy()

req = {"gene", "FDR", "logFC"}
if not req.issubset(set(de_df.columns)):
    raise KeyError(f"de_tables['{ZONE_KEY}'] must contain {sorted(req)}. Found: {list(de_df.columns)}")

de_df["gene"] = de_df["gene"].astype(str)
de_df["FDR"] = pd.to_numeric(de_df["FDR"], errors="coerce")
de_df["logFC"] = pd.to_numeric(de_df["logFC"], errors="coerce")

# DE-defined targets by direction (sign anchored to TIL_high_vs_low)
de_high = set(
    de_df.loc[
        (de_df["FDR"] <= float(FDR_CUTOFF)) & (de_df["logFC"] >= float(LOGFC_CUTOFF)),
        "gene"
    ].dropna()
)
de_low = set(
    de_df.loc[
        (de_df["FDR"] <= float(FDR_CUTOFF)) & (de_df["logFC"] <= -float(LOGFC_CUTOFF)),
        "gene"
    ].dropna()
)

# Gate to receiver-expressed background per group
targets_high = de_high.intersection(panel["TIL_high"]["background"])
targets_low = de_low.intersection(panel["TIL_low"]["background"])

targets_high_pre = set(targets_high)
targets_low_pre = set(targets_low)

# Optional pathway filter (apply after background gating)
if PATHWAY_FILTER is not None:
    pf = set(map(str, PATHWAY_FILTER))
    targets_high = targets_high.intersection(pf)
    targets_low = targets_low.intersection(pf)

universes = {
    "targets_high": sorted(map(str, targets_high)),
    "targets_low": sorted(map(str, targets_low)),
    "ligands_high": sorted(map(str, panel["TIL_high"]["ligands"])),
    "ligands_low": sorted(map(str, panel["TIL_low"]["ligands"])),
    "background_high": sorted(map(str, panel["TIL_high"]["background"])),
    "background_low": sorted(map(str, panel["TIL_low"]["background"])),
    "receptors_high": sorted(map(str, panel["TIL_high"]["receptors"])),
    "receptors_low": sorted(map(str, panel["TIL_low"]["receptors"])),
}

summary_df = pd.DataFrame(
    [
        {
            "group": g,
            "n_sender_cells": panel[g]["n_sender_cells"],
            "n_receiver_cells": panel[g]["n_receiver_cells"],
            "n_background": len(panel[g]["background"]),
            "n_ligands": len(panel[g]["ligands"]),
            "n_receptors": len(panel[g]["receptors"]),
        }
        for g in ["TIL_high", "TIL_low"]
    ]
)

print(summary_df.to_string(index=False))

print("\nTargets:")
print("  targets_high (DE∩background) pre pathway:", len(targets_high_pre), "post:", len(targets_high))
print("  targets_low  (DE∩background) pre pathway:", len(targets_low_pre),  "post:", len(targets_low))
print("  pathway filter enabled:", PATHWAY_FILTER is not None)

if PATHWAY_FILTER is not None:
    pf_set = set(map(str, PATHWAY_FILTER))
    print("  targets_high removed by PATHWAY_FILTER:", len(targets_high_pre - pf_set))
    print("  targets_low  removed by PATHWAY_FILTER:", len(targets_low_pre - pf_set))

if len(universes["targets_high"]) == 0 or len(universes["targets_low"]) == 0:
    raise ValueError(
        "Targets empty after DE/background gating and (optional) pathway filter. "
        "Check thresholds, DE orientation, and PATHWAY_FILTER."
    )
```

### Map targets to receiver cell types by expression fraction

```python
# %% [markdown]
# ### Map targets to receiver cell types by expression fraction

# %%
import numpy as np
import pandas as pd
import scipy.sparse as sp

# === CLEAR STALE VARIABLES ===
for _v in ["target_receiver_map_high", "target_receiver_map_low", "target_panel_coverage",
           "global_best_high", "global_best_low", "all_celltypes", "valid_celltypes", 
           "excluded_celltypes", "ct", "grp", "X", "var_names", "var_set"]:
    if _v in globals():
        del globals()[_v]

required = ["adata_sc", "targets_high", "targets_low", "RECEIVERS", "FRACTION_CUTOFF"]
missing = [x for x in required if x not in globals()]
if missing:
    raise NameError(f"Missing required objects/params: {missing}")

# === VERIFY INPUTS ===
print(f"INPUT CHECK:")
print(f"  targets_high: {len(targets_high)} genes")
print(f"  targets_low: {len(targets_low)} genes")
print(f"  RECEIVERS: {len(RECEIVERS)} cell types")
print(f"  FRACTION_CUTOFF: {FRACTION_CUTOFF}")

LAYER = "LogNormalize"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
SAMPLE_KEY = "sample_id"

MIN_SAMPLES_EXPRESSING = 3
PANEL_COVERAGE_THRESHOLD = 0.50

# Minimum cells required for a cell type to be considered in global best computation
# This prevents rare cell types (e.g., 1 cell) from dominating due to noise
MIN_CELLS_FOR_GLOBAL_BEST = 50

if LAYER not in adata_sc.layers:
    raise KeyError(f"adata_sc.layers['{LAYER}'] not found.")
for k in [CT_KEY, GROUP_KEY, SAMPLE_KEY]:
    if k not in adata_sc.obs.columns:
        raise KeyError(f"adata_sc.obs['{k}'] not found.")

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

def is_garbage_ct(ct_name: str) -> bool:
    """Exclude Doublet, Other, Unknown, Unassigned cell types."""
    s = str(ct_name)
    for bad in ["Doublet", "doublet", "Other", "other", "Unknown", "Unassigned"]:
        if bad in s:
            return True
    return False

receivers_std = sorted(set(std_ct(x) for x in RECEIVERS))

# Build fresh from adata_sc
ct = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)
sample_ids = adata_sc.obs[SAMPLE_KEY].astype(str)

X = adata_sc.layers[LAYER]
if sp.issparse(X):
    X = X.tocsr()
else:
    X = sp.csr_matrix(X)

var_names = pd.Index(adata_sc.var_names.astype(str))
var_set = set(var_names)

# Get valid cell types (excluding garbage)
all_celltypes_raw = sorted(ct.unique())
valid_celltypes = [c for c in all_celltypes_raw if not is_garbage_ct(c)]
excluded_celltypes = [c for c in all_celltypes_raw if is_garbage_ct(c)]

print(f"\nCELL TYPE FILTERING:")
print(f"  Raw cell types: {len(all_celltypes_raw)}")
print(f"  Valid cell types (no Doublet/Other): {len(valid_celltypes)}")
print(f"  Excluded: {len(excluded_celltypes)}")

# Compute cell counts per cell type per group
celltype_counts = {}
for group_label in ["TIL_high", "TIL_low"]:
    celltype_counts[group_label] = {}
    for celltype in valid_celltypes:
        m = (grp == group_label).to_numpy() & (ct == celltype).to_numpy()
        n_cells = m.sum()
        celltype_counts[group_label][celltype] = n_cells

# Filter valid cell types for global best by minimum cell count
valid_celltypes_for_global = {}
for group_label in ["TIL_high", "TIL_low"]:
    valid_celltypes_for_global[group_label] = [
        c for c in valid_celltypes 
        if celltype_counts[group_label][c] >= MIN_CELLS_FOR_GLOBAL_BEST
    ]
    n_excluded_by_count = len(valid_celltypes) - len(valid_celltypes_for_global[group_label])
    print(f"  {group_label}: {len(valid_celltypes_for_global[group_label])} cell types with >= {MIN_CELLS_FOR_GLOBAL_BEST} cells ({n_excluded_by_count} excluded)")

def build_trm_for_group(group_label: str, targets_panel) -> pd.DataFrame:
    """Build target-receiver map for panel receivers only."""
    targets_panel = [str(x) for x in list(targets_panel)]
    targets_present = [t for t in targets_panel if t in var_set]
    if len(targets_present) == 0:
        raise ValueError(f"No targets found in var_names for {group_label}.")

    print(f"  {group_label}: {len(targets_present)} targets in var_names")
    
    gene_idx = var_names.get_indexer(targets_present)
    rows = []

    for rct in receivers_std:
        m = (grp == group_label).to_numpy() & (ct == rct).to_numpy()
        idx = np.flatnonzero(m)
        if idx.size == 0:
            continue

        sample_sub = sample_ids.iloc[idx].values
        n_samples = len(np.unique(sample_sub))

        Xsub = X[idx, :][:, gene_idx]
        nnz = np.asarray(Xsub.getnnz(axis=0)).ravel()
        pct = nnz / float(idx.size)
        mean_expr = np.asarray(Xsub.mean(axis=0)).ravel()
        Xsub_dense = Xsub.toarray()

        for i, target in enumerate(targets_present):
            gene_expr = Xsub_dense[:, i]
            n_samples_expressing = len(set(sample_sub[gene_expr > 0]))
            sample_frac = n_samples_expressing / n_samples if n_samples > 0 else 0.0

            if pct[i] >= float(FRACTION_CUTOFF) and n_samples_expressing >= MIN_SAMPLES_EXPRESSING:
                rows.append({
                    "target": target,
                    "receiver_celltype": rct,
                    "target_pct_expr": float(pct[i]),
                    "target_mean_expr": float(mean_expr[i]),
                    "n_samples": n_samples,
                    "n_samples_expressing": n_samples_expressing,
                    "sample_frac": float(sample_frac),
                })

    if not rows:
        return pd.DataFrame(columns=[
            "target", "receiver_celltype", "target_pct_expr", "target_mean_expr",
            "n_samples", "n_samples_expressing", "sample_frac", "specificity", "receiver_rank_score"
        ])

    out = pd.DataFrame(rows)

    # Specificity within panel
    target_mean_all = out.groupby("target")["target_mean_expr"].transform("mean").replace(0, np.nan)
    out["specificity"] = (out["target_mean_expr"] / target_mean_all).fillna(0)

    # Receiver rank score
    out["_mr"] = out.groupby("target")["target_mean_expr"].rank(ascending=False)
    out["_pr"] = out.groupby("target")["target_pct_expr"].rank(ascending=False)
    out["_sfr"] = out.groupby("target")["sample_frac"].rank(ascending=False)
    out["_spr"] = out.groupby("target")["specificity"].rank(ascending=False)
    out["receiver_rank_score"] = 0.35/out["_mr"] + 0.25/out["_pr"] + 0.20/out["_sfr"] + 0.20/out["_spr"]
    out = out.drop(columns=["_mr", "_pr", "_sfr", "_spr"])

    return out

def compute_global_best(group_label: str, targets_list) -> pd.DataFrame:
    """Find global best cell type per target.
    
    Only considers cell types with >= MIN_CELLS_FOR_GLOBAL_BEST cells
    to avoid noise from rare populations.
    """
    targets_list = [str(x) for x in targets_list if str(x) in var_set]
    if not targets_list:
        return pd.DataFrame(columns=["target", "global_best_celltype", "global_best_mean_expr", "global_best_n_cells"])

    gene_idx = var_names.get_indexer(targets_list)
    best = {t: (None, 0.0, 0) for t in targets_list}  # (celltype, mean_expr, n_cells)

    # Only use cell types with sufficient cells
    celltypes_to_check = valid_celltypes_for_global[group_label]
    
    for celltype in celltypes_to_check:
        m = (grp == group_label).to_numpy() & (ct == celltype).to_numpy()
        idx = np.flatnonzero(m)
        if idx.size == 0:
            continue
        n_cells = idx.size
        Xsub = X[idx, :][:, gene_idx]
        mean_expr = np.asarray(Xsub.mean(axis=0)).ravel()
        for i, target in enumerate(targets_list):
            if mean_expr[i] > best[target][1]:
                best[target] = (celltype, float(mean_expr[i]), n_cells)

    results = [{"target": t, "global_best_celltype": v[0], "global_best_mean_expr": v[1], "global_best_n_cells": v[2]} 
               for t, v in best.items() if v[0] is not None]
    return pd.DataFrame(results)

def add_coverage(trm: pd.DataFrame, global_best: pd.DataFrame) -> pd.DataFrame:
    if trm.empty:
        return trm
    trm = trm.merge(global_best, on="target", how="left")
    panel_best = trm.groupby("target")["target_mean_expr"].transform("max")
    trm["panel_coverage_ratio"] = np.where(
        trm["global_best_mean_expr"] > 0,
        panel_best / trm["global_best_mean_expr"],
        1.0
    )
    trm["panel_is_adequate"] = trm["panel_coverage_ratio"] >= PANEL_COVERAGE_THRESHOLD
    trm["coverage_flag"] = np.where(trm["panel_is_adequate"], "OK", "LOW_COVERAGE")
    return trm

# === BUILD TRMs ===
print("\nBUILDING TARGET-RECEIVER MAPS:")
target_receiver_map_high = build_trm_for_group("TIL_high", targets_high)
target_receiver_map_low = build_trm_for_group("TIL_low", targets_low)

# === COMPUTE GLOBAL BEST ===
print(f"\nCOMPUTING GLOBAL BEST (min {MIN_CELLS_FOR_GLOBAL_BEST} cells, excluding Doublets/Other):")
global_best_high = compute_global_best("TIL_high", targets_high)
global_best_low = compute_global_best("TIL_low", targets_low)

# Verify no garbage in global_best
for name, gb in [("high", global_best_high), ("low", global_best_low)]:
    if not gb.empty:
        garbage = [ct for ct in gb["global_best_celltype"].dropna().unique() if is_garbage_ct(ct)]
        if garbage:
            raise RuntimeError(f"BUG: Garbage cell types in global_best_{name}: {garbage}")
        print(f"  global_best_{name}: {len(gb)} targets, no garbage cell types ✓")

# === ADD COVERAGE ===
target_receiver_map_high = add_coverage(target_receiver_map_high, global_best_high)
target_receiver_map_low = add_coverage(target_receiver_map_low, global_best_low)

# === SUMMARY ===
print("\nTARGET-RECEIVER MAP SUMMARY:")
for name, trm in [("TIL_high", target_receiver_map_high), ("TIL_low", target_receiver_map_low)]:
    print(f"\n  {name}:")
    print(f"    rows: {len(trm)}")
    print(f"    targets: {trm['target'].nunique()}")
    print(f"    receivers: {trm['receiver_celltype'].nunique()}")
    if "coverage_flag" in trm.columns:
        n_ok = (trm["coverage_flag"] == "OK").sum()
        n_low = (trm["coverage_flag"] == "LOW_COVERAGE").sum()
        print(f"    coverage: OK={n_ok}, LOW_COVERAGE={n_low}")

# Show LOW_COVERAGE targets
print("\nLOW_COVERAGE TARGETS:")
for name, trm in [("TIL_high", target_receiver_map_high), ("TIL_low", target_receiver_map_low)]:
    low = trm[trm["coverage_flag"] == "LOW_COVERAGE"][["target", "global_best_celltype", "global_best_n_cells", "panel_coverage_ratio"]].drop_duplicates()
    if low.empty:
        print(f"  {name}: None")
    else:
        print(f"  {name}:")
        for _, r in low.iterrows():
            print(f"    {r['target']}: global_best={r['global_best_celltype']} (n={r['global_best_n_cells']}), coverage={r['panel_coverage_ratio']:.1%}")

if target_receiver_map_high.empty or target_receiver_map_low.empty:
    raise ValueError("One or both target_receiver_maps are empty!")

# Store for downstream
target_panel_coverage = pd.concat([
    global_best_high.assign(direction="TIL_high"),
    global_best_low.assign(direction="TIL_low")
], ignore_index=True)

print("\n✓ Created: target_receiver_map_high, target_receiver_map_low, target_panel_coverage")
```

### Build SCE (python)

```python
from pathlib import Path

import numpy as np
import pandas as pd
import scipy.sparse as sp
from scipy.io import mmwrite

if "adata_sc" not in globals():
    raise NameError("adata_sc not found")
if "SENDERS" not in globals() or "RECEIVERS" not in globals():
    raise NameError("SENDERS and/or RECEIVERS not found")

COUNTS_LAYER = "unnormalized_counts"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
SAMPLE_KEY = "sample_id"
BATCH_KEY = "orig.ident"

if COUNTS_LAYER not in adata_sc.layers:
    raise KeyError(f"adata_sc.layers['{COUNTS_LAYER}'] not found. Available: {list(adata_sc.layers.keys())}")

for k in [CT_KEY, GROUP_KEY, SAMPLE_KEY, BATCH_KEY]:
    if k not in adata_sc.obs.columns:
        raise KeyError(f"adata_sc.obs['{k}'] not found. Available obs cols (first 50): {list(adata_sc.obs.columns)[:50]}")

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

def drop_bad_ct(ct_list):
    out = []
    for ct in ct_list:
        s = str(ct)
        if s.startswith("Doublet"):
            continue
        if s.startswith("Other"):
            continue
        out.append(s)
    return sorted(set(out))

senders_py = drop_bad_ct(list(SENDERS))
receivers_py = drop_bad_ct(list(RECEIVERS))

senders_std = [std_ct(x) for x in senders_py]
receivers_std = [std_ct(x) for x in receivers_py]
panel_ct_std = sorted(set(senders_std) | set(receivers_std))

ct_std = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)

m = grp.isin(["TIL_low", "TIL_high"]) & ct_std.isin(panel_ct_std)
n_keep = int(m.sum())
if n_keep == 0:
    raise ValueError("No cells left after filtering to (TIL_low/TIL_high) and sender/receiver panel celltypes.")

adata_sub = adata_sc[m].copy()

X = adata_sub.layers[COUNTS_LAYER]
if not sp.issparse(X):
    X = sp.csr_matrix(X)
else:
    X = X.tocsr()

EFS_ROOT = Path("/mnt/custom-file-systems/efs/fs-09913c1f7db79b6fd")
zone_dir = ZONE.replace("+", "plus").replace("-", "minus")  # CHANGED: derive from ZONE
out_dir = EFS_ROOT / "results" / "visium" / "til_zones" / "nichenet_filtering" / f"mnn_{zone_dir}_input"  # CHANGED: use zone_dir
out_dir.mkdir(parents=True, exist_ok=True)

for fname in ["counts.mtx", "features.tsv", "barcodes.tsv", "metadata.csv", "senders.txt", "receivers.txt"]:
    fpath = out_dir / fname
    if fpath.exists():
        fpath.unlink()

counts_gene_cell = X.T.tocsc()
mmwrite(str(out_dir / "counts.mtx"), counts_gene_cell)

genes = pd.Series(adata_sub.var_names.astype(str))
genes.to_csv(out_dir / "features.tsv", sep="\t", index=False, header=False)

cells = pd.Series(adata_sub.obs_names.astype(str))
cells.to_csv(out_dir / "barcodes.tsv", sep="\t", index=False, header=False)

ct_sub_std = adata_sub.obs[CT_KEY].astype(str).map(std_ct).values

meta = pd.DataFrame(
    {
        "cell": cells.values,
        "celltype": ct_sub_std,
        "sample": adata_sub.obs[SAMPLE_KEY].astype(str).values,
        "group": adata_sub.obs[GROUP_KEY].astype(str).values,
        "batch": adata_sub.obs[BATCH_KEY].astype(str).values,
    }
)
meta.to_csv(out_dir / "metadata.csv", index=False)

pd.Series(senders_std).to_csv(out_dir / "senders.txt", index=False, header=False)
pd.Series(receivers_std).to_csv(out_dir / "receivers.txt", index=False, header=False)

# ADDED: Export for R cell
SCE_INPUT_DIR = f"mnn_{zone_dir}_input"

print("Exported for SCE:")
print("  out_dir:", str(out_dir))
print("  genes x cells:", counts_gene_cell.shape[0], "x", counts_gene_cell.shape[1])
print("  nnz:", int(counts_gene_cell.nnz))
print("  groups:", dict(pd.Series(meta["group"]).value_counts()))
print("  celltypes in panel (senders, receivers):", len(senders_std), ",", len(receivers_std))
```

### Build SCE (R)

```python
%%R -i SCE_INPUT_DIR
suppressPackageStartupMessages({
  library(Matrix)
  library(SingleCellExperiment)
  library(S4Vectors)
  library(dplyr)
  library(readr)
  library(multinichenetr)
})

EFS_ROOT  <- "/mnt/custom-file-systems/efs/fs-09913c1f7db79b6fd"
in_dir <- file.path(EFS_ROOT, "results", "visium", "til_zones", "nichenet_filtering", SCE_INPUT_DIR)  # CHANGED: use variable
stopifnot(dir.exists(in_dir))

counts <- readMM(file.path(in_dir, "counts.mtx"))
counts <- as(counts, "dgCMatrix")

feat <- read_tsv(file.path(in_dir, "features.tsv"), col_names = FALSE, show_col_types = FALSE)
genes <- if (ncol(feat) == 1) feat[[1]] else feat[[2]]

cells <- read_tsv(file.path(in_dir, "barcodes.tsv"), col_names = FALSE, show_col_types = FALSE)[[1]]
meta  <- read_csv(file.path(in_dir, "metadata.csv"), show_col_types = FALSE)

stopifnot(nrow(counts) == length(genes))
stopifnot(ncol(counts) == length(cells))

rownames(counts) <- genes
colnames(counts) <- cells

stopifnot("cell" %in% colnames(meta))
meta <- meta %>% tibble::column_to_rownames("cell")
meta <- meta[colnames(counts), , drop = FALSE]

stopifnot(all(c("celltype", "sample", "group", "batch") %in% colnames(meta)))

meta$celltype_mnn <- make.names(meta$celltype)

sce <- SingleCellExperiment(
  assays = list(counts = counts),
  colData = S4Vectors::DataFrame(meta),
  rowData = S4Vectors::DataFrame(gene = rownames(counts))
)

stopifnot("sample" %in% colnames(colData(sce)))
stopifnot("celltype_mnn" %in% colnames(colData(sce)))
stopifnot("group" %in% colnames(colData(sce)))

colData(sce)$sample_id <- as.character(colData(sce)$sample)
colData(sce)$celltype  <- as.character(colData(sce)$celltype_mnn)

if (anyNA(colData(sce)$sample_id)) stop("sample_id has NA values (check sample in metadata.csv).")
if (anyNA(colData(sce)$celltype))  stop("celltype has NA values (check celltype in metadata.csv).")

senders_oi   <- make.names(readLines(file.path(in_dir, "senders.txt")))
receivers_oi <- make.names(readLines(file.path(in_dir, "receivers.txt")))

ct_levels <- unique(colData(sce)$celltype)
missing_senders   <- setdiff(senders_oi, ct_levels)
missing_receivers <- setdiff(receivers_oi, ct_levels)
if (length(missing_senders) > 0) {
  stop("Senders not found in sce$celltype: ", paste(missing_senders, collapse = ", "))
}
if (length(missing_receivers) > 0) {
  stop("Receivers not found in sce$celltype: ", paste(missing_receivers, collapse = ", "))
}

cat("SCE:", dim(sce)[1], "genes x", dim(sce)[2], "cells\n")
cat("Groups (group):\n")
print(table(colData(sce)$group, useNA = "ifany"))
cat("\nCelltypes (celltype) top 15:\n")
print(sort(table(colData(sce)$celltype), decreasing = TRUE)[1:15])

cat("\nSenders (n=", length(senders_oi), "):\n", sep = "")
print(senders_oi)
cat("\nReceivers (n=", length(receivers_oi), "):\n", sep = "")
print(receivers_oi)

stopifnot(exists("lr_network"), exists("ligand_target_matrix"), exists("weighted_networks"))
cat("\nPriors OK:\n")
cat("  lr_network dim:", paste(dim(lr_network), collapse = " x "), "\n")
cat("  ligand_target_matrix dim:", paste(dim(ligand_target_matrix), collapse = " x "), "\n")
cat("  weighted_networks class:", paste(class(weighted_networks), collapse = ","), "\n")
```

### Compute ligand activity per direction

```python
# Make R-safe copies (rpy2 cannot convert Python sets)
targets_high_r = sorted(list(targets_high))
targets_low_r  = sorted(list(targets_low))

background_high_r = list(background_high)
background_low_r  = list(background_low)
ligands_high_r    = list(ligands_high)
ligands_low_r     = list(ligands_low)

print("R-ready objects:")
print("  targets_high_r:", type(targets_high_r), len(targets_high_r))
print("  targets_low_r :", type(targets_low_r),  len(targets_low_r))
print("  background_high_r:", type(background_high_r), len(background_high_r))
print("  background_low_r :", type(background_low_r),  len(background_low_r))
print("  ligands_high_r:", type(ligands_high_r), len(ligands_high_r))
print("  ligands_low_r :", type(ligands_low_r),  len(ligands_low_r))
```

```python
%%R -i targets_high_r -i targets_low_r -i background_high_r -i background_low_r -i ligands_high_r -i ligands_low_r
suppressPackageStartupMessages({
  library(dplyr)
  library(tibble)
  library(nichenetr)
})

stopifnot(exists("ligand_target_matrix"))

to_chr_unique <- function(x) {
  unique(as.character(unlist(x, use.names = FALSE)))
}

targets_high_chr    <- to_chr_unique(targets_high_r)
targets_low_chr     <- to_chr_unique(targets_low_r)
background_high_chr <- to_chr_unique(background_high_r)
background_low_chr  <- to_chr_unique(background_low_r)
ligands_high_chr    <- to_chr_unique(ligands_high_r)
ligands_low_chr     <- to_chr_unique(ligands_low_r)

ltm <- ligand_target_matrix
ltm_ligands <- rownames(ltm)
ltm_targets <- colnames(ltm)

targets_high_use <- intersect(targets_high_chr, ltm_targets)
targets_low_use  <- intersect(targets_low_chr,  ltm_targets)

bg_high_use <- intersect(background_high_chr, ltm_targets)
bg_low_use  <- intersect(background_low_chr,  ltm_targets)

lig_high_use <- intersect(ligands_high_chr, ltm_ligands)
lig_low_use  <- intersect(ligands_low_chr,  ltm_ligands)

get_top_ltm_ligands <- function(targets_use, ltm, k_per_target) {
  out <- character()
  if (length(targets_use) == 0) return(out)

  for (t in targets_use) {
    w <- ltm[, t, drop = TRUE]
    w <- w[w > 0]
    if (length(w) == 0) next
    top <- sort(w, decreasing = TRUE)
    if (length(top) > k_per_target) top <- top[seq_len(k_per_target)]
    out <- c(out, names(top))
  }
  unique(out)
}

LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY <- 10

top_ltm_high <- get_top_ltm_ligands(targets_high_use, ltm, LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY)
top_ltm_low  <- get_top_ltm_ligands(targets_low_use,  ltm, LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY)

potential_high <- union(lig_high_use, top_ltm_high)
potential_low  <- union(lig_low_use,  top_ltm_low)

cat("Ligand activity inputs after LTM intersection + LTM augmentation:\n")
cat("  LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY=", LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY, "\n", sep = "")
cat("  TIL_high: targets=", length(targets_high_use),
    " background=", length(bg_high_use),
    " sender_panel_ligs=", length(lig_high_use),
    " ltm_top_ligs=", length(top_ltm_high),
    " potential_ligs=", length(potential_high), "\n", sep = "")
cat("  TIL_low : targets=", length(targets_low_use),
    " background=", length(bg_low_use),
    " sender_panel_ligs=", length(lig_low_use),
    " ltm_top_ligs=", length(top_ltm_low),
    " potential_ligs=", length(potential_low), "\n\n", sep = "")

if (length(targets_high_use) < 5) stop("Too few TIL_high targets after intersecting with ligand_target_matrix columns.")
if (length(targets_low_use)  < 5) stop("Too few TIL_low targets after intersecting with ligand_target_matrix columns.")
if (length(bg_high_use) < 50) stop("Too few TIL_high background genes after intersecting with ligand_target_matrix columns.")
if (length(bg_low_use)  < 50) stop("Too few TIL_low background genes after intersecting with ligand_target_matrix columns.")
if (length(potential_high) < 20) stop("Too few TIL_high potential ligands after augmentation.")
if (length(potential_low)  < 20) stop("Too few TIL_low potential ligands after augmentation.")

ligand_activities_high <- predict_ligand_activities(
  geneset = targets_high_use,
  background_expressed_genes = bg_high_use,
  potential_ligands = potential_high,
  ligand_target_matrix = ltm
) %>%
  as_tibble() %>%
  mutate(
    direction = "TIL_high",
    test_ligand = as.character(test_ligand),
    aupr_corrected = as.numeric(aupr_corrected),
    in_sender_panel = test_ligand %in% lig_high_use,
    in_ltm_top = test_ligand %in% top_ltm_high
  ) %>%
  arrange(desc(aupr_corrected), test_ligand) %>%
  mutate(rank = dplyr::row_number()) %>%
  select(test_ligand, direction, aupr_corrected, rank, in_sender_panel, in_ltm_top, everything())

ligand_activities_low <- predict_ligand_activities(
  geneset = targets_low_use,
  background_expressed_genes = bg_low_use,
  potential_ligands = potential_low,
  ligand_target_matrix = ltm
) %>%
  as_tibble() %>%
  mutate(
    direction = "TIL_low",
    test_ligand = as.character(test_ligand),
    aupr_corrected = as.numeric(aupr_corrected),
    in_sender_panel = test_ligand %in% lig_low_use,
    in_ltm_top = test_ligand %in% top_ltm_low
  ) %>%
  arrange(desc(aupr_corrected), test_ligand) %>%
  mutate(rank = dplyr::row_number()) %>%
  select(test_ligand, direction, aupr_corrected, rank, in_sender_panel, in_ltm_top, everything())

ligand_activities <- bind_rows(ligand_activities_high, ligand_activities_low)

cat("Ligand activities computed:\n")
cat("  rows:", nrow(ligand_activities), "\n")
cat("  TIL_high:", nrow(ligand_activities_high), "\n")
cat("  TIL_low :", nrow(ligand_activities_low), "\n\n")

cat("Top 10 (TIL_high):\n")
print(ligand_activities_high %>% slice_head(n = 10) %>% select(test_ligand, aupr_corrected, rank, in_sender_panel, in_ltm_top))
cat("\nTop 10 (TIL_low):\n")
print(ligand_activities_low %>% slice_head(n = 10) %>% select(test_ligand, aupr_corrected, rank, in_sender_panel, in_ltm_top))
```

### Ligand-receptor DE per cell type

```python
# Pass zone prefix to R for file naming
zone_prefix = ZONE.replace("+", "plus").replace("-", "minus")
print(f"Zone prefix for R: {zone_prefix}")
```

```python
%%R -i zone_prefix
# LR-DE: edgeR pseudobulk per celltype, TIL_high vs TIL_low
suppressPackageStartupMessages({
  library(Matrix)
  library(SingleCellExperiment)
  library(SummarizedExperiment)
  library(S4Vectors)
  library(dplyr)
  library(tibble)
  library(readr)
  library(edgeR)
})

stopifnot(exists("sce"), exists("senders_oi"), exists("receivers_oi"))
stopifnot(exists("lr_network"), exists("ligand_target_matrix"))
stopifnot(exists("ligand_activities"))
stopifnot("counts" %in% SummarizedExperiment::assayNames(sce))

EFS_ROOT  <- "/mnt/custom-file-systems/efs/fs-09913c1f7db79b6fd"
FILTER_DIR <- file.path(EFS_ROOT, "results", "visium", "til_zones", "nichenet_filtering")
dir.create(FILTER_DIR, recursive = TRUE, showWarnings = FALSE)

GROUPS <- c("TIL_low","TIL_high")

MIN_CELLS_PER_SAMPLE  <- 5
MIN_SAMPLES_PER_GROUP <- 3

CT_COL     <- "celltype"
GROUP_COL  <- "group"
SAMPLE_COL <- "sample_id"

cd0 <- as.data.frame(colData(sce))
stopifnot(CT_COL %in% colnames(cd0), GROUP_COL %in% colnames(cd0), SAMPLE_COL %in% colnames(cd0))

cd0[[CT_COL]]     <- as.character(cd0[[CT_COL]])
cd0[[GROUP_COL]]  <- as.character(cd0[[GROUP_COL]])
cd0[[SAMPLE_COL]] <- as.character(cd0[[SAMPLE_COL]])

bad_sample <- is.na(cd0[[SAMPLE_COL]]) | cd0[[SAMPLE_COL]] == ""
if (any(bad_sample)) {
  ex <- unique(cd0[[SAMPLE_COL]][bad_sample])[1:min(10, sum(bad_sample))]
  stop(paste0("sample_id contains NA/empty values (n=", sum(bad_sample), "). Examples: ", paste(ex, collapse=", ")))
}

bad_group <- is.na(cd0[[GROUP_COL]]) | !(cd0[[GROUP_COL]] %in% GROUPS)
if (any(bad_group)) {
  ex <- unique(cd0[[GROUP_COL]][bad_group])[1:min(10, sum(bad_group))]
  stop(paste0("group has NA or unexpected labels (n=", sum(bad_group), "). Examples: ", paste(ex, collapse=", ")))
}

cat("Using columns:\n")
cat("  celltype:", CT_COL, "\n")
cat("  group   :", GROUP_COL, "\n")
cat("  sample  :", SAMPLE_COL, "\n")
cat("Cells:", nrow(cd0), "\n\n")

cat("Group counts in sce:\n")
print(table(cd0[[GROUP_COL]], useNA = "ifany"))
cat("\n")

assays_avail <- assayNames(sce)
if (!("counts" %in% assays_avail)) {
  stop(paste0("sce is missing assay 'counts'. Available assays: ", paste(assays_avail, collapse=", ")))
}
counts_all <- assay(sce, "counts")
if (!is(counts_all, "dgCMatrix")) counts_all <- as(counts_all, "CsparseMatrix")
counts_all <- as(counts_all, "dgCMatrix")

cat("Counts matrix:\n")
cat("  class:", class(counts_all)[1], "\n")
cat("  dim  :", nrow(counts_all), "x", ncol(counts_all), "\n\n")

cat("Ligand activities directions:\n")
print(table(as.character(ligand_activities$direction), useNA="ifany"))
cat("\n")

# Harmonize lr_network columns
lr_cols <- colnames(lr_network)
lig_col <- if ("from" %in% lr_cols) "from" else if ("ligand" %in% lr_cols) "ligand" else stop("lr_network missing ligand column (from/ligand).")
rec_col <- if ("to" %in% lr_cols) "to" else if ("receptor" %in% lr_cols) "receptor" else stop("lr_network missing receptor column (to/receptor).")

lr_tbl <- lr_network %>%
  dplyr::distinct(.data[[lig_col]], .data[[rec_col]])
colnames(lr_tbl) <- c("from","to")

# Feasible ligands: use NicheNet activity list, constrained to LTM rows + sce genes
ltm <- ligand_target_matrix
ltm_ligands <- rownames(ltm)
ltm_targets <- colnames(ltm)

feasible_ligands <- intersect(
  unique(as.character(ligand_activities$test_ligand)),
  intersect(ltm_ligands, rownames(counts_all))
)

# Feasible receptors: receptors paired to feasible ligands, constrained to sce genes
feasible_receptors <- intersect(
  unique(lr_tbl$to[lr_tbl$from %in% feasible_ligands]),
  rownames(counts_all)
)

cat("DE universes:\n")
cat("  feasible ligands   :", length(feasible_ligands), "\n")
cat("  feasible receptors :", length(feasible_receptors), "\n\n")

if (length(feasible_ligands) < 20) stop("Too few feasible ligands after intersecting activity ligands with LTM rows and sce genes.")
if (length(feasible_receptors) < 20) stop("Too few feasible receptors after intersecting LR receptors with sce genes.")

# Fast pseudobulk using sparse model matrix
pseudobulk_by_sample <- function(counts_gene_cell, sample_vec) {
  sample_f <- factor(sample_vec)
  M <- Matrix::sparse.model.matrix(~0 + sample_f)
  colnames(M) <- levels(sample_f)
  pb <- counts_gene_cell %*% M
  as(pb, "dgCMatrix")
}

run_edger_for_celltype <- function(celltype_name, genes_keep, mode_label) {
  tryCatch({

    idx_ct <- which(cd0[[CT_COL]] == celltype_name & cd0[[GROUP_COL]] %in% GROUPS)
    if (length(idx_ct) == 0) {
      return(list(status="skip", reason="no cells after filtering", de=NULL, meta=NULL))
    }

    meta_ct <- cd0[idx_ct, , drop=FALSE]

    cells_per_sample <- meta_ct %>%
      dplyr::count(.data[[SAMPLE_COL]], name="n_cells")

    keep_samples <- cells_per_sample %>%
      dplyr::filter(n_cells >= MIN_CELLS_PER_SAMPLE) %>%
      dplyr::pull(.data[[SAMPLE_COL]])

    if (length(keep_samples) == 0) {
      return(list(status="skip", reason=paste0("no samples with >= ", MIN_CELLS_PER_SAMPLE, " cells"), de=NULL, meta=cells_per_sample))
    }

    keep_mask <- meta_ct[[SAMPLE_COL]] %in% keep_samples
    meta_ct <- meta_ct[keep_mask, , drop=FALSE]
    idx_ct  <- idx_ct[keep_mask]

    sample_group_tbl <- meta_ct %>%
      dplyr::group_by(.data[[SAMPLE_COL]]) %>%
      dplyr::summarise(
        n_cells = dplyr::n(),
        n_groups = dplyr::n_distinct(.data[[GROUP_COL]]),
        group = dplyr::first(.data[[GROUP_COL]]),
        .groups="drop"
      )

    bad <- sample_group_tbl %>% dplyr::filter(n_groups > 1)
    if (nrow(bad) > 0) {
      return(list(status="skip", reason="sample maps to >1 group within this celltype", de=NULL, meta=sample_group_tbl))
    }

    n_by_group <- sample_group_tbl %>% dplyr::count(group, name="n_samples")
    n_low  <- n_by_group %>% dplyr::filter(group=="TIL_low")  %>% dplyr::pull(n_samples)
    n_high <- n_by_group %>% dplyr::filter(group=="TIL_high") %>% dplyr::pull(n_samples)
    n_low  <- ifelse(length(n_low)==0, 0, n_low)
    n_high <- ifelse(length(n_high)==0, 0, n_high)

    if (n_low < MIN_SAMPLES_PER_GROUP || n_high < MIN_SAMPLES_PER_GROUP) {
      return(list(
        status="skip",
        reason=paste0("too few samples per group (low=", n_low, ", high=", n_high, ")"),
        de=NULL,
        meta=sample_group_tbl
      ))
    }

    genes_keep <- intersect(as.character(genes_keep), rownames(counts_all))
    if (length(genes_keep) < 10) {
      return(list(status="skip", reason="too few genes in universe after intersecting with sce genes", de=NULL, meta=sample_group_tbl))
    }

    counts_sub <- counts_all[genes_keep, idx_ct, drop=FALSE]
    pb <- pseudobulk_by_sample(counts_sub, meta_ct[[SAMPLE_COL]])

    group_vec <- sample_group_tbl$group
    names(group_vec) <- sample_group_tbl[[SAMPLE_COL]]
    group_vec <- group_vec[colnames(pb)]

    if (any(is.na(group_vec))) {
      na_samples <- colnames(pb)[is.na(group_vec)]
      return(list(
        status="error",
        reason=paste0("group_vec has NA after aligning to pseudobulk columns. Examples: ", paste(head(na_samples, 10), collapse=", ")),
        de=NULL,
        meta=sample_group_tbl
      ))
    }

    group_vec <- factor(group_vec, levels=GROUPS)

    y <- DGEList(counts = as.matrix(pb))
    design <- model.matrix(~ group_vec)

    keep <- filterByExpr(y, design=design)
    y <- y[keep, , keep.lib.sizes=FALSE]
    y <- calcNormFactors(y)

    y <- estimateDisp(y, design, robust=TRUE)
    fit <- glmQLFit(y, design, robust=TRUE)
    qlf <- glmQLFTest(fit, coef=2)

    tt <- topTags(qlf, n=Inf)$table %>%
      tibble::rownames_to_column("gene") %>%
      dplyr::as_tibble() %>%
      dplyr::transmute(
        gene,
        logFC,
        logCPM,
        F,
        PValue,
        FDR,
        celltype = celltype_name,
        mode = mode_label,
        n_samples_total = ncol(pb),
        n_TIL_low = sum(group_vec=="TIL_low"),
        n_TIL_high = sum(group_vec=="TIL_high")
      )

    meta_out <- sample_group_tbl %>%
      dplyr::rename(sample = .data[[SAMPLE_COL]]) %>%
      dplyr::mutate(celltype = celltype_name, mode = mode_label)

    list(status="ok", reason="", de=tt, meta=meta_out)

  }, error = function(e) {
    list(status="error", reason=paste0("edgeR failed: ", conditionMessage(e)), de=NULL, meta=NULL)
  })
}

run_panel <- function(celltypes_vec, genes_keep, mode_label) {
  de_list <- list()
  meta_list <- list()
  status_rows <- list()

  for (ct in celltypes_vec) {
    res <- run_edger_for_celltype(ct, genes_keep, mode_label)
    status_rows[[ct]] <- tibble(
      mode = mode_label,
      celltype = ct,
      status = res$status,
      reason = res$reason
    )
    if (identical(res$status, "ok")) {
      de_list[[ct]] <- res$de
      meta_list[[ct]] <- res$meta
    } else {
      if (!is.null(res$meta)) meta_list[[ct]] <- res$meta
    }
  }

  list(
    status = dplyr::bind_rows(status_rows),
    de = if (length(de_list)) dplyr::bind_rows(de_list) else tibble(),
    meta = if (length(meta_list)) dplyr::bind_rows(meta_list) else tibble()
  )
}

# Run
sender_res   <- run_panel(senders_oi,   feasible_ligands,   "sender_ligand")
receiver_res <- run_panel(receivers_oi, feasible_receptors, "receiver_receptor")

# CHANGED: Use zone_prefix for output file names
out_sender_de     <- file.path(FILTER_DIR, paste0(zone_prefix, "_sender_ligand_DE_edger_by_celltype_TILhigh_vs_TILlow_activityLigs_LTMrows.csv"))
out_receiver_de   <- file.path(FILTER_DIR, paste0(zone_prefix, "_receiver_receptor_DE_edger_by_celltype_TILhigh_vs_TILlow_LRreceptorsOfActivityLigs.csv"))
out_sender_meta   <- file.path(FILTER_DIR, paste0(zone_prefix, "_sender_pseudobulk_samples_by_celltype.csv"))
out_receiver_meta <- file.path(FILTER_DIR, paste0(zone_prefix, "_receiver_pseudobulk_samples_by_celltype.csv"))
out_status        <- file.path(FILTER_DIR, paste0(zone_prefix, "_LR_DE_status_by_celltype.csv"))

readr::write_csv(sender_res$de, out_sender_de)
readr::write_csv(receiver_res$de, out_receiver_de)
readr::write_csv(sender_res$meta, out_sender_meta)
readr::write_csv(receiver_res$meta, out_receiver_meta)
readr::write_csv(dplyr::bind_rows(sender_res$status, receiver_res$status), out_status)

cat("Wrote:\n")
cat(" ", out_sender_de, "\n")
cat(" ", out_receiver_de, "\n")
cat(" ", out_sender_meta, "\n")
cat(" ", out_receiver_meta, "\n")
cat(" ", out_status, "\n\n")

cat("Celltypes with OK DE (senders):", sum(sender_res$status$status=="ok"), "/", length(senders_oi), "\n")
cat("Celltypes with OK DE (receivers):", sum(receiver_res$status$status=="ok"), "/", length(receivers_oi), "\n\n")

if (any(sender_res$status$status=="error") || any(receiver_res$status$status=="error")) {
  cat("Errors detected in some celltypes:\n")
  print(dplyr::bind_rows(sender_res$status, receiver_res$status) %>% dplyr::filter(status=="error"))
  cat("\n")
}

if (nrow(sender_res$de) > 0) {
  cat("Top 10 ligand DE rows overall (by FDR):\n")
  print(sender_res$de %>% dplyr::arrange(FDR) %>% dplyr::slice_head(n=10))
  cat("\n")
}
if (nrow(receiver_res$de) > 0) {
  cat("Top 10 receptor DE rows overall (by FDR):\n")
  print(receiver_res$de %>% dplyr::arrange(FDR) %>% dplyr::slice_head(n=10))
  cat("\n")
}
```

### Construct axis scores

```python
# %% [markdown]
# ### Construct axis scores

# %%
import numpy as np
import pandas as pd
import scipy.sparse as sp
from pathlib import Path

# === CLEAR STALE VARIABLES ===
for _v in ["all_axes", "top_axes", "sender_candidates", "ligand_sender_gating", 
           "base", "axes", "trm_high", "trm_low", "ltm_df", "act_by_dir",
           "allowed_ligands_by_dir", "lr_use_by_dir", "receptors_by_dir"]:
    if _v in globals():
        del globals()[_v]

# === VERIFY INPUTS ===
required = ["target_receiver_map_high", "target_receiver_map_low", "SENDERS", "RECEIVERS",
            "ligand_activities", "sender_de", "receiver_de", "lr_network", 
            "ligand_target_matrix", "de_df", "OUT_DIR",
            "TOP_LIGANDS_PER_DIRECTION", "LTM_TOPK_PER_TARGET", "TOP_AXES_PER_TARGET",
            "TOP_LIGANDS_PER_TARGET", "TOP_AXES_PER_LIGAND_PER_TARGET", "SENDER_TOPK_PER_LIGAND"]
missing = [x for x in required if x not in globals()]
if missing:
    raise NameError(f"Missing: {missing}")

print("INPUT CHECK:")
print(f"  target_receiver_map_high: {len(target_receiver_map_high)} rows, {target_receiver_map_high['target'].nunique()} targets")
print(f"  target_receiver_map_low: {len(target_receiver_map_low)} rows, {target_receiver_map_low['target'].nunique()} targets")

LAYER = "LogNormalize"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
SAMPLE_KEY = "sample_id"

MIN_PCT_EXPRESSED_LIGAND = 0.10
MIN_SAMPLES_EXPRESSING_LIGAND = 3

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

senders_std = sorted({std_ct(x) for x in SENDERS})
receivers_std = sorted({std_ct(x) for x in RECEIVERS})

TOP_LIGANDS_PER_DIRECTION = int(TOP_LIGANDS_PER_DIRECTION)
LTM_TOPK_PER_TARGET = int(LTM_TOPK_PER_TARGET)
TOP_AXES_PER_TARGET = int(TOP_AXES_PER_TARGET)
TOP_LIGANDS_PER_TARGET = int(TOP_LIGANDS_PER_TARGET)
TOP_AXES_PER_LIGAND_PER_TARGET = int(TOP_AXES_PER_LIGAND_PER_TARGET)
SENDER_TOPK_PER_LIGAND = int(SENDER_TOPK_PER_LIGAND) if SENDER_TOPK_PER_LIGAND is not None else None

LIGANDS_FROM_LTM_PER_TARGET = 10
KEEP_LIGANDS = []

OUT_DIR = Path(OUT_DIR)
OUT_DIR.mkdir(parents=True, exist_ok=True)

print(f"\nSELECTION CONTROLS:")
print(f"  TOP_LIGANDS_PER_DIRECTION: {TOP_LIGANDS_PER_DIRECTION}")
print(f"  LTM_TOPK_PER_TARGET: {LTM_TOPK_PER_TARGET}")
print(f"  TOP_AXES_PER_TARGET: {TOP_AXES_PER_TARGET}")
print(f"  TOP_LIGANDS_PER_TARGET: {TOP_LIGANDS_PER_TARGET}")
print(f"  TOP_AXES_PER_LIGAND_PER_TARGET: {TOP_AXES_PER_LIGAND_PER_TARGET}")
print(f"  SENDER_TOPK_PER_LIGAND: {SENDER_TOPK_PER_LIGAND}")

# === COPY INPUTS ===
ligand_activities = pd.DataFrame(ligand_activities).copy()
sender_de = pd.DataFrame(sender_de).copy()
receiver_de = pd.DataFrame(receiver_de).copy()

ligand_activities["test_ligand"] = ligand_activities["test_ligand"].astype(str)
ligand_activities["direction"] = ligand_activities["direction"].astype(str)
ligand_activities["rank"] = pd.to_numeric(ligand_activities["rank"], errors="coerce")
ligand_activities["aupr_corrected"] = pd.to_numeric(ligand_activities["aupr_corrected"], errors="coerce")

for df_ in (sender_de, receiver_de):
    df_["gene"] = df_["gene"].astype(str)
    df_["celltype"] = df_["celltype"].astype(str)
    df_["logFC"] = pd.to_numeric(df_["logFC"], errors="coerce")
    df_["FDR"] = pd.to_numeric(df_["FDR"], errors="coerce")

lr_df0 = pd.DataFrame(lr_network).copy()
cols = set(lr_df0.columns.astype(str))
if {"from", "to"}.issubset(cols):
    lr_df = lr_df0.rename(columns={"from": "ligand", "to": "receptor"})[["ligand", "receptor"]].copy()
elif {"ligand", "receptor"}.issubset(cols):
    lr_df = lr_df0[["ligand", "receptor"]].copy()
else:
    raise KeyError(f"lr_network columns not recognized: {list(lr_df0.columns)}")
lr_df["ligand"] = lr_df["ligand"].astype(str)
lr_df["receptor"] = lr_df["receptor"].astype(str)
lr_df = lr_df.drop_duplicates()

# Fresh copy of TRMs
trm_high = pd.DataFrame(target_receiver_map_high).copy()
trm_low = pd.DataFrame(target_receiver_map_low).copy()

for trm in (trm_high, trm_low):
    trm["target"] = trm["target"].astype(str)
    trm["receiver_celltype"] = trm["receiver_celltype"].astype(str)
    for col in ["target_pct_expr", "target_mean_expr", "receiver_rank_score", "panel_coverage_ratio"]:
        if col in trm.columns:
            trm[col] = pd.to_numeric(trm[col], errors="coerce")

ltm_df_full = pd.DataFrame(
    ligand_target_matrix.values,
    index=pd.Index(ligand_target_matrix.index.astype(str), name="ligand"),
    columns=pd.Index(ligand_target_matrix.columns.astype(str), name="target"),
)

def get_ltm_ligands_for_targets(targets, ltm_df, k):
    targets = [t for t in targets if t in ltm_df.columns]
    ligs = set()
    for t in targets:
        w = ltm_df[t]
        w = w[w > 0].nlargest(k)
        ligs.update(w.index.astype(str))
    return ligs

# === ACTIVITY MAPS ===
act_by_dir = {}
for direction in ["TIL_high", "TIL_low"]:
    act = ligand_activities[ligand_activities["direction"] == direction].dropna(subset=["rank"])
    act = act.sort_values("rank")
    act_by_dir[direction] = {
        "top_ligs": act.nsmallest(TOP_LIGANDS_PER_DIRECTION, "rank")["test_ligand"].tolist(),
        "activity_map": act.set_index("test_ligand")["aupr_corrected"].to_dict(),
        "rank_map": act.set_index("test_ligand")["rank"].to_dict(),
    }

# === LIGAND UNIVERSE ===
allowed_ligands_by_dir = {}
for direction, trm in [("TIL_high", trm_high), ("TIL_low", trm_low)]:
    top_act = set(act_by_dir[direction]["top_ligs"])
    dir_targets = sorted(set(trm["target"]) & set(ltm_df_full.columns))
    top_ltm = get_ltm_ligands_for_targets(dir_targets, ltm_df_full, LIGANDS_FROM_LTM_PER_TARGET)
    allowed_ligands_by_dir[direction] = top_act | top_ltm | set(KEEP_LIGANDS)
    print(f"{direction} ligand universe: {len(allowed_ligands_by_dir[direction])}")

need_ligs = sorted(allowed_ligands_by_dir["TIL_high"] | allowed_ligands_by_dir["TIL_low"])
need_tgts = sorted(set(trm_high["target"]) | set(trm_low["target"]))
use_ligs = sorted(set(ltm_df_full.index) & set(need_ligs))
use_tgts = sorted(set(ltm_df_full.columns) & set(need_tgts))
ltm_df = ltm_df_full.loc[use_ligs, use_tgts].copy()

print(f"LTM subset: {len(use_ligs)} ligands x {len(use_tgts)} targets")

# === SENDER GATING ===
ct_std = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)
sample_ids = adata_sc.obs[SAMPLE_KEY].astype(str)

X = adata_sc.layers[LAYER]
if sp.issparse(X):
    X = X.tocsr()
else:
    X = sp.csr_matrix(X)

var_names = pd.Index(adata_sc.var_names.astype(str))
var_set = set(var_names)
ligs_in_var = sorted(set(use_ligs) & var_set)

sender_rows = []
for direction in ["TIL_high", "TIL_low"]:
    dir_ligs = sorted(allowed_ligands_by_dir[direction] & set(ligs_in_var))
    if not dir_ligs:
        continue
    lig_idx = var_names.get_indexer(dir_ligs)

    for sct in senders_std:
        m = (grp == direction).to_numpy() & (ct_std == sct).to_numpy()
        idx = np.flatnonzero(m)
        if idx.size == 0:
            continue

        sample_sub = sample_ids.iloc[idx].values
        n_samples = len(np.unique(sample_sub))
        Xsub = X[idx, :][:, lig_idx]
        pct = np.asarray(Xsub.getnnz(axis=0)).ravel() / idx.size
        mean_expr = np.asarray(Xsub.mean(axis=0)).ravel()
        Xsub_dense = Xsub.toarray()

        for i, lig in enumerate(dir_ligs):
            n_samp_expr = len(set(sample_sub[Xsub_dense[:, i] > 0]))
            if pct[i] >= MIN_PCT_EXPRESSED_LIGAND and n_samp_expr >= MIN_SAMPLES_EXPRESSING_LIGAND:
                sender_rows.append({
                    "direction": direction, "ligand": lig, "sender_celltype": sct,
                    "pct_expr": float(pct[i]), "mean_expr": float(mean_expr[i]),
                    "n_samples_expressing": n_samp_expr,
                    "sample_frac": n_samp_expr / n_samples if n_samples > 0 else 0,
                })

sender_gating = pd.DataFrame(sender_rows)

# Sender specificity and rank score
for direction in ["TIL_high", "TIL_low"]:
    mask = sender_gating["direction"] == direction
    if mask.sum() == 0:
        continue
    lig_mean = sender_gating.loc[mask].groupby("ligand")["mean_expr"].transform("mean").replace(0, np.nan)
    sender_gating.loc[mask, "specificity"] = (sender_gating.loc[mask, "mean_expr"] / lig_mean).fillna(0)

sender_gating["_mr"] = sender_gating.groupby(["direction", "ligand"])["mean_expr"].rank(ascending=False)
sender_gating["_pr"] = sender_gating.groupby(["direction", "ligand"])["pct_expr"].rank(ascending=False)
sender_gating["_sfr"] = sender_gating.groupby(["direction", "ligand"])["sample_frac"].rank(ascending=False)
sender_gating["_spr"] = sender_gating.groupby(["direction", "ligand"])["specificity"].rank(ascending=False)
sender_gating["sender_rank_score"] = 0.35/sender_gating["_mr"] + 0.25/sender_gating["_pr"] + 0.20/sender_gating["_sfr"] + 0.20/sender_gating["_spr"]
sender_gating = sender_gating.drop(columns=["_mr", "_pr", "_sfr", "_spr"])

# Merge sender DE
sender_de_use = sender_de.rename(columns={"gene": "ligand", "celltype": "sender_celltype", "logFC": "ligand_logFC", "FDR": "ligand_FDR"})
sender_candidates = sender_gating.merge(sender_de_use[["ligand", "sender_celltype", "ligand_logFC", "ligand_FDR"]], on=["ligand", "sender_celltype"], how="left")

# Concordance bonus
exp_sign = np.where(sender_candidates["direction"] == "TIL_high", 1, -1)
conc = sender_candidates["ligand_logFC"].notna() & ((exp_sign * sender_candidates["ligand_logFC"]) > 0)
sender_candidates.loc[conc, "sender_rank_score"] += 0.05

sender_candidates = sender_candidates.sort_values(["direction", "ligand", "sender_rank_score"], ascending=[True, True, False])
if SENDER_TOPK_PER_LIGAND is not None:
    sender_candidates = sender_candidates.groupby(["direction", "ligand"], as_index=False).head(SENDER_TOPK_PER_LIGAND)

print(f"Sender candidates: {len(sender_candidates)} rows")

# === LR BY DIRECTION ===
lr_by_dir = {}
for d in ["TIL_high", "TIL_low"]:
    lr_by_dir[d] = lr_df[lr_df["ligand"].isin(allowed_ligands_by_dir[d])].copy()
    lr_by_dir[d]["direction"] = d

# === LTM TOP-K ===
def ltm_topk(direction, trm):
    tgts = sorted(set(trm["target"]) & set(ltm_df.columns))
    elig_ligs = sorted(set(lr_by_dir[direction]["ligand"]) & set(ltm_df.index))
    if not tgts or not elig_ligs:
        return pd.DataFrame(columns=["direction", "target", "ligand", "ltm_weight"])
    sub = ltm_df.loc[elig_ligs, tgts]
    rows = []
    for t in tgts:
        w = sub[t]
        w = w[w > 0].nlargest(LTM_TOPK_PER_TARGET)
        for lig, wt in w.items():
            rows.append({"direction": direction, "target": t, "ligand": str(lig), "ltm_weight": float(wt)})
    return pd.DataFrame(rows)

topk_high = ltm_topk("TIL_high", trm_high)
topk_low = ltm_topk("TIL_low", trm_low)

# === MERGE TRM ===
trm_cols = ["target", "receiver_celltype", "target_pct_expr", "target_mean_expr", "receiver_rank_score",
            "global_best_celltype", "panel_coverage_ratio", "coverage_flag"]
trm_cols_high = [c for c in trm_cols if c in trm_high.columns]
trm_cols_low = [c for c in trm_cols if c in trm_low.columns]

base_high = topk_high.merge(trm_high[trm_cols_high], on="target", how="inner")
base_low = topk_low.merge(trm_low[trm_cols_low], on="target", how="inner")
base = pd.concat([base_high, base_low], ignore_index=True)

print(f"Base: {len(base)} rows (high={len(base_high)}, low={len(base_low)})")

# Attach activity
base["ligand_activity"] = np.nan
base["ligand_activity_rank"] = np.nan
for d in ["TIL_high", "TIL_low"]:
    m = base["direction"] == d
    base.loc[m, "ligand_activity"] = base.loc[m, "ligand"].map(act_by_dir[d]["activity_map"])
    base.loc[m, "ligand_activity_rank"] = base.loc[m, "ligand"].map(act_by_dir[d]["rank_map"])

# Attach sender
base = base.merge(sender_candidates, on=["direction", "ligand"], how="inner")
print(f"After sender merge: {len(base)} rows")

# Attach receptor
lr_dir = pd.concat([lr_by_dir["TIL_high"], lr_by_dir["TIL_low"]], ignore_index=True)
axes = base.merge(lr_dir, on=["direction", "ligand"], how="inner")

receiver_tbl = receiver_de.rename(columns={"celltype": "receiver_celltype", "gene": "receptor", "logFC": "receptor_logFC", "FDR": "receptor_FDR"})
axes = axes.merge(receiver_tbl[["receiver_celltype", "receptor", "receptor_logFC", "receptor_FDR"]], on=["receiver_celltype", "receptor"], how="left")

# Attach target DE
de_use = de_df[["gene", "logFC", "FDR"]].copy()
de_use.columns = ["target", "target_logFC", "target_FDR"]
axes = axes.merge(de_use, on="target", how="left")

axes = axes.dropna(subset=["receptor_logFC"])
print(f"After receptor filter: {len(axes)} axes")

# === SCORE ===
def score_axes(df):
    df = df.copy()
    for d in ["TIL_high", "TIL_low"]:
        m = df["direction"] == d
        if m.sum() == 0:
            continue
        ranks = df.loc[m, "ligand_activity_rank"]
        max_r = ranks.max()
        df.loc[m, "activity_pctl"] = 1 - (ranks - 1) / max_r if max_r > 1 else 1.0
        df.loc[m, "ltm_pctl"] = df.loc[m, "ltm_weight"].rank(pct=True)
        if "receiver_rank_score" in df.columns:
            df.loc[m, "receiver_rank_pctl"] = df.loc[m, "receiver_rank_score"].rank(pct=True)
        else:
            df.loc[m, "receiver_rank_pctl"] = 0.5

    exp_sign = np.where(df["direction"] == "TIL_high", 1, -1)
    df["ligand_concordant"] = (exp_sign * df["ligand_logFC"]) > 0
    df["receptor_concordant"] = (exp_sign * df["receptor_logFC"]) > 0

    df["axis_score"] = (
        0.35 * df["activity_pctl"].fillna(0) +
        0.25 * df["ltm_pctl"].fillna(0) +
        0.15 * df["receiver_rank_pctl"].fillna(0) +
        0.10 * df["ligand_concordant"].astype(float) +
        0.15 * df["receptor_concordant"].astype(float)
    )
    return df

all_axes = score_axes(axes)
all_axes = all_axes.sort_values(["direction", "target", "axis_score"], ascending=[True, True, False])
all_axes["axis_rank"] = all_axes.groupby(["direction", "target"]).cumcount() + 1
all_axes = all_axes.drop_duplicates(subset=["direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype"])

# === TOP AXES SELECTION ===
lig_best = all_axes.groupby(["direction", "target", "ligand"]).agg(
    best_score=("axis_score", "max"), best_ltm=("ltm_weight", "max"), best_act=("ligand_activity", "max"), n=("axis_score", "size")
).reset_index()
lig_best = lig_best.sort_values(["direction", "target", "best_score", "best_ltm", "best_act"], ascending=[True, True, False, False, False])
lig_best["lig_rank"] = lig_best.groupby(["direction", "target"]).cumcount() + 1
keep_ligs = lig_best[lig_best["lig_rank"] <= TOP_LIGANDS_PER_TARGET][["direction", "target", "ligand"]]

sel = all_axes.merge(keep_ligs, on=["direction", "target", "ligand"], how="inner")
sel = sel.sort_values(["direction", "target", "ligand", "axis_score"], ascending=[True, True, True, False])
sel["ax_in_lig"] = sel.groupby(["direction", "target", "ligand"]).cumcount() + 1
sel = sel[sel["ax_in_lig"] <= TOP_AXES_PER_LIGAND_PER_TARGET]
sel = sel.sort_values(["direction", "target", "axis_score"], ascending=[True, True, False])
sel["ax_final"] = sel.groupby(["direction", "target"]).cumcount() + 1
top_axes = sel[sel["ax_final"] <= TOP_AXES_PER_TARGET].copy()
top_axes["axis_rank"] = top_axes.groupby(["direction", "target"]).cumcount() + 1

# === SUMMARY ===
print(f"\nAXES BUILT:")
print(f"  all_axes: {len(all_axes)} rows, {all_axes['target'].nunique()} targets")
print(f"  top_axes: {len(top_axes)} rows, {top_axes['target'].nunique()} targets")

for d in ["TIL_high", "TIL_low"]:
    sub = all_axes[all_axes["direction"] == d]
    print(f"\n  {d}: {len(sub)} axes, {sub['target'].nunique()} targets, {sub['ligand'].nunique()} ligands")

# === SAVE ===
all_axes.to_csv(OUT_DIR / "all_axes.csv", index=False)
top_axes.to_csv(OUT_DIR / "top_axes.csv", index=False)
print(f"\n✓ Saved to {OUT_DIR}")
```

### Summarize axis scores for plot selection

```python
# %% [markdown]
# ### Summarize axis scores for plot selection

# %%
import pandas as pd
import numpy as np
from pathlib import Path
import hashlib

# === CLEAR STALE VARIABLES ===
for _v in ["df", "df_all", "df_top", "axes_catalog", "axes_shortlist", "short", "short2"]:
    if _v in globals():
        del globals()[_v]

# === VERIFY INPUTS ===
if "OUT_DIR" not in globals():
    raise NameError("OUT_DIR not found.")
if "top_axes" not in globals():
    raise NameError("top_axes not found. Run Cell 5 first.")

OUT_DIR = Path(OUT_DIR)

print("INPUT CHECK:")
print(f"  top_axes: {len(top_axes)} rows, {top_axes['target'].nunique()} targets")
if "all_axes" in globals():
    print(f"  all_axes: {len(all_axes)} rows (for reference, not used in catalog)")

# Use top_axes (filtered) for catalog, NOT all_axes
df = pd.DataFrame(top_axes).copy()

req_cols = {"direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype",
            "target_logFC", "target_FDR", "ligand_activity", "ltm_weight",
            "ligand_concordant", "receptor_concordant", "axis_score", "axis_rank"}
missing = req_cols - set(df.columns)
if missing:
    raise KeyError(f"top_axes missing columns: {sorted(missing)}")

# Normalize dtypes
for c in ["direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype"]:
    df[c] = df[c].astype(str)
for c in ["axis_score", "axis_rank", "target_logFC", "target_FDR", "ligand_activity", "ltm_weight"]:
    df[c] = pd.to_numeric(df[c], errors="coerce")
df["ligand_concordant"] = df["ligand_concordant"].astype(bool)
df["receptor_concordant"] = df["receptor_concordant"].astype(bool)

if "receiver_rank_score" in df.columns:
    df["receiver_rank_score"] = pd.to_numeric(df["receiver_rank_score"], errors="coerce")
if "panel_coverage_ratio" in df.columns:
    df["panel_coverage_ratio"] = pd.to_numeric(df["panel_coverage_ratio"], errors="coerce")

# === FILTER 1: positive activity ===
n_before = len(df)
df = df[df["ligand_activity"] > 0].copy()
print(f"\nFilter: ligand_activity > 0: {n_before} -> {len(df)} axes")

# === ABUNDANCE ANNOTATIONS ===
def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

abundance_available = False
if "EFS_ROOT" in globals() and "ZONE" in globals():
    ab_path = Path(EFS_ROOT) / "results/visium/til_zones/celltype_zone_specificity_curated/celltype_zone_specificity_level4_curated.csv"

    if ab_path.exists():
        ab = pd.read_csv(ab_path)
        ab["group"] = ab["group"].astype(str)
        ab["zone"] = ab["zone"].astype(str)
        ab["celltype_std"] = ab["celltype_level4_curated"].astype(str).map(std_ct)
        ab_zone = ab[ab["zone"] == str(ZONE)].copy()

        delta_tbl = ab_zone.pivot_table(index="celltype_std", columns="group", values="mean_zone", aggfunc="first").reset_index()
        if "TIL_high" in delta_tbl.columns and "TIL_low" in delta_tbl.columns:
            delta_tbl["mean_zone_high_minus_low"] = pd.to_numeric(delta_tbl["TIL_high"], errors="coerce") - pd.to_numeric(delta_tbl["TIL_low"], errors="coerce")
            delta_tbl = delta_tbl[["celltype_std", "mean_zone_high_minus_low"]]
        else:
            delta_tbl = pd.DataFrame({"celltype_std": ab_zone["celltype_std"].unique(), "mean_zone_high_minus_low": np.nan})

        df["sender_celltype_std"] = df["sender_celltype"].map(std_ct)
        df["receiver_celltype_std"] = df["receiver_celltype"].map(std_ct)

        # Sender abundance
        sender_ab = ab_zone.rename(columns={"celltype_std": "sender_celltype_std", "mean_zone": "sender_mean_zone"})
        sender_ab = sender_ab[["group", "sender_celltype_std", "sender_mean_zone"]].copy()
        df = df.merge(sender_ab, left_on=["direction", "sender_celltype_std"], right_on=["group", "sender_celltype_std"], how="left")
        df = df.drop(columns=["group"], errors="ignore")

        # Receiver abundance
        receiver_ab = ab_zone.rename(columns={"celltype_std": "receiver_celltype_std", "mean_zone": "receiver_mean_zone"})
        receiver_ab = receiver_ab[["group", "receiver_celltype_std", "receiver_mean_zone"]].copy()
        df = df.merge(receiver_ab, left_on=["direction", "receiver_celltype_std"], right_on=["group", "receiver_celltype_std"], how="left")
        df = df.drop(columns=["group"], errors="ignore")

        # Delta
        delta_sender = delta_tbl.rename(columns={"celltype_std": "sender_celltype_std", "mean_zone_high_minus_low": "sender_mean_zone_high_minus_low"})
        delta_receiver = delta_tbl.rename(columns={"celltype_std": "receiver_celltype_std", "mean_zone_high_minus_low": "receiver_mean_zone_high_minus_low"})
        df = df.merge(delta_sender, on="sender_celltype_std", how="left")
        df = df.merge(delta_receiver, on="receiver_celltype_std", how="left")

        print(f"Abundance annotations loaded from: {ab_path}")
        abundance_available = True
    else:
        print(f"Abundance CSV not found: {ab_path}")
else:
    print("EFS_ROOT or ZONE not set, skipping abundance annotations")

# === FILTER 2: concordance/abundance ===
ABUNDANCE_THRESHOLD = 0.05
n_before = len(df)

if abundance_available:
    high_mask = df["direction"] == "TIL_high"
    low_mask = df["direction"] == "TIL_low"
    high_keep = df["ligand_concordant"] | (df["sender_mean_zone_high_minus_low"] > ABUNDANCE_THRESHOLD)
    low_keep = df["ligand_concordant"] | (df["sender_mean_zone_high_minus_low"] < -ABUNDANCE_THRESHOLD)
    df = df[(high_mask & high_keep) | (low_mask & low_keep)].copy()
else:
    df = df[df["ligand_concordant"]].copy()

print(f"Filter: concordance/abundance: {n_before} -> {len(df)} axes")

# === AXIS ID ===
key_cols = ["direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype"]
df = df.drop_duplicates(subset=key_cols).copy()

def axis_hash_id(row):
    s = "|".join(str(row[c]) for c in key_cols)
    h = hashlib.md5(s.encode()).hexdigest()[:6].upper()
    return ("H" if row["direction"] == "TIL_high" else "L") + h

df["axis_id"] = df.apply(axis_hash_id, axis=1)

# === RANK SCORE ===
df["de_strength"] = -np.log10(df["target_FDR"].clip(lower=1e-300))
df["rank_score"] = df["axis_score"] * df["de_strength"] * df["target_logFC"].abs()

# === VERIFY AXES PER TARGET ===
axes_per_target = df.groupby(["direction", "target"]).size()
print(f"\nAxes per target: min={axes_per_target.min()}, max={axes_per_target.max()}, mean={axes_per_target.mean():.1f}")

# === CATALOG COLUMNS ===
catalog_cols = ["axis_id", "direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype",
                "axis_score", "axis_rank", "rank_score", "target_logFC", "target_FDR",
                "ligand_activity", "ltm_weight", "ligand_concordant", "receptor_concordant"]

# Add optional columns if present
for c in ["receiver_rank_score", "global_best_celltype", "global_best_n_cells", "panel_coverage_ratio", "coverage_flag",
          "sender_mean_zone", "receiver_mean_zone", "sender_mean_zone_high_minus_low", "receiver_mean_zone_high_minus_low"]:
    if c in df.columns:
        catalog_cols.append(c)

axes_catalog = df[[c for c in catalog_cols if c in df.columns]].copy()
axes_catalog.to_csv(OUT_DIR / "axes_catalog.csv", index=False)

# === SHORTLIST (further filtered subset) ===
AXIS_SCORE_MIN = 0.5
MAX_AXIS_RANK_PER_TARGET = 3
TARGET_FDR_MAX = 0.05
TARGET_LOGFC_MIN = 0.25
AXES_PER_TARGET_DE_BAL = 2

short = df[(df["axis_score"] > AXIS_SCORE_MIN) & (df["axis_rank"] <= MAX_AXIS_RANK_PER_TARGET)].copy()
short = short.dropna(subset=["target_FDR", "target_logFC", "rank_score"])
short = short[short["target_FDR"] <= TARGET_FDR_MAX]

high = short[(short["direction"] == "TIL_high") & (short["target_logFC"] >= TARGET_LOGFC_MIN)]
low = short[(short["direction"] == "TIL_low") & (short["target_logFC"] <= -TARGET_LOGFC_MIN)]
short = pd.concat([high, low])
short = short.sort_values(["direction", "target", "rank_score"], ascending=[True, True, False])
axes_shortlist = short.groupby(["direction", "target"], as_index=False).head(AXES_PER_TARGET_DE_BAL)
axes_shortlist = axes_shortlist[[c for c in catalog_cols if c in axes_shortlist.columns]]
axes_shortlist.to_csv(OUT_DIR / "axes_shortlist.csv", index=False)

# === SUMMARY ===
print(f"\nCATALOG SUMMARY:")
for d in ["TIL_high", "TIL_low"]:
    sub = df[df["direction"] == d]
    print(f"  {d}: {len(sub)} axes, {sub['target'].nunique()} targets, {sub['ligand'].nunique()} ligands")

if "coverage_flag" in df.columns:
    print(f"\nPANEL COVERAGE:")
    for d in ["TIL_high", "TIL_low"]:
        sub = df[df["direction"] == d]
        if sub.empty:
            continue
        n_ok = (sub["coverage_flag"] == "OK").sum()
        n_low = (sub["coverage_flag"] == "LOW_COVERAGE").sum()
        print(f"  {d}: OK={n_ok}, LOW_COVERAGE={n_low}")

print(f"\nSHORTLIST: {len(axes_shortlist)} axes, {axes_shortlist['target'].nunique()} targets")

print(f"\nWROTE:")
print(f"  {OUT_DIR / 'axes_catalog.csv'}")
print(f"  {OUT_DIR / 'axes_shortlist.csv'}")

# === DISPLAY ===
display_cols = ["axis_id", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype",
                "axis_score", "rank_score", "target_logFC"]
if "coverage_flag" in axes_catalog.columns:
    display_cols.append("coverage_flag")
display_cols = [c for c in display_cols if c in axes_catalog.columns]

print(f"\nTOP 15 TIL_high:")
display(axes_catalog[axes_catalog["direction"] == "TIL_high"].nlargest(15, "rank_score")[display_cols])

print(f"\nTOP 15 TIL_low:")
display(axes_catalog[axes_catalog["direction"] == "TIL_low"].nlargest(15, "rank_score")[display_cols])

# === WARNINGS ===
if "coverage_flag" in axes_shortlist.columns:
    low_cov = axes_shortlist[axes_shortlist["coverage_flag"] == "LOW_COVERAGE"]
    if len(low_cov) > 0:
        print(f"\n⚠️ WARNING: {len(low_cov)} shortlist axes have LOW_COVERAGE:")
        for _, r in low_cov.iterrows():
            gb_ct = r.get('global_best_celltype', 'N/A')
            gb_n = r.get('global_best_n_cells', 'N/A')
            print(f"  {r['axis_id']}: {r['target']} (global_best={gb_ct}, n={gb_n})")

print("\n✓ Done")
```

### Circos plotting

```python
# Pass ZONE to R for circos plotting
ZONE_FOR_R = str(ZONE)
print(f"Passing ZONE to R: {ZONE_FOR_R}")
```

```python
%%R -i ZONE_FOR_R
suppressPackageStartupMessages({
  library(dplyr)
  library(readr)
  library(circlize)
})

options(bitmapType = "cairo")

EFS_ROOT <- "/home/sagemaker-user/custom-file-systems/efs/fs-09913c1f7db79b6fd"
ZONE <- ZONE_FOR_R
zone_dir <- gsub("\\+", "plus", gsub("-", "minus", ZONE))
OUT_DIR <- file.path(EFS_ROOT, "results", "visium", "til_zones", "nichenet_analysis", zone_dir)
stopifnot(dir.exists(OUT_DIR))

AXES_SELECTED_CSV <- file.path(OUT_DIR, "axes_selected.csv")
stopifnot(file.exists(AXES_SELECTED_CSV))

MAX_EDGES <- Inf

LINK_ALPHA <- 0.5

BORDER_COL <- "#777777"
BORDER_LWD <- 0.25

TITLE_CEX <- 1.00

START_DEGREE <- -8
SMALL_GAP <- 1
BIG_GAP <- 5

WIDTH_MIN <- 0.10
WIDTH_MAX <- 2.50

COLOR_CLIP_LO <- 0.02
COLOR_CLIP_HI <- 0.98

CANVAS_PAD <- 0.40

LABEL_CEX <- 0.46
LABEL_RADIUS <- 1.0
LABEL_NICEFACING <- TRUE
LABEL_MIN_DEG <- 0

OUT_PNG <- file.path(OUT_DIR, "circos_selected_axes_senders_receivers_only.png")
PNG_SIZE <- 2600
PNG_RES <- 240

axes <- readr::read_csv(AXES_SELECTED_CSV, show_col_types = FALSE)

required_cols <- c(
  "axis_id","direction",
  "sender_celltype","receiver_celltype",
  "ligand","target",
  "rank_score","ligand_activity"
)
missing <- setdiff(required_cols, colnames(axes))
if (length(missing) > 0) stop(paste0("axes_selected.csv missing columns: ", paste(missing, collapse = ", ")))

axes <- axes %>%
  transmute(
    axis_id = as.character(axis_id),
    direction = as.character(direction),
    sender = as.character(sender_celltype),
    receiver = as.character(receiver_celltype),
    ligand = as.character(ligand),
    target = as.character(target),
    rank_score = as.numeric(rank_score),
    ligand_activity = as.numeric(ligand_activity)
  ) %>%
  filter(
    direction %in% c("TIL_high","TIL_low"),
    is.finite(rank_score),
    is.finite(ligand_activity)
  )

if (nrow(axes) == 0) stop("No rows left after filtering. Check direction/rank_score/ligand_activity.")

if (is.finite(MAX_EDGES) && nrow(axes) > MAX_EDGES) {
  axes <- axes %>% slice_head(n = MAX_EDGES)
}

rescale01 <- function(x) {
  x <- as.numeric(x)
  ok <- is.finite(x)
  if (sum(ok) == 0) return(rep(0.5, length(x)))
  lo <- min(x[ok]); hi <- max(x[ok])
  if (!is.finite(lo) || !is.finite(hi) || lo == hi) return(rep(0.5, length(x)))
  y <- (x - lo) / (hi - lo)
  y[!is.finite(y)] <- 0.5
  pmin(pmax(y, 0), 1)
}

clip01_quantile <- function(x, qlo = 0.02, qhi = 0.98) {
  x <- as.numeric(x)
  ok <- is.finite(x)
  if (sum(ok) == 0) return(rep(0.5, length(x)))
  lo <- suppressWarnings(as.numeric(stats::quantile(x[ok], probs = qlo, na.rm = TRUE, names = FALSE)))
  hi <- suppressWarnings(as.numeric(stats::quantile(x[ok], probs = qhi, na.rm = TRUE, names = FALSE)))
  if (!is.finite(lo) || !is.finite(hi) || lo == hi) return(rescale01(x))
  y <- (x - lo) / (hi - lo)
  y[!is.finite(y)] <- 0.5
  pmin(pmax(y, 0), 1)
}

axes <- axes %>%
  mutate(
    width_norm = rescale01(ligand_activity),
    w = WIDTH_MIN + width_norm * (WIDTH_MAX - WIDTH_MIN)
  )

rank_norm <- clip01_quantile(axes$rank_score, COLOR_CLIP_LO, COLOR_CLIP_HI)

col_high <- circlize::colorRamp2(c(0, 1), c("#FEE6CE", "#D55E00"))
col_low  <- circlize::colorRamp2(c(0, 1), c("#DEEBF7", "#0072B2"))

axes$link_col <- vapply(seq_len(nrow(axes)), function(i) {
  if (axes$direction[i] == "TIL_high") col_high(rank_norm[i]) else col_low(rank_norm[i])
}, character(1))
axes$link_col <- adjustcolor(axes$link_col, alpha.f = LINK_ALPHA)

S_PREFIX <- "S|"
R_PREFIX <- "R|"
clean_label <- function(x) sub("^[SR]\\|", "", x)

senders <- axes %>%
  group_by(sender) %>%
  summarise(total = sum(w), .groups = "drop") %>%
  arrange(desc(total)) %>%
  pull(sender)

receivers <- axes %>%
  group_by(receiver) %>%
  summarise(total = sum(w), .groups = "drop") %>%
  arrange(desc(total)) %>%
  pull(receiver)

sender_sectors <- paste0(S_PREFIX, senders)
receiver_sectors <- paste0(R_PREFIX, receivers)
sectors <- c(sender_sectors, receiver_sectors)

ct_names <- sort(unique(c(senders, receivers)))
ct_cols <- setNames(grDevices::hcl.colors(length(ct_names), palette = "Dark 3"), ct_names)

grid_cols <- c(
  setNames(ct_cols[clean_label(sender_sectors)], sender_sectors),
  setNames(ct_cols[clean_label(receiver_sectors)], receiver_sectors)
)

axes <- axes %>%
  mutate(
    s_sector = paste0(S_PREFIX, sender),
    r_sector = paste0(R_PREFIX, receiver),
    row_id = seq_len(n())
  )

sender_totals <- tapply(axes$w, axes$s_sector, sum)
receiver_totals <- tapply(axes$w, axes$r_sector, sum)

totals <- setNames(rep(1, length(sectors)), sectors)
if (length(sender_totals) > 0) totals[names(sender_totals)] <- sender_totals
if (length(receiver_totals) > 0) totals[names(receiver_totals)] <- receiver_totals
totals[!is.finite(totals) | totals <= 0] <- 1

xlim_mat <- cbind(rep(0, length(sectors)), as.numeric(totals[sectors]))
rownames(xlim_mat) <- sectors

# Collapse ligands per sender:
# Allocate one contiguous sender-side block per (sender, ligand), then subdivide within block for links.
lig_groups <- axes %>%
  group_by(s_sector, ligand) %>%
  summarise(
    group_w = sum(w),
    group_order = min(row_id),
    .groups = "drop"
  ) %>%
  arrange(s_sector, group_order)

sender_cursor <- setNames(rep(0, length(sender_sectors)), sender_sectors)
lig_groups$lig_start <- NA_real_
lig_groups$lig_end <- NA_real_

for (k in seq_len(nrow(lig_groups))) {
  ss <- lig_groups$s_sector[k]
  ww <- lig_groups$group_w[k]
  st <- sender_cursor[ss]
  en <- st + ww
  lig_groups$lig_start[k] <- st
  lig_groups$lig_end[k] <- en
  sender_cursor[ss] <- en
}

axes <- axes %>%
  left_join(lig_groups %>% select(s_sector, ligand, lig_start, lig_end), by = c("s_sector", "ligand")) %>%
  group_by(s_sector, ligand) %>%
  arrange(row_id, .by_group = TRUE) %>%
  mutate(
    s_x1 = lig_start + dplyr::lag(cumsum(w), default = 0),
    s_x2 = lig_start + cumsum(w)
  ) %>%
  ungroup()

# Receiver allocation stays per-edge (unchanged)
receiver_cursor <- setNames(rep(0, length(receiver_sectors)), receiver_sectors)
r_x1 <- numeric(nrow(axes)); r_x2 <- numeric(nrow(axes))

for (i in seq_len(nrow(axes))) {
  rr <- axes$r_sector[i]
  ww <- axes$w[i]
  r_x1[i] <- receiver_cursor[rr]
  r_x2[i] <- receiver_cursor[rr] + ww
  receiver_cursor[rr] <- r_x2[i]
}

axes$r_x1 <- r_x1; axes$r_x2 <- r_x2
axes$s_mid <- (axes$s_x1 + axes$s_x2) / 2
axes$r_mid <- (axes$r_x1 + axes$r_x2) / 2

gap_after <- c(
  rep(SMALL_GAP, max(length(sender_sectors) - 1, 0)), BIG_GAP,
  rep(SMALL_GAP, max(length(receiver_sectors) - 1, 0)), BIG_GAP
)

thin_by_min_deg <- function(theta, min_deg) {
  if (!is.finite(min_deg) || min_deg <= 0) return(rep(TRUE, length(theta)))
  ord <- order(theta)
  keep <- rep(FALSE, length(theta))
  last <- NA_real_
  for (j in ord) {
    if (!is.finite(theta[j])) next
    if (is.na(last) || abs(theta[j] - last) >= min_deg) {
      keep[j] <- TRUE
      last <- theta[j]
    }
  }
  keep
}

plot_one <- function() {
  circos.clear()
  circos.par(
    start.degree = START_DEGREE,
    gap.after = gap_after,
    cell.padding = c(0, 0, 0, 0),
    track.margin = c(0.01, 0.01),
    canvas.xlim = c(-(1 + CANVAS_PAD), 1 + CANVAS_PAD),
    canvas.ylim = c(-(1 + CANVAS_PAD), 1 + CANVAS_PAD)
  )

  circos.initialize(factors = sectors, xlim = xlim_mat)

  circos.trackPlotRegion(
    factors = sectors,
    ylim = c(0, 1),
    track.height = 0.08,
    bg.border = NA,
    panel.fun = function(x, y) {
      si <- get.cell.meta.data("sector.index")
      xl <- get.cell.meta.data("xlim")
      circos.rect(xl[1], 0, xl[2], 1, col = grid_cols[si], border = NA)
    }
  )

  for (i in seq_len(nrow(axes))) {
    circos.link(
      sector.index1 = axes$s_sector[i],
      point1 = c(axes$s_x1[i], axes$s_x2[i]),
      sector.index2 = axes$r_sector[i],
      point2 = c(axes$r_x1[i], axes$r_x2[i]),
      col = axes$link_col[i],
      border = BORDER_COL,
      lwd = BORDER_LWD,
      directional = 1,
      arr.type = "big.arrow",
      arr.length = 0.04
    )
  }

  # Sender-side ligand labels: one per (sender, ligand), positioned at the center of the ligand block
  ligand_labels <- lig_groups %>%
    mutate(lig_mid = (lig_start + lig_end) / 2)

  theta_lig <- rep(NA_real_, nrow(ligand_labels))
  for (k in seq_len(nrow(ligand_labels))) {
    ss <- ligand_labels$s_sector[k]
    m <- circlize(ligand_labels$lig_mid[k], 0.5, sector.index = ss, track.index = 1)
    theta_lig[k] <- m[1, 1]
  }

  keep_lig <- rep(TRUE, nrow(ligand_labels))
  if (LABEL_MIN_DEG > 0) {
    for (ss in unique(ligand_labels$s_sector)) {
      idx <- which(ligand_labels$s_sector == ss)
      keep_lig[idx] <- thin_by_min_deg(theta_lig[idx], LABEL_MIN_DEG)
    }
  }

  # Receiver-side target labels remain per-edge (unchanged)
  theta_r <- rep(NA_real_, nrow(axes))
  for (rr in unique(axes$r_sector)) {
    idx <- which(is.finite(axes$r_mid) & axes$r_sector == rr)
    if (length(idx) == 0) next
    m <- circlize(axes$r_mid[idx], rep(0.5, length(idx)), sector.index = rr, track.index = 1)
    theta_r[idx] <- m[, 1]
  }

  keep_r <- rep(TRUE, nrow(axes))
  if (LABEL_MIN_DEG > 0) {
    for (rr in unique(axes$r_sector)) {
      idx <- which(axes$r_sector == rr)
      keep_r[idx] <- thin_by_min_deg(theta_r[idx], LABEL_MIN_DEG)
    }
  }

  op <- par(xpd = NA)

  for (k in seq_len(nrow(ligand_labels))) {
    if (!keep_lig[k] || !is.finite(theta_lig[k])) next

    ang <- theta_lig[k] * pi / 180
    x <- LABEL_RADIUS * cos(ang)
    y <- LABEL_RADIUS * sin(ang)

    srt <- theta_lig[k]
    adjx <- 0
    if (LABEL_NICEFACING && theta_lig[k] > 90 && theta_lig[k] < 270) {
      srt <- theta_lig[k] + 180
      adjx <- 1
    }

    text(
      x = x, y = y,
      labels = ligand_labels$ligand[k],
      srt = srt,
      adj = c(adjx, 0.5),
      cex = LABEL_CEX
    )
  }

  for (i in seq_len(nrow(axes))) {
    if (keep_r[i] && is.finite(theta_r[i])) {
      ang <- theta_r[i] * pi / 180
      x <- LABEL_RADIUS * cos(ang)
      y <- LABEL_RADIUS * sin(ang)

      srt <- theta_r[i]
      adjx <- 0
      if (LABEL_NICEFACING && theta_r[i] > 90 && theta_r[i] < 270) {
        srt <- theta_r[i] + 180
        adjx <- 1
      }

      text(x = x, y = y, labels = axes$target[i], srt = srt, adj = c(adjx, 0.5), cex = LABEL_CEX)
    }
  }

  par(op)

  legend(
    "topleft",
    legend = ct_names,
    fill = ct_cols[ct_names],
    border = NA,
    cex = 0.75,
    bty = "n",
    title = "Cell types"
  )

  legend(
    "bottomleft",
    legend = c("TIL_high axes", "TIL_low axes"),
    fill = c("#D55E00", "#0072B2"),
    border = NA,
    cex = 0.95,
    bty = "n",
    title = "Axis direction"
  )

  title(
    main = paste0(
      "Selected axes circos | ", zone_dir,
      " | edges=", nrow(axes),
      " | width ~ ligand_activity, color ~ rank_score"
    ),
    cex.main = TITLE_CEX
  )

  circos.clear()
}

# Write PNG once
png(OUT_PNG, width = PNG_SIZE, height = PNG_SIZE, res = PNG_RES, type = "cairo")
plot_one()
dev.off()

OUT_PDF <- sub("\\.png$", ".pdf", OUT_PNG)

pdf(OUT_PDF, width = PNG_SIZE / PNG_RES, height = PNG_SIZE / PNG_RES, useDingbats = FALSE)
plot_one()
dev.off()

cat("Wrote:", OUT_PNG, "\n")
cat("Wrote:", OUT_PDF, "\n")
cat("Edges:", nrow(axes), "\n")
cat("ligand_activity range:", min(axes$ligand_activity), "to", max(axes$ligand_activity), "\n")
cat("rank_score range:", min(axes$rank_score), "to", max(axes$rank_score), "\n")
cat("rank_score clip quantiles:", COLOR_CLIP_LO, "to", COLOR_CLIP_HI, "\n")
cat("Label radius:", LABEL_RADIUS, " Label cex:", LABEL_CEX, " Label min deg:", LABEL_MIN_DEG, "\n")
```

### GSEA of HIGH and LOW BZ2 programs

```python
# %% [markdown]
# ### ULM pathway enrichment scoring

# %%
import numpy as np
import pandas as pd
from pathlib import Path
import requests
import decoupler as dc

# User config
ZONE = "BZ2"  # Options: "TIL+", "BZ1", "BZ2"

# Paths
cache_dir = Path.home() / ".cache" / "msigdb"
cache_dir.mkdir(parents=True, exist_ok=True)

EFS_ROOT = Path("/mnt/custom-file-systems/efs/fs-09913c1f7db79b6fd")
DE_DIR = EFS_ROOT / "results/visium/til_zones/deg_pseudobulk_edger"
OUT_DIR = EFS_ROOT / "results/visium/til_zones/nichenet_filtering"
OUT_DIR.mkdir(parents=True, exist_ok=True)

# Zone-specific DE file
visium_de_file = DE_DIR / f"DE_edger_zone_{ZONE}_TIL_high_vs_TIL_low.csv"

print(f"ZONE: {ZONE}")
print(f"DE file: {visium_de_file.name}")

if not visium_de_file.exists():
    raise FileNotFoundError(f"Missing: {visium_de_file}")

print("Input file found")

# MSigDB GMT download and parse
urls = {
    "hallmark": "https://data.broadinstitute.org/gsea-msigdb/msigdb/release/2023.2.Hs/h.all.v2023.2.Hs.symbols.gmt",
    "kegg":     "https://data.broadinstitute.org/gsea-msigdb/msigdb/release/2023.2.Hs/c2.cp.kegg_medicus.v2023.2.Hs.symbols.gmt",
    "biocarta": "https://data.broadinstitute.org/gsea-msigdb/msigdb/release/2023.2.Hs/c2.cp.biocarta.v2023.2.Hs.symbols.gmt",
}
paths = {k: cache_dir / Path(v).name for k, v in urls.items()}

def download_text(url: str, out_path: Path, timeout: int = 120) -> None:
    if out_path.exists() and out_path.stat().st_size > 0:
        return
    r = requests.get(url, timeout=timeout)
    r.raise_for_status()
    out_path.write_text(r.text)

def parse_gmt(gmt_path: Path) -> dict:
    gene_sets = {}
    with gmt_path.open("r") as f:
        for line in f:
            parts = line.rstrip("\n").split("\t")
            if len(parts) >= 3:
                gene_sets[parts[0]] = [g for g in parts[2:] if g]
    return gene_sets

for k, url in urls.items():
    download_text(url, paths[k])

hallmark_sets = parse_gmt(paths["hallmark"])
kegg_sets = parse_gmt(paths["kegg"])
biocarta_sets = parse_gmt(paths["biocarta"])

def gmt_to_net(gene_sets: dict) -> pd.DataFrame:
    rows = [(gs, g, 1.0) for gs, genes in gene_sets.items() for g in genes]
    return pd.DataFrame(rows, columns=["source", "target", "weight"])

net = pd.concat(
    [gmt_to_net(hallmark_sets), gmt_to_net(kegg_sets), gmt_to_net(biocarta_sets)],
    ignore_index=True
)

# Load full DE results (all genes)
visium_de = pd.read_csv(visium_de_file)
for c in ["gene", "logFC"]:
    if c not in visium_de.columns:
        raise ValueError(f"Missing column '{c}' in {visium_de_file.name}")

visium_de["gene"] = visium_de["gene"].astype(str)
visium_de["logFC"] = pd.to_numeric(visium_de["logFC"], errors="coerce")
visium_de = visium_de.dropna(subset=["logFC"])
visium_de = visium_de.drop_duplicates(subset=["gene"], keep="first")

print(f"Genes in DE results: {len(visium_de)}")

# BH FDR correction
def bh_fdr(pvals: np.ndarray) -> np.ndarray:
    p = np.asarray(pvals, dtype=float)
    q = np.full_like(p, np.nan, dtype=float)
    ok = np.isfinite(p)
    if ok.sum() == 0:
        return q
    p0 = p[ok]
    n = p0.size
    order = np.argsort(p0)
    ranked = p0[order]
    adj = ranked * n / (np.arange(1, n + 1))
    adj = np.minimum.accumulate(adj[::-1])[::-1]
    out = np.empty_like(p0)
    out[order] = np.clip(adj, 0, 1)
    q[ok] = out
    return q

# Run ULM on all genes
# Positive logFC = upregulated in TIL_high
# Negative logFC = upregulated in TIL_low
def run_ulm_all_genes(de_df: pd.DataFrame, label: str) -> pd.DataFrame:
    w = de_df.set_index("gene")["logFC"].astype(float)
    mat = w.to_frame().T
    mat.index = [label]

    est, pvs = dc.run_ulm(
        mat=mat,
        net=net,
        source="source",
        target="target",
        weight="weight",
        min_n=5,
        verbose=False,
    )

    res = pd.DataFrame({
        "gene_set": est.columns.astype(str),
        "score": est.iloc[0].to_numpy(dtype=float),
        "pval": pvs.iloc[0].to_numpy(dtype=float),
    })
    res["fdr"] = bh_fdr(res["pval"].to_numpy())
    res["collection"] = np.select(
        [
            res["gene_set"].str.startswith("HALLMARK_"),
            res["gene_set"].str.startswith("KEGG_"),
            res["gene_set"].str.startswith("BIOCARTA_"),
        ],
        ["Hallmark", "KEGG", "BioCarta"],
        default="Other",
    )
    return res

ulm_results = run_ulm_all_genes(visium_de, label=f"{ZONE}_TIL_high_vs_low")

# Positive score = enriched in TIL_high direction
# Negative score = enriched in TIL_low direction
ulm_high = ulm_results[ulm_results["score"] > 0].sort_values("score", ascending=False).copy()
ulm_low = ulm_results[ulm_results["score"] < 0].sort_values("score", ascending=True).copy()

# Save outputs
zone_file_prefix = ZONE.replace("+", "plus").replace("-", "minus")
out_all = OUT_DIR / f"{zone_file_prefix}_ULM_all_pathways.csv"
out_high = OUT_DIR / f"{zone_file_prefix}_ULM_TIL_high_enriched.csv"
out_low = OUT_DIR / f"{zone_file_prefix}_ULM_TIL_low_enriched.csv"

ulm_results.sort_values("score", ascending=False).to_csv(out_all, index=False)
ulm_high.to_csv(out_high, index=False)
ulm_low.to_csv(out_low, index=False)

print(f"\nGene sets: Hallmark={len(hallmark_sets)}, KEGG={len(kegg_sets)}, BioCarta={len(biocarta_sets)}")
print(f"Network: {net['source'].nunique()} sets, {net['target'].nunique()} genes")
print(f"\nWrote:")
print(f"  {out_all}")
print(f"  {out_high}")
print(f"  {out_low}")

# Display top results
def show_top(res: pd.DataFrame, title: str, n: int = 20) -> None:
    if res.empty:
        print(f"\n{title}: no results")
        return
    top = res.head(n).copy()
    top["score"] = top["score"].round(3)
    top["pval"] = top["pval"].map(lambda x: f"{x:.2e}" if np.isfinite(x) else "nan")
    top["fdr"] = top["fdr"].map(lambda x: f"{x:.2e}" if np.isfinite(x) else "nan")
    print(f"\n{title}")
    print(top[["gene_set", "collection", "score", "pval", "fdr"]].to_string(index=False))

show_top(ulm_high, f"Top pathways enriched in TIL_high ({ZONE})")
show_top(ulm_low, f"Top pathways enriched in TIL_low ({ZONE})")
```

### Define pathway filters

```python
selected_pathways = [
    "HALLMARK_ANGIOGENESIS",
    "HALLMARK_COMPLEMENT",
    "HALLMARK_EPITHELIAL_MESENCHYMAL_TRANSITION",
    "HALLMARK_HYPOXIA",
    "HALLMARK_APOPTOSIS",
    "HALLMARK_UV_RESPONSE_DN",
    "HALLMARK_IL6_JAK_STAT3_SIGNALING",
    "HALLMARK_INFLAMMATORY_RESPONSE",
    "HALLMARK_INTERFERON_ALPHA_RESPONSE",
    "HALLMARK_INTERFERON_GAMMA_RESPONSE",
    "HALLMARK_TGF_BETA_SIGNALING",
    "HALLMARK_TNFA_SIGNALING_VIA_NFKB",
    "HALLMARK_IL2_STAT5_SIGNALING",
    "HALLMARK_GLYCOLYSIS",
    "HALLMARK_MTORC1_SIGNALING",
    "HALLMARK_MYOGENESIS",
    "HALLMARK_PROTEIN_SECRETION",
    "KEGG_MEDICUS_REFERENCE_WNT_SIGNALING_MODULATION_WNT_ACYLATION",
    "HALLMARK_COAGULATION"
]

ecm_genes_brain = [
    "COL1A1","COL1A2","COL3A1","COL4A1","COL4A2","COL5A1","COL6A1","COL6A2","COL6A3","COL7A1",
    "COL8A1","COL10A1","COL11A1","COL12A1","COL14A1","COL15A1","COL16A1","FN1","FBN1","FBN2",
    "TNC","SPARC","POSTN","LAMA1","LAMA2","LAMA3","LAMB1","LAMB2","LAMC1","THBS1","THBS2","THBS4",
    "VTN","NID1","NID2","VCAN","DCN","BGN","LUM","HSPG2","AGRN","GPC1","GPC3","MMP1","MMP2","MMP9",
    "MMP11","MMP14","ADAMTS1","ADAMTS4","ADAMTS5","TIMP1","TIMP2","LOX","LOXL1","SERPINE1","PLAU",
    "PLAT","CSPG4","BCAN","NCAN","HAPLN1","HAPLN2","PRELP","FMOD","COL18A1","LOXL2","LOXL3","HPSE",
    "SDC1","GPC2","CSPG5","PTPRZ1","COL5A2","COL5A3","LAMA4","LAMA5","ACAN","TNR","HAPLN3","HAPLN4",
    "FBLN1","FBLN2","FBLN5","SDC4","SPP1","IGFBP2","IGFBP5","SRGN","MMP3","MMP7","MMP12","ADAMTS8",
    "ADAM8","TIMP3","LOXL4",
]

# Combine all pathway sets into one lookup
all_pathway_sets = {}
if "hallmark_sets" in globals():
    all_pathway_sets.update(hallmark_sets)
if "kegg_sets" in globals():
    all_pathway_sets.update(kegg_sets)
if "biocarta_sets" in globals():
    all_pathway_sets.update(biocarta_sets)

if not all_pathway_sets:
    raise RuntimeError("No pathway sets found. Run ULM scoring cell first.")

print(f"Available pathway sets: {len(all_pathway_sets)}")

# Validate selected pathways exist
missing = [p for p in selected_pathways if p not in all_pathway_sets]
if missing:
    print("\nMissing pathways:")
    for m in missing:
        print(f"  {m}")
    raise RuntimeError(f"{len(missing)} selected pathways not found in available sets")

# Build gene filter
pathway_genes = set(ecm_genes_brain)
for pathway in selected_pathways:
    pathway_genes.update(all_pathway_sets[pathway])

PATHWAY_FILTER = sorted(pathway_genes)

print(f"\nSelected pathways: {len(selected_pathways)}")
print(f"Custom gene lists: 1 (ecm_genes_brain: {len(ecm_genes_brain)} genes)")
print(f"PATHWAY_FILTER: {len(PATHWAY_FILTER)} unique genes")
```

### Zone background gene universe

```python
import numpy as np
import pandas as pd
import scipy.sparse as sp

# Defines these as sorted Python lists of gene symbols for downstream %%R cells:
#   background_high, background_low
#   ligands_high, ligands_low
#   receptors_high, receptors_low
#
# Also keeps a structured dict for inspection:
#   universes_panel

LAYER = "LogNormalize"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
MIN_PCT_EXPRESSED = 0.10

required = ["adata_sc", "nichenet_priors", "SENDERS", "RECEIVERS"]
missing = [x for x in required if x not in globals()]
if missing:
    raise NameError(f"Missing required variables: {missing}")

if LAYER not in adata_sc.layers:
    raise KeyError(
        f"adata_sc.layers['{LAYER}'] not found. Available layers: {list(adata_sc.layers.keys())}"
    )
for k in [CT_KEY, GROUP_KEY]:
    if k not in adata_sc.obs.columns:
        raise KeyError(
            f"adata_sc.obs['{k}'] not found. Available obs columns (first 60): {list(adata_sc.obs.columns)[:60]}"
        )

for k in ["ltm_targets", "ltm_ligands", "lr_ligands", "lr_receptors"]:
    if k not in nichenet_priors:
        raise KeyError(f"nichenet_priors['{k}'] not found. Available keys: {list(nichenet_priors.keys())}")

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

senders_std = set(std_ct(x) for x in SENDERS)
receivers_std = set(std_ct(x) for x in RECEIVERS)

ct = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)

expected_groups = {"TIL_high", "TIL_low"}
groups_present = set(grp.unique())
if not expected_groups.issubset(groups_present):
    raise ValueError(f"Expected {expected_groups} in {GROUP_KEY}, found: {sorted(groups_present)}")

var_names = pd.Index(adata_sc.var_names.astype(str))
var_set = set(var_names)

X = adata_sc.layers[LAYER]
if sp.issparse(X):
    X = X.tocsr()
else:
    X = sp.csr_matrix(X)

ltm_targets = set(map(str, nichenet_priors["ltm_targets"]))
ltm_ligands = set(map(str, nichenet_priors["ltm_ligands"]))
lr_ligands = set(map(str, nichenet_priors["lr_ligands"]))
lr_receptors = set(map(str, nichenet_priors["lr_receptors"]))

candidate_background = ltm_targets
candidate_ligands = ltm_ligands.intersection(lr_ligands)
candidate_receptors = lr_receptors

def expressed_genes(mask_cells: np.ndarray, candidate_genes: set, min_pct: float) -> set:
    idx = np.flatnonzero(mask_cells)
    if idx.size == 0:
        return set()

    genes_present = sorted(candidate_genes.intersection(var_set))
    if not genes_present:
        return set()

    gene_idx = var_names.get_indexer(genes_present)
    Xsub = X[idx, :][:, gene_idx]

    nnz = np.asarray(Xsub.getnnz(axis=0)).ravel()
    pct = nnz / float(idx.size)

    keep = pct >= float(min_pct)
    return set(np.asarray(genes_present, dtype=object)[keep])

universes_panel = {}

for g in ["TIL_high", "TIL_low"]:
    sender_mask = (grp == g).to_numpy() & ct.isin(senders_std).to_numpy()
    receiver_mask = (grp == g).to_numpy() & ct.isin(receivers_std).to_numpy()

    bg = expressed_genes(receiver_mask, candidate_background, MIN_PCT_EXPRESSED)
    lig = expressed_genes(sender_mask, candidate_ligands, MIN_PCT_EXPRESSED)
    rec = expressed_genes(receiver_mask, candidate_receptors, MIN_PCT_EXPRESSED)

    universes_panel[g] = {
        "background": bg,
        "ligands": lig,
        "receptors": rec,
        "n_sender_cells": int(sender_mask.sum()),
        "n_receiver_cells": int(receiver_mask.sum()),
    }

background_high = sorted(map(str, universes_panel["TIL_high"]["background"]))
background_low = sorted(map(str, universes_panel["TIL_low"]["background"]))
ligands_high = sorted(map(str, universes_panel["TIL_high"]["ligands"]))
ligands_low = sorted(map(str, universes_panel["TIL_low"]["ligands"]))
receptors_high = sorted(map(str, universes_panel["TIL_high"]["receptors"]))
receptors_low = sorted(map(str, universes_panel["TIL_low"]["receptors"]))

summary_df = pd.DataFrame(
    [
        {
            "group": g,
            "n_sender_cells": universes_panel[g]["n_sender_cells"],
            "n_receiver_cells": universes_panel[g]["n_receiver_cells"],
            "n_background": len(universes_panel[g]["background"]),
            "n_ligands": len(universes_panel[g]["ligands"]),
            "n_receptors": len(universes_panel[g]["receptors"]),
        }
        for g in ["TIL_high", "TIL_low"]
    ]
)

print(summary_df.to_string(index=False))

print("\nUniverse integrity checks:")
print("  candidate_background (ltm_targets):", len(candidate_background))
print("  candidate_ligands (ltm_ligands ∩ lr_ligands):", len(candidate_ligands))
print("  candidate_receptors (lr_receptors):", len(candidate_receptors))
print("  var genes:", len(var_names))
print("  overlap candidate_ligands with var:", len(candidate_ligands.intersection(var_set)))
print("  overlap candidate_receptors with var:", len(candidate_receptors.intersection(var_set)))

if len(background_high) == 0 or len(background_low) == 0:
    raise ValueError("Background is empty for at least one group. Check receiver panel, group labels, and MIN_PCT_EXPRESSED.")
if len(ligands_high) == 0 or len(ligands_low) == 0:
    raise ValueError("Ligands are empty for at least one group. Check sender panel, group labels, and MIN_PCT_EXPRESSED.")
if len(receptors_high) == 0 or len(receptors_low) == 0:
    raise ValueError("Receptors are empty for at least one group. Check receiver panel, group labels, and MIN_PCT_EXPRESSED.")

print("\nExported lists ready for R:")
print("  background_high:", len(background_high), "background_low:", len(background_low))
print("  ligands_high   :", len(ligands_high), "ligands_low   :", len(ligands_low))
print("  receptors_high :", len(receptors_high), "receptors_low :", len(receptors_low))
```

### Target gene universe

```python
import numpy as np
import pandas as pd
import scipy.sparse as sp

# Builds canonical objects for this session:
#   targets_high, targets_low
#   universes (dict with targets/ligands/background/receptors per direction)
# Keeps de_df for downstream axis construction (Cell 5 expects de_df).

LAYER = "LogNormalize"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
ZONE_KEY = ZONE  

MIN_PCT_EXPRESSED = 0.10

required = [
    "adata_sc", "nichenet_priors", "SENDERS", "RECEIVERS",
    "de_tables", "PATHWAY_FILTER", "FDR_CUTOFF", "LOGFC_CUTOFF"
]
missing = [x for x in required if x not in globals()]
if missing:
    raise NameError(f"Missing required objects/params (run upstream cells): {missing}")

if LAYER not in adata_sc.layers:
    raise KeyError(f"adata_sc.layers['{LAYER}'] not found. Available: {list(adata_sc.layers.keys())}")

for k in [CT_KEY, GROUP_KEY]:
    if k not in adata_sc.obs.columns:
        raise KeyError(f"adata_sc.obs['{k}'] not found. Available (first 60): {list(adata_sc.obs.columns)[:60]}")

if ZONE_KEY not in de_tables:
    raise KeyError(f"de_tables['{ZONE_KEY}'] not found. Available keys: {list(de_tables.keys())}")

for k in ["ltm_targets", "ltm_ligands", "lr_ligands", "lr_receptors"]:
    if k not in nichenet_priors:
        raise KeyError(f"nichenet_priors['{k}'] not found. Available keys: {list(nichenet_priors.keys())}")

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

senders_std = set(std_ct(x) for x in SENDERS)
receivers_std = set(std_ct(x) for x in RECEIVERS)

ct = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)

expected_groups = {"TIL_high", "TIL_low"}
groups_present = set(grp.unique())
if not expected_groups.issubset(groups_present):
    raise ValueError(f"Expected {expected_groups} in {GROUP_KEY}, found: {sorted(groups_present)}")

var_names = pd.Index(adata_sc.var_names.astype(str))
var_set = set(var_names)

X = adata_sc.layers[LAYER]
if sp.issparse(X):
    X = X.tocsr()
else:
    X = sp.csr_matrix(X)

ltm_targets = set(map(str, nichenet_priors["ltm_targets"]))
ltm_ligands = set(map(str, nichenet_priors["ltm_ligands"]))
lr_ligands = set(map(str, nichenet_priors["lr_ligands"]))
lr_receptors = set(map(str, nichenet_priors["lr_receptors"]))

candidate_background = ltm_targets
candidate_ligands = ltm_ligands.intersection(lr_ligands)
candidate_receptors = lr_receptors

def expressed_genes(mask_cells: np.ndarray, candidate_genes: set, min_pct: float) -> set:
    idx = np.flatnonzero(mask_cells)
    if idx.size == 0:
        return set()

    genes_present = sorted(candidate_genes.intersection(var_set))
    if not genes_present:
        return set()

    gene_idx = var_names.get_indexer(genes_present)
    Xsub = X[idx, :][:, gene_idx]

    nnz = np.asarray(Xsub.getnnz(axis=0)).ravel()
    pct = nnz / float(idx.size)

    keep = pct >= float(min_pct)
    return set(np.asarray(genes_present, dtype=object)[keep])

panel = {}
for g in ["TIL_high", "TIL_low"]:
    sender_mask = (grp == g).to_numpy() & ct.isin(senders_std).to_numpy()
    receiver_mask = (grp == g).to_numpy() & ct.isin(receivers_std).to_numpy()

    panel[g] = {
        "n_sender_cells": int(sender_mask.sum()),
        "n_receiver_cells": int(receiver_mask.sum()),
        "background": expressed_genes(receiver_mask, candidate_background, MIN_PCT_EXPRESSED),
        "ligands": expressed_genes(sender_mask, candidate_ligands, MIN_PCT_EXPRESSED),
        "receptors": expressed_genes(receiver_mask, candidate_receptors, MIN_PCT_EXPRESSED),
    }

de_df = de_tables[ZONE_KEY].copy()

req = {"gene", "FDR", "logFC"}
if not req.issubset(set(de_df.columns)):
    raise KeyError(f"de_tables['{ZONE_KEY}'] must contain {sorted(req)}. Found: {list(de_df.columns)}")

de_df["gene"] = de_df["gene"].astype(str)
de_df["FDR"] = pd.to_numeric(de_df["FDR"], errors="coerce")
de_df["logFC"] = pd.to_numeric(de_df["logFC"], errors="coerce")

# DE-defined targets by direction (sign anchored to TIL_high_vs_low)
de_high = set(
    de_df.loc[
        (de_df["FDR"] <= float(FDR_CUTOFF)) & (de_df["logFC"] >= float(LOGFC_CUTOFF)),
        "gene"
    ].dropna()
)
de_low = set(
    de_df.loc[
        (de_df["FDR"] <= float(FDR_CUTOFF)) & (de_df["logFC"] <= -float(LOGFC_CUTOFF)),
        "gene"
    ].dropna()
)

# Gate to receiver-expressed background per group
targets_high = de_high.intersection(panel["TIL_high"]["background"])
targets_low = de_low.intersection(panel["TIL_low"]["background"])

targets_high_pre = set(targets_high)
targets_low_pre = set(targets_low)

# Optional pathway filter (apply after background gating)
if PATHWAY_FILTER is not None:
    pf = set(map(str, PATHWAY_FILTER))
    targets_high = targets_high.intersection(pf)
    targets_low = targets_low.intersection(pf)

universes = {
    "targets_high": sorted(map(str, targets_high)),
    "targets_low": sorted(map(str, targets_low)),
    "ligands_high": sorted(map(str, panel["TIL_high"]["ligands"])),
    "ligands_low": sorted(map(str, panel["TIL_low"]["ligands"])),
    "background_high": sorted(map(str, panel["TIL_high"]["background"])),
    "background_low": sorted(map(str, panel["TIL_low"]["background"])),
    "receptors_high": sorted(map(str, panel["TIL_high"]["receptors"])),
    "receptors_low": sorted(map(str, panel["TIL_low"]["receptors"])),
}

summary_df = pd.DataFrame(
    [
        {
            "group": g,
            "n_sender_cells": panel[g]["n_sender_cells"],
            "n_receiver_cells": panel[g]["n_receiver_cells"],
            "n_background": len(panel[g]["background"]),
            "n_ligands": len(panel[g]["ligands"]),
            "n_receptors": len(panel[g]["receptors"]),
        }
        for g in ["TIL_high", "TIL_low"]
    ]
)

print(summary_df.to_string(index=False))

print("\nTargets:")
print("  targets_high (DE∩background) pre pathway:", len(targets_high_pre), "post:", len(targets_high))
print("  targets_low  (DE∩background) pre pathway:", len(targets_low_pre),  "post:", len(targets_low))
print("  pathway filter enabled:", PATHWAY_FILTER is not None)

if PATHWAY_FILTER is not None:
    pf_set = set(map(str, PATHWAY_FILTER))
    print("  targets_high removed by PATHWAY_FILTER:", len(targets_high_pre - pf_set))
    print("  targets_low  removed by PATHWAY_FILTER:", len(targets_low_pre - pf_set))

if len(universes["targets_high"]) == 0 or len(universes["targets_low"]) == 0:
    raise ValueError(
        "Targets empty after DE/background gating and (optional) pathway filter. "
        "Check thresholds, DE orientation, and PATHWAY_FILTER."
    )
```

### Map targets to receiver cell types by expression fraction

```python
# %% [markdown]
# ### Map targets to receiver cell types by expression fraction

# %%
import numpy as np
import pandas as pd
import scipy.sparse as sp

# === CLEAR STALE VARIABLES ===
for _v in ["target_receiver_map_high", "target_receiver_map_low", "target_panel_coverage",
           "global_best_high", "global_best_low", "all_celltypes", "valid_celltypes", 
           "excluded_celltypes", "ct", "grp", "X", "var_names", "var_set"]:
    if _v in globals():
        del globals()[_v]

required = ["adata_sc", "targets_high", "targets_low", "RECEIVERS", "FRACTION_CUTOFF"]
missing = [x for x in required if x not in globals()]
if missing:
    raise NameError(f"Missing required objects/params: {missing}")

# === VERIFY INPUTS ===
print(f"INPUT CHECK:")
print(f"  targets_high: {len(targets_high)} genes")
print(f"  targets_low: {len(targets_low)} genes")
print(f"  RECEIVERS: {len(RECEIVERS)} cell types")
print(f"  FRACTION_CUTOFF: {FRACTION_CUTOFF}")

LAYER = "LogNormalize"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
SAMPLE_KEY = "sample_id"

MIN_SAMPLES_EXPRESSING = 3
PANEL_COVERAGE_THRESHOLD = 0.50

# Minimum cells required for a cell type to be considered in global best computation
# This prevents rare cell types (e.g., 1 cell) from dominating due to noise
MIN_CELLS_FOR_GLOBAL_BEST = 50

if LAYER not in adata_sc.layers:
    raise KeyError(f"adata_sc.layers['{LAYER}'] not found.")
for k in [CT_KEY, GROUP_KEY, SAMPLE_KEY]:
    if k not in adata_sc.obs.columns:
        raise KeyError(f"adata_sc.obs['{k}'] not found.")

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

def is_garbage_ct(ct_name: str) -> bool:
    """Exclude Doublet, Other, Unknown, Unassigned cell types."""
    s = str(ct_name)
    for bad in ["Doublet", "doublet", "Other", "other", "Unknown", "Unassigned"]:
        if bad in s:
            return True
    return False

receivers_std = sorted(set(std_ct(x) for x in RECEIVERS))

# Build fresh from adata_sc
ct = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)
sample_ids = adata_sc.obs[SAMPLE_KEY].astype(str)

X = adata_sc.layers[LAYER]
if sp.issparse(X):
    X = X.tocsr()
else:
    X = sp.csr_matrix(X)

var_names = pd.Index(adata_sc.var_names.astype(str))
var_set = set(var_names)

# Get valid cell types (excluding garbage)
all_celltypes_raw = sorted(ct.unique())
valid_celltypes = [c for c in all_celltypes_raw if not is_garbage_ct(c)]
excluded_celltypes = [c for c in all_celltypes_raw if is_garbage_ct(c)]

print(f"\nCELL TYPE FILTERING:")
print(f"  Raw cell types: {len(all_celltypes_raw)}")
print(f"  Valid cell types (no Doublet/Other): {len(valid_celltypes)}")
print(f"  Excluded: {len(excluded_celltypes)}")

# Compute cell counts per cell type per group
celltype_counts = {}
for group_label in ["TIL_high", "TIL_low"]:
    celltype_counts[group_label] = {}
    for celltype in valid_celltypes:
        m = (grp == group_label).to_numpy() & (ct == celltype).to_numpy()
        n_cells = m.sum()
        celltype_counts[group_label][celltype] = n_cells

# Filter valid cell types for global best by minimum cell count
valid_celltypes_for_global = {}
for group_label in ["TIL_high", "TIL_low"]:
    valid_celltypes_for_global[group_label] = [
        c for c in valid_celltypes 
        if celltype_counts[group_label][c] >= MIN_CELLS_FOR_GLOBAL_BEST
    ]
    n_excluded_by_count = len(valid_celltypes) - len(valid_celltypes_for_global[group_label])
    print(f"  {group_label}: {len(valid_celltypes_for_global[group_label])} cell types with >= {MIN_CELLS_FOR_GLOBAL_BEST} cells ({n_excluded_by_count} excluded)")

def build_trm_for_group(group_label: str, targets_panel) -> pd.DataFrame:
    """Build target-receiver map for panel receivers only."""
    targets_panel = [str(x) for x in list(targets_panel)]
    targets_present = [t for t in targets_panel if t in var_set]
    if len(targets_present) == 0:
        raise ValueError(f"No targets found in var_names for {group_label}.")

    print(f"  {group_label}: {len(targets_present)} targets in var_names")
    
    gene_idx = var_names.get_indexer(targets_present)
    rows = []

    for rct in receivers_std:
        m = (grp == group_label).to_numpy() & (ct == rct).to_numpy()
        idx = np.flatnonzero(m)
        if idx.size == 0:
            continue

        sample_sub = sample_ids.iloc[idx].values
        n_samples = len(np.unique(sample_sub))

        Xsub = X[idx, :][:, gene_idx]
        nnz = np.asarray(Xsub.getnnz(axis=0)).ravel()
        pct = nnz / float(idx.size)
        mean_expr = np.asarray(Xsub.mean(axis=0)).ravel()
        Xsub_dense = Xsub.toarray()

        for i, target in enumerate(targets_present):
            gene_expr = Xsub_dense[:, i]
            n_samples_expressing = len(set(sample_sub[gene_expr > 0]))
            sample_frac = n_samples_expressing / n_samples if n_samples > 0 else 0.0

            if pct[i] >= float(FRACTION_CUTOFF) and n_samples_expressing >= MIN_SAMPLES_EXPRESSING:
                rows.append({
                    "target": target,
                    "receiver_celltype": rct,
                    "target_pct_expr": float(pct[i]),
                    "target_mean_expr": float(mean_expr[i]),
                    "n_samples": n_samples,
                    "n_samples_expressing": n_samples_expressing,
                    "sample_frac": float(sample_frac),
                })

    if not rows:
        return pd.DataFrame(columns=[
            "target", "receiver_celltype", "target_pct_expr", "target_mean_expr",
            "n_samples", "n_samples_expressing", "sample_frac", "specificity", "receiver_rank_score"
        ])

    out = pd.DataFrame(rows)

    # Specificity within panel
    target_mean_all = out.groupby("target")["target_mean_expr"].transform("mean").replace(0, np.nan)
    out["specificity"] = (out["target_mean_expr"] / target_mean_all).fillna(0)

    # Receiver rank score
    out["_mr"] = out.groupby("target")["target_mean_expr"].rank(ascending=False)
    out["_pr"] = out.groupby("target")["target_pct_expr"].rank(ascending=False)
    out["_sfr"] = out.groupby("target")["sample_frac"].rank(ascending=False)
    out["_spr"] = out.groupby("target")["specificity"].rank(ascending=False)
    out["receiver_rank_score"] = 0.35/out["_mr"] + 0.25/out["_pr"] + 0.20/out["_sfr"] + 0.20/out["_spr"]
    out = out.drop(columns=["_mr", "_pr", "_sfr", "_spr"])

    return out

def compute_global_best(group_label: str, targets_list) -> pd.DataFrame:
    """Find global best cell type per target.
    
    Only considers cell types with >= MIN_CELLS_FOR_GLOBAL_BEST cells
    to avoid noise from rare populations.
    """
    targets_list = [str(x) for x in targets_list if str(x) in var_set]
    if not targets_list:
        return pd.DataFrame(columns=["target", "global_best_celltype", "global_best_mean_expr", "global_best_n_cells"])

    gene_idx = var_names.get_indexer(targets_list)
    best = {t: (None, 0.0, 0) for t in targets_list}  # (celltype, mean_expr, n_cells)

    # Only use cell types with sufficient cells
    celltypes_to_check = valid_celltypes_for_global[group_label]
    
    for celltype in celltypes_to_check:
        m = (grp == group_label).to_numpy() & (ct == celltype).to_numpy()
        idx = np.flatnonzero(m)
        if idx.size == 0:
            continue
        n_cells = idx.size
        Xsub = X[idx, :][:, gene_idx]
        mean_expr = np.asarray(Xsub.mean(axis=0)).ravel()
        for i, target in enumerate(targets_list):
            if mean_expr[i] > best[target][1]:
                best[target] = (celltype, float(mean_expr[i]), n_cells)

    results = [{"target": t, "global_best_celltype": v[0], "global_best_mean_expr": v[1], "global_best_n_cells": v[2]} 
               for t, v in best.items() if v[0] is not None]
    return pd.DataFrame(results)

def add_coverage(trm: pd.DataFrame, global_best: pd.DataFrame) -> pd.DataFrame:
    if trm.empty:
        return trm
    trm = trm.merge(global_best, on="target", how="left")
    panel_best = trm.groupby("target")["target_mean_expr"].transform("max")
    trm["panel_coverage_ratio"] = np.where(
        trm["global_best_mean_expr"] > 0,
        panel_best / trm["global_best_mean_expr"],
        1.0
    )
    trm["panel_is_adequate"] = trm["panel_coverage_ratio"] >= PANEL_COVERAGE_THRESHOLD
    trm["coverage_flag"] = np.where(trm["panel_is_adequate"], "OK", "LOW_COVERAGE")
    return trm

# === BUILD TRMs ===
print("\nBUILDING TARGET-RECEIVER MAPS:")
target_receiver_map_high = build_trm_for_group("TIL_high", targets_high)
target_receiver_map_low = build_trm_for_group("TIL_low", targets_low)

# === COMPUTE GLOBAL BEST ===
print(f"\nCOMPUTING GLOBAL BEST (min {MIN_CELLS_FOR_GLOBAL_BEST} cells, excluding Doublets/Other):")
global_best_high = compute_global_best("TIL_high", targets_high)
global_best_low = compute_global_best("TIL_low", targets_low)

# Verify no garbage in global_best
for name, gb in [("high", global_best_high), ("low", global_best_low)]:
    if not gb.empty:
        garbage = [ct for ct in gb["global_best_celltype"].dropna().unique() if is_garbage_ct(ct)]
        if garbage:
            raise RuntimeError(f"BUG: Garbage cell types in global_best_{name}: {garbage}")
        print(f"  global_best_{name}: {len(gb)} targets, no garbage cell types ✓")

# === ADD COVERAGE ===
target_receiver_map_high = add_coverage(target_receiver_map_high, global_best_high)
target_receiver_map_low = add_coverage(target_receiver_map_low, global_best_low)

# === SUMMARY ===
print("\nTARGET-RECEIVER MAP SUMMARY:")
for name, trm in [("TIL_high", target_receiver_map_high), ("TIL_low", target_receiver_map_low)]:
    print(f"\n  {name}:")
    print(f"    rows: {len(trm)}")
    print(f"    targets: {trm['target'].nunique()}")
    print(f"    receivers: {trm['receiver_celltype'].nunique()}")
    if "coverage_flag" in trm.columns:
        n_ok = (trm["coverage_flag"] == "OK").sum()
        n_low = (trm["coverage_flag"] == "LOW_COVERAGE").sum()
        print(f"    coverage: OK={n_ok}, LOW_COVERAGE={n_low}")

# Show LOW_COVERAGE targets
print("\nLOW_COVERAGE TARGETS:")
for name, trm in [("TIL_high", target_receiver_map_high), ("TIL_low", target_receiver_map_low)]:
    low = trm[trm["coverage_flag"] == "LOW_COVERAGE"][["target", "global_best_celltype", "global_best_n_cells", "panel_coverage_ratio"]].drop_duplicates()
    if low.empty:
        print(f"  {name}: None")
    else:
        print(f"  {name}:")
        for _, r in low.iterrows():
            print(f"    {r['target']}: global_best={r['global_best_celltype']} (n={r['global_best_n_cells']}), coverage={r['panel_coverage_ratio']:.1%}")

if target_receiver_map_high.empty or target_receiver_map_low.empty:
    raise ValueError("One or both target_receiver_maps are empty!")

# Store for downstream
target_panel_coverage = pd.concat([
    global_best_high.assign(direction="TIL_high"),
    global_best_low.assign(direction="TIL_low")
], ignore_index=True)

print("\n✓ Created: target_receiver_map_high, target_receiver_map_low, target_panel_coverage")
```

### Build SCE (python)

```python
from pathlib import Path

import numpy as np
import pandas as pd
import scipy.sparse as sp
from scipy.io import mmwrite

if "adata_sc" not in globals():
    raise NameError("adata_sc not found")
if "SENDERS" not in globals() or "RECEIVERS" not in globals():
    raise NameError("SENDERS and/or RECEIVERS not found")

COUNTS_LAYER = "unnormalized_counts"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
SAMPLE_KEY = "sample_id"
BATCH_KEY = "orig.ident"

if COUNTS_LAYER not in adata_sc.layers:
    raise KeyError(f"adata_sc.layers['{COUNTS_LAYER}'] not found. Available: {list(adata_sc.layers.keys())}")

for k in [CT_KEY, GROUP_KEY, SAMPLE_KEY, BATCH_KEY]:
    if k not in adata_sc.obs.columns:
        raise KeyError(f"adata_sc.obs['{k}'] not found. Available obs cols (first 50): {list(adata_sc.obs.columns)[:50]}")

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

def drop_bad_ct(ct_list):
    out = []
    for ct in ct_list:
        s = str(ct)
        if s.startswith("Doublet"):
            continue
        if s.startswith("Other"):
            continue
        out.append(s)
    return sorted(set(out))

senders_py = drop_bad_ct(list(SENDERS))
receivers_py = drop_bad_ct(list(RECEIVERS))

senders_std = [std_ct(x) for x in senders_py]
receivers_std = [std_ct(x) for x in receivers_py]
panel_ct_std = sorted(set(senders_std) | set(receivers_std))

ct_std = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)

m = grp.isin(["TIL_low", "TIL_high"]) & ct_std.isin(panel_ct_std)
n_keep = int(m.sum())
if n_keep == 0:
    raise ValueError("No cells left after filtering to (TIL_low/TIL_high) and sender/receiver panel celltypes.")

adata_sub = adata_sc[m].copy()

X = adata_sub.layers[COUNTS_LAYER]
if not sp.issparse(X):
    X = sp.csr_matrix(X)
else:
    X = X.tocsr()

EFS_ROOT = Path("/mnt/custom-file-systems/efs/fs-09913c1f7db79b6fd")
zone_dir = ZONE.replace("+", "plus").replace("-", "minus")  # CHANGED: derive from ZONE
out_dir = EFS_ROOT / "results" / "visium" / "til_zones" / "nichenet_filtering" / f"mnn_{zone_dir}_input"  # CHANGED: use zone_dir
out_dir.mkdir(parents=True, exist_ok=True)

for fname in ["counts.mtx", "features.tsv", "barcodes.tsv", "metadata.csv", "senders.txt", "receivers.txt"]:
    fpath = out_dir / fname
    if fpath.exists():
        fpath.unlink()

counts_gene_cell = X.T.tocsc()
mmwrite(str(out_dir / "counts.mtx"), counts_gene_cell)

genes = pd.Series(adata_sub.var_names.astype(str))
genes.to_csv(out_dir / "features.tsv", sep="\t", index=False, header=False)

cells = pd.Series(adata_sub.obs_names.astype(str))
cells.to_csv(out_dir / "barcodes.tsv", sep="\t", index=False, header=False)

ct_sub_std = adata_sub.obs[CT_KEY].astype(str).map(std_ct).values

meta = pd.DataFrame(
    {
        "cell": cells.values,
        "celltype": ct_sub_std,
        "sample": adata_sub.obs[SAMPLE_KEY].astype(str).values,
        "group": adata_sub.obs[GROUP_KEY].astype(str).values,
        "batch": adata_sub.obs[BATCH_KEY].astype(str).values,
    }
)
meta.to_csv(out_dir / "metadata.csv", index=False)

pd.Series(senders_std).to_csv(out_dir / "senders.txt", index=False, header=False)
pd.Series(receivers_std).to_csv(out_dir / "receivers.txt", index=False, header=False)

# ADDED: Export for R cell
SCE_INPUT_DIR = f"mnn_{zone_dir}_input"

print("Exported for SCE:")
print("  out_dir:", str(out_dir))
print("  genes x cells:", counts_gene_cell.shape[0], "x", counts_gene_cell.shape[1])
print("  nnz:", int(counts_gene_cell.nnz))
print("  groups:", dict(pd.Series(meta["group"]).value_counts()))
print("  celltypes in panel (senders, receivers):", len(senders_std), ",", len(receivers_std))
```

### Build SCE (R)

```python
%%R -i SCE_INPUT_DIR
suppressPackageStartupMessages({
  library(Matrix)
  library(SingleCellExperiment)
  library(S4Vectors)
  library(dplyr)
  library(readr)
  library(multinichenetr)
})

EFS_ROOT  <- "/mnt/custom-file-systems/efs/fs-09913c1f7db79b6fd"
in_dir <- file.path(EFS_ROOT, "results", "visium", "til_zones", "nichenet_filtering", SCE_INPUT_DIR)  # CHANGED: use variable
stopifnot(dir.exists(in_dir))

counts <- readMM(file.path(in_dir, "counts.mtx"))
counts <- as(counts, "dgCMatrix")

feat <- read_tsv(file.path(in_dir, "features.tsv"), col_names = FALSE, show_col_types = FALSE)
genes <- if (ncol(feat) == 1) feat[[1]] else feat[[2]]

cells <- read_tsv(file.path(in_dir, "barcodes.tsv"), col_names = FALSE, show_col_types = FALSE)[[1]]
meta  <- read_csv(file.path(in_dir, "metadata.csv"), show_col_types = FALSE)

stopifnot(nrow(counts) == length(genes))
stopifnot(ncol(counts) == length(cells))

rownames(counts) <- genes
colnames(counts) <- cells

stopifnot("cell" %in% colnames(meta))
meta <- meta %>% tibble::column_to_rownames("cell")
meta <- meta[colnames(counts), , drop = FALSE]

stopifnot(all(c("celltype", "sample", "group", "batch") %in% colnames(meta)))

meta$celltype_mnn <- make.names(meta$celltype)

sce <- SingleCellExperiment(
  assays = list(counts = counts),
  colData = S4Vectors::DataFrame(meta),
  rowData = S4Vectors::DataFrame(gene = rownames(counts))
)

stopifnot("sample" %in% colnames(colData(sce)))
stopifnot("celltype_mnn" %in% colnames(colData(sce)))
stopifnot("group" %in% colnames(colData(sce)))

colData(sce)$sample_id <- as.character(colData(sce)$sample)
colData(sce)$celltype  <- as.character(colData(sce)$celltype_mnn)

if (anyNA(colData(sce)$sample_id)) stop("sample_id has NA values (check sample in metadata.csv).")
if (anyNA(colData(sce)$celltype))  stop("celltype has NA values (check celltype in metadata.csv).")

senders_oi   <- make.names(readLines(file.path(in_dir, "senders.txt")))
receivers_oi <- make.names(readLines(file.path(in_dir, "receivers.txt")))

ct_levels <- unique(colData(sce)$celltype)
missing_senders   <- setdiff(senders_oi, ct_levels)
missing_receivers <- setdiff(receivers_oi, ct_levels)
if (length(missing_senders) > 0) {
  stop("Senders not found in sce$celltype: ", paste(missing_senders, collapse = ", "))
}
if (length(missing_receivers) > 0) {
  stop("Receivers not found in sce$celltype: ", paste(missing_receivers, collapse = ", "))
}

cat("SCE:", dim(sce)[1], "genes x", dim(sce)[2], "cells\n")
cat("Groups (group):\n")
print(table(colData(sce)$group, useNA = "ifany"))
cat("\nCelltypes (celltype) top 15:\n")
print(sort(table(colData(sce)$celltype), decreasing = TRUE)[1:15])

cat("\nSenders (n=", length(senders_oi), "):\n", sep = "")
print(senders_oi)
cat("\nReceivers (n=", length(receivers_oi), "):\n", sep = "")
print(receivers_oi)

stopifnot(exists("lr_network"), exists("ligand_target_matrix"), exists("weighted_networks"))
cat("\nPriors OK:\n")
cat("  lr_network dim:", paste(dim(lr_network), collapse = " x "), "\n")
cat("  ligand_target_matrix dim:", paste(dim(ligand_target_matrix), collapse = " x "), "\n")
cat("  weighted_networks class:", paste(class(weighted_networks), collapse = ","), "\n")
```

### Compute ligand activity per direction

```python
# Make R-safe copies (rpy2 cannot convert Python sets)
targets_high_r = sorted(list(targets_high))
targets_low_r  = sorted(list(targets_low))

background_high_r = list(background_high)
background_low_r  = list(background_low)
ligands_high_r    = list(ligands_high)
ligands_low_r     = list(ligands_low)

print("R-ready objects:")
print("  targets_high_r:", type(targets_high_r), len(targets_high_r))
print("  targets_low_r :", type(targets_low_r),  len(targets_low_r))
print("  background_high_r:", type(background_high_r), len(background_high_r))
print("  background_low_r :", type(background_low_r),  len(background_low_r))
print("  ligands_high_r:", type(ligands_high_r), len(ligands_high_r))
print("  ligands_low_r :", type(ligands_low_r),  len(ligands_low_r))
```

```python
%%R -i targets_high_r -i targets_low_r -i background_high_r -i background_low_r -i ligands_high_r -i ligands_low_r
suppressPackageStartupMessages({
  library(dplyr)
  library(tibble)
  library(nichenetr)
})

stopifnot(exists("ligand_target_matrix"))

to_chr_unique <- function(x) {
  unique(as.character(unlist(x, use.names = FALSE)))
}

targets_high_chr    <- to_chr_unique(targets_high_r)
targets_low_chr     <- to_chr_unique(targets_low_r)
background_high_chr <- to_chr_unique(background_high_r)
background_low_chr  <- to_chr_unique(background_low_r)
ligands_high_chr    <- to_chr_unique(ligands_high_r)
ligands_low_chr     <- to_chr_unique(ligands_low_r)

ltm <- ligand_target_matrix
ltm_ligands <- rownames(ltm)
ltm_targets <- colnames(ltm)

targets_high_use <- intersect(targets_high_chr, ltm_targets)
targets_low_use  <- intersect(targets_low_chr,  ltm_targets)

bg_high_use <- intersect(background_high_chr, ltm_targets)
bg_low_use  <- intersect(background_low_chr,  ltm_targets)

lig_high_use <- intersect(ligands_high_chr, ltm_ligands)
lig_low_use  <- intersect(ligands_low_chr,  ltm_ligands)

get_top_ltm_ligands <- function(targets_use, ltm, k_per_target) {
  out <- character()
  if (length(targets_use) == 0) return(out)

  for (t in targets_use) {
    w <- ltm[, t, drop = TRUE]
    w <- w[w > 0]
    if (length(w) == 0) next
    top <- sort(w, decreasing = TRUE)
    if (length(top) > k_per_target) top <- top[seq_len(k_per_target)]
    out <- c(out, names(top))
  }
  unique(out)
}

LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY <- 10

top_ltm_high <- get_top_ltm_ligands(targets_high_use, ltm, LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY)
top_ltm_low  <- get_top_ltm_ligands(targets_low_use,  ltm, LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY)

potential_high <- union(lig_high_use, top_ltm_high)
potential_low  <- union(lig_low_use,  top_ltm_low)

cat("Ligand activity inputs after LTM intersection + LTM augmentation:\n")
cat("  LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY=", LIGANDS_FROM_LTM_PER_TARGET_ACTIVITY, "\n", sep = "")
cat("  TIL_high: targets=", length(targets_high_use),
    " background=", length(bg_high_use),
    " sender_panel_ligs=", length(lig_high_use),
    " ltm_top_ligs=", length(top_ltm_high),
    " potential_ligs=", length(potential_high), "\n", sep = "")
cat("  TIL_low : targets=", length(targets_low_use),
    " background=", length(bg_low_use),
    " sender_panel_ligs=", length(lig_low_use),
    " ltm_top_ligs=", length(top_ltm_low),
    " potential_ligs=", length(potential_low), "\n\n", sep = "")

if (length(targets_high_use) < 5) stop("Too few TIL_high targets after intersecting with ligand_target_matrix columns.")
if (length(targets_low_use)  < 5) stop("Too few TIL_low targets after intersecting with ligand_target_matrix columns.")
if (length(bg_high_use) < 50) stop("Too few TIL_high background genes after intersecting with ligand_target_matrix columns.")
if (length(bg_low_use)  < 50) stop("Too few TIL_low background genes after intersecting with ligand_target_matrix columns.")
if (length(potential_high) < 20) stop("Too few TIL_high potential ligands after augmentation.")
if (length(potential_low)  < 20) stop("Too few TIL_low potential ligands after augmentation.")

ligand_activities_high <- predict_ligand_activities(
  geneset = targets_high_use,
  background_expressed_genes = bg_high_use,
  potential_ligands = potential_high,
  ligand_target_matrix = ltm
) %>%
  as_tibble() %>%
  mutate(
    direction = "TIL_high",
    test_ligand = as.character(test_ligand),
    aupr_corrected = as.numeric(aupr_corrected),
    in_sender_panel = test_ligand %in% lig_high_use,
    in_ltm_top = test_ligand %in% top_ltm_high
  ) %>%
  arrange(desc(aupr_corrected), test_ligand) %>%
  mutate(rank = dplyr::row_number()) %>%
  select(test_ligand, direction, aupr_corrected, rank, in_sender_panel, in_ltm_top, everything())

ligand_activities_low <- predict_ligand_activities(
  geneset = targets_low_use,
  background_expressed_genes = bg_low_use,
  potential_ligands = potential_low,
  ligand_target_matrix = ltm
) %>%
  as_tibble() %>%
  mutate(
    direction = "TIL_low",
    test_ligand = as.character(test_ligand),
    aupr_corrected = as.numeric(aupr_corrected),
    in_sender_panel = test_ligand %in% lig_low_use,
    in_ltm_top = test_ligand %in% top_ltm_low
  ) %>%
  arrange(desc(aupr_corrected), test_ligand) %>%
  mutate(rank = dplyr::row_number()) %>%
  select(test_ligand, direction, aupr_corrected, rank, in_sender_panel, in_ltm_top, everything())

ligand_activities <- bind_rows(ligand_activities_high, ligand_activities_low)

cat("Ligand activities computed:\n")
cat("  rows:", nrow(ligand_activities), "\n")
cat("  TIL_high:", nrow(ligand_activities_high), "\n")
cat("  TIL_low :", nrow(ligand_activities_low), "\n\n")

cat("Top 10 (TIL_high):\n")
print(ligand_activities_high %>% slice_head(n = 10) %>% select(test_ligand, aupr_corrected, rank, in_sender_panel, in_ltm_top))
cat("\nTop 10 (TIL_low):\n")
print(ligand_activities_low %>% slice_head(n = 10) %>% select(test_ligand, aupr_corrected, rank, in_sender_panel, in_ltm_top))
```

### Ligand-receptor DE per cell type

```python
# Pass zone prefix to R for file naming
zone_prefix = ZONE.replace("+", "plus").replace("-", "minus")
print(f"Zone prefix for R: {zone_prefix}")
```

```python
%%R -i zone_prefix
# LR-DE: edgeR pseudobulk per celltype, TIL_high vs TIL_low
suppressPackageStartupMessages({
  library(Matrix)
  library(SingleCellExperiment)
  library(SummarizedExperiment)
  library(S4Vectors)
  library(dplyr)
  library(tibble)
  library(readr)
  library(edgeR)
})

stopifnot(exists("sce"), exists("senders_oi"), exists("receivers_oi"))
stopifnot(exists("lr_network"), exists("ligand_target_matrix"))
stopifnot(exists("ligand_activities"))
stopifnot("counts" %in% SummarizedExperiment::assayNames(sce))

EFS_ROOT  <- "/mnt/custom-file-systems/efs/fs-09913c1f7db79b6fd"
FILTER_DIR <- file.path(EFS_ROOT, "results", "visium", "til_zones", "nichenet_filtering")
dir.create(FILTER_DIR, recursive = TRUE, showWarnings = FALSE)

GROUPS <- c("TIL_low","TIL_high")

MIN_CELLS_PER_SAMPLE  <- 5
MIN_SAMPLES_PER_GROUP <- 3

CT_COL     <- "celltype"
GROUP_COL  <- "group"
SAMPLE_COL <- "sample_id"

cd0 <- as.data.frame(colData(sce))
stopifnot(CT_COL %in% colnames(cd0), GROUP_COL %in% colnames(cd0), SAMPLE_COL %in% colnames(cd0))

cd0[[CT_COL]]     <- as.character(cd0[[CT_COL]])
cd0[[GROUP_COL]]  <- as.character(cd0[[GROUP_COL]])
cd0[[SAMPLE_COL]] <- as.character(cd0[[SAMPLE_COL]])

bad_sample <- is.na(cd0[[SAMPLE_COL]]) | cd0[[SAMPLE_COL]] == ""
if (any(bad_sample)) {
  ex <- unique(cd0[[SAMPLE_COL]][bad_sample])[1:min(10, sum(bad_sample))]
  stop(paste0("sample_id contains NA/empty values (n=", sum(bad_sample), "). Examples: ", paste(ex, collapse=", ")))
}

bad_group <- is.na(cd0[[GROUP_COL]]) | !(cd0[[GROUP_COL]] %in% GROUPS)
if (any(bad_group)) {
  ex <- unique(cd0[[GROUP_COL]][bad_group])[1:min(10, sum(bad_group))]
  stop(paste0("group has NA or unexpected labels (n=", sum(bad_group), "). Examples: ", paste(ex, collapse=", ")))
}

cat("Using columns:\n")
cat("  celltype:", CT_COL, "\n")
cat("  group   :", GROUP_COL, "\n")
cat("  sample  :", SAMPLE_COL, "\n")
cat("Cells:", nrow(cd0), "\n\n")

cat("Group counts in sce:\n")
print(table(cd0[[GROUP_COL]], useNA = "ifany"))
cat("\n")

assays_avail <- assayNames(sce)
if (!("counts" %in% assays_avail)) {
  stop(paste0("sce is missing assay 'counts'. Available assays: ", paste(assays_avail, collapse=", ")))
}
counts_all <- assay(sce, "counts")
if (!is(counts_all, "dgCMatrix")) counts_all <- as(counts_all, "CsparseMatrix")
counts_all <- as(counts_all, "dgCMatrix")

cat("Counts matrix:\n")
cat("  class:", class(counts_all)[1], "\n")
cat("  dim  :", nrow(counts_all), "x", ncol(counts_all), "\n\n")

cat("Ligand activities directions:\n")
print(table(as.character(ligand_activities$direction), useNA="ifany"))
cat("\n")

# Harmonize lr_network columns
lr_cols <- colnames(lr_network)
lig_col <- if ("from" %in% lr_cols) "from" else if ("ligand" %in% lr_cols) "ligand" else stop("lr_network missing ligand column (from/ligand).")
rec_col <- if ("to" %in% lr_cols) "to" else if ("receptor" %in% lr_cols) "receptor" else stop("lr_network missing receptor column (to/receptor).")

lr_tbl <- lr_network %>%
  dplyr::distinct(.data[[lig_col]], .data[[rec_col]])
colnames(lr_tbl) <- c("from","to")

# Feasible ligands: use NicheNet activity list, constrained to LTM rows + sce genes
ltm <- ligand_target_matrix
ltm_ligands <- rownames(ltm)
ltm_targets <- colnames(ltm)

feasible_ligands <- intersect(
  unique(as.character(ligand_activities$test_ligand)),
  intersect(ltm_ligands, rownames(counts_all))
)

# Feasible receptors: receptors paired to feasible ligands, constrained to sce genes
feasible_receptors <- intersect(
  unique(lr_tbl$to[lr_tbl$from %in% feasible_ligands]),
  rownames(counts_all)
)

cat("DE universes:\n")
cat("  feasible ligands   :", length(feasible_ligands), "\n")
cat("  feasible receptors :", length(feasible_receptors), "\n\n")

if (length(feasible_ligands) < 20) stop("Too few feasible ligands after intersecting activity ligands with LTM rows and sce genes.")
if (length(feasible_receptors) < 20) stop("Too few feasible receptors after intersecting LR receptors with sce genes.")

# Fast pseudobulk using sparse model matrix
pseudobulk_by_sample <- function(counts_gene_cell, sample_vec) {
  sample_f <- factor(sample_vec)
  M <- Matrix::sparse.model.matrix(~0 + sample_f)
  colnames(M) <- levels(sample_f)
  pb <- counts_gene_cell %*% M
  as(pb, "dgCMatrix")
}

run_edger_for_celltype <- function(celltype_name, genes_keep, mode_label) {
  tryCatch({

    idx_ct <- which(cd0[[CT_COL]] == celltype_name & cd0[[GROUP_COL]] %in% GROUPS)
    if (length(idx_ct) == 0) {
      return(list(status="skip", reason="no cells after filtering", de=NULL, meta=NULL))
    }

    meta_ct <- cd0[idx_ct, , drop=FALSE]

    cells_per_sample <- meta_ct %>%
      dplyr::count(.data[[SAMPLE_COL]], name="n_cells")

    keep_samples <- cells_per_sample %>%
      dplyr::filter(n_cells >= MIN_CELLS_PER_SAMPLE) %>%
      dplyr::pull(.data[[SAMPLE_COL]])

    if (length(keep_samples) == 0) {
      return(list(status="skip", reason=paste0("no samples with >= ", MIN_CELLS_PER_SAMPLE, " cells"), de=NULL, meta=cells_per_sample))
    }

    keep_mask <- meta_ct[[SAMPLE_COL]] %in% keep_samples
    meta_ct <- meta_ct[keep_mask, , drop=FALSE]
    idx_ct  <- idx_ct[keep_mask]

    sample_group_tbl <- meta_ct %>%
      dplyr::group_by(.data[[SAMPLE_COL]]) %>%
      dplyr::summarise(
        n_cells = dplyr::n(),
        n_groups = dplyr::n_distinct(.data[[GROUP_COL]]),
        group = dplyr::first(.data[[GROUP_COL]]),
        .groups="drop"
      )

    bad <- sample_group_tbl %>% dplyr::filter(n_groups > 1)
    if (nrow(bad) > 0) {
      return(list(status="skip", reason="sample maps to >1 group within this celltype", de=NULL, meta=sample_group_tbl))
    }

    n_by_group <- sample_group_tbl %>% dplyr::count(group, name="n_samples")
    n_low  <- n_by_group %>% dplyr::filter(group=="TIL_low")  %>% dplyr::pull(n_samples)
    n_high <- n_by_group %>% dplyr::filter(group=="TIL_high") %>% dplyr::pull(n_samples)
    n_low  <- ifelse(length(n_low)==0, 0, n_low)
    n_high <- ifelse(length(n_high)==0, 0, n_high)

    if (n_low < MIN_SAMPLES_PER_GROUP || n_high < MIN_SAMPLES_PER_GROUP) {
      return(list(
        status="skip",
        reason=paste0("too few samples per group (low=", n_low, ", high=", n_high, ")"),
        de=NULL,
        meta=sample_group_tbl
      ))
    }

    genes_keep <- intersect(as.character(genes_keep), rownames(counts_all))
    if (length(genes_keep) < 10) {
      return(list(status="skip", reason="too few genes in universe after intersecting with sce genes", de=NULL, meta=sample_group_tbl))
    }

    counts_sub <- counts_all[genes_keep, idx_ct, drop=FALSE]
    pb <- pseudobulk_by_sample(counts_sub, meta_ct[[SAMPLE_COL]])

    group_vec <- sample_group_tbl$group
    names(group_vec) <- sample_group_tbl[[SAMPLE_COL]]
    group_vec <- group_vec[colnames(pb)]

    if (any(is.na(group_vec))) {
      na_samples <- colnames(pb)[is.na(group_vec)]
      return(list(
        status="error",
        reason=paste0("group_vec has NA after aligning to pseudobulk columns. Examples: ", paste(head(na_samples, 10), collapse=", ")),
        de=NULL,
        meta=sample_group_tbl
      ))
    }

    group_vec <- factor(group_vec, levels=GROUPS)

    y <- DGEList(counts = as.matrix(pb))
    design <- model.matrix(~ group_vec)

    keep <- filterByExpr(y, design=design)
    y <- y[keep, , keep.lib.sizes=FALSE]
    y <- calcNormFactors(y)

    y <- estimateDisp(y, design, robust=TRUE)
    fit <- glmQLFit(y, design, robust=TRUE)
    qlf <- glmQLFTest(fit, coef=2)

    tt <- topTags(qlf, n=Inf)$table %>%
      tibble::rownames_to_column("gene") %>%
      dplyr::as_tibble() %>%
      dplyr::transmute(
        gene,
        logFC,
        logCPM,
        F,
        PValue,
        FDR,
        celltype = celltype_name,
        mode = mode_label,
        n_samples_total = ncol(pb),
        n_TIL_low = sum(group_vec=="TIL_low"),
        n_TIL_high = sum(group_vec=="TIL_high")
      )

    meta_out <- sample_group_tbl %>%
      dplyr::rename(sample = .data[[SAMPLE_COL]]) %>%
      dplyr::mutate(celltype = celltype_name, mode = mode_label)

    list(status="ok", reason="", de=tt, meta=meta_out)

  }, error = function(e) {
    list(status="error", reason=paste0("edgeR failed: ", conditionMessage(e)), de=NULL, meta=NULL)
  })
}

run_panel <- function(celltypes_vec, genes_keep, mode_label) {
  de_list <- list()
  meta_list <- list()
  status_rows <- list()

  for (ct in celltypes_vec) {
    res <- run_edger_for_celltype(ct, genes_keep, mode_label)
    status_rows[[ct]] <- tibble(
      mode = mode_label,
      celltype = ct,
      status = res$status,
      reason = res$reason
    )
    if (identical(res$status, "ok")) {
      de_list[[ct]] <- res$de
      meta_list[[ct]] <- res$meta
    } else {
      if (!is.null(res$meta)) meta_list[[ct]] <- res$meta
    }
  }

  list(
    status = dplyr::bind_rows(status_rows),
    de = if (length(de_list)) dplyr::bind_rows(de_list) else tibble(),
    meta = if (length(meta_list)) dplyr::bind_rows(meta_list) else tibble()
  )
}

# Run
sender_res   <- run_panel(senders_oi,   feasible_ligands,   "sender_ligand")
receiver_res <- run_panel(receivers_oi, feasible_receptors, "receiver_receptor")

# CHANGED: Use zone_prefix for output file names
out_sender_de     <- file.path(FILTER_DIR, paste0(zone_prefix, "_sender_ligand_DE_edger_by_celltype_TILhigh_vs_TILlow_activityLigs_LTMrows.csv"))
out_receiver_de   <- file.path(FILTER_DIR, paste0(zone_prefix, "_receiver_receptor_DE_edger_by_celltype_TILhigh_vs_TILlow_LRreceptorsOfActivityLigs.csv"))
out_sender_meta   <- file.path(FILTER_DIR, paste0(zone_prefix, "_sender_pseudobulk_samples_by_celltype.csv"))
out_receiver_meta <- file.path(FILTER_DIR, paste0(zone_prefix, "_receiver_pseudobulk_samples_by_celltype.csv"))
out_status        <- file.path(FILTER_DIR, paste0(zone_prefix, "_LR_DE_status_by_celltype.csv"))

readr::write_csv(sender_res$de, out_sender_de)
readr::write_csv(receiver_res$de, out_receiver_de)
readr::write_csv(sender_res$meta, out_sender_meta)
readr::write_csv(receiver_res$meta, out_receiver_meta)
readr::write_csv(dplyr::bind_rows(sender_res$status, receiver_res$status), out_status)

cat("Wrote:\n")
cat(" ", out_sender_de, "\n")
cat(" ", out_receiver_de, "\n")
cat(" ", out_sender_meta, "\n")
cat(" ", out_receiver_meta, "\n")
cat(" ", out_status, "\n\n")

cat("Celltypes with OK DE (senders):", sum(sender_res$status$status=="ok"), "/", length(senders_oi), "\n")
cat("Celltypes with OK DE (receivers):", sum(receiver_res$status$status=="ok"), "/", length(receivers_oi), "\n\n")

if (any(sender_res$status$status=="error") || any(receiver_res$status$status=="error")) {
  cat("Errors detected in some celltypes:\n")
  print(dplyr::bind_rows(sender_res$status, receiver_res$status) %>% dplyr::filter(status=="error"))
  cat("\n")
}

if (nrow(sender_res$de) > 0) {
  cat("Top 10 ligand DE rows overall (by FDR):\n")
  print(sender_res$de %>% dplyr::arrange(FDR) %>% dplyr::slice_head(n=10))
  cat("\n")
}
if (nrow(receiver_res$de) > 0) {
  cat("Top 10 receptor DE rows overall (by FDR):\n")
  print(receiver_res$de %>% dplyr::arrange(FDR) %>% dplyr::slice_head(n=10))
  cat("\n")
}
```

### Construct axis scores

```python
# %% [markdown]
# ### Construct axis scores

# %%
import numpy as np
import pandas as pd
import scipy.sparse as sp
from pathlib import Path

# === CLEAR STALE VARIABLES ===
for _v in ["all_axes", "top_axes", "sender_candidates", "ligand_sender_gating", 
           "base", "axes", "trm_high", "trm_low", "ltm_df", "act_by_dir",
           "allowed_ligands_by_dir", "lr_use_by_dir", "receptors_by_dir"]:
    if _v in globals():
        del globals()[_v]

# === VERIFY INPUTS ===
required = ["target_receiver_map_high", "target_receiver_map_low", "SENDERS", "RECEIVERS",
            "ligand_activities", "sender_de", "receiver_de", "lr_network", 
            "ligand_target_matrix", "de_df", "OUT_DIR",
            "TOP_LIGANDS_PER_DIRECTION", "LTM_TOPK_PER_TARGET", "TOP_AXES_PER_TARGET",
            "TOP_LIGANDS_PER_TARGET", "TOP_AXES_PER_LIGAND_PER_TARGET", "SENDER_TOPK_PER_LIGAND"]
missing = [x for x in required if x not in globals()]
if missing:
    raise NameError(f"Missing: {missing}")

print("INPUT CHECK:")
print(f"  target_receiver_map_high: {len(target_receiver_map_high)} rows, {target_receiver_map_high['target'].nunique()} targets")
print(f"  target_receiver_map_low: {len(target_receiver_map_low)} rows, {target_receiver_map_low['target'].nunique()} targets")

LAYER = "LogNormalize"
CT_KEY = "celltype_level4_curated"
GROUP_KEY = "TIL_group"
SAMPLE_KEY = "sample_id"

MIN_PCT_EXPRESSED_LIGAND = 0.10
MIN_SAMPLES_EXPRESSING_LIGAND = 3

def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

senders_std = sorted({std_ct(x) for x in SENDERS})
receivers_std = sorted({std_ct(x) for x in RECEIVERS})

TOP_LIGANDS_PER_DIRECTION = int(TOP_LIGANDS_PER_DIRECTION)
LTM_TOPK_PER_TARGET = int(LTM_TOPK_PER_TARGET)
TOP_AXES_PER_TARGET = int(TOP_AXES_PER_TARGET)
TOP_LIGANDS_PER_TARGET = int(TOP_LIGANDS_PER_TARGET)
TOP_AXES_PER_LIGAND_PER_TARGET = int(TOP_AXES_PER_LIGAND_PER_TARGET)
SENDER_TOPK_PER_LIGAND = int(SENDER_TOPK_PER_LIGAND) if SENDER_TOPK_PER_LIGAND is not None else None

LIGANDS_FROM_LTM_PER_TARGET = 10
KEEP_LIGANDS = []

OUT_DIR = Path(OUT_DIR)
OUT_DIR.mkdir(parents=True, exist_ok=True)

print(f"\nSELECTION CONTROLS:")
print(f"  TOP_LIGANDS_PER_DIRECTION: {TOP_LIGANDS_PER_DIRECTION}")
print(f"  LTM_TOPK_PER_TARGET: {LTM_TOPK_PER_TARGET}")
print(f"  TOP_AXES_PER_TARGET: {TOP_AXES_PER_TARGET}")
print(f"  TOP_LIGANDS_PER_TARGET: {TOP_LIGANDS_PER_TARGET}")
print(f"  TOP_AXES_PER_LIGAND_PER_TARGET: {TOP_AXES_PER_LIGAND_PER_TARGET}")
print(f"  SENDER_TOPK_PER_LIGAND: {SENDER_TOPK_PER_LIGAND}")

# === COPY INPUTS ===
ligand_activities = pd.DataFrame(ligand_activities).copy()
sender_de = pd.DataFrame(sender_de).copy()
receiver_de = pd.DataFrame(receiver_de).copy()

ligand_activities["test_ligand"] = ligand_activities["test_ligand"].astype(str)
ligand_activities["direction"] = ligand_activities["direction"].astype(str)
ligand_activities["rank"] = pd.to_numeric(ligand_activities["rank"], errors="coerce")
ligand_activities["aupr_corrected"] = pd.to_numeric(ligand_activities["aupr_corrected"], errors="coerce")

for df_ in (sender_de, receiver_de):
    df_["gene"] = df_["gene"].astype(str)
    df_["celltype"] = df_["celltype"].astype(str)
    df_["logFC"] = pd.to_numeric(df_["logFC"], errors="coerce")
    df_["FDR"] = pd.to_numeric(df_["FDR"], errors="coerce")

lr_df0 = pd.DataFrame(lr_network).copy()
cols = set(lr_df0.columns.astype(str))
if {"from", "to"}.issubset(cols):
    lr_df = lr_df0.rename(columns={"from": "ligand", "to": "receptor"})[["ligand", "receptor"]].copy()
elif {"ligand", "receptor"}.issubset(cols):
    lr_df = lr_df0[["ligand", "receptor"]].copy()
else:
    raise KeyError(f"lr_network columns not recognized: {list(lr_df0.columns)}")
lr_df["ligand"] = lr_df["ligand"].astype(str)
lr_df["receptor"] = lr_df["receptor"].astype(str)
lr_df = lr_df.drop_duplicates()

# Fresh copy of TRMs
trm_high = pd.DataFrame(target_receiver_map_high).copy()
trm_low = pd.DataFrame(target_receiver_map_low).copy()

for trm in (trm_high, trm_low):
    trm["target"] = trm["target"].astype(str)
    trm["receiver_celltype"] = trm["receiver_celltype"].astype(str)
    for col in ["target_pct_expr", "target_mean_expr", "receiver_rank_score", "panel_coverage_ratio"]:
        if col in trm.columns:
            trm[col] = pd.to_numeric(trm[col], errors="coerce")

ltm_df_full = pd.DataFrame(
    ligand_target_matrix.values,
    index=pd.Index(ligand_target_matrix.index.astype(str), name="ligand"),
    columns=pd.Index(ligand_target_matrix.columns.astype(str), name="target"),
)

def get_ltm_ligands_for_targets(targets, ltm_df, k):
    targets = [t for t in targets if t in ltm_df.columns]
    ligs = set()
    for t in targets:
        w = ltm_df[t]
        w = w[w > 0].nlargest(k)
        ligs.update(w.index.astype(str))
    return ligs

# === ACTIVITY MAPS ===
act_by_dir = {}
for direction in ["TIL_high", "TIL_low"]:
    act = ligand_activities[ligand_activities["direction"] == direction].dropna(subset=["rank"])
    act = act.sort_values("rank")
    act_by_dir[direction] = {
        "top_ligs": act.nsmallest(TOP_LIGANDS_PER_DIRECTION, "rank")["test_ligand"].tolist(),
        "activity_map": act.set_index("test_ligand")["aupr_corrected"].to_dict(),
        "rank_map": act.set_index("test_ligand")["rank"].to_dict(),
    }

# === LIGAND UNIVERSE ===
allowed_ligands_by_dir = {}
for direction, trm in [("TIL_high", trm_high), ("TIL_low", trm_low)]:
    top_act = set(act_by_dir[direction]["top_ligs"])
    dir_targets = sorted(set(trm["target"]) & set(ltm_df_full.columns))
    top_ltm = get_ltm_ligands_for_targets(dir_targets, ltm_df_full, LIGANDS_FROM_LTM_PER_TARGET)
    allowed_ligands_by_dir[direction] = top_act | top_ltm | set(KEEP_LIGANDS)
    print(f"{direction} ligand universe: {len(allowed_ligands_by_dir[direction])}")

need_ligs = sorted(allowed_ligands_by_dir["TIL_high"] | allowed_ligands_by_dir["TIL_low"])
need_tgts = sorted(set(trm_high["target"]) | set(trm_low["target"]))
use_ligs = sorted(set(ltm_df_full.index) & set(need_ligs))
use_tgts = sorted(set(ltm_df_full.columns) & set(need_tgts))
ltm_df = ltm_df_full.loc[use_ligs, use_tgts].copy()

print(f"LTM subset: {len(use_ligs)} ligands x {len(use_tgts)} targets")

# === SENDER GATING ===
ct_std = adata_sc.obs[CT_KEY].astype(str).map(std_ct)
grp = adata_sc.obs[GROUP_KEY].astype(str)
sample_ids = adata_sc.obs[SAMPLE_KEY].astype(str)

X = adata_sc.layers[LAYER]
if sp.issparse(X):
    X = X.tocsr()
else:
    X = sp.csr_matrix(X)

var_names = pd.Index(adata_sc.var_names.astype(str))
var_set = set(var_names)
ligs_in_var = sorted(set(use_ligs) & var_set)

sender_rows = []
for direction in ["TIL_high", "TIL_low"]:
    dir_ligs = sorted(allowed_ligands_by_dir[direction] & set(ligs_in_var))
    if not dir_ligs:
        continue
    lig_idx = var_names.get_indexer(dir_ligs)

    for sct in senders_std:
        m = (grp == direction).to_numpy() & (ct_std == sct).to_numpy()
        idx = np.flatnonzero(m)
        if idx.size == 0:
            continue

        sample_sub = sample_ids.iloc[idx].values
        n_samples = len(np.unique(sample_sub))
        Xsub = X[idx, :][:, lig_idx]
        pct = np.asarray(Xsub.getnnz(axis=0)).ravel() / idx.size
        mean_expr = np.asarray(Xsub.mean(axis=0)).ravel()
        Xsub_dense = Xsub.toarray()

        for i, lig in enumerate(dir_ligs):
            n_samp_expr = len(set(sample_sub[Xsub_dense[:, i] > 0]))
            if pct[i] >= MIN_PCT_EXPRESSED_LIGAND and n_samp_expr >= MIN_SAMPLES_EXPRESSING_LIGAND:
                sender_rows.append({
                    "direction": direction, "ligand": lig, "sender_celltype": sct,
                    "pct_expr": float(pct[i]), "mean_expr": float(mean_expr[i]),
                    "n_samples_expressing": n_samp_expr,
                    "sample_frac": n_samp_expr / n_samples if n_samples > 0 else 0,
                })

sender_gating = pd.DataFrame(sender_rows)

# Sender specificity and rank score
for direction in ["TIL_high", "TIL_low"]:
    mask = sender_gating["direction"] == direction
    if mask.sum() == 0:
        continue
    lig_mean = sender_gating.loc[mask].groupby("ligand")["mean_expr"].transform("mean").replace(0, np.nan)
    sender_gating.loc[mask, "specificity"] = (sender_gating.loc[mask, "mean_expr"] / lig_mean).fillna(0)

sender_gating["_mr"] = sender_gating.groupby(["direction", "ligand"])["mean_expr"].rank(ascending=False)
sender_gating["_pr"] = sender_gating.groupby(["direction", "ligand"])["pct_expr"].rank(ascending=False)
sender_gating["_sfr"] = sender_gating.groupby(["direction", "ligand"])["sample_frac"].rank(ascending=False)
sender_gating["_spr"] = sender_gating.groupby(["direction", "ligand"])["specificity"].rank(ascending=False)
sender_gating["sender_rank_score"] = 0.35/sender_gating["_mr"] + 0.25/sender_gating["_pr"] + 0.20/sender_gating["_sfr"] + 0.20/sender_gating["_spr"]
sender_gating = sender_gating.drop(columns=["_mr", "_pr", "_sfr", "_spr"])

# Merge sender DE
sender_de_use = sender_de.rename(columns={"gene": "ligand", "celltype": "sender_celltype", "logFC": "ligand_logFC", "FDR": "ligand_FDR"})
sender_candidates = sender_gating.merge(sender_de_use[["ligand", "sender_celltype", "ligand_logFC", "ligand_FDR"]], on=["ligand", "sender_celltype"], how="left")

# Concordance bonus
exp_sign = np.where(sender_candidates["direction"] == "TIL_high", 1, -1)
conc = sender_candidates["ligand_logFC"].notna() & ((exp_sign * sender_candidates["ligand_logFC"]) > 0)
sender_candidates.loc[conc, "sender_rank_score"] += 0.05

sender_candidates = sender_candidates.sort_values(["direction", "ligand", "sender_rank_score"], ascending=[True, True, False])
if SENDER_TOPK_PER_LIGAND is not None:
    sender_candidates = sender_candidates.groupby(["direction", "ligand"], as_index=False).head(SENDER_TOPK_PER_LIGAND)

print(f"Sender candidates: {len(sender_candidates)} rows")

# === LR BY DIRECTION ===
lr_by_dir = {}
for d in ["TIL_high", "TIL_low"]:
    lr_by_dir[d] = lr_df[lr_df["ligand"].isin(allowed_ligands_by_dir[d])].copy()
    lr_by_dir[d]["direction"] = d

# === LTM TOP-K ===
def ltm_topk(direction, trm):
    tgts = sorted(set(trm["target"]) & set(ltm_df.columns))
    elig_ligs = sorted(set(lr_by_dir[direction]["ligand"]) & set(ltm_df.index))
    if not tgts or not elig_ligs:
        return pd.DataFrame(columns=["direction", "target", "ligand", "ltm_weight"])
    sub = ltm_df.loc[elig_ligs, tgts]
    rows = []
    for t in tgts:
        w = sub[t]
        w = w[w > 0].nlargest(LTM_TOPK_PER_TARGET)
        for lig, wt in w.items():
            rows.append({"direction": direction, "target": t, "ligand": str(lig), "ltm_weight": float(wt)})
    return pd.DataFrame(rows)

topk_high = ltm_topk("TIL_high", trm_high)
topk_low = ltm_topk("TIL_low", trm_low)

# === MERGE TRM ===
trm_cols = ["target", "receiver_celltype", "target_pct_expr", "target_mean_expr", "receiver_rank_score",
            "global_best_celltype", "panel_coverage_ratio", "coverage_flag"]
trm_cols_high = [c for c in trm_cols if c in trm_high.columns]
trm_cols_low = [c for c in trm_cols if c in trm_low.columns]

base_high = topk_high.merge(trm_high[trm_cols_high], on="target", how="inner")
base_low = topk_low.merge(trm_low[trm_cols_low], on="target", how="inner")
base = pd.concat([base_high, base_low], ignore_index=True)

print(f"Base: {len(base)} rows (high={len(base_high)}, low={len(base_low)})")

# Attach activity
base["ligand_activity"] = np.nan
base["ligand_activity_rank"] = np.nan
for d in ["TIL_high", "TIL_low"]:
    m = base["direction"] == d
    base.loc[m, "ligand_activity"] = base.loc[m, "ligand"].map(act_by_dir[d]["activity_map"])
    base.loc[m, "ligand_activity_rank"] = base.loc[m, "ligand"].map(act_by_dir[d]["rank_map"])

# Attach sender
base = base.merge(sender_candidates, on=["direction", "ligand"], how="inner")
print(f"After sender merge: {len(base)} rows")

# Attach receptor
lr_dir = pd.concat([lr_by_dir["TIL_high"], lr_by_dir["TIL_low"]], ignore_index=True)
axes = base.merge(lr_dir, on=["direction", "ligand"], how="inner")

receiver_tbl = receiver_de.rename(columns={"celltype": "receiver_celltype", "gene": "receptor", "logFC": "receptor_logFC", "FDR": "receptor_FDR"})
axes = axes.merge(receiver_tbl[["receiver_celltype", "receptor", "receptor_logFC", "receptor_FDR"]], on=["receiver_celltype", "receptor"], how="left")

# Attach target DE
de_use = de_df[["gene", "logFC", "FDR"]].copy()
de_use.columns = ["target", "target_logFC", "target_FDR"]
axes = axes.merge(de_use, on="target", how="left")

axes = axes.dropna(subset=["receptor_logFC"])
print(f"After receptor filter: {len(axes)} axes")

# === SCORE ===
def score_axes(df):
    df = df.copy()
    for d in ["TIL_high", "TIL_low"]:
        m = df["direction"] == d
        if m.sum() == 0:
            continue
        ranks = df.loc[m, "ligand_activity_rank"]
        max_r = ranks.max()
        df.loc[m, "activity_pctl"] = 1 - (ranks - 1) / max_r if max_r > 1 else 1.0
        df.loc[m, "ltm_pctl"] = df.loc[m, "ltm_weight"].rank(pct=True)
        if "receiver_rank_score" in df.columns:
            df.loc[m, "receiver_rank_pctl"] = df.loc[m, "receiver_rank_score"].rank(pct=True)
        else:
            df.loc[m, "receiver_rank_pctl"] = 0.5

    exp_sign = np.where(df["direction"] == "TIL_high", 1, -1)
    df["ligand_concordant"] = (exp_sign * df["ligand_logFC"]) > 0
    df["receptor_concordant"] = (exp_sign * df["receptor_logFC"]) > 0

    df["axis_score"] = (
        0.35 * df["activity_pctl"].fillna(0) +
        0.25 * df["ltm_pctl"].fillna(0) +
        0.15 * df["receiver_rank_pctl"].fillna(0) +
        0.10 * df["ligand_concordant"].astype(float) +
        0.15 * df["receptor_concordant"].astype(float)
    )
    return df

all_axes = score_axes(axes)
all_axes = all_axes.sort_values(["direction", "target", "axis_score"], ascending=[True, True, False])
all_axes["axis_rank"] = all_axes.groupby(["direction", "target"]).cumcount() + 1
all_axes = all_axes.drop_duplicates(subset=["direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype"])

# === TOP AXES SELECTION ===
lig_best = all_axes.groupby(["direction", "target", "ligand"]).agg(
    best_score=("axis_score", "max"), best_ltm=("ltm_weight", "max"), best_act=("ligand_activity", "max"), n=("axis_score", "size")
).reset_index()
lig_best = lig_best.sort_values(["direction", "target", "best_score", "best_ltm", "best_act"], ascending=[True, True, False, False, False])
lig_best["lig_rank"] = lig_best.groupby(["direction", "target"]).cumcount() + 1
keep_ligs = lig_best[lig_best["lig_rank"] <= TOP_LIGANDS_PER_TARGET][["direction", "target", "ligand"]]

sel = all_axes.merge(keep_ligs, on=["direction", "target", "ligand"], how="inner")
sel = sel.sort_values(["direction", "target", "ligand", "axis_score"], ascending=[True, True, True, False])
sel["ax_in_lig"] = sel.groupby(["direction", "target", "ligand"]).cumcount() + 1
sel = sel[sel["ax_in_lig"] <= TOP_AXES_PER_LIGAND_PER_TARGET]
sel = sel.sort_values(["direction", "target", "axis_score"], ascending=[True, True, False])
sel["ax_final"] = sel.groupby(["direction", "target"]).cumcount() + 1
top_axes = sel[sel["ax_final"] <= TOP_AXES_PER_TARGET].copy()
top_axes["axis_rank"] = top_axes.groupby(["direction", "target"]).cumcount() + 1

# === SUMMARY ===
print(f"\nAXES BUILT:")
print(f"  all_axes: {len(all_axes)} rows, {all_axes['target'].nunique()} targets")
print(f"  top_axes: {len(top_axes)} rows, {top_axes['target'].nunique()} targets")

for d in ["TIL_high", "TIL_low"]:
    sub = all_axes[all_axes["direction"] == d]
    print(f"\n  {d}: {len(sub)} axes, {sub['target'].nunique()} targets, {sub['ligand'].nunique()} ligands")

# === SAVE ===
all_axes.to_csv(OUT_DIR / "all_axes.csv", index=False)
top_axes.to_csv(OUT_DIR / "top_axes.csv", index=False)
print(f"\n✓ Saved to {OUT_DIR}")
```

### Summarize axis scores for plot selection

```python
# %% [markdown]
# ### Summarize axis scores for plot selection

# %%
import pandas as pd
import numpy as np
from pathlib import Path
import hashlib

# === CLEAR STALE VARIABLES ===
for _v in ["df", "df_all", "df_top", "axes_catalog", "axes_shortlist", "short", "short2"]:
    if _v in globals():
        del globals()[_v]

# === VERIFY INPUTS ===
if "OUT_DIR" not in globals():
    raise NameError("OUT_DIR not found.")
if "top_axes" not in globals():
    raise NameError("top_axes not found. Run Cell 5 first.")

OUT_DIR = Path(OUT_DIR)

print("INPUT CHECK:")
print(f"  top_axes: {len(top_axes)} rows, {top_axes['target'].nunique()} targets")
if "all_axes" in globals():
    print(f"  all_axes: {len(all_axes)} rows (for reference, not used in catalog)")

# Use top_axes (filtered) for catalog, NOT all_axes
df = pd.DataFrame(top_axes).copy()

req_cols = {"direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype",
            "target_logFC", "target_FDR", "ligand_activity", "ltm_weight",
            "ligand_concordant", "receptor_concordant", "axis_score", "axis_rank"}
missing = req_cols - set(df.columns)
if missing:
    raise KeyError(f"top_axes missing columns: {sorted(missing)}")

# Normalize dtypes
for c in ["direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype"]:
    df[c] = df[c].astype(str)
for c in ["axis_score", "axis_rank", "target_logFC", "target_FDR", "ligand_activity", "ltm_weight"]:
    df[c] = pd.to_numeric(df[c], errors="coerce")
df["ligand_concordant"] = df["ligand_concordant"].astype(bool)
df["receptor_concordant"] = df["receptor_concordant"].astype(bool)

if "receiver_rank_score" in df.columns:
    df["receiver_rank_score"] = pd.to_numeric(df["receiver_rank_score"], errors="coerce")
if "panel_coverage_ratio" in df.columns:
    df["panel_coverage_ratio"] = pd.to_numeric(df["panel_coverage_ratio"], errors="coerce")

# === FILTER 1: positive activity ===
n_before = len(df)
df = df[df["ligand_activity"] > 0].copy()
print(f"\nFilter: ligand_activity > 0: {n_before} -> {len(df)} axes")

# === ABUNDANCE ANNOTATIONS ===
def std_ct(x: str) -> str:
    return str(x).replace("-", ".")

abundance_available = False
if "EFS_ROOT" in globals() and "ZONE" in globals():
    ab_path = Path(EFS_ROOT) / "results/visium/til_zones/celltype_zone_specificity_curated/celltype_zone_specificity_level4_curated.csv"

    if ab_path.exists():
        ab = pd.read_csv(ab_path)
        ab["group"] = ab["group"].astype(str)
        ab["zone"] = ab["zone"].astype(str)
        ab["celltype_std"] = ab["celltype_level4_curated"].astype(str).map(std_ct)
        ab_zone = ab[ab["zone"] == str(ZONE)].copy()

        delta_tbl = ab_zone.pivot_table(index="celltype_std", columns="group", values="mean_zone", aggfunc="first").reset_index()
        if "TIL_high" in delta_tbl.columns and "TIL_low" in delta_tbl.columns:
            delta_tbl["mean_zone_high_minus_low"] = pd.to_numeric(delta_tbl["TIL_high"], errors="coerce") - pd.to_numeric(delta_tbl["TIL_low"], errors="coerce")
            delta_tbl = delta_tbl[["celltype_std", "mean_zone_high_minus_low"]]
        else:
            delta_tbl = pd.DataFrame({"celltype_std": ab_zone["celltype_std"].unique(), "mean_zone_high_minus_low": np.nan})

        df["sender_celltype_std"] = df["sender_celltype"].map(std_ct)
        df["receiver_celltype_std"] = df["receiver_celltype"].map(std_ct)

        # Sender abundance
        sender_ab = ab_zone.rename(columns={"celltype_std": "sender_celltype_std", "mean_zone": "sender_mean_zone"})
        sender_ab = sender_ab[["group", "sender_celltype_std", "sender_mean_zone"]].copy()
        df = df.merge(sender_ab, left_on=["direction", "sender_celltype_std"], right_on=["group", "sender_celltype_std"], how="left")
        df = df.drop(columns=["group"], errors="ignore")

        # Receiver abundance
        receiver_ab = ab_zone.rename(columns={"celltype_std": "receiver_celltype_std", "mean_zone": "receiver_mean_zone"})
        receiver_ab = receiver_ab[["group", "receiver_celltype_std", "receiver_mean_zone"]].copy()
        df = df.merge(receiver_ab, left_on=["direction", "receiver_celltype_std"], right_on=["group", "receiver_celltype_std"], how="left")
        df = df.drop(columns=["group"], errors="ignore")

        # Delta
        delta_sender = delta_tbl.rename(columns={"celltype_std": "sender_celltype_std", "mean_zone_high_minus_low": "sender_mean_zone_high_minus_low"})
        delta_receiver = delta_tbl.rename(columns={"celltype_std": "receiver_celltype_std", "mean_zone_high_minus_low": "receiver_mean_zone_high_minus_low"})
        df = df.merge(delta_sender, on="sender_celltype_std", how="left")
        df = df.merge(delta_receiver, on="receiver_celltype_std", how="left")

        print(f"Abundance annotations loaded from: {ab_path}")
        abundance_available = True
    else:
        print(f"Abundance CSV not found: {ab_path}")
else:
    print("EFS_ROOT or ZONE not set, skipping abundance annotations")

# === FILTER 2: concordance/abundance ===
ABUNDANCE_THRESHOLD = 0.05
n_before = len(df)

if abundance_available:
    high_mask = df["direction"] == "TIL_high"
    low_mask = df["direction"] == "TIL_low"
    high_keep = df["ligand_concordant"] | (df["sender_mean_zone_high_minus_low"] > ABUNDANCE_THRESHOLD)
    low_keep = df["ligand_concordant"] | (df["sender_mean_zone_high_minus_low"] < -ABUNDANCE_THRESHOLD)
    df = df[(high_mask & high_keep) | (low_mask & low_keep)].copy()
else:
    df = df[df["ligand_concordant"]].copy()

print(f"Filter: concordance/abundance: {n_before} -> {len(df)} axes")

# === AXIS ID ===
key_cols = ["direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype"]
df = df.drop_duplicates(subset=key_cols).copy()

def axis_hash_id(row):
    s = "|".join(str(row[c]) for c in key_cols)
    h = hashlib.md5(s.encode()).hexdigest()[:6].upper()
    return ("H" if row["direction"] == "TIL_high" else "L") + h

df["axis_id"] = df.apply(axis_hash_id, axis=1)

# === RANK SCORE ===
df["de_strength"] = -np.log10(df["target_FDR"].clip(lower=1e-300))
df["rank_score"] = df["axis_score"] * df["de_strength"] * df["target_logFC"].abs()

# === VERIFY AXES PER TARGET ===
axes_per_target = df.groupby(["direction", "target"]).size()
print(f"\nAxes per target: min={axes_per_target.min()}, max={axes_per_target.max()}, mean={axes_per_target.mean():.1f}")

# === CATALOG COLUMNS ===
catalog_cols = ["axis_id", "direction", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype",
                "axis_score", "axis_rank", "rank_score", "target_logFC", "target_FDR",
                "ligand_activity", "ltm_weight", "ligand_concordant", "receptor_concordant"]

# Add optional columns if present
for c in ["receiver_rank_score", "global_best_celltype", "global_best_n_cells", "panel_coverage_ratio", "coverage_flag",
          "sender_mean_zone", "receiver_mean_zone", "sender_mean_zone_high_minus_low", "receiver_mean_zone_high_minus_low"]:
    if c in df.columns:
        catalog_cols.append(c)

axes_catalog = df[[c for c in catalog_cols if c in df.columns]].copy()
axes_catalog.to_csv(OUT_DIR / "axes_catalog.csv", index=False)

# === SHORTLIST (further filtered subset) ===
AXIS_SCORE_MIN = 0.5
MAX_AXIS_RANK_PER_TARGET = 3
TARGET_FDR_MAX = 0.05
TARGET_LOGFC_MIN = 0.25
AXES_PER_TARGET_DE_BAL = 2

short = df[(df["axis_score"] > AXIS_SCORE_MIN) & (df["axis_rank"] <= MAX_AXIS_RANK_PER_TARGET)].copy()
short = short.dropna(subset=["target_FDR", "target_logFC", "rank_score"])
short = short[short["target_FDR"] <= TARGET_FDR_MAX]

high = short[(short["direction"] == "TIL_high") & (short["target_logFC"] >= TARGET_LOGFC_MIN)]
low = short[(short["direction"] == "TIL_low") & (short["target_logFC"] <= -TARGET_LOGFC_MIN)]
short = pd.concat([high, low])
short = short.sort_values(["direction", "target", "rank_score"], ascending=[True, True, False])
axes_shortlist = short.groupby(["direction", "target"], as_index=False).head(AXES_PER_TARGET_DE_BAL)
axes_shortlist = axes_shortlist[[c for c in catalog_cols if c in axes_shortlist.columns]]
axes_shortlist.to_csv(OUT_DIR / "axes_shortlist.csv", index=False)

# === SUMMARY ===
print(f"\nCATALOG SUMMARY:")
for d in ["TIL_high", "TIL_low"]:
    sub = df[df["direction"] == d]
    print(f"  {d}: {len(sub)} axes, {sub['target'].nunique()} targets, {sub['ligand'].nunique()} ligands")

if "coverage_flag" in df.columns:
    print(f"\nPANEL COVERAGE:")
    for d in ["TIL_high", "TIL_low"]:
        sub = df[df["direction"] == d]
        if sub.empty:
            continue
        n_ok = (sub["coverage_flag"] == "OK").sum()
        n_low = (sub["coverage_flag"] == "LOW_COVERAGE").sum()
        print(f"  {d}: OK={n_ok}, LOW_COVERAGE={n_low}")

print(f"\nSHORTLIST: {len(axes_shortlist)} axes, {axes_shortlist['target'].nunique()} targets")

print(f"\nWROTE:")
print(f"  {OUT_DIR / 'axes_catalog.csv'}")
print(f"  {OUT_DIR / 'axes_shortlist.csv'}")

# === DISPLAY ===
display_cols = ["axis_id", "target", "ligand", "receptor", "sender_celltype", "receiver_celltype",
                "axis_score", "rank_score", "target_logFC"]
if "coverage_flag" in axes_catalog.columns:
    display_cols.append("coverage_flag")
display_cols = [c for c in display_cols if c in axes_catalog.columns]

print(f"\nTOP 15 TIL_high:")
display(axes_catalog[axes_catalog["direction"] == "TIL_high"].nlargest(15, "rank_score")[display_cols])

print(f"\nTOP 15 TIL_low:")
display(axes_catalog[axes_catalog["direction"] == "TIL_low"].nlargest(15, "rank_score")[display_cols])

# === WARNINGS ===
if "coverage_flag" in axes_shortlist.columns:
    low_cov = axes_shortlist[axes_shortlist["coverage_flag"] == "LOW_COVERAGE"]
    if len(low_cov) > 0:
        print(f"\n⚠️ WARNING: {len(low_cov)} shortlist axes have LOW_COVERAGE:")
        for _, r in low_cov.iterrows():
            gb_ct = r.get('global_best_celltype', 'N/A')
            gb_n = r.get('global_best_n_cells', 'N/A')
            print(f"  {r['axis_id']}: {r['target']} (global_best={gb_ct}, n={gb_n})")

print("\n✓ Done")
```

### Circos plotting

```python
# Pass ZONE to R for circos plotting
ZONE_FOR_R = str(ZONE)
print(f"Passing ZONE to R: {ZONE_FOR_R}")
```

```python
%%R -i ZONE_FOR_R
suppressPackageStartupMessages({
  library(dplyr)
  library(readr)
  library(circlize)
})

options(bitmapType = "cairo")

EFS_ROOT <- "/home/sagemaker-user/custom-file-systems/efs/fs-09913c1f7db79b6fd"
ZONE <- ZONE_FOR_R
zone_dir <- gsub("\\+", "plus", gsub("-", "minus", ZONE))
OUT_DIR <- file.path(EFS_ROOT, "results", "visium", "til_zones", "nichenet_analysis", zone_dir)
stopifnot(dir.exists(OUT_DIR))

AXES_SELECTED_CSV <- file.path(OUT_DIR, "axes_selected.csv")
stopifnot(file.exists(AXES_SELECTED_CSV))

MAX_EDGES <- Inf

LINK_ALPHA <- 0.5

BORDER_COL <- "#777777"
BORDER_LWD <- 0.25

TITLE_CEX <- 1.00

START_DEGREE <- -8
SMALL_GAP <- 1
BIG_GAP <- 5

WIDTH_MIN <- 0.10
WIDTH_MAX <- 2.50

COLOR_CLIP_LO <- 0.02
COLOR_CLIP_HI <- 0.98

CANVAS_PAD <- 0.40

LABEL_CEX <- 0.46
LABEL_RADIUS <- 1.0
LABEL_NICEFACING <- TRUE
LABEL_MIN_DEG <- 0

OUT_PNG <- file.path(OUT_DIR, "circos_selected_axes_senders_receivers_only.png")
PNG_SIZE <- 2600
PNG_RES <- 240

axes <- readr::read_csv(AXES_SELECTED_CSV, show_col_types = FALSE)

required_cols <- c(
  "axis_id","direction",
  "sender_celltype","receiver_celltype",
  "ligand","target",
  "rank_score","ligand_activity"
)
missing <- setdiff(required_cols, colnames(axes))
if (length(missing) > 0) stop(paste0("axes_selected.csv missing columns: ", paste(missing, collapse = ", ")))

axes <- axes %>%
  transmute(
    axis_id = as.character(axis_id),
    direction = as.character(direction),
    sender = as.character(sender_celltype),
    receiver = as.character(receiver_celltype),
    ligand = as.character(ligand),
    target = as.character(target),
    rank_score = as.numeric(rank_score),
    ligand_activity = as.numeric(ligand_activity)
  ) %>%
  filter(
    direction %in% c("TIL_high","TIL_low"),
    is.finite(rank_score),
    is.finite(ligand_activity)
  )

if (nrow(axes) == 0) stop("No rows left after filtering. Check direction/rank_score/ligand_activity.")

if (is.finite(MAX_EDGES) && nrow(axes) > MAX_EDGES) {
  axes <- axes %>% slice_head(n = MAX_EDGES)
}

rescale01 <- function(x) {
  x <- as.numeric(x)
  ok <- is.finite(x)
  if (sum(ok) == 0) return(rep(0.5, length(x)))
  lo <- min(x[ok]); hi <- max(x[ok])
  if (!is.finite(lo) || !is.finite(hi) || lo == hi) return(rep(0.5, length(x)))
  y <- (x - lo) / (hi - lo)
  y[!is.finite(y)] <- 0.5
  pmin(pmax(y, 0), 1)
}

clip01_quantile <- function(x, qlo = 0.02, qhi = 0.98) {
  x <- as.numeric(x)
  ok <- is.finite(x)
  if (sum(ok) == 0) return(rep(0.5, length(x)))
  lo <- suppressWarnings(as.numeric(stats::quantile(x[ok], probs = qlo, na.rm = TRUE, names = FALSE)))
  hi <- suppressWarnings(as.numeric(stats::quantile(x[ok], probs = qhi, na.rm = TRUE, names = FALSE)))
  if (!is.finite(lo) || !is.finite(hi) || lo == hi) return(rescale01(x))
  y <- (x - lo) / (hi - lo)
  y[!is.finite(y)] <- 0.5
  pmin(pmax(y, 0), 1)
}

axes <- axes %>%
  mutate(
    width_norm = rescale01(ligand_activity),
    w = WIDTH_MIN + width_norm * (WIDTH_MAX - WIDTH_MIN)
  )

rank_norm <- clip01_quantile(axes$rank_score, COLOR_CLIP_LO, COLOR_CLIP_HI)

col_high <- circlize::colorRamp2(c(0, 1), c("#FEE6CE", "#D55E00"))
col_low  <- circlize::colorRamp2(c(0, 1), c("#DEEBF7", "#0072B2"))

axes$link_col <- vapply(seq_len(nrow(axes)), function(i) {
  if (axes$direction[i] == "TIL_high") col_high(rank_norm[i]) else col_low(rank_norm[i])
}, character(1))
axes$link_col <- adjustcolor(axes$link_col, alpha.f = LINK_ALPHA)

S_PREFIX <- "S|"
R_PREFIX <- "R|"
clean_label <- function(x) sub("^[SR]\\|", "", x)

senders <- axes %>%
  group_by(sender) %>%
  summarise(total = sum(w), .groups = "drop") %>%
  arrange(desc(total)) %>%
  pull(sender)

receivers <- axes %>%
  group_by(receiver) %>%
  summarise(total = sum(w), .groups = "drop") %>%
  arrange(desc(total)) %>%
  pull(receiver)

sender_sectors <- paste0(S_PREFIX, senders)
receiver_sectors <- paste0(R_PREFIX, receivers)
sectors <- c(sender_sectors, receiver_sectors)

ct_names <- sort(unique(c(senders, receivers)))
ct_cols <- setNames(grDevices::hcl.colors(length(ct_names), palette = "Dark 3"), ct_names)

grid_cols <- c(
  setNames(ct_cols[clean_label(sender_sectors)], sender_sectors),
  setNames(ct_cols[clean_label(receiver_sectors)], receiver_sectors)
)

axes <- axes %>%
  mutate(
    s_sector = paste0(S_PREFIX, sender),
    r_sector = paste0(R_PREFIX, receiver),
    row_id = seq_len(n())
  )

sender_totals <- tapply(axes$w, axes$s_sector, sum)
receiver_totals <- tapply(axes$w, axes$r_sector, sum)

totals <- setNames(rep(1, length(sectors)), sectors)
if (length(sender_totals) > 0) totals[names(sender_totals)] <- sender_totals
if (length(receiver_totals) > 0) totals[names(receiver_totals)] <- receiver_totals
totals[!is.finite(totals) | totals <= 0] <- 1

xlim_mat <- cbind(rep(0, length(sectors)), as.numeric(totals[sectors]))
rownames(xlim_mat) <- sectors

# Collapse ligands per sender:
# Allocate one contiguous sender-side block per (sender, ligand), then subdivide within block for links.
lig_groups <- axes %>%
  group_by(s_sector, ligand) %>%
  summarise(
    group_w = sum(w),
    group_order = min(row_id),
    .groups = "drop"
  ) %>%
  arrange(s_sector, group_order)

sender_cursor <- setNames(rep(0, length(sender_sectors)), sender_sectors)
lig_groups$lig_start <- NA_real_
lig_groups$lig_end <- NA_real_

for (k in seq_len(nrow(lig_groups))) {
  ss <- lig_groups$s_sector[k]
  ww <- lig_groups$group_w[k]
  st <- sender_cursor[ss]
  en <- st + ww
  lig_groups$lig_start[k] <- st
  lig_groups$lig_end[k] <- en
  sender_cursor[ss] <- en
}

axes <- axes %>%
  left_join(lig_groups %>% select(s_sector, ligand, lig_start, lig_end), by = c("s_sector", "ligand")) %>%
  group_by(s_sector, ligand) %>%
  arrange(row_id, .by_group = TRUE) %>%
  mutate(
    s_x1 = lig_start + dplyr::lag(cumsum(w), default = 0),
    s_x2 = lig_start + cumsum(w)
  ) %>%
  ungroup()

# Receiver allocation stays per-edge (unchanged)
receiver_cursor <- setNames(rep(0, length(receiver_sectors)), receiver_sectors)
r_x1 <- numeric(nrow(axes)); r_x2 <- numeric(nrow(axes))

for (i in seq_len(nrow(axes))) {
  rr <- axes$r_sector[i]
  ww <- axes$w[i]
  r_x1[i] <- receiver_cursor[rr]
  r_x2[i] <- receiver_cursor[rr] + ww
  receiver_cursor[rr] <- r_x2[i]
}

axes$r_x1 <- r_x1; axes$r_x2 <- r_x2
axes$s_mid <- (axes$s_x1 + axes$s_x2) / 2
axes$r_mid <- (axes$r_x1 + axes$r_x2) / 2

gap_after <- c(
  rep(SMALL_GAP, max(length(sender_sectors) - 1, 0)), BIG_GAP,
  rep(SMALL_GAP, max(length(receiver_sectors) - 1, 0)), BIG_GAP
)

thin_by_min_deg <- function(theta, min_deg) {
  if (!is.finite(min_deg) || min_deg <= 0) return(rep(TRUE, length(theta)))
  ord <- order(theta)
  keep <- rep(FALSE, length(theta))
  last <- NA_real_
  for (j in ord) {
    if (!is.finite(theta[j])) next
    if (is.na(last) || abs(theta[j] - last) >= min_deg) {
      keep[j] <- TRUE
      last <- theta[j]
    }
  }
  keep
}

plot_one <- function() {
  circos.clear()
  circos.par(
    start.degree = START_DEGREE,
    gap.after = gap_after,
    cell.padding = c(0, 0, 0, 0),
    track.margin = c(0.01, 0.01),
    canvas.xlim = c(-(1 + CANVAS_PAD), 1 + CANVAS_PAD),
    canvas.ylim = c(-(1 + CANVAS_PAD), 1 + CANVAS_PAD)
  )

  circos.initialize(factors = sectors, xlim = xlim_mat)

  circos.trackPlotRegion(
    factors = sectors,
    ylim = c(0, 1),
    track.height = 0.08,
    bg.border = NA,
    panel.fun = function(x, y) {
      si <- get.cell.meta.data("sector.index")
      xl <- get.cell.meta.data("xlim")
      circos.rect(xl[1], 0, xl[2], 1, col = grid_cols[si], border = NA)
    }
  )

  for (i in seq_len(nrow(axes))) {
    circos.link(
      sector.index1 = axes$s_sector[i],
      point1 = c(axes$s_x1[i], axes$s_x2[i]),
      sector.index2 = axes$r_sector[i],
      point2 = c(axes$r_x1[i], axes$r_x2[i]),
      col = axes$link_col[i],
      border = BORDER_COL,
      lwd = BORDER_LWD,
      directional = 1,
      arr.type = "big.arrow",
      arr.length = 0.04
    )
  }

  # Sender-side ligand labels: one per (sender, ligand), positioned at the center of the ligand block
  ligand_labels <- lig_groups %>%
    mutate(lig_mid = (lig_start + lig_end) / 2)

  theta_lig <- rep(NA_real_, nrow(ligand_labels))
  for (k in seq_len(nrow(ligand_labels))) {
    ss <- ligand_labels$s_sector[k]
    m <- circlize(ligand_labels$lig_mid[k], 0.5, sector.index = ss, track.index = 1)
    theta_lig[k] <- m[1, 1]
  }

  keep_lig <- rep(TRUE, nrow(ligand_labels))
  if (LABEL_MIN_DEG > 0) {
    for (ss in unique(ligand_labels$s_sector)) {
      idx <- which(ligand_labels$s_sector == ss)
      keep_lig[idx] <- thin_by_min_deg(theta_lig[idx], LABEL_MIN_DEG)
    }
  }

  # Receiver-side target labels remain per-edge (unchanged)
  theta_r <- rep(NA_real_, nrow(axes))
  for (rr in unique(axes$r_sector)) {
    idx <- which(is.finite(axes$r_mid) & axes$r_sector == rr)
    if (length(idx) == 0) next
    m <- circlize(axes$r_mid[idx], rep(0.5, length(idx)), sector.index = rr, track.index = 1)
    theta_r[idx] <- m[, 1]
  }

  keep_r <- rep(TRUE, nrow(axes))
  if (LABEL_MIN_DEG > 0) {
    for (rr in unique(axes$r_sector)) {
      idx <- which(axes$r_sector == rr)
      keep_r[idx] <- thin_by_min_deg(theta_r[idx], LABEL_MIN_DEG)
    }
  }

  op <- par(xpd = NA)

  for (k in seq_len(nrow(ligand_labels))) {
    if (!keep_lig[k] || !is.finite(theta_lig[k])) next

    ang <- theta_lig[k] * pi / 180
    x <- LABEL_RADIUS * cos(ang)
    y <- LABEL_RADIUS * sin(ang)

    srt <- theta_lig[k]
    adjx <- 0
    if (LABEL_NICEFACING && theta_lig[k] > 90 && theta_lig[k] < 270) {
      srt <- theta_lig[k] + 180
      adjx <- 1
    }

    text(
      x = x, y = y,
      labels = ligand_labels$ligand[k],
      srt = srt,
      adj = c(adjx, 0.5),
      cex = LABEL_CEX
    )
  }

  for (i in seq_len(nrow(axes))) {
    if (keep_r[i] && is.finite(theta_r[i])) {
      ang <- theta_r[i] * pi / 180
      x <- LABEL_RADIUS * cos(ang)
      y <- LABEL_RADIUS * sin(ang)

      srt <- theta_r[i]
      adjx <- 0
      if (LABEL_NICEFACING && theta_r[i] > 90 && theta_r[i] < 270) {
        srt <- theta_r[i] + 180
        adjx <- 1
      }

      text(x = x, y = y, labels = axes$target[i], srt = srt, adj = c(adjx, 0.5), cex = LABEL_CEX)
    }
  }

  par(op)

  legend(
    "topleft",
    legend = ct_names,
    fill = ct_cols[ct_names],
    border = NA,
    cex = 0.75,
    bty = "n",
    title = "Cell types"
  )

  legend(
    "bottomleft",
    legend = c("TIL_high axes", "TIL_low axes"),
    fill = c("#D55E00", "#0072B2"),
    border = NA,
    cex = 0.95,
    bty = "n",
    title = "Axis direction"
  )

  title(
    main = paste0(
      "Selected axes circos | ", zone_dir,
      " | edges=", nrow(axes),
      " | width ~ ligand_activity, color ~ rank_score"
    ),
    cex.main = TITLE_CEX
  )

  circos.clear()
}

# Write PNG once
png(OUT_PNG, width = PNG_SIZE, height = PNG_SIZE, res = PNG_RES, type = "cairo")
plot_one()
dev.off()

OUT_PDF <- sub("\\.png$", ".pdf", OUT_PNG)

pdf(OUT_PDF, width = PNG_SIZE / PNG_RES, height = PNG_SIZE / PNG_RES, useDingbats = FALSE)
plot_one()
dev.off()

cat("Wrote:", OUT_PNG, "\n")
cat("Wrote:", OUT_PDF, "\n")
cat("Edges:", nrow(axes), "\n")
cat("ligand_activity range:", min(axes$ligand_activity), "to", max(axes$ligand_activity), "\n")
cat("rank_score range:", min(axes$rank_score), "to", max(axes$rank_score), "\n")
cat("rank_score clip quantiles:", COLOR_CLIP_LO, "to", COLOR_CLIP_HI, "\n")
cat("Label radius:", LABEL_RADIUS, " Label cex:", LABEL_CEX, " Label min deg:", LABEL_MIN_DEG, "\n")
```

```python
# ============================================================
# Check why ECM genes aren't appearing as TARGETS in CCC
# ============================================================

ECM_GENES = ["MMP12", "LAMA5", "HAPLN4", "CSPG4", "LOXL3"]

print("=" * 70)
print("ECM GENES AS TARGETS - WHY MISSING FROM CCC?")
print("=" * 70)

# 1. Are they in the target sets that went into NicheNet?
print("\n1. PRESENCE IN TARGET SETS (targets_high / targets_low)")
print("-" * 70)
for gene in ECM_GENES:
    in_high = gene in targets_high if 'targets_high' in dir() else "N/A"
    in_low = gene in targets_low if 'targets_low' in dir() else "N/A"
    print(f"   {gene:8s}  targets_high={str(in_high):5s}  targets_low={str(in_low):5s}")

# 2. Check the filtering steps - are they in the ligand-target matrix as targets?
print("\n2. IN NICHENET LIGAND-TARGET MATRIX (as targets)")
print("-" * 70)
ltm = nichenet_priors["ligand_target_matrix"]
ltm_targets = set(ltm.columns)
for gene in ECM_GENES:
    in_ltm = gene in ltm_targets
    print(f"   {gene:8s}  in_ltm_targets={in_ltm}")

# 3. Are they expressed in receivers? (required to be a valid target)
print("\n3. EXPRESSION IN RECEIVER CELL TYPES")
print("-" * 70)
ct_col = "celltype_level4_curated"

for gene in ECM_GENES:
    print(f"\n   {gene}:")
    if gene not in adata_sc.var_names:
        print("      NOT FOUND in adata_sc.var_names")
        continue
    
    gene_idx = adata_sc.var_names.get_loc(gene)
    expr = adata_sc.X[:, gene_idx]
    if hasattr(expr, "toarray"):
        expr = expr.toarray().flatten()
    
    # Check expression specifically in RECEIVER cell types
    receiver_expression = []
    for ct in adata_sc.obs[ct_col].unique():
        ct_std = ct.replace("-", ".")
        if ct_std in RECEIVERS_STD:
            mask = adata_sc.obs[ct_col] == ct
            ct_expr = expr[mask]
            frac = (ct_expr > 0).sum() / len(ct_expr)
            receiver_expression.append((ct, frac))
    
    if receiver_expression:
        receiver_expression.sort(key=lambda x: -x[1])
        for ct, frac in receiver_expression[:5]:
            flag = "OK" if frac >= 0.10 else "LOW"
            print(f"      {ct:40s} {frac*100:5.1f}%  [{flag}]")
    else:
        print("      Not checked in any receiver cell type")

# 4. The key filter: Were they DE + in pathway filter + in receiver background?
print("\n\n4. FILTERING SUMMARY FOR TIL+ (why not in final target set)")
print("-" * 70)

for gene in ECM_GENES:
    print(f"\n   {gene}:")
    
    # Step A: In DE table?
    if gene not in de_tables["TIL+"]["gene"].values:
        print("      ✗ FAILED: Not in TIL+ DE table at all")
        continue
    
    row = de_tables["TIL+"][de_tables["TIL+"]["gene"] == gene].iloc[0]
    
    # Step B: Passes DE threshold?
    passes_de = row["FDR"] <= 0.05 and abs(row["logFC"]) >= 0.25
    if not passes_de:
        print(f"      ✗ FAILED: Not significant (FDR={row['FDR']:.2e}, logFC={row['logFC']:+.3f})")
        print(f"               Need: FDR≤0.05 AND |logFC|≥0.25")
        continue
    
    # Step C: In pathway filter?
    if gene not in PATHWAY_FILTER:
        print("      ✗ FAILED: Not in PATHWAY_FILTER")
        continue
    
    # Step D: In ligand-target matrix as target?
    if gene not in ltm_targets:
        print("      ✗ FAILED: Not in NicheNet ligand-target matrix")
        continue
    
    # Step E: In receiver background?
    in_bg_high = gene in background_high if 'background_high' in dir() else "?"
    in_bg_low = gene in background_low if 'background_low' in dir() else "?"
    print(f"      DE: PASS | PathwayFilter: PASS | LTM: PASS")
    print(f"      background_high={in_bg_high}  background_low={in_bg_low}")

print("\n" + "=" * 70)
```

```python
# %% [markdown]
# ### UMAP Visualization by TIL Group

# %%
# UMAP visualization: All cells, TIL_low only, TIL_high only
# Colored by celltype_level3_figure annotations

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from pathlib import Path

# -----------------------------------------------------------------------------
# USER CONFIGURATION
# -----------------------------------------------------------------------------

# Annotation column
CT_KEY = "celltype_level3_figure"
GROUP_KEY = "TIL_group"
UMAP_KEY = "X_umap"

# Plot appearance
DOT_SIZE = 0.5          # Size of each point
DOT_ALPHA = 0.3         # Transparency (0-1)
SHOW_LABELS = False     # Add text labels on cluster centroids
LABEL_FONTSIZE = 6      # Font size for labels if enabled

# Figure settings
FIG_WIDTH = 8           # Width of each UMAP panel
FIG_HEIGHT = 8          # Height of each UMAP panel
DPI = 150               # Resolution for saved figures
TITLE_FONTSIZE = 12

# Output
OUT_DIR = Path("/home/sagemaker-user/gbm-collaboration_2/analysis_sc_merged/umap_figures")
OUT_DIR.mkdir(parents=True, exist_ok=True)

# -----------------------------------------------------------------------------
# VALIDATION
# -----------------------------------------------------------------------------
required = ["adata_sc"]
missing = [x for x in required if x not in globals()]
if missing:
    raise NameError(f"Missing required objects: {missing}")

if CT_KEY not in adata_sc.obs.columns:
    raise KeyError(f"'{CT_KEY}' not found in adata_sc.obs. Available: {list(adata_sc.obs.columns)[:30]}")
if GROUP_KEY not in adata_sc.obs.columns:
    raise KeyError(f"'{GROUP_KEY}' not found in adata_sc.obs. Available: {list(adata_sc.obs.columns)[:30]}")
if UMAP_KEY not in adata_sc.obsm:
    raise KeyError(f"'{UMAP_KEY}' not found in adata_sc.obsm. Available: {list(adata_sc.obsm.keys())}")

# -----------------------------------------------------------------------------
# PREPARE DATA
# -----------------------------------------------------------------------------
umap_coords = adata_sc.obsm[UMAP_KEY]
celltypes = adata_sc.obs[CT_KEY].astype(str)
groups = adata_sc.obs[GROUP_KEY].astype(str)

# Get unique cell types and create color palette
unique_cts = sorted(celltypes.unique())
n_colors = len(unique_cts)

# Use a qualitative colormap with enough colors
if n_colors <= 20:
    cmap = plt.cm.get_cmap("tab20", n_colors)
elif n_colors <= 40:
    # Combine tab20 and tab20b
    colors1 = plt.cm.tab20(np.linspace(0, 1, 20))
    colors2 = plt.cm.tab20b(np.linspace(0, 1, 20))
    all_colors = np.vstack([colors1, colors2])
    cmap = lambda i: all_colors[i % 40]
else:
    cmap = plt.cm.get_cmap("gist_ncar", n_colors)

ct_to_color = {ct: cmap(i) if n_colors <= 40 else cmap(i / n_colors) 
               for i, ct in enumerate(unique_cts)}

print(f"Cell types: {n_colors}")
print(f"Total cells: {len(celltypes)}")
print(f"TIL groups: {groups.value_counts().to_dict()}")

# -----------------------------------------------------------------------------
# PLOTTING FUNCTION
# -----------------------------------------------------------------------------
def plot_umap(mask, title, ax, show_labels=False):
    """Plot UMAP for cells matching mask."""
    coords = umap_coords[mask]
    cts = celltypes[mask]
    
    # Shuffle for better visualization (avoid overplotting bias)
    idx = np.random.permutation(len(coords))
    coords = coords[idx]
    cts = cts.iloc[idx]
    
    # Plot each cell type
    for ct in unique_cts:
        ct_mask = cts == ct
        if ct_mask.sum() == 0:
            continue
        ax.scatter(
            coords[ct_mask, 0],
            coords[ct_mask, 1],
            c=[ct_to_color[ct]],
            s=DOT_SIZE,
            alpha=DOT_ALPHA,
            linewidths=0,
            rasterized=True,
        )
    
    # Add labels at centroids
    if show_labels:
        for ct in unique_cts:
            ct_mask = cts == ct
            if ct_mask.sum() < 10:
                continue
            centroid_x = coords[ct_mask, 0].mean()
            centroid_y = coords[ct_mask, 1].mean()
            ax.text(
                centroid_x, centroid_y, ct,
                fontsize=LABEL_FONTSIZE,
                ha='center', va='center',
                fontweight='bold',
                bbox=dict(boxstyle='round,pad=0.2', facecolor='white', alpha=0.7, edgecolor='none')
            )
    
    ax.set_title(title, fontsize=TITLE_FONTSIZE, fontweight='bold')
    ax.set_xlabel("UMAP1")
    ax.set_ylabel("UMAP2")
    ax.set_xticks([])
    ax.set_yticks([])
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_visible(False)
    ax.spines['left'].set_visible(False)
    
    return ax

# -----------------------------------------------------------------------------
# CREATE MASKS
# -----------------------------------------------------------------------------
mask_all = np.ones(len(celltypes), dtype=bool)
mask_til_low = (groups == "TIL_low").values
mask_til_high = (groups == "TIL_high").values

print(f"\nCell counts:")
print(f"  All: {mask_all.sum()}")
print(f"  TIL_low: {mask_til_low.sum()}")
print(f"  TIL_high: {mask_til_high.sum()}")

# -----------------------------------------------------------------------------
# PLOT 1: ALL CELLS
# -----------------------------------------------------------------------------
fig1, ax1 = plt.subplots(1, 1, figsize=(FIG_WIDTH, FIG_HEIGHT), dpi=DPI)
plot_umap(mask_all, f"All Cells (n={mask_all.sum():,})", ax1, show_labels=SHOW_LABELS)
plt.tight_layout()
fig1.savefig(OUT_DIR / "umap_all_cells.png", dpi=DPI, bbox_inches='tight', facecolor='white')
fig1.savefig(OUT_DIR / "umap_all_cells.pdf", bbox_inches='tight', facecolor='white')
plt.show()
print(f"Saved: {OUT_DIR / 'umap_all_cells.png'}")

# -----------------------------------------------------------------------------
# PLOT 2: TIL_LOW ONLY
# -----------------------------------------------------------------------------
fig2, ax2 = plt.subplots(1, 1, figsize=(FIG_WIDTH, FIG_HEIGHT), dpi=DPI)
plot_umap(mask_til_low, f"TIL_low Only (n={mask_til_low.sum():,})", ax2, show_labels=SHOW_LABELS)
plt.tight_layout()
fig2.savefig(OUT_DIR / "umap_til_low.png", dpi=DPI, bbox_inches='tight', facecolor='white')
fig2.savefig(OUT_DIR / "umap_til_low.pdf", bbox_inches='tight', facecolor='white')
plt.show()
print(f"Saved: {OUT_DIR / 'umap_til_low.png'}")

# -----------------------------------------------------------------------------
# PLOT 3: TIL_HIGH ONLY
# -----------------------------------------------------------------------------
fig3, ax3 = plt.subplots(1, 1, figsize=(FIG_WIDTH, FIG_HEIGHT), dpi=DPI)
plot_umap(mask_til_high, f"TIL_high Only (n={mask_til_high.sum():,})", ax3, show_labels=SHOW_LABELS)
plt.tight_layout()
fig3.savefig(OUT_DIR / "umap_til_high.png", dpi=DPI, bbox_inches='tight', facecolor='white')
fig3.savefig(OUT_DIR / "umap_til_high.pdf", bbox_inches='tight', facecolor='white')
plt.show()
print(f"Saved: {OUT_DIR / 'umap_til_high.png'}")

# -----------------------------------------------------------------------------
# PLOT 4: LEGEND ONLY (separate figure)
# -----------------------------------------------------------------------------
fig_legend, ax_legend = plt.subplots(1, 1, figsize=(4, max(2, n_colors * 0.25)), dpi=DPI)
ax_legend.axis('off')

# Create legend patches
patches = [mpatches.Patch(color=ct_to_color[ct], label=ct) for ct in unique_cts]

# Add legend
legend = ax_legend.legend(
    handles=patches,
    loc='center left',
    frameon=False,
    fontsize=8,
    ncol=1 if n_colors <= 25 else 2,
    title=CT_KEY,
    title_fontsize=10,
)

plt.tight_layout()
fig_legend.savefig(OUT_DIR / "umap_legend.png", dpi=DPI, bbox_inches='tight', facecolor='white')
fig_legend.savefig(OUT_DIR / "umap_legend.pdf", bbox_inches='tight', facecolor='white')
plt.show()
print(f"Saved: {OUT_DIR / 'umap_legend.png'}")

# -----------------------------------------------------------------------------
# SUMMARY
# -----------------------------------------------------------------------------
print("\n" + "=" * 70)
print("UMAP FIGURES SAVED")
print("=" * 70)
print(f"Output directory: {OUT_DIR}")
print(f"Files:")
print(f"  - umap_all_cells.png/pdf")
print(f"  - umap_til_low.png/pdf")
print(f"  - umap_til_high.png/pdf")
print(f"  - umap_legend.png/pdf")
print(f"\nSettings used:")
print(f"  DOT_SIZE: {DOT_SIZE}")
print(f"  DOT_ALPHA: {DOT_ALPHA}")
print(f"  SHOW_LABELS: {SHOW_LABELS}")
print(f"  Annotation: {CT_KEY}")
```

### Stack Bar Plot

```python
# %% [markdown]
# ### Stacked barplot: Malignant cell type composition by sample

# %%
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.cluster.hierarchy import linkage, leaves_list
from scipy.spatial.distance import pdist

# User config
CT_KEY = "celltype_level4_curated"
SAMPLE_KEY = "orig.ident"

EXCLUDE_CELLTYPES = [
    "Malignant_other_1",
]

FIG_WIDTH = 15
FIG_HEIGHT = 5
TITLE_FONTSIZE = 12
LABEL_FONTSIZE = 10
LEGEND_FONTSIZE = 8

# Get malignant cell types (excluding specified)
all_cts = adata_sc.obs[CT_KEY].astype(str).unique()
malignant_cts = sorted([ct for ct in all_cts if ct.startswith("Malignant") and ct not in EXCLUDE_CELLTYPES])
print(f"Malignant cell types: {len(malignant_cts)}")
if EXCLUDE_CELLTYPES:
    print(f"Excluded: {EXCLUDE_CELLTYPES}")

# Subset to malignant cells only
mask = adata_sc.obs[CT_KEY].astype(str).isin(malignant_cts)
obs_mal = adata_sc.obs.loc[mask, [CT_KEY, SAMPLE_KEY]].copy()
obs_mal[SAMPLE_KEY] = obs_mal[SAMPLE_KEY].astype(str)
obs_mal[CT_KEY] = obs_mal[CT_KEY].astype(str)

print(f"Malignant cells: {len(obs_mal)}")

# Count matrix: samples x cell types
counts = obs_mal.groupby([SAMPLE_KEY, CT_KEY]).size().unstack(fill_value=0)
counts = counts.reindex(columns=malignant_cts, fill_value=0)

# Normalize to proportions per sample
props = counts.div(counts.sum(axis=1), axis=0)

# Cluster samples by composition
if len(props) > 2:
    dist = pdist(props.values, metric="euclidean")
    Z = linkage(dist, method="ward")
    sample_order = props.index[leaves_list(Z)].tolist()
else:
    sample_order = props.index.tolist()

props_ordered = props.loc[sample_order]

# Color palette for cell types
n_cts = len(malignant_cts)
ct_colors = dict(zip(malignant_cts, sns.color_palette("tab20", n_cts)))

# Plot
fig, ax = plt.subplots(figsize=(FIG_WIDTH, FIG_HEIGHT))

bottom = np.zeros(len(sample_order))
for ct in malignant_cts:
    vals = props_ordered[ct].values
    ax.bar(range(len(sample_order)), vals, bottom=bottom, label=ct, color=ct_colors[ct], width=1.0, edgecolor="none")
    bottom += vals

ax.set_xlim(-0.5, len(sample_order) - 0.5)
ax.set_ylim(0, 1)
ax.set_ylabel("Proportion", fontsize=LABEL_FONTSIZE)
ax.set_xlabel("Samples (clustered)", fontsize=LABEL_FONTSIZE)
ax.set_title("Malignant cell type composition by sample", fontsize=TITLE_FONTSIZE)

ax.set_xticks([])
ax.grid(False)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.legend(loc="upper left", bbox_to_anchor=(1.01, 1), title="Cell type", fontsize=LEGEND_FONTSIZE)

plt.tight_layout()
plt.show()

# Summary
print(f"\nSamples: {len(props)}")
print(f"\nMean proportion per cell type:")
print(props.mean().sort_values(ascending=False).round(3).to_string())
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scanpy as sc
import bbknn

# User config
CT_KEY = "celltype_level4_curated"
SAMPLE_KEY = "orig.ident"

EXCLUDE_CELLTYPES = [
    "Malignant_other_1",
]

N_PCS = 30
N_NEIGHBORS = 15

FIG_WIDTH = 8
FIG_HEIGHT = 6
DOT_SIZE = 3
DOT_ALPHA = 0.5
TITLE_FONTSIZE = 12
LEGEND_FONTSIZE = 8

# Get malignant cell types (excluding specified)
all_cts = adata_sc.obs[CT_KEY].astype(str).unique()
malignant_cts = sorted([ct for ct in all_cts if ct.startswith("Malignant") and ct not in EXCLUDE_CELLTYPES])
print(f"Malignant cell types: {len(malignant_cts)}")
if EXCLUDE_CELLTYPES:
    print(f"Excluded: {EXCLUDE_CELLTYPES}")

# Subset to malignant cells
mask = adata_sc.obs[CT_KEY].astype(str).isin(malignant_cts)
adata_mal = adata_sc[mask].copy()
print(f"Malignant cells (pre-filter): {adata_mal.n_obs}")
print(f"Samples (pre-filter): {adata_mal.obs[SAMPLE_KEY].nunique()}")

# Filter out batches with too few cells for BBKNN
batch_counts = adata_mal.obs[SAMPLE_KEY].value_counts()
offending_batches = batch_counts[batch_counts < N_NEIGHBORS]

if offending_batches.shape[0] > 0:
    print(f"\nFiltering out batches with < {N_NEIGHBORS} malignant cells:")
    print(offending_batches.sort_values().to_string())
    keep_batches = batch_counts[batch_counts >= N_NEIGHBORS].index
    adata_mal = adata_mal[adata_mal.obs[SAMPLE_KEY].isin(keep_batches)].copy()
else:
    print(f"\nNo batches have < {N_NEIGHBORS} malignant cells. No filtering applied.")

print(f"\nMalignant cells (post-filter): {adata_mal.n_obs}")
print(f"Samples (post-filter): {adata_mal.obs[SAMPLE_KEY].nunique()}")

# Run PCA
print("\nRunning PCA...")
sc.pp.pca(adata_mal, n_comps=N_PCS)

# Run BBKNN with sample as batch key
print("Running BBKNN...")
bbknn.bbknn(adata_mal, batch_key=SAMPLE_KEY, n_pcs=N_PCS, neighbors_within_batch=N_NEIGHBORS)

# Compute UMAP on BBKNN graph
print("Computing UMAP...")
sc.tl.umap(adata_mal)

# Get coordinates and labels
umap_coords = adata_mal.obsm["X_umap"]
ct_labels = adata_mal.obs[CT_KEY].astype(str).values

# Color palette
n_cts = len(malignant_cts)
palette = dict(zip(malignant_cts, sns.color_palette("tab20", n_cts)))

# Plot
fig, ax = plt.subplots(figsize=(FIG_WIDTH, FIG_HEIGHT))

for ct in malignant_cts:
    ct_mask = ct_labels == ct
    ax.scatter(
        umap_coords[ct_mask, 0],
        umap_coords[ct_mask, 1],
        c=[palette[ct]],
        label=ct,
        s=DOT_SIZE,
        alpha=DOT_ALPHA,
        edgecolors="none"
    )

ax.set_xlabel("UMAP1")
ax.set_ylabel("UMAP2")
ax.set_title("Malignant cells (BBKNN corrected)", fontsize=TITLE_FONTSIZE)

ax.grid(False)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.legend(
    loc="upper left",
    bbox_to_anchor=(1.01, 1),
    title="Cell type",
    fontsize=LEGEND_FONTSIZE,
    markerscale=3
)

plt.tight_layout()
plt.show()

# Summary
print(f"\nCells per type (post-filter):")
print(pd.Series(ct_labels).value_counts().sort_index().to_string())
```

### Corelative cell type co-occurance analysis

```python
# %% [markdown]
# ### Cell type correlation analysis with cluster detection

# %%
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, fcluster
from scipy.spatial.distance import squareform

# User config
CT_KEY_L3 = "celltype_level3_curated"
CT_KEY_L4 = "celltype_level4_curated"
SAMPLE_KEY = "orig.ident"

EXCLUDE_PATTERNS = ["Doublet", "doublet"]

FIG_SIZE_L3 = 8
FIG_SIZE_L4 = 12
LABEL_FONTSIZE_L3 = 7
LABEL_FONTSIZE_L4 = 6
TITLE_FONTSIZE = 12

CMAP = "seismic"
VMIN = -1
VMAX = 1

CLUSTER_METHOD = "ward" # or "average", "complete"
CLUSTER_THRESHOLD_L3 = 1
CLUSTER_THRESHOLD_L4 = 1

SHOW_CBAR = False

OUT_DIR = "/home/sagemaker-user/gbm-collaboration_2/analysis_sc_merged"

def is_excluded(ct_name):
    for pattern in EXCLUDE_PATTERNS:
        if pattern in str(ct_name):
            return True
    return False

def build_proportion_matrix(adata, ct_key, sample_key):
    obs = adata.obs[[sample_key, ct_key]].copy()
    obs[sample_key] = obs[sample_key].astype(str)
    obs[ct_key] = obs[ct_key].astype(str)
    
    counts = obs.groupby([sample_key, ct_key]).size().unstack(fill_value=0)
    keep_cols = [c for c in counts.columns if not is_excluded(c)]
    counts = counts[keep_cols]
    props = counts.div(counts.sum(axis=1), axis=0)
    
    return props

def clean_corr(corr):
    corr = corr.replace([np.inf, -np.inf], np.nan)
    corr = corr.dropna(axis=0, how="all").dropna(axis=1, how="all")
    common = corr.index.intersection(corr.columns)
    corr = corr.loc[common, common]
    return corr.fillna(0.0)

def compute_linkage(corr, method="ward"):
    dist_matrix = 1 - corr.values
    np.fill_diagonal(dist_matrix, 0)
    dist_matrix = (dist_matrix + dist_matrix.T) / 2
    dist_matrix = np.clip(dist_matrix, 0, None)
    condensed = squareform(dist_matrix, checks=False)
    Z = linkage(condensed, method=method)
    return Z

def plot_corr_clustermap_with_clusters(corr, threshold, method, fig_size, label_fontsize, title):
    Z = compute_linkage(corr, method)
    cluster_labels = fcluster(Z, t=threshold, criterion="distance")
    
    cluster_df = pd.DataFrame({
        "celltype": corr.index,
        "cluster": cluster_labels
    })
    
    n_clusters = cluster_df["cluster"].nunique()
    cluster_palette = dict(zip(
        sorted(cluster_df["cluster"].unique()),
        sns.color_palette("tab20", n_clusters)
    ))
    
    cluster_map = cluster_df.set_index("celltype")["cluster"].to_dict()
    row_colors = pd.Series(
        [cluster_palette[cluster_map[ct]] for ct in corr.index],
        index=corr.index,
        name="Cluster"
    )
    
    cbar_pos = (0.02, 0.8, 0.03, 0.15) if SHOW_CBAR else None
    
    g = sns.clustermap(
        corr,
        cmap=CMAP,
        vmin=VMIN,
        vmax=VMAX,
        linewidths=0,
        figsize=(fig_size, fig_size),
        dendrogram_ratio=0.1,
        cbar_pos=cbar_pos,
        tree_kws={"linewidths": 0.5},
        row_colors=row_colors,
        col_colors=row_colors,
        row_linkage=Z,
        col_linkage=Z,
    )
    
    ax = g.ax_heatmap
    ax.grid(False)
    
    rows = g.data2d.index
    cols = g.data2d.columns
    
    ax.set_yticks(np.arange(len(rows)) + 0.5)
    ax.set_yticklabels(rows, rotation=0, fontsize=label_fontsize)
    
    ax.set_xticks(np.arange(len(cols)) + 0.5)
    ax.set_xticklabels(cols, rotation=90, fontsize=label_fontsize)
    
    ax.set_aspect("equal")
    
    g.fig.suptitle(title, y=0.98, fontsize=TITLE_FONTSIZE)
    
    return g, cluster_df

def build_cluster_dict(cluster_df):
    """Build dict mapping cluster number to list of cell types."""
    cluster_dict = {}
    for c in sorted(cluster_df["cluster"].unique()):
        members = cluster_df[cluster_df["cluster"] == c]["celltype"].tolist()
        cluster_dict[c] = members
    return cluster_dict

def print_clusters(cluster_dict, level_name):
    print(f"\n{level_name} clusters ({len(cluster_dict)} total):")
    for c, members in cluster_dict.items():
        print(f"  Cluster {c} ({len(members)}): {', '.join(members)}")

# Build proportion matrices
print("Building proportion matrices...")
props_l3 = build_proportion_matrix(adata_sc, CT_KEY_L3, SAMPLE_KEY)
props_l4 = build_proportion_matrix(adata_sc, CT_KEY_L4, SAMPLE_KEY)

print(f"Level 3: {props_l3.shape[0]} samples x {props_l3.shape[1]} cell types")
print(f"Level 4: {props_l4.shape[0]} samples x {props_l4.shape[1]} cell types")

# Compute correlations
print("\nComputing correlations...")
corr_l3 = clean_corr(props_l3.corr(method="pearson"))
corr_l4 = clean_corr(props_l4.corr(method="pearson"))

# Save correlations
corr_l3.to_csv(f"{OUT_DIR}/sc_celltype_level3_correlations.csv")
corr_l4.to_csv(f"{OUT_DIR}/sc_celltype_level4_correlations.csv")

# Plot level 3 with clusters
print("\nPlotting level 3...")
g3, clusters_l3_df = plot_corr_clustermap_with_clusters(
    corr_l3,
    threshold=CLUSTER_THRESHOLD_L3,
    method=CLUSTER_METHOD,
    fig_size=FIG_SIZE_L3, 
    label_fontsize=LABEL_FONTSIZE_L3,
    title=f"Cell type correlations (level 3)"
)
clusters_l3 = build_cluster_dict(clusters_l3_df)
print_clusters(clusters_l3, "Level 3")
plt.show()

# Plot level 4 with clusters
print("\nPlotting level 4...")
g4, clusters_l4_df = plot_corr_clustermap_with_clusters(
    corr_l4,
    threshold=CLUSTER_THRESHOLD_L4,
    method=CLUSTER_METHOD,
    fig_size=FIG_SIZE_L4, 
    label_fontsize=LABEL_FONTSIZE_L4,
    title=f"Cell type correlations (level 4)"
)
clusters_l4 = build_cluster_dict(clusters_l4_df)
print_clusters(clusters_l4, "Level 4")
plt.show()

# Save clusters
clusters_l3_df.to_csv(f"{OUT_DIR}/sc_celltype_level3_clusters.csv", index=False)
clusters_l4_df.to_csv(f"{OUT_DIR}/sc_celltype_level4_clusters.csv", index=False)
print(f"\nSaved to {OUT_DIR}")

# Store proportion matrices for survival analysis
props_by_sample_l3 = props_l3
props_by_sample_l4 = props_l4

print("\nOutputs available for survival analysis:")
print("  clusters_l3: dict mapping cluster # -> cell types (level 3)")
print("  clusters_l4: dict mapping cluster # -> cell types (level 4)")
print("  props_by_sample_l3: sample x celltype proportion matrix (level 3)")
print("  props_by_sample_l4: sample x celltype proportion matrix (level 4)")
```

### Sweep parmaters to get significant survival sratification from cell matrix clusters

```python
# %% [markdown]
# ### Grid search: Find optimal survival stratification

# %%
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

import rpy2.robjects as ro
from rpy2.robjects import pandas2ri
from rpy2.robjects.conversion import localconverter

# User config
CLUSTER_LEVEL = "l4"  # "l3" or "l4"
CLUSTER_NUMBER = 7  # cluster to analyze

SURVIVAL_TYPE = "OS"  # "OS" or "PFS"

# Grid search options
SCORING_METHODS = ["sum", "mean", "max", "pca"]
STRATIFICATIONS = ["median", "tertile", "quartile_vs_rest", "quartile_vs_quartile"]
COVARIATE_OPTIONS = [
    #[],  # univariate
    #["age_at_diagnosis"],
    ["MGMT promoter methylation"],
    ["age_at_diagnosis", "MGMT promoter methylation"],
]

# Column mappings
SAMPLE_ID_COL = "Subject Id"
CHRONOLOGY_COL = "Sample collection chronology"
TIME_COL_OS = "os_years"
EVENT_COL_OS = "os_event"
TIME_COL_PFS = "pfs_years"
EVENT_COL_PFS = "pfs_event"
AGE_COL = "age_at_diagnosis"
MGMT_COL = "MGMT promoter methylation"

FIG_WIDTH = 8
FIG_HEIGHT = 6

# Select cluster and proportions
if CLUSTER_LEVEL == "l3":
    cluster_dict = clusters_l3
    props_orig = props_by_sample_l3.copy()
elif CLUSTER_LEVEL == "l4":
    cluster_dict = clusters_l4
    props_orig = props_by_sample_l4.copy()

cluster_celltypes = cluster_dict[CLUSTER_NUMBER]
cluster_cols = [c for c in cluster_celltypes if c in props_orig.columns]

print(f"Cluster {CLUSTER_NUMBER} cell types ({len(cluster_celltypes)}):")
for ct in cluster_celltypes:
    print(f"  {ct}")
print()

# Prepare clinical data once
clin = clin_out.copy()
clin["sample_id"] = clin[SAMPLE_ID_COL].astype(str)
clin["_priority"] = clin[CHRONOLOGY_COL].map({"Baseline": 0, "At recurrence": 1}).fillna(2)
clin = clin.sort_values(["sample_id", "_priority"])
clin = clin.drop_duplicates(subset=["sample_id"], keep="first")
clin = clin.drop(columns=["_priority"])

# Select survival columns
if SURVIVAL_TYPE == "OS":
    time_col, event_col = TIME_COL_OS, EVENT_COL_OS
else:
    time_col, event_col = TIME_COL_PFS, EVENT_COL_PFS

# Initialize R
ro.r("suppressPackageStartupMessages(library(survival))")

def compute_score(props, cluster_cols, method):
    """Compute cluster score using specified method."""
    if method == "sum":
        return props[cluster_cols].sum(axis=1)
    elif method == "mean":
        return props[cluster_cols].mean(axis=1)
    elif method == "max":
        return props[cluster_cols].max(axis=1)
    elif method == "median":
        return props[cluster_cols].median(axis=1)
    elif method == "geometric_mean":
        eps = 1e-6
        return np.exp(np.log(props[cluster_cols] + eps).mean(axis=1))
    elif method == "pca":
        from sklearn.preprocessing import StandardScaler
        from sklearn.decomposition import PCA
        X = props[cluster_cols].values
        X_scaled = StandardScaler().fit_transform(X)
        pca = PCA(n_components=1)
        return pca.fit_transform(X_scaled).flatten()
    else:
        raise ValueError(f"Unknown method: {method}")

def stratify(scores, method):
    """Stratify scores into High/Low groups."""
    if method == "median":
        threshold = scores.median()
        groups = np.where(scores >= threshold, "High", "Low")
        return groups, None  # No samples excluded
    elif method == "tertile":
        t33 = scores.quantile(0.33)
        t67 = scores.quantile(0.67)
        groups = np.where(scores >= t67, "High", np.where(scores <= t33, "Low", "Middle"))
        return groups, "Middle"
    elif method == "quartile_vs_rest":
        q75 = scores.quantile(0.75)
        groups = np.where(scores >= q75, "High", "Low")
        return groups, None
    elif method == "quartile_vs_quartile":
        q25 = scores.quantile(0.25)
        q75 = scores.quantile(0.75)
        groups = np.where(scores >= q75, "High", np.where(scores <= q25, "Low", "Middle"))
        return groups, "Middle"
    else:
        raise ValueError(f"Unknown stratification: {method}")

def run_analysis(props_orig, cluster_cols, clin, scoring_method, stratification, covariates):
    """Run a single survival analysis and return results."""
    props = props_orig.copy()
    
    # Compute score
    props["cluster_score"] = compute_score(props, cluster_cols, scoring_method)
    
    # Prepare sample IDs
    props = props.reset_index()
    props.columns = ["sample_id_raw"] + list(props.columns[1:])
    props = props[props["sample_id_raw"].astype(str).str.endswith("a")].copy()
    props["sample_id"] = props["sample_id_raw"].astype(str).str[:-1]
    
    # Merge
    merged = props[["sample_id", "cluster_score"]].merge(clin, on="sample_id", how="inner")
    
    # Build survival df
    surv_df = merged[["sample_id", "cluster_score", time_col, event_col]].copy()
    
    # Add covariates
    cov_cols = []
    for cov in covariates:
        if cov == AGE_COL or cov == "age_at_diagnosis":
            surv_df["age"] = pd.to_numeric(merged[AGE_COL], errors="coerce").values
            cov_cols.append("age")
        elif cov == MGMT_COL or cov == "MGMT promoter methylation":
            surv_df["mgmt"] = (merged[MGMT_COL].astype(str) == "Methylated").astype(int).values
            cov_cols.append("mgmt")
    
    surv_df = surv_df.rename(columns={time_col: "time", event_col: "event"})
    surv_df["time"] = pd.to_numeric(surv_df["time"], errors="coerce")
    surv_df["event"] = pd.to_numeric(surv_df["event"], errors="coerce")
    surv_df = surv_df.dropna(subset=["time", "event", "cluster_score"] + cov_cols)
    surv_df = surv_df[surv_df["time"] > 0]
    
    if len(surv_df) < 20:
        return None
    
    # Stratify
    groups, exclude = stratify(surv_df["cluster_score"], stratification)
    surv_df["group"] = groups
    
    if exclude:
        surv_df = surv_df[surv_df["group"] != exclude].copy()
    
    if len(surv_df) < 20:
        return None
    
    surv_df["group_numeric"] = (surv_df["group"] == "High").astype(int)
    
    n_high = (surv_df["group"] == "High").sum()
    n_low = (surv_df["group"] == "Low").sum()
    
    if n_high < 5 or n_low < 5:
        return None
    
    # Run R analysis
    with localconverter(ro.default_converter + pandas2ri.converter):
        r_df = ro.conversion.py2rpy(surv_df)
    
    ro.globalenv["surv_data"] = r_df
    
    # Log-rank test
    try:
        ro.r("""
        logrank <- survdiff(Surv(time, event) ~ group_numeric, data = surv_data)
        logrank_p <- 1 - pchisq(logrank$chisq, df = 1)
        """)
        logrank_p = ro.r("logrank_p")[0]
    except:
        return None
    
    # Cox model
    if cov_cols:
        cov_formula = " + ".join(cov_cols)
        cox_formula = f"Surv(time, event) ~ group_numeric + {cov_formula}"
    else:
        cox_formula = "Surv(time, event) ~ group_numeric"
    
    try:
        ro.r(f"""
        cox_fit <- coxph({cox_formula}, data = surv_data)
        cox_summary <- summary(cox_fit)
        cox_hr <- exp(coef(cox_fit)["group_numeric"])
        cox_p <- cox_summary$coefficients["group_numeric", "Pr(>|z|)"]
        cox_ci <- exp(confint(cox_fit)["group_numeric",])
        """)
        cox_hr = ro.r("cox_hr")[0]
        cox_p = ro.r("cox_p")[0]
        cox_ci_low = ro.r("cox_ci")[0]
        cox_ci_high = ro.r("cox_ci")[1]
    except:
        return None
    
    # Median survival times
    try:
        ro.r("""
        km_fit <- survfit(Surv(time, event) ~ group_numeric, data = surv_data)
        med_surv <- summary(km_fit)$table[, "median"]
        """)
        med_surv = ro.r("med_surv")
        median_low = med_surv[0] if not np.isnan(med_surv[0]) else None
        median_high = med_surv[1] if not np.isnan(med_surv[1]) else None
    except:
        median_low = None
        median_high = None
    
    return {
        "scoring": scoring_method,
        "stratification": stratification,
        "covariates": ", ".join(covariates) if covariates else "none",
        "n_total": len(surv_df),
        "n_high": n_high,
        "n_low": n_low,
        "logrank_p": logrank_p,
        "cox_hr": cox_hr,
        "cox_p": cox_p,
        "cox_ci_low": cox_ci_low,
        "cox_ci_high": cox_ci_high,
        "median_high": median_high,
        "median_low": median_low,
        "median_diff": (median_low - median_high) if (median_low and median_high) else None,
    }

# Run grid search
print("Running grid search...")
print(f"  Scoring methods: {SCORING_METHODS}")
print(f"  Stratifications: {STRATIFICATIONS}")
print(f"  Covariate options: {len(COVARIATE_OPTIONS)}")
print()

results = []
total = len(SCORING_METHODS) * len(STRATIFICATIONS) * len(COVARIATE_OPTIONS)
i = 0

for scoring in SCORING_METHODS:
    for strat in STRATIFICATIONS:
        for covs in COVARIATE_OPTIONS:
            i += 1
            result = run_analysis(props_orig, cluster_cols, clin, scoring, strat, covs)
            if result:
                results.append(result)

print(f"Completed {i}/{total} combinations")
print(f"Valid results: {len(results)}")

if not results:
    raise ValueError("No valid results from grid search")

# Convert to dataframe
results_df = pd.DataFrame(results)

# Rank by combined metric: low p-value and high HR (or low HR if protective)
# Use -log10(cox_p) * |log(HR)| as ranking metric
results_df["effect_size"] = np.abs(np.log(results_df["cox_hr"]))
results_df["neg_log_p"] = -np.log10(results_df["cox_p"].clip(lower=1e-300))
results_df["rank_score"] = results_df["neg_log_p"] * results_df["effect_size"]

# Sort by rank score
results_df = results_df.sort_values("rank_score", ascending=False)

# Display top 10
print("\n" + "="*80)
print("TOP 10 RESULTS (ranked by statistical support × effect size)")
print("="*80)

display_cols = ["scoring", "stratification", "covariates", "n_high", "n_low", 
                "cox_hr", "cox_p", "logrank_p", "median_high", "median_low"]
print(results_df[display_cols].head(10).to_string(index=False))

# Best result
best = results_df.iloc[0]

print("\n" + "="*80)
print("BEST RESULT")
print("="*80)
print(f"Scoring method: {best['scoring']}")
print(f"Stratification: {best['stratification']}")
print(f"Covariates: {best['covariates']}")
print(f"N (High/Low): {best['n_high']}/{best['n_low']}")
print(f"Cox HR: {best['cox_hr']:.3f} (95% CI: {best['cox_ci_low']:.3f} - {best['cox_ci_high']:.3f})")
print(f"Cox p-value: {best['cox_p']:.4e}")
print(f"Log-rank p-value: {best['logrank_p']:.4e}")
if best['median_high'] and best['median_low']:
    print(f"Median survival - High: {best['median_high']:.2f} yrs, Low: {best['median_low']:.2f} yrs")
    print(f"Median survival difference: {best['median_diff']:.2f} yrs")

# Now run the best configuration and plot
print("\n" + "="*80)
print("GENERATING PLOT FOR BEST RESULT")
print("="*80)

# Re-run best analysis to get KM data
props = props_orig.copy()
props["cluster_score"] = compute_score(props, cluster_cols, best["scoring"])

props = props.reset_index()
props.columns = ["sample_id_raw"] + list(props.columns[1:])
props = props[props["sample_id_raw"].astype(str).str.endswith("a")].copy()
props["sample_id"] = props["sample_id_raw"].astype(str).str[:-1]

merged = props[["sample_id", "cluster_score"]].merge(clin, on="sample_id", how="inner")

surv_df = merged[["sample_id", "cluster_score", time_col, event_col]].copy()

cov_cols = []
covariates = best["covariates"].split(", ") if best["covariates"] != "none" else []
for cov in covariates:
    if cov == AGE_COL or cov == "age_at_diagnosis":
        surv_df["age"] = pd.to_numeric(merged[AGE_COL], errors="coerce").values
        cov_cols.append("age")
    elif cov == MGMT_COL or cov == "MGMT promoter methylation":
        surv_df["mgmt"] = (merged[MGMT_COL].astype(str) == "Methylated").astype(int).values
        cov_cols.append("mgmt")

surv_df = surv_df.rename(columns={time_col: "time", event_col: "event"})
surv_df["time"] = pd.to_numeric(surv_df["time"], errors="coerce")
surv_df["event"] = pd.to_numeric(surv_df["event"], errors="coerce")
surv_df = surv_df.dropna(subset=["time", "event", "cluster_score"] + cov_cols)
surv_df = surv_df[surv_df["time"] > 0]

groups, exclude = stratify(surv_df["cluster_score"], best["stratification"])
surv_df["group"] = groups
if exclude:
    surv_df = surv_df[surv_df["group"] != exclude].copy()
surv_df["group_numeric"] = (surv_df["group"] == "High").astype(int)

with localconverter(ro.default_converter + pandas2ri.converter):
    r_df = ro.conversion.py2rpy(surv_df)
ro.globalenv["surv_data"] = r_df

ro.r("""
km_fit <- survfit(Surv(time, event) ~ group_numeric, data = surv_data)
""")

km_times_0 = np.array(ro.r("km_fit[1]$time"))
km_surv_0 = np.array(ro.r("km_fit[1]$surv"))
km_lower_0 = np.array(ro.r("km_fit[1]$lower"))
km_upper_0 = np.array(ro.r("km_fit[1]$upper"))

km_times_1 = np.array(ro.r("km_fit[2]$time"))
km_surv_1 = np.array(ro.r("km_fit[2]$surv"))
km_lower_1 = np.array(ro.r("km_fit[2]$lower"))
km_upper_1 = np.array(ro.r("km_fit[2]$upper"))

# Plot
fig, ax = plt.subplots(figsize=(FIG_WIDTH, FIG_HEIGHT))

times_0 = np.concatenate([[0], km_times_0])
surv_0 = np.concatenate([[1], km_surv_0])
lower_0 = np.concatenate([[1], km_lower_0])
upper_0 = np.concatenate([[1], km_upper_0])

ax.step(times_0, surv_0, where="post", color="#0072B2", label="Low", linewidth=2)
ax.fill_between(times_0, lower_0, upper_0, step="post", alpha=0.2, color="#0072B2")

times_1 = np.concatenate([[0], km_times_1])
surv_1 = np.concatenate([[1], km_surv_1])
lower_1 = np.concatenate([[1], km_lower_1])
upper_1 = np.concatenate([[1], km_upper_1])

ax.step(times_1, surv_1, where="post", color="#D55E00", label="High", linewidth=2)
ax.fill_between(times_1, lower_1, upper_1, step="post", alpha=0.2, color="#D55E00")

# Build title
cov_str = f", adjusted for {best['covariates']}" if best['covariates'] != "none" else ""
title = (
    f"{SURVIVAL_TYPE} by Cluster {CLUSTER_NUMBER} immune infiltrate\n"
    f"Scoring: {best['scoring']}, Stratification: {best['stratification']}{cov_str}\n"
    f"HR={best['cox_hr']:.2f} (p={best['cox_p']:.3e}), Log-rank p={best['logrank_p']:.3e}"
)

ax.set_xlabel("Time (years)", fontsize=11)
ax.set_ylabel("Survival probability", fontsize=11)
ax.set_title(title, fontsize=10)
ax.legend(title="Cluster score", loc="lower left")
ax.set_ylim(0, 1.05)
ax.set_xlim(0, None)
ax.grid(False)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()

# Save results
results_df.to_csv(f"{OUT_DIR}/survival_grid_search_cluster{CLUSTER_NUMBER}.csv" if "OUT_DIR" in dir() else f"survival_grid_search_cluster{CLUSTER_NUMBER}.csv", index=False)
print(f"\nResults saved to survival_grid_search_cluster{CLUSTER_NUMBER}.csv")
```

```python
# =============================================================================
# KM + RMST PLOT — select cluster below
# =============================================================================

SEL_THRESHOLD  = 0.9
SEL_CLUSTER    = 7
SEL_SCORING    = "sum"
SEL_STRAT      = "median"
SEL_COVARIATES = []
RMST_TAU       = 1.5   # None = 75th percentile of event times

# -----------------------------------------------------------------------
sel_cluster_dict, _ = get_clusters_at_threshold(corr_l4, SEL_THRESHOLD, CLUSTER_METHOD)

if SEL_CLUSTER not in sel_cluster_dict:
    raise ValueError(f"Cluster {SEL_CLUSTER} not found at threshold {SEL_THRESHOLD}. "
                     f"Available: {sorted(sel_cluster_dict.keys())}")

sel_cols = [c for c in sel_cluster_dict[SEL_CLUSTER] if c in props_l4.columns]
print(f"Cluster {SEL_CLUSTER} @ threshold {SEL_THRESHOLD}: {sel_cols}")

props = props_l4.copy()
props["cluster_score"] = compute_score(props, sel_cols, SEL_SCORING)
props = props.reset_index()
props.columns = ["sample_id_raw"] + list(props.columns[1:])
props = props[props["sample_id_raw"].astype(str).str.endswith("a")].copy()
props["sample_id"] = props["sample_id_raw"].astype(str).str[:-1]

merged = props[["sample_id", "cluster_score"]].merge(clin, on="sample_id", how="inner")
surv_df = merged[["sample_id", "cluster_score", time_col, event_col]].copy()

cov_cols = []
for cov in SEL_COVARIATES:
    if cov in ("age_at_diagnosis", AGE_COL):
        surv_df["age"] = pd.to_numeric(merged[AGE_COL], errors="coerce").values
        cov_cols.append("age")
    elif cov in ("MGMT promoter methylation", MGMT_COL):
        surv_df["mgmt"] = (merged[MGMT_COL].astype(str) == "Methylated").astype(int).values
        cov_cols.append("mgmt")

surv_df = surv_df.rename(columns={time_col: "time", event_col: "event"})
surv_df["time"]  = pd.to_numeric(surv_df["time"],  errors="coerce")
surv_df["event"] = pd.to_numeric(surv_df["event"], errors="coerce")
surv_df = surv_df.dropna(subset=["time", "event", "cluster_score"] + cov_cols)
surv_df = surv_df[surv_df["time"] > 0]

# Auto-fallback for zero-inflated clusters
nonzero_frac = (surv_df["cluster_score"] > 0).mean()
if nonzero_frac < 0.5 and SEL_STRAT in ("median", "tertile"):
    print(f"  {nonzero_frac:.1%} non-zero scores — switching to 'quartile_vs_rest'")
    SEL_STRAT = "quartile_vs_rest"

groups, exclude = stratify(surv_df["cluster_score"], SEL_STRAT)
surv_df["group"] = groups
if exclude:
    surv_df = surv_df[surv_df["group"] != exclude].copy()
surv_df["group_numeric"] = (surv_df["group"] == "High").astype(int)

n_high = (surv_df["group"] == "High").sum()
n_low  = (surv_df["group"] == "Low").sum()
print(f"  N High={n_high}, Low={n_low}, Total={len(surv_df)}")

group_counts = surv_df["group"].value_counts()
if len(group_counts) < 2 or group_counts.min() < 5:
    raise ValueError(
        f"Invalid groups after stratification '{SEL_STRAT}'.\n"
        f"  Counts: {group_counts.to_dict()}\n"
        f"  Score range: {surv_df['cluster_score'].min():.4f} – {surv_df['cluster_score'].max():.4f}"
    )

# -----------------------------------------------------------------------
# R block
tau_val = float(np.quantile(surv_df.loc[surv_df["event"] == 1, "time"], 0.75)) \
          if RMST_TAU is None else float(RMST_TAU)

with localconverter(ro.default_converter + pandas2ri.converter):
    r_df = ro.conversion.py2rpy(surv_df)
ro.globalenv["surv_data"] = r_df
ro.globalenv["rmst_tau"]  = ro.FloatVector([tau_val])

cov_formula = (" + " + " + ".join(cov_cols)) if cov_cols else ""
ro.r(f"""
suppressPackageStartupMessages({{library(survival); library(survRM2)}})

km_fit     <- survfit(Surv(time, event) ~ group_numeric, data = surv_data)
logrank    <- survdiff(Surv(time, event) ~ group_numeric, data = surv_data)
logrank_p  <- 1 - pchisq(logrank$chisq, df = 1)
cox_fit    <- coxph(Surv(time, event) ~ group_numeric{cov_formula}, data = surv_data)
cox_hr     <- exp(coef(cox_fit)["group_numeric"])
cox_p      <- summary(cox_fit)$coefficients["group_numeric", "Pr(>|z|)"]
cox_ci     <- exp(confint(cox_fit)["group_numeric",])
rmst_res   <- rmst2(surv_data$time, surv_data$event, surv_data$group_numeric, tau = rmst_tau)
rmst_high  <- rmst_res$RMST.arm1$result["RMST", "Est."]
rmst_low   <- rmst_res$RMST.arm0$result["RMST", "Est."]
rmst_diff  <- rmst_res$unadjusted.result["RMST (arm=1)-(arm=0)", "Est."]
rmst_diff_p  <- rmst_res$unadjusted.result["RMST (arm=1)-(arm=0)", "p"]
rmst_diff_lo <- rmst_res$unadjusted.result["RMST (arm=1)-(arm=0)", "lower .95"]
rmst_diff_hi <- rmst_res$unadjusted.result["RMST (arm=1)-(arm=0)", "upper .95"]
med_surv   <- summary(km_fit)$table[, "median"]
""")

logrank_p    = ro.r("logrank_p")[0]
cox_hr       = ro.r("cox_hr")[0]
cox_p        = ro.r("cox_p")[0]
cox_ci_low   = ro.r("cox_ci")[0]
cox_ci_high  = ro.r("cox_ci")[1]
rmst_high    = ro.r("rmst_high")[0]
rmst_low_val = ro.r("rmst_low")[0]
rmst_diff    = ro.r("rmst_diff")[0]
rmst_diff_p  = ro.r("rmst_diff_p")[0]
rmst_diff_lo = ro.r("rmst_diff_lo")[0]
rmst_diff_hi = ro.r("rmst_diff_hi")[0]
med_surv     = ro.r("med_surv")
median_low   = float(med_surv[0]) if not np.isnan(med_surv[0]) else None
median_high  = float(med_surv[1]) if not np.isnan(med_surv[1]) else None

km_times_0 = np.array(ro.r("km_fit[1]$time"))
km_surv_0  = np.array(ro.r("km_fit[1]$surv"))
km_lower_0 = np.array(ro.r("km_fit[1]$lower"))
km_upper_0 = np.array(ro.r("km_fit[1]$upper"))
km_times_1 = np.array(ro.r("km_fit[2]$time"))
km_surv_1  = np.array(ro.r("km_fit[2]$surv"))
km_lower_1 = np.array(ro.r("km_fit[2]$lower"))
km_upper_1 = np.array(ro.r("km_fit[2]$upper"))

# -----------------------------------------------------------------------
# Plot — same style as grid search cell
fig, ax = plt.subplots(figsize=(FIG_WIDTH, FIG_HEIGHT))

times_0 = np.concatenate([[0], km_times_0])
surv_0  = np.concatenate([[1], km_surv_0])
lower_0 = np.concatenate([[1], km_lower_0])
upper_0 = np.concatenate([[1], km_upper_0])
ax.step(times_0, surv_0, where="post", color="#0072B2", label=f"Low (n={n_low})", linewidth=2)
ax.fill_between(times_0, lower_0, upper_0, step="post", alpha=0.2, color="#0072B2")

times_1 = np.concatenate([[0], km_times_1])
surv_1  = np.concatenate([[1], km_surv_1])
lower_1 = np.concatenate([[1], km_lower_1])
upper_1 = np.concatenate([[1], km_upper_1])
ax.step(times_1, surv_1, where="post", color="#D55E00", label=f"High (n={n_high})", linewidth=2)
ax.fill_between(times_1, lower_1, upper_1, step="post", alpha=0.2, color="#D55E00")

ax.axvline(tau_val, color="gray", linestyle=":", linewidth=1, alpha=0.7)
ax.text(tau_val + 0.03, 0.97, f"τ={tau_val:.1f}y", color="gray", fontsize=8, va="top")

cov_str = f", adjusted for {', '.join(SEL_COVARIATES)}" if SEL_COVARIATES else ""
med_str = (f"Median: High={median_high:.2f} yrs, Low={median_low:.2f} yrs\n"
           if (median_high and median_low) else "")

title = (
    f"{SURVIVAL_TYPE} by Level 4 Cluster (threshold={SEL_THRESHOLD}, cluster={SEL_CLUSTER})\n"
    f"Scoring: {SEL_SCORING}, Stratification: {SEL_STRAT}{cov_str}\n"
    f"HR={cox_hr:.2f} (95% CI: {cox_ci_low:.2f}-{cox_ci_high:.2f}), "
    f"Cox p={cox_p:.2e}, Log-rank p={logrank_p:.2e}\n"
    f"{med_str}"
    f"RMST(τ={tau_val:.1f}y): High={rmst_high:.2f} yrs, Low={rmst_low_val:.2f} yrs, "
    f"ΔRMST={rmst_diff:.2f} (95% CI: {rmst_diff_lo:.2f}-{rmst_diff_hi:.2f}), p={rmst_diff_p:.2e}"
)

ax.set_xlabel("Time (years)", fontsize=11)
ax.set_ylabel("Survival probability", fontsize=11)
ax.set_title(title, fontsize=9)
ax.legend(title="Cluster score", loc="lower left")
ax.set_ylim(0, 1.05)
ax.set_xlim(0, None)
ax.grid(False)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.savefig(
    f"{OUT_DIR}/km_rmst_t{SEL_THRESHOLD}_c{SEL_CLUSTER}_{SEL_SCORING}_{SEL_STRAT}.pdf",
    bbox_inches="tight")
plt.show()
```

```python
# %% [markdown]
# ### Focused analysis: T cell clusters, HR < 1, MGMT-independent

# %%
# Filter comprehensive results for:
# 1. Clusters containing T_cell subtypes
# 2. HR < 1 (protective effect)
# 3. MGMT-only adjustment

# First, let's see what T cell subtypes exist in our data
print("T cell subtypes in Level 4 annotations:")
tcell_types = [c for c in props_l4.columns if "T_cell" in c or "T_reg" in c or "Treg" in c]
for t in sorted(tcell_types):
    print(f"  {t}")

print(f"\nTotal: {len(tcell_types)} T cell subtypes")

# %%
# Filter results for T cell-containing clusters with MGMT adjustment only

# Identify rows where:
# 1. celltypes contains any T_cell subtype
# 2. covariates == "MGMT promoter methylation" (MGMT only, no age)
# 3. cox_hr < 1

def contains_tcell(celltypes_str):
    """Check if cluster contains any T cell subtype."""
    celltypes = celltypes_str.split("; ")
    for ct in celltypes:
        if "T_cell" in ct or "T_reg" in ct or "Treg" in ct:
            return True
    return False

# Filter
tcell_results = comprehensive_results[
    (comprehensive_results["celltypes"].apply(contains_tcell)) &
    (comprehensive_results["covariates"] == "MGMT promoter methylation") &
    (comprehensive_results["cox_hr"] < 1)
].copy()

print(f"Results matching criteria: {len(tcell_results)}")
print(f"  - Contains T cell subtype")
print(f"  - MGMT-only adjustment")
print(f"  - HR < 1 (protective)")

# %%
# Sort by p-value and display
tcell_results = tcell_results.sort_values("cox_p")

print("\n" + "="*80)
print("T CELL CLUSTERS WITH PROTECTIVE EFFECT (HR < 1), MGMT-INDEPENDENT")
print("="*80)

if len(tcell_results) > 0:
    # Show top results
    display_cols = ["threshold", "cluster", "n_celltypes", "scoring", "stratification",
                    "n_high", "n_low", "cox_hr", "cox_ci_low", "cox_ci_high", "cox_p", "logrank_p"]
    
    print("\n--- Top 20 Results ---")
    print(tcell_results[display_cols].head(20).to_string(index=False))
    
    # Show unique clusters
    print("\n--- Unique T Cell Clusters ---")
    unique_clusters = tcell_results.drop_duplicates(subset=["threshold", "cluster"]).sort_values("cox_p")
    
    for _, row in unique_clusters.iterrows():
        sig_marker = "**" if row["cox_p"] < 0.05 else ""
        print(f"\n{sig_marker}Threshold {row['threshold']}, Cluster {row['cluster']} ({row['n_celltypes']} cell types){sig_marker}")
        print(f"  Best HR: {row['cox_hr']:.3f} (95% CI: {row['cox_ci_low']:.3f}-{row['cox_ci_high']:.3f})")
        print(f"  Best p-value: {row['cox_p']:.4f}")
        print(f"  Cell types: {row['celltypes']}")
        
        # Identify T cell subtypes in this cluster
        celltypes = row['celltypes'].split("; ")
        tcells_in_cluster = [c for c in celltypes if "T_cell" in c or "T_reg" in c or "Treg" in c]
        print(f"  T cell subtypes: {', '.join(tcells_in_cluster)}")
    
    # Summary of significant results
    sig_tcell = tcell_results[tcell_results["cox_p"] < 0.05]
    print(f"\n\n--- Summary ---")
    print(f"Total T cell cluster results (HR<1, MGMT-adjusted): {len(tcell_results)}")
    print(f"Significant at p<0.05: {len(sig_tcell)}")
    
    if len(sig_tcell) > 0:
        print(f"\nSignificant clusters:")
        sig_unique = sig_tcell.drop_duplicates(subset=["threshold", "cluster"])
        for _, row in sig_unique.iterrows():
            print(f"  Threshold {row['threshold']}, Cluster {row['cluster']}: HR={row['cox_hr']:.3f}, p={row['cox_p']:.4f}")
else:
    print("\nNo results found matching criteria.")
    
# %%
# Let's also check what's close to significance (p < 0.1)
print("\n" + "="*80)
print("BORDERLINE SIGNIFICANT (p < 0.1)")
print("="*80)

borderline = tcell_results[(tcell_results["cox_p"] >= 0.05) & (tcell_results["cox_p"] < 0.1)]
if len(borderline) > 0:
    unique_borderline = borderline.drop_duplicates(subset=["threshold", "cluster"]).sort_values("cox_p")
    for _, row in unique_borderline.iterrows():
        print(f"\nThreshold {row['threshold']}, Cluster {row['cluster']} ({row['n_celltypes']} cell types)")
        print(f"  HR: {row['cox_hr']:.3f} (95% CI: {row['cox_ci_low']:.3f}-{row['cox_ci_high']:.3f}), p={row['cox_p']:.4f}")
        print(f"  Cell types: {row['celltypes']}")
else:
    print("None found.")

# %%
# Store for further analysis
tcell_mgmt_results = tcell_results
```

```python
# %% [markdown]
# ### RMST grid search using native survRM2 covariate adjustment (Tian's IPCW method)
# 
# This uses the `covariates` argument in `survRM2::rmst2()` which implements 
# Tian et al.'s ANCOVA-type adjusted analysis using inverse probability of 
# censoring weighting (IPCW). This is the best-supported approach in survRM2.

# %%
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, fcluster
from scipy.spatial.distance import squareform
import warnings
warnings.filterwarnings("ignore")

import rpy2.robjects as ro
from rpy2.robjects import pandas2ri
from rpy2.robjects.conversion import localconverter

# =============================================================================
# USER CONFIG
# =============================================================================

CT_KEY_L4 = "celltype_level4_curated"
SAMPLE_KEY = "orig.ident"
EXCLUDE_PATTERNS = ["Doublet", "doublet"]

CLUSTER_THRESHOLDS = [0.7, 0.8, 0.9, 1.0, 1.2, 1.3, 1.5]
CLUSTER_METHOD = "ward"

SURVIVAL_TYPE = "OS"

SCORING_METHODS = ["sum", "mean", "max", "pca"]
STRATIFICATIONS = ["median", "tertile", "quartile_vs_rest", "quartile_vs_quartile"]

# RMST tau options (years)
TAU_OPTIONS = [1.5, 2.0, 2.5, 3.0]

P_THRESHOLD = 0.05

SAMPLE_ID_COL = "Subject Id"
CHRONOLOGY_COL = "Sample collection chronology"
TIME_COL_OS = "os_years"
EVENT_COL_OS = "os_event"
TIME_COL_PFS = "pfs_years"
EVENT_COL_PFS = "pfs_event"
MGMT_COL = "MGMT promoter methylation"

FIG_WIDTH = 8
FIG_HEIGHT = 6

OUT_DIR = "/home/sagemaker-user/gbm-collaboration_2/analysis_sc_merged"

# =============================================================================
# HELPER FUNCTIONS
# =============================================================================

def is_excluded(ct_name):
    for pattern in EXCLUDE_PATTERNS:
        if pattern in str(ct_name):
            return True
    return False

def build_proportion_matrix(adata, ct_key, sample_key):
    obs = adata.obs[[sample_key, ct_key]].copy()
    obs[sample_key] = obs[sample_key].astype(str)
    obs[ct_key] = obs[ct_key].astype(str)
    counts = obs.groupby([sample_key, ct_key]).size().unstack(fill_value=0)
    keep_cols = [c for c in counts.columns if not is_excluded(c)]
    counts = counts[keep_cols]
    props = counts.div(counts.sum(axis=1), axis=0)
    return props

def clean_corr(corr):
    corr = corr.replace([np.inf, -np.inf], np.nan)
    corr = corr.dropna(axis=0, how="all").dropna(axis=1, how="all")
    common = corr.index.intersection(corr.columns)
    corr = corr.loc[common, common]
    return corr.fillna(0.0)

def compute_linkage(corr, method="ward"):
    dist_matrix = 1 - corr.values
    np.fill_diagonal(dist_matrix, 0)
    dist_matrix = (dist_matrix + dist_matrix.T) / 2
    dist_matrix = np.clip(dist_matrix, 0, None)
    condensed = squareform(dist_matrix, checks=False)
    return linkage(condensed, method=method)

def get_clusters_at_threshold(corr, threshold, method="ward"):
    Z = compute_linkage(corr, method)
    cluster_labels = fcluster(Z, t=threshold, criterion="distance")
    cluster_df = pd.DataFrame({"celltype": corr.index, "cluster": cluster_labels})
    cluster_dict = {}
    for c in sorted(cluster_df["cluster"].unique()):
        members = cluster_df[cluster_df["cluster"] == c]["celltype"].tolist()
        cluster_dict[c] = members
    return cluster_dict, cluster_df

def compute_score(props, cluster_cols, method):
    if method == "sum":
        return props[cluster_cols].sum(axis=1)
    elif method == "mean":
        return props[cluster_cols].mean(axis=1)
    elif method == "max":
        return props[cluster_cols].max(axis=1)
    elif method == "pca":
        from sklearn.preprocessing import StandardScaler
        from sklearn.decomposition import PCA
        X = props[cluster_cols].values
        if X.shape[1] < 2:
            return props[cluster_cols].iloc[:, 0]
        X_scaled = StandardScaler().fit_transform(X)
        pca = PCA(n_components=1)
        return pca.fit_transform(X_scaled).flatten()
    else:
        raise ValueError(f"Unknown method: {method}")

def stratify(scores, method):
    if method == "median":
        threshold = scores.median()
        groups = np.where(scores >= threshold, "High", "Low")
        return groups, None
    elif method == "tertile":
        t33 = scores.quantile(0.33)
        t67 = scores.quantile(0.67)
        groups = np.where(scores >= t67, "High", np.where(scores <= t33, "Low", "Middle"))
        return groups, "Middle"
    elif method == "quartile_vs_rest":
        q75 = scores.quantile(0.75)
        groups = np.where(scores >= q75, "High", "Low")
        return groups, None
    elif method == "quartile_vs_quartile":
        q25 = scores.quantile(0.25)
        q75 = scores.quantile(0.75)
        groups = np.where(scores >= q75, "High", np.where(scores <= q25, "Low", "Middle"))
        return groups, "Middle"
    else:
        raise ValueError(f"Unknown stratification: {method}")

def contains_tcell(celltypes_str):
    celltypes = celltypes_str.split("; ")
    for ct in celltypes:
        if "T_cell" in ct or "T_reg" in ct or "Treg" in ct:
            return True
    return False

# =============================================================================
# RMST ANALYSIS using native survRM2 (Tian's method)
# =============================================================================

def run_rmst_survRM2(props_orig, cluster_cols, clin, scoring_method, stratification,
                     time_col, event_col, tau, adjust_mgmt=True):
    """
    Run RMST analysis using survRM2::rmst2 with native covariate adjustment.
    
    When covariates are passed, survRM2 uses Tian et al.'s IPCW-based ANCOVA method.
    The output changes:
    - adjusted.result: summary of adjusted between-group contrasts
    - RMST.difference.adjusted: full model output for difference
    """
    props = props_orig.copy()
    
    try:
        props["cluster_score"] = compute_score(props, cluster_cols, scoring_method)
    except Exception:
        return None
    
    props = props.reset_index()
    props.columns = ["sample_id_raw"] + list(props.columns[1:])
    props = props[props["sample_id_raw"].astype(str).str.endswith("a")].copy()
    props["sample_id"] = props["sample_id_raw"].astype(str).str[:-1]
    
    merged = props[["sample_id", "cluster_score"]].merge(clin, on="sample_id", how="inner")
    
    surv_df = merged[["sample_id", "cluster_score", time_col, event_col]].copy()
    surv_df["mgmt"] = (merged[MGMT_COL].astype(str) == "Methylated").astype(int).values
    
    surv_df = surv_df.rename(columns={time_col: "time", event_col: "event"})
    surv_df["time"] = pd.to_numeric(surv_df["time"], errors="coerce")
    surv_df["event"] = pd.to_numeric(surv_df["event"], errors="coerce")
    surv_df = surv_df.dropna(subset=["time", "event", "cluster_score", "mgmt"])
    surv_df = surv_df[surv_df["time"] > 0]
    
    if len(surv_df) < 20:
        return None
    
    groups, exclude = stratify(surv_df["cluster_score"], stratification)
    surv_df["group"] = groups
    
    if exclude:
        surv_df = surv_df[surv_df["group"] != exclude].copy()
    
    if len(surv_df) < 20:
        return None
    
    # arm: 0 = Low, 1 = High
    surv_df["arm"] = (surv_df["group"] == "High").astype(int)
    
    n_high = (surv_df["group"] == "High").sum()
    n_low = (surv_df["group"] == "Low").sum()
    
    if n_high < 5 or n_low < 5:
        return None
    
    max_time = surv_df["time"].max()
    if tau > max_time:
        return None
    
    with localconverter(ro.default_converter + pandas2ri.converter):
        r_df = ro.conversion.py2rpy(surv_df)
    
    ro.globalenv["surv_data"] = r_df
    ro.globalenv["tau"] = tau
    
    try:
        if adjust_mgmt:
            # ADJUSTED analysis using survRM2's native covariate support
            # Pass mgmt as covariate matrix
            ro.r("""
            library(survRM2)
            
            # Create covariate matrix (just MGMT)
            cov_matrix <- as.matrix(surv_data$mgmt)
            
            # Run adjusted RMST analysis
            rmst_adj <- rmst2(
                time = surv_data$time,
                status = surv_data$event,
                arm = surv_data$arm,
                tau = tau,
                covariates = cov_matrix
            )
            
            # Extract adjusted results
            # adjusted.result has rows: RMST (arm=1)-(arm=0), RMST ratio, RMTL ratio
            # columns: Est., lower.95, upper.95, p
            adj_diff <- rmst_adj$adjusted.result[1, "Est."]
            adj_diff_lower <- rmst_adj$adjusted.result[1, "lower .95"]
            adj_diff_upper <- rmst_adj$adjusted.result[1, "upper .95"]
            adj_p <- rmst_adj$adjusted.result[1, "p"]
            
            # Also get unadjusted RMSTs per arm for reference/plotting
            rmst_unadj <- rmst2(
                time = surv_data$time,
                status = surv_data$event,
                arm = surv_data$arm,
                tau = tau
            )
            rmst_high <- rmst_unadj$RMST.arm1$rmst["Est."]
            rmst_low <- rmst_unadj$RMST.arm0$rmst["Est."]
            unadj_p <- rmst_unadj$unadjusted.result[1, "p"]
            """)
            
            rmst_diff = ro.r("adj_diff")[0]
            rmst_diff_lower = ro.r("adj_diff_lower")[0]
            rmst_diff_upper = ro.r("adj_diff_upper")[0]
            rmst_p = ro.r("adj_p")[0]
            rmst_high = ro.r("rmst_high")[0]
            rmst_low = ro.r("rmst_low")[0]
            unadj_p = ro.r("unadj_p")[0]
            method_used = "survRM2_MGMT_adjusted"
            
        else:
            # Unadjusted RMST
            ro.r("""
            library(survRM2)
            
            rmst_result <- rmst2(
                time = surv_data$time,
                status = surv_data$event,
                arm = surv_data$arm,
                tau = tau
            )
            
            rmst_diff <- rmst_result$unadjusted.result[1, "Est."]
            rmst_diff_lower <- rmst_result$unadjusted.result[1, "lower .95"]
            rmst_diff_upper <- rmst_result$unadjusted.result[1, "upper .95"]
            rmst_p <- rmst_result$unadjusted.result[1, "p"]
            rmst_high <- rmst_result$RMST.arm1$rmst["Est."]
            rmst_low <- rmst_result$RMST.arm0$rmst["Est."]
            """)
            
            rmst_diff = ro.r("rmst_diff")[0]
            rmst_diff_lower = ro.r("rmst_diff_lower")[0]
            rmst_diff_upper = ro.r("rmst_diff_upper")[0]
            rmst_p = ro.r("rmst_p")[0]
            rmst_high = ro.r("rmst_high")[0]
            rmst_low = ro.r("rmst_low")[0]
            unadj_p = rmst_p
            method_used = "survRM2_unadjusted"
            
    except Exception as e:
        return None
    
    return {
        "n_total": len(surv_df),
        "n_high": n_high,
        "n_low": n_low,
        "tau": tau,
        "rmst_high": rmst_high,
        "rmst_low": rmst_low,
        "rmst_diff": rmst_diff,  # High - Low (positive = High lives longer)
        "rmst_diff_lower": rmst_diff_lower,
        "rmst_diff_upper": rmst_diff_upper,
        "rmst_p": rmst_p,
        "rmst_p_unadjusted": unadj_p if adjust_mgmt else rmst_p,
        "method": method_used,
    }

# =============================================================================
# MAIN ANALYSIS
# =============================================================================

print("Loading R packages...")
ro.r("""
suppressPackageStartupMessages({
    library(survival)
    library(survRM2)
})
""")
print("  survival and survRM2 loaded")

if SURVIVAL_TYPE == "OS":
    time_col, event_col = TIME_COL_OS, EVENT_COL_OS
else:
    time_col, event_col = TIME_COL_PFS, EVENT_COL_PFS

# Build data if not present
if 'props_l4' not in dir():
    print("\nBuilding proportion matrix for level 4...")
    props_l4 = build_proportion_matrix(adata_sc, CT_KEY_L4, SAMPLE_KEY)
    print(f"  {props_l4.shape[0]} samples x {props_l4.shape[1]} cell types")
    print("Computing correlation matrix...")
    corr_l4 = clean_corr(props_l4.corr(method="pearson"))

if 'clin' not in dir():
    print("Preparing clinical data...")
    clin = clin_out.copy()
    clin["sample_id"] = clin[SAMPLE_ID_COL].astype(str)
    clin["_priority"] = clin[CHRONOLOGY_COL].map({"Baseline": 0, "At recurrence": 1}).fillna(2)
    clin = clin.sort_values(["sample_id", "_priority"])
    clin = clin.drop_duplicates(subset=["sample_id"], keep="first")
    clin = clin.drop(columns=["_priority"])

# =============================================================================
# GRID SEARCH
# =============================================================================

print("\n" + "="*80)
print("RMST GRID SEARCH (survRM2 native MGMT adjustment)")
print("="*80)
print(f"Thresholds: {CLUSTER_THRESHOLDS}")
print(f"Tau values: {TAU_OPTIONS}")
print(f"Scoring: {SCORING_METHODS}")
print(f"Stratifications: {STRATIFICATIONS}")
print()

all_results = []
cluster_info = {}

for threshold in CLUSTER_THRESHOLDS:
    print(f"\n--- Threshold {threshold} ---")
    
    cluster_dict, _ = get_clusters_at_threshold(corr_l4, threshold, CLUSTER_METHOD)
    cluster_info[threshold] = cluster_dict
    count = 0
    
    for cluster_num, cluster_celltypes in cluster_dict.items():
        cluster_cols = [c for c in cluster_celltypes if c in props_l4.columns]
        
        if len(cluster_cols) < 2:
            continue
        
        for tau in TAU_OPTIONS:
            for scoring in SCORING_METHODS:
                for strat in STRATIFICATIONS:
                    result = run_rmst_survRM2(
                        props_l4, cluster_cols, clin,
                        scoring, strat, time_col, event_col, tau,
                        adjust_mgmt=True
                    )
                    
                    if result:
                        result["threshold"] = threshold
                        result["cluster"] = cluster_num
                        result["n_celltypes"] = len(cluster_cols)
                        result["celltypes"] = "; ".join(cluster_cols)
                        result["scoring"] = scoring
                        result["stratification"] = strat
                        all_results.append(result)
                        count += 1
    
    print(f"  Valid results: {count}")

print(f"\n\nTotal valid results: {len(all_results)}")

if not all_results:
    raise ValueError("No valid results")

results_df = pd.DataFrame(all_results)
results_df = results_df.sort_values("rmst_p")

# =============================================================================
# T CELL FOCUS
# =============================================================================

print("\n" + "="*80)
print("T CELL CLUSTERS - RMST (MGMT-adjusted via survRM2)")
print("="*80)

tcell_results = results_df[
    (results_df["celltypes"].apply(contains_tcell)) &
    (results_df["rmst_diff"] > 0)  # High lives longer (protective)
].copy().sort_values("rmst_p")

print(f"\nT cell clusters with protective effect: {len(tcell_results)}")

if len(tcell_results) > 0:
    print("\n--- Top 20 Results ---")
    display_cols = ["threshold", "cluster", "n_celltypes", "tau", "scoring", "stratification",
                    "n_high", "n_low", "rmst_high", "rmst_low", "rmst_diff", "rmst_p", "rmst_p_unadjusted"]
    top20 = tcell_results.head(20).copy()
    top20["rmst_high"] = top20["rmst_high"].round(3)
    top20["rmst_low"] = top20["rmst_low"].round(3)
    top20["rmst_diff"] = top20["rmst_diff"].round(3)
    top20["rmst_p"] = top20["rmst_p"].round(4)
    top20["rmst_p_unadjusted"] = top20["rmst_p_unadjusted"].round(4)
    print(top20[display_cols].to_string(index=False))
    
    # Significant
    sig = tcell_results[tcell_results["rmst_p"] < 0.05]
    print(f"\n--- Significant at p < 0.05 (MGMT-adjusted): {len(sig)} ---")
    
    if len(sig) > 0:
        unique_sig = sig.drop_duplicates(subset=["threshold", "cluster"]).sort_values("rmst_p")
        for _, row in unique_sig.iterrows():
            diff_mo = row['rmst_diff'] * 12
            print(f"\n**Threshold {row['threshold']}, Cluster {row['cluster']}**")
            print(f"  Adjusted p = {row['rmst_p']:.4f}, Unadjusted p = {row['rmst_p_unadjusted']:.4f}")
            print(f"  RMST diff = +{diff_mo:.1f} months (tau={row['tau']}yr)")
            print(f"  Cell types: {row['celltypes']}")
            tcells = [c for c in row['celltypes'].split("; ") if "T_cell" in c or "T_reg" in c]
            print(f"  T cells: {', '.join(tcells)}")
    
    # Borderline
    borderline = tcell_results[(tcell_results["rmst_p"] >= 0.05) & (tcell_results["rmst_p"] < 0.1)]
    print(f"\n--- Borderline (0.05 <= p < 0.1): {len(borderline)} ---")
    if len(borderline) > 0:
        unique_bl = borderline.drop_duplicates(subset=["threshold", "cluster"]).sort_values("rmst_p")
        for _, row in unique_bl.head(5).iterrows():
            diff_mo = row['rmst_diff'] * 12
            print(f"\nThreshold {row['threshold']}, Cluster {row['cluster']}: adj_p={row['rmst_p']:.4f}, +{diff_mo:.1f} mo")
            print(f"  {row['celltypes']}")

# =============================================================================
# BEST RESULT
# =============================================================================

if len(tcell_results) > 0:
    best = tcell_results.iloc[0]
    
    print("\n" + "="*80)
    print("BEST T CELL RESULT")
    print("="*80)
    print(f"Threshold: {best['threshold']}, Cluster: {best['cluster']}")
    print(f"Tau: {best['tau']} years")
    print(f"Cell types: {best['celltypes']}")
    print(f"Method: {best['method']}")
    print(f"N (High/Low): {best['n_high']}/{best['n_low']}")
    print(f"RMST High: {best['rmst_high']:.3f} yrs, Low: {best['rmst_low']:.3f} yrs")
    diff_mo = best['rmst_diff'] * 12
    print(f"RMST Difference: +{best['rmst_diff']:.3f} yrs (+{diff_mo:.1f} months)")
    print(f"95% CI: [{best['rmst_diff_lower']:.3f}, {best['rmst_diff_upper']:.3f}]")
    print(f"P-value (MGMT-adjusted): {best['rmst_p']:.4f}")
    print(f"P-value (unadjusted): {best['rmst_p_unadjusted']:.4f}")
    
    # Plot
    if best['rmst_p'] < 0.15:
        print("\n--- Generating plot ---")
        
        best_cluster_dict, _ = get_clusters_at_threshold(corr_l4, best["threshold"], CLUSTER_METHOD)
        best_cols = [c for c in best_cluster_dict[best["cluster"]] if c in props_l4.columns]
        
        props = props_l4.copy()
        props["cluster_score"] = compute_score(props, best_cols, best["scoring"])
        props = props.reset_index()
        props.columns = ["sample_id_raw"] + list(props.columns[1:])
        props = props[props["sample_id_raw"].astype(str).str.endswith("a")].copy()
        props["sample_id"] = props["sample_id_raw"].astype(str).str[:-1]
        
        merged = props[["sample_id", "cluster_score"]].merge(clin, on="sample_id", how="inner")
        surv_df = merged[["sample_id", "cluster_score", time_col, event_col]].copy()
        surv_df = surv_df.rename(columns={time_col: "time", event_col: "event"})
        surv_df["time"] = pd.to_numeric(surv_df["time"], errors="coerce")
        surv_df["event"] = pd.to_numeric(surv_df["event"], errors="coerce")
        surv_df = surv_df.dropna()
        surv_df = surv_df[surv_df["time"] > 0]
        
        groups, exclude = stratify(surv_df["cluster_score"], best["stratification"])
        surv_df["group"] = groups
        if exclude:
            surv_df = surv_df[surv_df["group"] != exclude].copy()
        surv_df["arm"] = (surv_df["group"] == "High").astype(int)
        
        with localconverter(ro.default_converter + pandas2ri.converter):
            r_df = ro.conversion.py2rpy(surv_df)
        ro.globalenv["surv_data"] = r_df
        
        ro.r("km_fit <- survfit(Surv(time, event) ~ arm, data = surv_data)")
        
        km_times_0 = np.array(ro.r("km_fit[1]$time"))
        km_surv_0 = np.array(ro.r("km_fit[1]$surv"))
        km_lower_0 = np.array(ro.r("km_fit[1]$lower"))
        km_upper_0 = np.array(ro.r("km_fit[1]$upper"))
        
        km_times_1 = np.array(ro.r("km_fit[2]$time"))
        km_surv_1 = np.array(ro.r("km_fit[2]$surv"))
        km_lower_1 = np.array(ro.r("km_fit[2]$lower"))
        km_upper_1 = np.array(ro.r("km_fit[2]$upper"))
        
        fig, ax = plt.subplots(figsize=(FIG_WIDTH, FIG_HEIGHT))
        
        times_0 = np.concatenate([[0], km_times_0])
        surv_0 = np.concatenate([[1], km_surv_0])
        lower_0 = np.concatenate([[1], km_lower_0])
        upper_0 = np.concatenate([[1], km_upper_0])
        ax.step(times_0, surv_0, where="post", color="#0072B2", label="Low", linewidth=2)
        ax.fill_between(times_0, lower_0, upper_0, step="post", alpha=0.2, color="#0072B2")
        
        times_1 = np.concatenate([[0], km_times_1])
        surv_1 = np.concatenate([[1], km_surv_1])
        lower_1 = np.concatenate([[1], km_lower_1])
        upper_1 = np.concatenate([[1], km_upper_1])
        ax.step(times_1, surv_1, where="post", color="#D55E00", label="High", linewidth=2)
        ax.fill_between(times_1, lower_1, upper_1, step="post", alpha=0.2, color="#D55E00")
        
        ax.axvline(x=best["tau"], color="gray", linestyle="--", alpha=0.7, label=f"τ={best['tau']}yr")
        
        rmst_text = (
            f"RMST (τ={best['tau']}yr):\n"
            f"  High: {best['rmst_high']:.2f} yrs\n"
            f"  Low: {best['rmst_low']:.2f} yrs\n"
            f"  Δ: +{diff_mo:.1f} months\n"
            f"  p(adj) = {best['rmst_p']:.4f}\n"
            f"  p(unadj) = {best['rmst_p_unadjusted']:.4f}"
        )
        ax.text(0.98, 0.98, rmst_text, transform=ax.transAxes, fontsize=9,
                verticalalignment='top', horizontalalignment='right',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))
        
        tcells = [c for c in best['celltypes'].split("; ") if "T_cell" in c or "T_reg" in c]
        title = (
            f"{SURVIVAL_TYPE} by T cell cluster (thresh={best['threshold']}, clust={best['cluster']})\n"
            f"T cells: {', '.join(tcells)}\n"
            f"{best['scoring']}, {best['stratification']}, MGMT-adjusted (survRM2)"
        )
        
        ax.set_xlabel("Time (years)", fontsize=11)
        ax.set_ylabel("Survival probability", fontsize=11)
        ax.set_title(title, fontsize=9)
        ax.legend(loc="lower left")
        ax.set_ylim(0, 1.05)
        ax.set_xlim(0, None)
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)
        
        plt.tight_layout()
        plt.show()

# =============================================================================
# SAVE
# =============================================================================

results_df.to_csv(f"{OUT_DIR}/rmst_survRM2_mgmt_adjusted.csv", index=False)
if len(tcell_results) > 0:
    tcell_results.to_csv(f"{OUT_DIR}/rmst_tcell_survRM2_mgmt_adjusted.csv", index=False)
print(f"\nSaved to {OUT_DIR}")

rmst_all_results = results_df
rmst_tcell_results = tcell_results if len(tcell_results) > 0 else None
```

```python
# =============================================================================
# PLOT — select result by cluster
# =============================================================================

SEL_THRESHOLD    = 1
SEL_CLUSTER      = 7
SEL_TAU          = 1.5
SEL_SCORING      = "max"
SEL_STRAT        = "quartile_vs_rest"

# -----------------------------------------------------------------------
mask = (
    (rmst_all_results["threshold"]      == SEL_THRESHOLD) &
    (rmst_all_results["cluster"]        == SEL_CLUSTER) &
    (rmst_all_results["tau"]            == SEL_TAU) &
    (rmst_all_results["scoring"]        == SEL_SCORING) &
    (rmst_all_results["stratification"] == SEL_STRAT)
)

matches = rmst_all_results[mask].sort_values("rmst_p")
if len(matches) == 0:
    print(f"No matching rows. Available clusters at threshold {SEL_THRESHOLD}:")
    avail = rmst_all_results[rmst_all_results["threshold"] == SEL_THRESHOLD][["cluster", "celltypes"]].drop_duplicates()
    print(avail.to_string(index=False))
    raise ValueError("No rows matched — check SEL_THRESHOLD, SEL_CLUSTER, SEL_SCORING, SEL_STRAT, SEL_TAU")

row = matches.iloc[0]
print(f"Using: threshold={row['threshold']}, cluster={row['cluster']}, "
      f"strat={row['stratification']}, scoring={row['scoring']}, tau={row['tau']}")
print(f"Expected: rmst_p={row['rmst_p']:.4f}, diff={row['rmst_diff']*12:.1f} mo, "
      f"n_high={row['n_high']}, n_low={row['n_low']}")

# -----------------------------------------------------------------------
cluster_cols = [c for c in row["celltypes"].split("; ") if c in props_l4.columns]
print(f"Cell types: {cluster_cols}")

props = props_l4.copy()
props["cluster_score"] = compute_score(props, cluster_cols, SEL_SCORING)
props = props.reset_index()
props.columns = ["sample_id_raw"] + list(props.columns[1:])
props = props[props["sample_id_raw"].astype(str).str.endswith("a")].copy()
props["sample_id"] = props["sample_id_raw"].astype(str).str[:-1]

merged  = props[["sample_id", "cluster_score"]].merge(clin, on="sample_id", how="inner")
surv_df = merged[["sample_id", "cluster_score", time_col, event_col]].copy()
surv_df["mgmt"] = (merged[MGMT_COL].astype(str) == "Methylated").astype(int).values
surv_df = surv_df.rename(columns={time_col: "time", event_col: "event"})
surv_df["time"]  = pd.to_numeric(surv_df["time"],  errors="coerce")
surv_df["event"] = pd.to_numeric(surv_df["event"], errors="coerce")
surv_df = surv_df.dropna(subset=["time", "event", "cluster_score", "mgmt"])
surv_df = surv_df[surv_df["time"] > 0]

groups, exclude = stratify(surv_df["cluster_score"], SEL_STRAT)
surv_df["group"] = groups
if exclude:
    surv_df = surv_df[surv_df["group"] != exclude].copy()
surv_df["arm"] = (surv_df["group"] == "High").astype(int)

n_high = (surv_df["group"] == "High").sum()
n_low  = (surv_df["group"] == "Low").sum()
print(f"Actual:   n_high={n_high}, n_low={n_low}")

# -----------------------------------------------------------------------
with localconverter(ro.default_converter + pandas2ri.converter):
    r_df = ro.conversion.py2rpy(surv_df)
ro.globalenv["surv_data"] = r_df
ro.globalenv["tau"]       = float(SEL_TAU)

ro.r("""
library(survRM2)
cov_matrix     <- as.matrix(surv_data$mgmt)
rmst_adj       <- rmst2(surv_data$time, surv_data$event, surv_data$arm,
                        tau = tau, covariates = cov_matrix)
adj_diff       <- rmst_adj$adjusted.result[1, "Est."]
adj_diff_lower <- rmst_adj$adjusted.result[1, "lower .95"]
adj_diff_upper <- rmst_adj$adjusted.result[1, "upper .95"]
adj_p          <- rmst_adj$adjusted.result[1, "p"]
rmst_unadj     <- rmst2(surv_data$time, surv_data$event, surv_data$arm, tau = tau)
rmst_high      <- rmst_unadj$RMST.arm1$rmst["Est."]
rmst_low       <- rmst_unadj$RMST.arm0$rmst["Est."]
unadj_p        <- rmst_unadj$unadjusted.result[1, "p"]
km_fit         <- survfit(Surv(time, event) ~ arm, data = surv_data)
""")

rmst_diff       = ro.r("adj_diff")[0]
rmst_diff_lower = ro.r("adj_diff_lower")[0]
rmst_diff_upper = ro.r("adj_diff_upper")[0]
rmst_p          = ro.r("adj_p")[0]
rmst_high_val   = ro.r("rmst_high")[0]
rmst_low_val    = ro.r("rmst_low")[0]
unadj_p         = ro.r("unadj_p")[0]
print(f"Reproduced: rmst_p={rmst_p:.4f}, diff={rmst_diff*12:.1f} mo")

km_times_0 = np.array(ro.r("km_fit[1]$time"))
km_surv_0  = np.array(ro.r("km_fit[1]$surv"))
km_lower_0 = np.array(ro.r("km_fit[1]$lower"))
km_upper_0 = np.array(ro.r("km_fit[1]$upper"))
km_times_1 = np.array(ro.r("km_fit[2]$time"))
km_surv_1  = np.array(ro.r("km_fit[2]$surv"))
km_lower_1 = np.array(ro.r("km_fit[2]$lower"))
km_upper_1 = np.array(ro.r("km_fit[2]$upper"))

# -----------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(FIG_WIDTH, FIG_HEIGHT))

times_0 = np.concatenate([[0], km_times_0])
surv_0  = np.concatenate([[1], km_surv_0])
lower_0 = np.concatenate([[1], km_lower_0])
upper_0 = np.concatenate([[1], km_upper_0])
ax.step(times_0, surv_0, where="post", color="#0072B2", label="Low", linewidth=2)
ax.fill_between(times_0, lower_0, upper_0, step="post", alpha=0.2, color="#0072B2")

times_1 = np.concatenate([[0], km_times_1])
surv_1  = np.concatenate([[1], km_surv_1])
lower_1 = np.concatenate([[1], km_lower_1])
upper_1 = np.concatenate([[1], km_upper_1])
ax.step(times_1, surv_1, where="post", color="#D55E00", label="High", linewidth=2)
ax.fill_between(times_1, lower_1, upper_1, step="post", alpha=0.2, color="#D55E00")

ax.axvline(x=SEL_TAU, color="gray", linestyle="--", alpha=0.7, label=f"τ={SEL_TAU}yr")

diff_mo   = rmst_diff * 12
rmst_text = (
    f"RMST (τ={SEL_TAU}yr):\n"
    f"  High: {rmst_high_val:.2f} yrs\n"
    f"  Low:  {rmst_low_val:.2f} yrs\n"
    f"  Δ: +{diff_mo:.1f} months\n"
    f"  p(adj) = {rmst_p:.4f}\n"
    f"  p(unadj) = {unadj_p:.4f}"
)
ax.text(0.98, 0.98, rmst_text, transform=ax.transAxes, fontsize=9,
        verticalalignment="top", horizontalalignment="right",
        bbox=dict(boxstyle="round", facecolor="white", alpha=0.8))

tcells = [c for c in cluster_cols if "T_cell" in c or "T_reg" in c or "Treg" in c]
title = (
    f"{SURVIVAL_TYPE} by T cell cluster "
    f"(thresh={SEL_THRESHOLD}, clust={SEL_CLUSTER})\n"
    f"T cells: {', '.join(tcells)}\n"
    f"{SEL_SCORING}, {SEL_STRAT}, MGMT-adjusted (survRM2)"
)

ax.set_xlabel("Time (years)", fontsize=11)
ax.set_ylabel("Survival probability", fontsize=11)
ax.set_title(title, fontsize=9)
ax.legend(loc="lower left")
ax.set_ylim(0, 1.05)
ax.set_xlim(0, None)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.savefig(
    f"{OUT_DIR}/km_rmst_t{SEL_THRESHOLD}_c{SEL_CLUSTER}_{SEL_SCORING}_{SEL_STRAT}.pdf",
    bbox_inches="tight")
plt.show()
```

```python
# %% [markdown]
# ### Find optimal threshold and tau for maximum significant OS associations

# %%
# Use the full results from the previous RMST analysis
# results_df contains ALL clusters (not just T cell)

print("="*80)
print("UNBIASED ANALYSIS: ALL CLUSTERS")
print("="*80)

# Top 20 results across ALL clusters
print("\n--- TOP 20 RESULTS (ALL CLUSTERS) ---")
display_cols = ["threshold", "cluster", "n_celltypes", "tau", "scoring", "stratification",
                "n_high", "n_low", "rmst_diff", "rmst_p", "rmst_p_unadjusted"]

top20_all = rmst_all_results.head(20).copy()
top20_all["rmst_diff"] = top20_all["rmst_diff"].round(3)
top20_all["rmst_p"] = top20_all["rmst_p"].round(4)
top20_all["rmst_p_unadjusted"] = top20_all["rmst_p_unadjusted"].round(4)
print(top20_all[display_cols].to_string(index=False))

# Show cell types for top 10
print("\n--- CELL TYPES IN TOP 10 CLUSTERS ---")
for i, row in rmst_all_results.head(10).iterrows():
    diff_mo = row['rmst_diff'] * 12
    direction = "protective" if row['rmst_diff'] > 0 else "harmful"
    print(f"\n{i+1}. Threshold {row['threshold']}, Cluster {row['cluster']} (p={row['rmst_p']:.4f}, {direction})")
    print(f"   RMST diff: {diff_mo:+.1f} months, tau={row['tau']}yr")
    print(f"   Cells: {row['celltypes']}")

# %%
# Summary by threshold: how many significant results?
print("\n" + "="*80)
print("SIGNIFICANT RESULTS BY THRESHOLD")
print("="*80)

sig_all = rmst_all_results[rmst_all_results["rmst_p"] < 0.05]

print(f"\nTotal significant results (p < 0.05): {len(sig_all)}")

# Count by threshold
print("\n--- By Threshold ---")
thresh_summary = sig_all.groupby("threshold").agg({
    "rmst_p": ["count", "min"],
    "cluster": "nunique"
}).reset_index()
thresh_summary.columns = ["Threshold", "N_significant", "Best_p", "N_unique_clusters"]
thresh_summary = thresh_summary.sort_values("N_significant", ascending=False)
print(thresh_summary.to_string(index=False))

# Count by tau
print("\n--- By Tau ---")
tau_summary = sig_all.groupby("tau").agg({
    "rmst_p": ["count", "min"],
    "cluster": "nunique"
}).reset_index()
tau_summary.columns = ["Tau", "N_significant", "Best_p", "N_unique_clusters"]
tau_summary = tau_summary.sort_values("N_significant", ascending=False)
print(tau_summary.to_string(index=False))

# Count by threshold AND tau
print("\n--- By Threshold × Tau ---")
thresh_tau = sig_all.groupby(["threshold", "tau"]).agg({
    "rmst_p": ["count", "min"],
    "cluster": "nunique"
}).reset_index()
thresh_tau.columns = ["Threshold", "Tau", "N_significant", "Best_p", "N_unique_clusters"]
thresh_tau = thresh_tau.sort_values("N_significant", ascending=False)
print(thresh_tau.head(15).to_string(index=False))

# %%
# Unique significant clusters at each threshold
print("\n" + "="*80)
print("UNIQUE SIGNIFICANT CLUSTERS BY THRESHOLD")
print("="*80)

for thresh in sorted(rmst_all_results["threshold"].unique()):
    sig_at_thresh = sig_all[sig_all["threshold"] == thresh]
    unique_clusters = sig_at_thresh.drop_duplicates(subset=["cluster"]).sort_values("rmst_p")
    
    print(f"\n--- Threshold {thresh}: {len(unique_clusters)} unique significant clusters ---")
    
    for _, row in unique_clusters.iterrows():
        diff_mo = row['rmst_diff'] * 12
        direction = "↑" if row['rmst_diff'] > 0 else "↓"
        print(f"  Cluster {row['cluster']:2d}: p={row['rmst_p']:.4f}, {direction}{abs(diff_mo):4.1f}mo | {row['celltypes'][:80]}...")

# %%
# Find the BEST threshold/tau combination
print("\n" + "="*80)
print("OPTIMAL THRESHOLD AND TAU SELECTION")
print("="*80)

# Criteria: maximize number of significant unique clusters
# with consideration for effect sizes and biological interpretability

# For each threshold, find best tau
print("\n--- Best Tau for Each Threshold ---")
best_configs = []

for thresh in sorted(rmst_all_results["threshold"].unique()):
    thresh_data = rmst_all_results[rmst_all_results["threshold"] == thresh]
    
    for tau in sorted(thresh_data["tau"].unique()):
        subset = thresh_data[thresh_data["tau"] == tau]
        sig_subset = subset[subset["rmst_p"] < 0.05]
        
        n_sig = len(sig_subset)
        n_clusters_sig = sig_subset["cluster"].nunique() if n_sig > 0 else 0
        best_p = sig_subset["rmst_p"].min() if n_sig > 0 else 1.0
        
        # Count protective (positive diff) vs harmful (negative diff)
        n_protective = len(sig_subset[sig_subset["rmst_diff"] > 0]) if n_sig > 0 else 0
        n_harmful = len(sig_subset[sig_subset["rmst_diff"] < 0]) if n_sig > 0 else 0
        
        # Mean absolute effect size among significant
        mean_effect = sig_subset["rmst_diff"].abs().mean() * 12 if n_sig > 0 else 0
        
        best_configs.append({
            "threshold": thresh,
            "tau": tau,
            "n_significant": n_sig,
            "n_clusters_sig": n_clusters_sig,
            "best_p": best_p,
            "n_protective": n_protective,
            "n_harmful": n_harmful,
            "mean_effect_months": mean_effect
        })

config_df = pd.DataFrame(best_configs)
config_df = config_df.sort_values(["n_clusters_sig", "n_significant"], ascending=False)

print(config_df.head(20).to_string(index=False))

# %%
# Recommend optimal configuration
print("\n" + "="*80)
print("RECOMMENDATION")
print("="*80)

# Best by number of unique significant clusters
best_config = config_df.iloc[0]
print(f"\nBest configuration by # unique significant clusters:")
print(f"  Threshold: {best_config['threshold']}")
print(f"  Tau: {best_config['tau']} years")
print(f"  Significant results: {best_config['n_significant']}")
print(f"  Unique significant clusters: {best_config['n_clusters_sig']}")
print(f"  Best p-value: {best_config['best_p']:.4f}")
print(f"  Protective associations: {best_config['n_protective']}")
print(f"  Harmful associations: {best_config['n_harmful']}")
print(f"  Mean effect size: {best_config['mean_effect_months']:.1f} months")

# %%
# Show all significant clusters at the optimal configuration
optimal_thresh = best_config['threshold']
optimal_tau = best_config['tau']

print(f"\n" + "="*80)
print(f"ALL SIGNIFICANT CLUSTERS AT OPTIMAL CONFIG (threshold={optimal_thresh}, tau={optimal_tau})")
print("="*80)

optimal_results = rmst_all_results[
    (rmst_all_results["threshold"] == optimal_thresh) &
    (rmst_all_results["tau"] == optimal_tau) &
    (rmst_all_results["rmst_p"] < 0.05)
].copy()

# Get best result per cluster
optimal_by_cluster = optimal_results.sort_values("rmst_p").drop_duplicates(subset=["cluster"])
optimal_by_cluster = optimal_by_cluster.sort_values("rmst_p")

print(f"\n{len(optimal_by_cluster)} unique significant clusters:\n")

for _, row in optimal_by_cluster.iterrows():
    diff_mo = row['rmst_diff'] * 12
    direction = "PROTECTIVE" if row['rmst_diff'] > 0 else "HARMFUL"
    ci_lo = row['rmst_diff_lower'] * 12
    ci_hi = row['rmst_diff_upper'] * 12
    
    print(f"Cluster {row['cluster']} ({row['n_celltypes']} cell types) - {direction}")
    print(f"  p = {row['rmst_p']:.4f} (unadj: {row['rmst_p_unadjusted']:.4f})")
    print(f"  RMST diff: {diff_mo:+.1f} months (95% CI: {ci_lo:+.1f} to {ci_hi:+.1f})")
    print(f"  High: {row['rmst_high']:.2f} yrs, Low: {row['rmst_low']:.2f} yrs")
    print(f"  Best params: {row['scoring']}, {row['stratification']}")
    print(f"  Cell types: {row['celltypes']}")
    print()

# %%
# Summary table for publication
print("\n" + "="*80)
print("SUMMARY TABLE FOR PUBLICATION")
print("="*80)

summary_table = optimal_by_cluster[[
    "cluster", "n_celltypes", "rmst_diff", "rmst_diff_lower", "rmst_diff_upper", 
    "rmst_p", "rmst_p_unadjusted", "scoring", "stratification", "celltypes"
]].copy()

summary_table["rmst_diff_months"] = summary_table["rmst_diff"] * 12
summary_table["ci_lower_months"] = summary_table["rmst_diff_lower"] * 12
summary_table["ci_upper_months"] = summary_table["rmst_diff_upper"] * 12
summary_table["direction"] = summary_table["rmst_diff"].apply(lambda x: "Protective" if x > 0 else "Harmful")

# Reorder columns
pub_table = summary_table[[
    "cluster", "n_celltypes", "direction", "rmst_diff_months", 
    "ci_lower_months", "ci_upper_months", "rmst_p", "celltypes"
]].copy()
pub_table.columns = ["Cluster", "N_celltypes", "Direction", "RMST_diff_months", 
                     "CI_lower", "CI_upper", "P_adjusted", "Cell_types"]

print(f"\nOptimal configuration: Threshold={optimal_thresh}, Tau={optimal_tau} years\n")
print(pub_table.to_string(index=False))

# Save
pub_table.to_csv(f"{OUT_DIR}/rmst_significant_clusters_optimal_config.csv", index=False)
print(f"\nSaved to {OUT_DIR}/rmst_significant_clusters_optimal_config.csv")

# %%
# Store optimal configuration
optimal_threshold = optimal_thresh
optimal_tau_value = optimal_tau
optimal_significant_clusters = optimal_by_cluster
```

### Adjust RMST for Age and MGMT

```python
# %% [markdown]
# ### RMST grid search with MGMT + Age adjustment
# 
# This extends the previous analysis to adjust for BOTH major clinical covariates:
# - MGMT promoter methylation status
# - Age at diagnosis
#
# This gives the strongest statement: survival associations independent of 
# the two biggest prognostic factors in GBM.

# %%
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, fcluster
from scipy.spatial.distance import squareform
import warnings
warnings.filterwarnings("ignore")

import rpy2.robjects as ro
from rpy2.robjects import pandas2ri
from rpy2.robjects.conversion import localconverter

# =============================================================================
# USER CONFIG
# =============================================================================

CT_KEY_L4 = "celltype_level4_curated"
SAMPLE_KEY = "orig.ident"
EXCLUDE_PATTERNS = ["Doublet", "doublet"]

CLUSTER_THRESHOLDS = [0.7, 0.8, 0.9, 1.0, 1.2, 1.3, 1.5]
CLUSTER_METHOD = "ward"

SURVIVAL_TYPE = "OS"

SCORING_METHODS = ["sum", "mean", "max", "pca"]
STRATIFICATIONS = ["median", "tertile", "quartile_vs_rest", "quartile_vs_quartile"]

TAU_OPTIONS = [1.5, 2.0, 2.5, 3.0]

P_THRESHOLD = 0.05

SAMPLE_ID_COL = "Subject Id"
CHRONOLOGY_COL = "Sample collection chronology"
TIME_COL_OS = "os_years"
EVENT_COL_OS = "os_event"
TIME_COL_PFS = "pfs_years"
EVENT_COL_PFS = "pfs_event"
MGMT_COL = "MGMT promoter methylation"
AGE_COL = "age_at_diagnosis"

FIG_WIDTH = 8
FIG_HEIGHT = 6

OUT_DIR = "/home/sagemaker-user/gbm-collaboration_2/analysis_sc_merged"

# =============================================================================
# HELPER FUNCTIONS (same as before)
# =============================================================================

def is_excluded(ct_name):
    for pattern in EXCLUDE_PATTERNS:
        if pattern in str(ct_name):
            return True
    return False

def build_proportion_matrix(adata, ct_key, sample_key):
    obs = adata.obs[[sample_key, ct_key]].copy()
    obs[sample_key] = obs[sample_key].astype(str)
    obs[ct_key] = obs[ct_key].astype(str)
    counts = obs.groupby([sample_key, ct_key]).size().unstack(fill_value=0)
    keep_cols = [c for c in counts.columns if not is_excluded(c)]
    counts = counts[keep_cols]
    props = counts.div(counts.sum(axis=1), axis=0)
    return props

def clean_corr(corr):
    corr = corr.replace([np.inf, -np.inf], np.nan)
    corr = corr.dropna(axis=0, how="all").dropna(axis=1, how="all")
    common = corr.index.intersection(corr.columns)
    corr = corr.loc[common, common]
    return corr.fillna(0.0)

def compute_linkage(corr, method="ward"):
    dist_matrix = 1 - corr.values
    np.fill_diagonal(dist_matrix, 0)
    dist_matrix = (dist_matrix + dist_matrix.T) / 2
    dist_matrix = np.clip(dist_matrix, 0, None)
    condensed = squareform(dist_matrix, checks=False)
    return linkage(condensed, method=method)

def get_clusters_at_threshold(corr, threshold, method="ward"):
    Z = compute_linkage(corr, method)
    cluster_labels = fcluster(Z, t=threshold, criterion="distance")
    cluster_df = pd.DataFrame({"celltype": corr.index, "cluster": cluster_labels})
    cluster_dict = {}
    for c in sorted(cluster_df["cluster"].unique()):
        members = cluster_df[cluster_df["cluster"] == c]["celltype"].tolist()
        cluster_dict[c] = members
    return cluster_dict, cluster_df

def compute_score(props, cluster_cols, method):
    if method == "sum":
        return props[cluster_cols].sum(axis=1)
    elif method == "mean":
        return props[cluster_cols].mean(axis=1)
    elif method == "max":
        return props[cluster_cols].max(axis=1)
    elif method == "pca":
        from sklearn.preprocessing import StandardScaler
        from sklearn.decomposition import PCA
        X = props[cluster_cols].values
        if X.shape[1] < 2:
            return props[cluster_cols].iloc[:, 0]
        X_scaled = StandardScaler().fit_transform(X)
        pca = PCA(n_components=1)
        return pca.fit_transform(X_scaled).flatten()
    else:
        raise ValueError(f"Unknown method: {method}")

def stratify(scores, method):
    if method == "median":
        threshold = scores.median()
        groups = np.where(scores >= threshold, "High", "Low")
        return groups, None
    elif method == "tertile":
        t33 = scores.quantile(0.33)
        t67 = scores.quantile(0.67)
        groups = np.where(scores >= t67, "High", np.where(scores <= t33, "Low", "Middle"))
        return groups, "Middle"
    elif method == "quartile_vs_rest":
        q75 = scores.quantile(0.75)
        groups = np.where(scores >= q75, "High", "Low")
        return groups, None
    elif method == "quartile_vs_quartile":
        q25 = scores.quantile(0.25)
        q75 = scores.quantile(0.75)
        groups = np.where(scores >= q75, "High", np.where(scores <= q25, "Low", "Middle"))
        return groups, "Middle"
    else:
        raise ValueError(f"Unknown stratification: {method}")

def contains_tcell(celltypes_str):
    celltypes = celltypes_str.split("; ")
    for ct in celltypes:
        if "T_cell" in ct or "T_reg" in ct or "Treg" in ct:
            return True
    return False

# =============================================================================
# RMST ANALYSIS with MGMT + AGE adjustment
# =============================================================================

def run_rmst_mgmt_age(props_orig, cluster_cols, clin, scoring_method, stratification,
                      time_col, event_col, tau):
    """
    Run RMST analysis using survRM2 with BOTH MGMT and Age adjustment.
    """
    props = props_orig.copy()
    
    try:
        props["cluster_score"] = compute_score(props, cluster_cols, scoring_method)
    except Exception:
        return None
    
    props = props.reset_index()
    props.columns = ["sample_id_raw"] + list(props.columns[1:])
    props = props[props["sample_id_raw"].astype(str).str.endswith("a")].copy()
    props["sample_id"] = props["sample_id_raw"].astype(str).str[:-1]
    
    merged = props[["sample_id", "cluster_score"]].merge(clin, on="sample_id", how="inner")
    
    surv_df = merged[["sample_id", "cluster_score", time_col, event_col]].copy()
    surv_df["mgmt"] = (merged[MGMT_COL].astype(str) == "Methylated").astype(int).values
    surv_df["age"] = pd.to_numeric(merged[AGE_COL], errors="coerce").values
    
    surv_df = surv_df.rename(columns={time_col: "time", event_col: "event"})
    surv_df["time"] = pd.to_numeric(surv_df["time"], errors="coerce")
    surv_df["event"] = pd.to_numeric(surv_df["event"], errors="coerce")
    
    # Drop rows with missing covariates
    surv_df = surv_df.dropna(subset=["time", "event", "cluster_score", "mgmt", "age"])
    surv_df = surv_df[surv_df["time"] > 0]
    
    if len(surv_df) < 20:
        return None
    
    groups, exclude = stratify(surv_df["cluster_score"], stratification)
    surv_df["group"] = groups
    
    if exclude:
        surv_df = surv_df[surv_df["group"] != exclude].copy()
    
    if len(surv_df) < 20:
        return None
    
    surv_df["arm"] = (surv_df["group"] == "High").astype(int)
    
    n_high = (surv_df["group"] == "High").sum()
    n_low = (surv_df["group"] == "Low").sum()
    
    if n_high < 5 or n_low < 5:
        return None
    
    max_time = surv_df["time"].max()
    if tau > max_time:
        return None
    
    with localconverter(ro.default_converter + pandas2ri.converter):
        r_df = ro.conversion.py2rpy(surv_df)
    
    ro.globalenv["surv_data"] = r_df
    ro.globalenv["tau"] = tau
    
    try:
        # Adjusted analysis with BOTH MGMT and Age
        ro.r("""
        library(survRM2)
        
        # Create covariate matrix (MGMT and Age)
        # Standardize age for numerical stability
        surv_data$age_scaled <- scale(surv_data$age)[,1]
        cov_matrix <- cbind(surv_data$mgmt, surv_data$age_scaled)
        
        # Run adjusted RMST analysis
        rmst_adj <- rmst2(
            time = surv_data$time,
            status = surv_data$event,
            arm = surv_data$arm,
            tau = tau,
            covariates = cov_matrix
        )
        
        # Extract adjusted results
        adj_diff <- rmst_adj$adjusted.result[1, "Est."]
        adj_diff_lower <- rmst_adj$adjusted.result[1, "lower .95"]
        adj_diff_upper <- rmst_adj$adjusted.result[1, "upper .95"]
        adj_p <- rmst_adj$adjusted.result[1, "p"]
        
        # Also get unadjusted for comparison
        rmst_unadj <- rmst2(
            time = surv_data$time,
            status = surv_data$event,
            arm = surv_data$arm,
            tau = tau
        )
        rmst_high <- rmst_unadj$RMST.arm1$rmst["Est."]
        rmst_low <- rmst_unadj$RMST.arm0$rmst["Est."]
        unadj_diff <- rmst_unadj$unadjusted.result[1, "Est."]
        unadj_p <- rmst_unadj$unadjusted.result[1, "p"]
        
        # Get MGMT-only adjusted for comparison
        cov_mgmt_only <- as.matrix(surv_data$mgmt)
        rmst_mgmt <- rmst2(
            time = surv_data$time,
            status = surv_data$event,
            arm = surv_data$arm,
            tau = tau,
            covariates = cov_mgmt_only
        )
        mgmt_adj_p <- rmst_mgmt$adjusted.result[1, "p"]
        """)
        
        rmst_diff = ro.r("adj_diff")[0]
        rmst_diff_lower = ro.r("adj_diff_lower")[0]
        rmst_diff_upper = ro.r("adj_diff_upper")[0]
        rmst_p = ro.r("adj_p")[0]
        rmst_high = ro.r("rmst_high")[0]
        rmst_low = ro.r("rmst_low")[0]
        unadj_diff = ro.r("unadj_diff")[0]
        unadj_p = ro.r("unadj_p")[0]
        mgmt_only_p = ro.r("mgmt_adj_p")[0]
        
    except Exception as e:
        return None
    
    return {
        "n_total": len(surv_df),
        "n_high": n_high,
        "n_low": n_low,
        "tau": tau,
        "rmst_high": rmst_high,
        "rmst_low": rmst_low,
        "rmst_diff": rmst_diff,
        "rmst_diff_lower": rmst_diff_lower,
        "rmst_diff_upper": rmst_diff_upper,
        "rmst_p_mgmt_age": rmst_p,  # Adjusted for BOTH
        "rmst_p_mgmt_only": mgmt_only_p,  # Adjusted for MGMT only
        "rmst_p_unadjusted": unadj_p,  # No adjustment
    }

# =============================================================================
# MAIN ANALYSIS
# =============================================================================

print("Loading R packages...")
ro.r("""
suppressPackageStartupMessages({
    library(survival)
    library(survRM2)
})
""")
print("  survival and survRM2 loaded")

if SURVIVAL_TYPE == "OS":
    time_col, event_col = TIME_COL_OS, EVENT_COL_OS
else:
    time_col, event_col = TIME_COL_PFS, EVENT_COL_PFS

# Build data if not present
if 'props_l4' not in dir():
    print("\nBuilding proportion matrix for level 4...")
    props_l4 = build_proportion_matrix(adata_sc, CT_KEY_L4, SAMPLE_KEY)
    print(f"  {props_l4.shape[0]} samples x {props_l4.shape[1]} cell types")
    print("Computing correlation matrix...")
    corr_l4 = clean_corr(props_l4.corr(method="pearson"))

if 'clin' not in dir():
    print("Preparing clinical data...")
    clin = clin_out.copy()
    clin["sample_id"] = clin[SAMPLE_ID_COL].astype(str)
    clin["_priority"] = clin[CHRONOLOGY_COL].map({"Baseline": 0, "At recurrence": 1}).fillna(2)
    clin = clin.sort_values(["sample_id", "_priority"])
    clin = clin.drop_duplicates(subset=["sample_id"], keep="first")
    clin = clin.drop(columns=["_priority"])

# =============================================================================
# GRID SEARCH WITH MGMT + AGE ADJUSTMENT
# =============================================================================

print("\n" + "="*80)
print("RMST GRID SEARCH (MGMT + Age adjustment)")
print("="*80)
print(f"Thresholds: {CLUSTER_THRESHOLDS}")
print(f"Tau values: {TAU_OPTIONS}")
print(f"Scoring: {SCORING_METHODS}")
print(f"Stratifications: {STRATIFICATIONS}")
print(f"Covariates: MGMT methylation + Age at diagnosis")
print()

all_results = []
cluster_info = {}

for threshold in CLUSTER_THRESHOLDS:
    print(f"\n--- Threshold {threshold} ---")
    
    cluster_dict, _ = get_clusters_at_threshold(corr_l4, threshold, CLUSTER_METHOD)
    cluster_info[threshold] = cluster_dict
    count = 0
    
    for cluster_num, cluster_celltypes in cluster_dict.items():
        cluster_cols = [c for c in cluster_celltypes if c in props_l4.columns]
        
        if len(cluster_cols) < 2:
            continue
        
        for tau in TAU_OPTIONS:
            for scoring in SCORING_METHODS:
                for strat in STRATIFICATIONS:
                    result = run_rmst_mgmt_age(
                        props_l4, cluster_cols, clin,
                        scoring, strat, time_col, event_col, tau
                    )
                    
                    if result:
                        result["threshold"] = threshold
                        result["cluster"] = cluster_num
                        result["n_celltypes"] = len(cluster_cols)
                        result["celltypes"] = "; ".join(cluster_cols)
                        result["scoring"] = scoring
                        result["stratification"] = strat
                        all_results.append(result)
                        count += 1
    
    print(f"  Valid results: {count}")

print(f"\n\nTotal valid results: {len(all_results)}")

if not all_results:
    raise ValueError("No valid results")

results_df = pd.DataFrame(all_results)
results_df = results_df.sort_values("rmst_p_mgmt_age")

# =============================================================================
# COMPREHENSIVE SUMMARY
# =============================================================================

print("\n" + "="*80)
print("TOP 20 RESULTS (Adjusted for MGMT + Age)")
print("="*80)

display_cols = ["threshold", "cluster", "n_celltypes", "tau", "scoring", "stratification",
                "n_high", "n_low", "rmst_diff", "rmst_p_mgmt_age", "rmst_p_mgmt_only", "rmst_p_unadjusted"]

top20 = results_df.head(20).copy()
top20["rmst_diff"] = top20["rmst_diff"].round(3)
top20["rmst_p_mgmt_age"] = top20["rmst_p_mgmt_age"].round(4)
top20["rmst_p_mgmt_only"] = top20["rmst_p_mgmt_only"].round(4)
top20["rmst_p_unadjusted"] = top20["rmst_p_unadjusted"].round(4)
print(top20[display_cols].to_string(index=False))

# Show cell types for top 10
print("\n--- CELL TYPES IN TOP 10 ---")
for i, (_, row) in enumerate(results_df.head(10).iterrows()):
    diff_mo = row['rmst_diff'] * 12
    direction = "protective" if row['rmst_diff'] > 0 else "harmful"
    print(f"\n{i+1}. Threshold {row['threshold']}, Cluster {row['cluster']} ({direction})")
    print(f"   RMST diff: {diff_mo:+.1f} months, tau={row['tau']}yr")
    print(f"   p(MGMT+Age): {row['rmst_p_mgmt_age']:.4f}, p(MGMT): {row['rmst_p_mgmt_only']:.4f}, p(unadj): {row['rmst_p_unadjusted']:.4f}")
    print(f"   Cells: {row['celltypes']}")

# =============================================================================
# SIGNIFICANT RESULTS SUMMARY
# =============================================================================

print("\n" + "="*80)
print("SIGNIFICANT RESULTS BY THRESHOLD (p < 0.05, MGMT+Age adjusted)")
print("="*80)

sig_all = results_df[results_df["rmst_p_mgmt_age"] < 0.05]

print(f"\nTotal significant results: {len(sig_all)}")

# Count by threshold
print("\n--- By Threshold ---")
thresh_summary = sig_all.groupby("threshold").agg({
    "rmst_p_mgmt_age": ["count", "min"],
    "cluster": "nunique"
}).reset_index()
thresh_summary.columns = ["Threshold", "N_significant", "Best_p", "N_unique_clusters"]
thresh_summary = thresh_summary.sort_values("N_significant", ascending=False)
print(thresh_summary.to_string(index=False))

# Count by tau
print("\n--- By Tau ---")
tau_summary = sig_all.groupby("tau").agg({
    "rmst_p_mgmt_age": ["count", "min"],
    "cluster": "nunique"
}).reset_index()
tau_summary.columns = ["Tau", "N_significant", "Best_p", "N_unique_clusters"]
tau_summary = tau_summary.sort_values("N_significant", ascending=False)
print(tau_summary.to_string(index=False))

# Threshold x Tau
print("\n--- By Threshold × Tau ---")
thresh_tau = sig_all.groupby(["threshold", "tau"]).agg({
    "rmst_p_mgmt_age": ["count", "min"],
    "cluster": "nunique"
}).reset_index()
thresh_tau.columns = ["Threshold", "Tau", "N_significant", "Best_p", "N_unique_clusters"]
thresh_tau = thresh_tau.sort_values("N_unique_clusters", ascending=False)
print(thresh_tau.head(15).to_string(index=False))

# =============================================================================
# OPTIMAL CONFIGURATION
# =============================================================================

print("\n" + "="*80)
print("OPTIMAL CONFIGURATION SELECTION")
print("="*80)

best_configs = []
for thresh in sorted(results_df["threshold"].unique()):
    thresh_data = results_df[results_df["threshold"] == thresh]
    for tau in sorted(thresh_data["tau"].unique()):
        subset = thresh_data[thresh_data["tau"] == tau]
        sig_subset = subset[subset["rmst_p_mgmt_age"] < 0.05]
        
        n_sig = len(sig_subset)
        n_clusters_sig = sig_subset["cluster"].nunique() if n_sig > 0 else 0
        best_p = sig_subset["rmst_p_mgmt_age"].min() if n_sig > 0 else 1.0
        n_protective = len(sig_subset[sig_subset["rmst_diff"] > 0]) if n_sig > 0 else 0
        n_harmful = len(sig_subset[sig_subset["rmst_diff"] < 0]) if n_sig > 0 else 0
        mean_effect = sig_subset["rmst_diff"].abs().mean() * 12 if n_sig > 0 else 0
        
        best_configs.append({
            "threshold": thresh,
            "tau": tau,
            "n_significant": n_sig,
            "n_clusters_sig": n_clusters_sig,
            "best_p": best_p,
            "n_protective": n_protective,
            "n_harmful": n_harmful,
            "mean_effect_months": mean_effect
        })

config_df = pd.DataFrame(best_configs)
config_df = config_df.sort_values(["n_clusters_sig", "n_significant"], ascending=False)

print("\n--- Top Configurations ---")
print(config_df.head(15).to_string(index=False))

best_config = config_df.iloc[0]
print(f"\n*** RECOMMENDED CONFIGURATION ***")
print(f"  Threshold: {best_config['threshold']}")
print(f"  Tau: {best_config['tau']} years")
print(f"  Significant results: {best_config['n_significant']}")
print(f"  Unique significant clusters: {best_config['n_clusters_sig']}")
print(f"  Best p-value: {best_config['best_p']:.4f}")
print(f"  Protective: {best_config['n_protective']}, Harmful: {best_config['n_harmful']}")
print(f"  Mean effect: {best_config['mean_effect_months']:.1f} months")

# =============================================================================
# ALL SIGNIFICANT CLUSTERS AT OPTIMAL CONFIG
# =============================================================================

optimal_thresh = best_config['threshold']
optimal_tau = best_config['tau']

print(f"\n" + "="*80)
print(f"SIGNIFICANT CLUSTERS AT OPTIMAL CONFIG (threshold={optimal_thresh}, tau={optimal_tau})")
print(f"Adjusted for MGMT methylation + Age at diagnosis")
print("="*80)

optimal_results = results_df[
    (results_df["threshold"] == optimal_thresh) &
    (results_df["tau"] == optimal_tau) &
    (results_df["rmst_p_mgmt_age"] < 0.05)
].copy()

optimal_by_cluster = optimal_results.sort_values("rmst_p_mgmt_age").drop_duplicates(subset=["cluster"])
optimal_by_cluster = optimal_by_cluster.sort_values("rmst_p_mgmt_age")

print(f"\n{len(optimal_by_cluster)} unique significant clusters:\n")

for _, row in optimal_by_cluster.iterrows():
    diff_mo = row['rmst_diff'] * 12
    direction = "PROTECTIVE" if row['rmst_diff'] > 0 else "HARMFUL"
    ci_lo = row['rmst_diff_lower'] * 12
    ci_hi = row['rmst_diff_upper'] * 12
    
    print(f"Cluster {row['cluster']} ({row['n_celltypes']} cell types) - {direction}")
    print(f"  p(MGMT+Age) = {row['rmst_p_mgmt_age']:.4f}")
    print(f"  p(MGMT only) = {row['rmst_p_mgmt_only']:.4f}")
    print(f"  p(unadjusted) = {row['rmst_p_unadjusted']:.4f}")
    print(f"  RMST diff: {diff_mo:+.1f} months (95% CI: {ci_lo:+.1f} to {ci_hi:+.1f})")
    print(f"  High: {row['rmst_high']:.2f} yrs, Low: {row['rmst_low']:.2f} yrs")
    print(f"  Best params: {row['scoring']}, {row['stratification']}")
    print(f"  Cell types: {row['celltypes']}")
    print()

# =============================================================================
# PUBLICATION TABLE
# =============================================================================

print("\n" + "="*80)
print("SUMMARY TABLE FOR PUBLICATION")
print("="*80)

pub_table = optimal_by_cluster[[
    "cluster", "n_celltypes", "rmst_diff", "rmst_diff_lower", "rmst_diff_upper",
    "rmst_p_mgmt_age", "rmst_p_mgmt_only", "rmst_p_unadjusted", "celltypes"
]].copy()

pub_table["direction"] = pub_table["rmst_diff"].apply(lambda x: "Protective" if x > 0 else "Harmful")
pub_table["rmst_diff_months"] = pub_table["rmst_diff"] * 12
pub_table["ci_lower_months"] = pub_table["rmst_diff_lower"] * 12
pub_table["ci_upper_months"] = pub_table["rmst_diff_upper"] * 12

pub_table_final = pub_table[[
    "cluster", "n_celltypes", "direction", "rmst_diff_months", 
    "ci_lower_months", "ci_upper_months", 
    "rmst_p_mgmt_age", "rmst_p_mgmt_only", "rmst_p_unadjusted", "celltypes"
]].copy()

pub_table_final.columns = [
    "Cluster", "N_celltypes", "Direction", "RMST_diff_months",
    "CI_lower", "CI_upper", 
    "P_MGMT_Age", "P_MGMT_only", "P_unadjusted", "Cell_types"
]

print(f"\nOptimal configuration: Threshold={optimal_thresh}, Tau={optimal_tau} years")
print(f"Covariates: MGMT methylation + Age at diagnosis\n")
print(pub_table_final.to_string(index=False))

# Save
pub_table_final.to_csv(f"{OUT_DIR}/rmst_significant_clusters_mgmt_age_adjusted.csv", index=False)
print(f"\nSaved to {OUT_DIR}/rmst_significant_clusters_mgmt_age_adjusted.csv")

# =============================================================================
# COMPARISON: How many clusters remain significant with full adjustment?
# =============================================================================

print("\n" + "="*80)
print("ROBUSTNESS CHECK: Significance across adjustment levels")
print("="*80)

# For optimal config, compare significance at each level
comparison_data = optimal_results.sort_values("rmst_p_mgmt_age").drop_duplicates(subset=["cluster"])

print(f"\nAt threshold={optimal_thresh}, tau={optimal_tau}:")
print(f"  Significant (p<0.05) unadjusted: {(comparison_data['rmst_p_unadjusted'] < 0.05).sum()}")
print(f"  Significant (p<0.05) MGMT-adjusted: {(comparison_data['rmst_p_mgmt_only'] < 0.05).sum()}")
print(f"  Significant (p<0.05) MGMT+Age-adjusted: {(comparison_data['rmst_p_mgmt_age'] < 0.05).sum()}")

# Store results
rmst_mgmt_age_results = results_df
rmst_mgmt_age_optimal = optimal_by_cluster
optimal_threshold_mgmt_age = optimal_thresh
optimal_tau_mgmt_age = optimal_tau
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import rpy2.robjects as ro
from rpy2.robjects import pandas2ri
from rpy2.robjects.conversion import localconverter
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# =============================================================================
# CONFIG
# =============================================================================

ADJUST_MODE   = "all"   # "mgmt", "mgmt_age", "all", or "none"
SEL_TAU       = 1.5
SEL_THRESHOLD = optimal_threshold_mgmt_age

SAMPLE_ID_COL  = "Subject Id"
CHRONOLOGY_COL = "Sample collection chronology"
TIME_COL       = "os_years"
EVENT_COL      = "os_event"
MGMT_COL       = "MGMT promoter methylation"
AGE_COL        = "age_at_diagnosis"
SEX_COL        = "Administrative gender"
P53_COL        = "p53"

SAVE_DIR     = "/home/sagemaker-user/gbm-collaboration_2/analysis_sc_merged"
FP_SAVE_STEM = f"{SAVE_DIR}/forest_rmst_modules"

# =============================================================================
# AESTHETICS
# =============================================================================

FP_FIG_W     = 12
FP_FIG_H     = 4
FP_DPI       = 300
FP_FS_TITLE  = 9
FP_FS_AXIS   = 8
FP_FS_HDR    = 8
FP_FS_ROW    = 8
FP_MARKER    = "D"
FP_MARKER_SZ = 5
FP_CAP_SZ    = 3
FP_LW        = 1.3
FP_ELW       = 1.1
FP_COL_POS   = "#2166AC"
FP_COL_NEG   = "#D6604D"
FP_COL_NS    = "#888888"
FP_SIG_ALPHA = 0.05
X_LABEL      = -0.02
X_EST        =  1.02
X_P          =  1.28

# =============================================================================
# PREPARE CLINICAL DATA
# =============================================================================

clin_fp = clin_out.copy()
clin_fp["sample_id"] = clin_fp[SAMPLE_ID_COL].astype(str)
clin_fp["_priority"] = clin_fp[CHRONOLOGY_COL].map(
    {"Baseline": 0, "At recurrence": 1}).fillna(2)
clin_fp = clin_fp.sort_values(["sample_id", "_priority"])
clin_fp = clin_fp.drop_duplicates(subset=["sample_id"], keep="first")
clin_fp = clin_fp.drop(columns=["_priority"])

clin_fp["mgmt_num"] = (clin_fp[MGMT_COL].astype(str) == "Methylated").astype(float)
clin_fp["sex_num"]  = (clin_fp[SEX_COL].astype(str) == "Female").astype(float)
clin_fp["p53_num"]  = (clin_fp[P53_COL].astype(str) == "Positive").astype(float)
age_vals = pd.to_numeric(clin_fp[AGE_COL], errors="coerce")
clin_fp["age_z"] = (age_vals - age_vals.mean()) / age_vals.std()

print(f"Clinical data: {len(clin_fp)} patients")

# =============================================================================
# BUILD PROPORTION MATRIX AND CLUSTERS
# =============================================================================

print("Building proportion matrix...")
props_fp = props_l4.copy().reset_index()
props_fp.columns = ["sample_id_raw"] + list(props_fp.columns[1:])
props_fp = props_fp[props_fp["sample_id_raw"].astype(str).str.endswith("a")].copy()
props_fp["sample_id"] = props_fp["sample_id_raw"].astype(str).str[:-1]
props_fp = props_fp.drop(columns=["sample_id_raw"]).set_index("sample_id")
print(f"Proportion matrix: {props_fp.shape[0]} samples x {props_fp.shape[1]} cell types")

print("Rebuilding clusters at selected threshold...")
cluster_dict_fp, _ = get_clusters_at_threshold(corr_l4, SEL_THRESHOLD, CLUSTER_METHOD)
sig_cluster_ids = rmst_mgmt_age_optimal["cluster"].tolist()
print(f"Significant modules: {sig_cluster_ids}")

# =============================================================================
# R SETUP
# =============================================================================

ro.r("suppressPackageStartupMessages({library(survival); library(survRM2)})")

# =============================================================================
# RMST FUNCTION
# =============================================================================

def run_rmst_module(score_series, clin_df, tau):
    score_df = score_series.reset_index()
    score_df.columns = ["sample_id", "score"]

    merged = score_df.merge(clin_df, on="sample_id", how="inner")
    merged["time"]  = pd.to_numeric(merged[TIME_COL],  errors="coerce")
    merged["event"] = pd.to_numeric(merged[EVENT_COL], errors="coerce")
    merged = merged.dropna(subset=[
        "time", "event", "score",
        "mgmt_num", "age_z", "sex_num", "p53_num"
    ])
    merged = merged[merged["time"] > 0]

    if len(merged) < 20 or tau > merged["time"].max():
        return None

    q75 = merged["score"].quantile(0.75)
    merged["arm"] = (merged["score"] >= q75).astype(int)
    n_high = int(merged["arm"].sum())
    n_low  = int((merged["arm"] == 0).sum())
    if n_high < 5 or n_low < 5:
        return None

    with localconverter(ro.default_converter + pandas2ri.converter):
        r_df = ro.conversion.py2rpy(merged)
    ro.globalenv["surv_data"] = r_df
    ro.globalenv["tau_val"]   = float(tau)

    try:
        ro.r("""
        # Unadjusted
        rmst_u <- rmst2(surv_data$time, surv_data$event, surv_data$arm, tau=tau_val)
        u_diff <- rmst_u$unadjusted.result[1, "Est."]
        u_lo   <- rmst_u$unadjusted.result[1, "lower .95"]
        u_hi   <- rmst_u$unadjusted.result[1, "upper .95"]
        u_p    <- rmst_u$unadjusted.result[1, "p"]

        # MGMT only
        cov_mgmt <- as.matrix(surv_data$mgmt_num)
        rmst_m <- rmst2(surv_data$time, surv_data$event, surv_data$arm,
                        tau=tau_val, covariates=cov_mgmt)
        m_diff <- rmst_m$adjusted.result[1, "Est."]
        m_lo   <- rmst_m$adjusted.result[1, "lower .95"]
        m_hi   <- rmst_m$adjusted.result[1, "upper .95"]
        m_p    <- rmst_m$adjusted.result[1, "p"]

        # MGMT + Age
        surv_data$age_z_col <- scale(surv_data$age_z)[,1]
        cov_mgmt_age <- cbind(surv_data$mgmt_num, surv_data$age_z_col)
        rmst_ma <- rmst2(surv_data$time, surv_data$event, surv_data$arm,
                         tau=tau_val, covariates=cov_mgmt_age)
        ma_diff <- rmst_ma$adjusted.result[1, "Est."]
        ma_lo   <- rmst_ma$adjusted.result[1, "lower .95"]
        ma_hi   <- rmst_ma$adjusted.result[1, "upper .95"]
        ma_p    <- rmst_ma$adjusted.result[1, "p"]

        # All: MGMT + Age + Sex + p53
        cov_all <- cbind(surv_data$mgmt_num, surv_data$age_z_col,
                         surv_data$sex_num, surv_data$p53_num)
        rmst_a <- rmst2(surv_data$time, surv_data$event, surv_data$arm,
                        tau=tau_val, covariates=cov_all)
        a_diff <- rmst_a$adjusted.result[1, "Est."]
        a_lo   <- rmst_a$adjusted.result[1, "lower .95"]
        a_hi   <- rmst_a$adjusted.result[1, "upper .95"]
        a_p    <- rmst_a$adjusted.result[1, "p"]
        """)

        return {
            "n_high":     n_high,
            "n_low":      n_low,
            "unadj_diff": float(ro.r("u_diff")[0]),
            "unadj_lo":   float(ro.r("u_lo")[0]),
            "unadj_hi":   float(ro.r("u_hi")[0]),
            "unadj_p":    float(ro.r("u_p")[0]),
            "mgmt_diff":  float(ro.r("m_diff")[0]),
            "mgmt_lo":    float(ro.r("m_lo")[0]),
            "mgmt_hi":    float(ro.r("m_hi")[0]),
            "mgmt_p":     float(ro.r("m_p")[0]),
            "mgmt_age_diff": float(ro.r("ma_diff")[0]),
            "mgmt_age_lo":   float(ro.r("ma_lo")[0]),
            "mgmt_age_hi":   float(ro.r("ma_hi")[0]),
            "mgmt_age_p":    float(ro.r("ma_p")[0]),
            "all_diff":   float(ro.r("a_diff")[0]),
            "all_lo":     float(ro.r("a_lo")[0]),
            "all_hi":     float(ro.r("a_hi")[0]),
            "all_p":      float(ro.r("a_p")[0]),
        }

    except Exception as e:
        print(f"  survRM2 error: {e}")
        return None

# =============================================================================
# RUN RMST PER MODULE
# =============================================================================

print(f"\nRunning RMST (tau={SEL_TAU}y, adjustment={ADJUST_MODE})...")

cluster_results = {}

for cid in sig_cluster_ids:
    members = cluster_dict_fp.get(cid, [])
    cols    = [c for c in members if c in props_fp.columns]
    if not cols:
        print(f"  Module {cid}: no matching columns — skipping")
        continue

    row_opt = rmst_mgmt_age_optimal[rmst_mgmt_age_optimal["cluster"] == cid]
    scoring = row_opt["scoring"].iloc[0] if len(row_opt) else "mean"

    if scoring == "mean":
        score = props_fp[cols].mean(axis=1)
    elif scoring == "sum":
        score = props_fp[cols].sum(axis=1)
    elif scoring == "max":
        score = props_fp[cols].max(axis=1)
    elif scoring == "pca" and len(cols) >= 2:
        X_s   = StandardScaler().fit_transform(props_fp[cols].values)
        score = pd.Series(
            PCA(n_components=1).fit_transform(X_s).flatten(),
            index=props_fp.index)
    else:
        score = props_fp[cols].mean(axis=1)

    res = run_rmst_module(score, clin_fp, SEL_TAU)

    if res:
        cluster_results[cid] = res
        d, p = res[f"{ADJUST_MODE}_diff"]*12, res[f"{ADJUST_MODE}_p"]
        print(f"  Module {cid}: diff={d:+.2f}mo  p={p:.4f}  "
              f"n_high={res['n_high']}  n_low={res['n_low']}")
    else:
        print(f"  Module {cid}: RMST failed")

print(f"\nResults: {len(cluster_results)}/{len(sig_cluster_ids)} modules")

# =============================================================================
# BUILD ROWS
# =============================================================================

def fmt_p(p):
    if np.isnan(p):
        return "—"
    if p < 0.0001:
        return f"{p:.2e}"
    return f"{p:.4f}"

def get_vals(res, mode):
    return (
        res[f"{mode}_diff"]*12,
        res[f"{mode}_lo"]*12,
        res[f"{mode}_hi"]*12,
        res[f"{mode}_p"],
    )

rows = []
for cid, res in cluster_results.items():
    d, lo, hi, p = get_vals(res, ADJUST_MODE)
    sig = p < FP_SIG_ALPHA
    rows.append({
        "label": f"Module {cid}",
        "diff":  d,
        "lo":    lo,
        "hi":    hi,
        "p":     p,
        "sig":   sig,
        "dir":   "pos" if d >= 0 else "neg",
    })

rows_pos = sorted([r for r in rows if r["dir"] == "pos"],
                   key=lambda x: x["diff"], reverse=True)
rows_neg = sorted([r for r in rows if r["dir"] == "neg"],
                   key=lambda x: x["diff"], reverse=True)
fp_rows  = rows_pos + rows_neg
n_rows   = len(fp_rows)
y_pos    = np.arange(n_rows - 1, -1, -1, dtype=float)

# =============================================================================
# PLOT
# =============================================================================

fig, ax = plt.subplots(figsize=(FP_FIG_W, FP_FIG_H))

for i, row in enumerate(fp_rows):
    y   = y_pos[i]
    col = (FP_COL_POS if row["dir"] == "pos" else FP_COL_NEG) if row["sig"] else FP_COL_NS
    ax.errorbar(
        x=row["diff"], y=y,
        xerr=[[row["diff"] - row["lo"]], [row["hi"] - row["diff"]]],
        fmt=FP_MARKER, color=col,
        markersize=FP_MARKER_SZ, capsize=FP_CAP_SZ,
        linewidth=FP_LW, elinewidth=FP_ELW, zorder=3,
    )

ax.axvline(x=0, color="#7F8C8D", linewidth=0.9, linestyle="--", zorder=2)

all_lo = [r["lo"] for r in fp_rows]
all_hi = [r["hi"] for r in fp_rows]
x_pad  = max((max(all_hi) - min(all_lo)) * 0.08, 0.5)
ax.set_xlim(min(all_lo) - x_pad, max(all_hi) + x_pad)
ax.set_ylim(-0.7, n_rows - 0.3)
ax.set_yticks([])
ax.set_xlabel(
    f"RMST difference (months, τ={SEL_TAU}y)  |  high vs low module abundance",
    fontsize=FP_FS_AXIS)
adj_labels = {
    "mgmt_only": "MGMT",
    "mgmt_age":  "MGMT + Age",
    "all":       "MGMT + Age + Sex + p53",
    "none":      "unadjusted",
}
ax.set_title(
    f"Co-occurrence module survival associations (τ={SEL_TAU} years)\n"
    f"Adjusted for {adj_labels.get(ADJUST_MODE, ADJUST_MODE)}  |  "
    f"blue = protective, red = harmful, gray = n.s.",
    fontsize=FP_FS_TITLE, pad=8)
for sp in ["top", "right", "left"]:
    ax.spines[sp].set_visible(False)
ax.spines["bottom"].set_linewidth(0.6)
ax.tick_params(axis="x", labelsize=FP_FS_AXIS, width=0.6, length=3)
ax.grid(axis="x", linewidth=0.3, alpha=0.4, zorder=0)

trans      = ax.transAxes
y_lo, y_hi = ax.get_ylim()

def y_frac(y_data):
    return (y_data - y_lo) / (y_hi - y_lo)

header_frac = y_frac(n_rows - 0.1)
ax.text(X_LABEL, header_frac, "Module",
        transform=trans, fontsize=FP_FS_HDR, fontweight="bold",
        ha="right", va="bottom")
ax.text(X_EST, header_frac, "RMST diff (months)",
        transform=trans, fontsize=FP_FS_HDR, fontweight="bold",
        ha="left", va="bottom")
ax.text(X_P, header_frac, "adj p",
        transform=trans, fontsize=FP_FS_HDR, fontweight="bold",
        ha="left", va="bottom")

for i, row in enumerate(fp_rows):
    yf  = y_frac(y_pos[i])
    col = (FP_COL_POS if row["dir"] == "pos" else FP_COL_NEG) if row["sig"] else FP_COL_NS
    ax.text(X_LABEL, yf, row["label"],
            transform=trans, fontsize=FP_FS_ROW,
            ha="right", va="center", color="#2C3E50")
    ax.text(X_EST, yf, f"{row['diff']:+.1f}",
            transform=trans, fontsize=FP_FS_ROW,
            ha="left", va="center", color=col)
    ax.text(X_P, yf, fmt_p(row["p"]) + (" *" if row["sig"] else ""),
            transform=trans, fontsize=FP_FS_ROW,
            ha="left", va="center", color=col)

fig.subplots_adjust(left=0.20, right=0.80)
fig.savefig(f"{FP_SAVE_STEM}.png", dpi=FP_DPI, bbox_inches="tight")
fig.savefig(f"{FP_SAVE_STEM}.pdf", bbox_inches="tight")
plt.show()
plt.close(fig)
print(f"Saved {FP_SAVE_STEM}.png/.pdf")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import rpy2.robjects as ro
from rpy2.robjects import pandas2ri
from rpy2.robjects.conversion import localconverter
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# =============================================================================
# CONFIG
# =============================================================================

ADJUST_MODE   = "all"   # "mgmt", "mgmt_age", "all", or "none"
SEL_TAU       = 1.5
SEL_THRESHOLD = optimal_threshold_mgmt_age

SAMPLE_ID_COL  = "Subject Id"
CHRONOLOGY_COL = "Sample collection chronology"
TIME_COL       = "os_years"
EVENT_COL      = "os_event"
MGMT_COL       = "MGMT promoter methylation"
AGE_COL        = "age_at_diagnosis"
SEX_COL        = "Administrative gender"
P53_COL        = "p53"

SAVE_DIR     = "/home/sagemaker-user/gbm-collaboration_2/analysis_sc_merged"
FP_SAVE_STEM = f"{SAVE_DIR}/forest_rmst_modules"

# =============================================================================
# AESTHETICS
# =============================================================================

FP_FIG_W     = 12
FP_FIG_H     = 4
FP_DPI       = 300
FP_FS_TITLE  = 9
FP_FS_AXIS   = 8
FP_FS_HDR    = 8
FP_FS_ROW    = 8
FP_MARKER    = "D"
FP_MARKER_SZ = 5
FP_CAP_SZ    = 3
FP_LW        = 1.3
FP_ELW       = 1.1
FP_COL_POS   = "#2166AC"
FP_COL_NEG   = "#D6604D"
FP_COL_NS    = "#888888"
FP_SIG_ALPHA = 0.05
X_LABEL      = -0.02
X_EST        =  1.02
X_P          =  1.28

# =============================================================================
# PREPARE CLINICAL DATA
# =============================================================================

clin_fp = clin_out.copy()
clin_fp["sample_id"] = clin_fp[SAMPLE_ID_COL].astype(str)
clin_fp["_priority"] = clin_fp[CHRONOLOGY_COL].map(
    {"Baseline": 0, "At recurrence": 1}).fillna(2)
clin_fp = clin_fp.sort_values(["sample_id", "_priority"])
clin_fp = clin_fp.drop_duplicates(subset=["sample_id"], keep="first")
clin_fp = clin_fp.drop(columns=["_priority"])
clin_fp["mgmt_num"] = (clin_fp[MGMT_COL].astype(str) == "Methylated").astype(float)
clin_fp["sex_num"]  = (clin_fp[SEX_COL].astype(str) == "Female").astype(float)
clin_fp["p53_num"]  = (clin_fp[P53_COL].astype(str) == "Positive").astype(float)
age_vals = pd.to_numeric(clin_fp[AGE_COL], errors="coerce")
clin_fp["age_z"] = (age_vals - age_vals.mean()) / age_vals.std()

print(f"Clinical data: {len(clin_fp)} patients")

# =============================================================================
# BUILD PROPORTION MATRIX AND CLUSTERS
# =============================================================================

print("Building proportion matrix...")
props_fp = props_l4.copy().reset_index()
props_fp.columns = ["sample_id_raw"] + list(props_fp.columns[1:])
props_fp = props_fp[props_fp["sample_id_raw"].astype(str).str.endswith("a")].copy()
props_fp["sample_id"] = props_fp["sample_id_raw"].astype(str).str[:-1]
props_fp = props_fp.drop(columns=["sample_id_raw"]).set_index("sample_id")
print(f"Proportion matrix: {props_fp.shape[0]} samples x {props_fp.shape[1]} cell types")

print("Rebuilding clusters at selected threshold...")
cluster_dict_fp, _ = get_clusters_at_threshold(corr_l4, SEL_THRESHOLD, CLUSTER_METHOD)
sig_cluster_ids = rmst_mgmt_age_optimal["cluster"].tolist()
print(f"Significant modules: {sig_cluster_ids}")

# =============================================================================
# R SETUP
# =============================================================================

ro.r("suppressPackageStartupMessages({library(survival); library(survRM2)})")

# =============================================================================
# STRATIFICATION HELPER
# =============================================================================

def apply_stratification(scores, method):
    """
    Returns arm assignment (1=high, 0=low) and mask for rows to keep.
    Middle group is excluded for tertile and quartile_vs_quartile.
    """
    if method == "median":
        threshold = scores.median()
        arm  = (scores >= threshold).astype(int)
        keep = pd.Series(True, index=scores.index)
    elif method == "quartile_vs_rest":
        q75  = scores.quantile(0.75)
        arm  = (scores >= q75).astype(int)
        keep = pd.Series(True, index=scores.index)
    elif method == "tertile":
        t33  = scores.quantile(0.33)
        t67  = scores.quantile(0.67)
        arm  = (scores >= t67).astype(int)
        keep = (scores >= t67) | (scores <= t33)
    elif method == "quartile_vs_quartile":
        q25  = scores.quantile(0.25)
        q75  = scores.quantile(0.75)
        arm  = (scores >= q75).astype(int)
        keep = (scores >= q75) | (scores <= q25)
    else:
        threshold = scores.median()
        arm  = (scores >= threshold).astype(int)
        keep = pd.Series(True, index=scores.index)
    return arm, keep

# =============================================================================
# RMST FUNCTION
# =============================================================================

def run_rmst_module(score_series, clin_df, tau, strat_method):
    score_df = score_series.reset_index()
    score_df.columns = ["sample_id", "score"]

    merged = score_df.merge(clin_df, on="sample_id", how="inner")
    merged["time"]  = pd.to_numeric(merged[TIME_COL],  errors="coerce")
    merged["event"] = pd.to_numeric(merged[EVENT_COL], errors="coerce")
    merged = merged.dropna(subset=[
        "time", "event", "score",
        "mgmt_num", "age_z", "sex_num", "p53_num"
    ])
    merged = merged[merged["time"] > 0].copy()

    if len(merged) < 20 or tau > merged["time"].max():
        return None

    arm, keep = apply_stratification(merged["score"], strat_method)
    merged["arm"] = arm
    merged = merged[keep].copy()

    n_high = int(merged["arm"].sum())
    n_low  = int((merged["arm"] == 0).sum())
    if n_high < 5 or n_low < 5:
        return None

    with localconverter(ro.default_converter + pandas2ri.converter):
        r_df = ro.conversion.py2rpy(merged)
    ro.globalenv["surv_data"] = r_df
    ro.globalenv["tau_val"]   = float(tau)

    try:
        ro.r("""
        # Unadjusted
        rmst_u <- rmst2(surv_data$time, surv_data$event, surv_data$arm, tau=tau_val)
        u_diff <- rmst_u$unadjusted.result[1, "Est."]
        u_lo   <- rmst_u$unadjusted.result[1, "lower .95"]
        u_hi   <- rmst_u$unadjusted.result[1, "upper .95"]
        u_p    <- rmst_u$unadjusted.result[1, "p"]

        # MGMT only
        cov_mgmt <- as.matrix(surv_data$mgmt_num)
        rmst_m <- rmst2(surv_data$time, surv_data$event, surv_data$arm,
                        tau=tau_val, covariates=cov_mgmt)
        m_diff <- rmst_m$adjusted.result[1, "Est."]
        m_lo   <- rmst_m$adjusted.result[1, "lower .95"]
        m_hi   <- rmst_m$adjusted.result[1, "upper .95"]
        m_p    <- rmst_m$adjusted.result[1, "p"]

        # MGMT + Age
        surv_data$age_z_col <- scale(surv_data$age_z)[,1]
        cov_mgmt_age <- cbind(surv_data$mgmt_num, surv_data$age_z_col)
        rmst_ma <- rmst2(surv_data$time, surv_data$event, surv_data$arm,
                         tau=tau_val, covariates=cov_mgmt_age)
        ma_diff <- rmst_ma$adjusted.result[1, "Est."]
        ma_lo   <- rmst_ma$adjusted.result[1, "lower .95"]
        ma_hi   <- rmst_ma$adjusted.result[1, "upper .95"]
        ma_p    <- rmst_ma$adjusted.result[1, "p"]

        # All: MGMT + Age + Sex + p53
        cov_all <- cbind(surv_data$mgmt_num, surv_data$age_z_col,
                         surv_data$sex_num, surv_data$p53_num)
        rmst_a <- rmst2(surv_data$time, surv_data$event, surv_data$arm,
                        tau=tau_val, covariates=cov_all)
        a_diff <- rmst_a$adjusted.result[1, "Est."]
        a_lo   <- rmst_a$adjusted.result[1, "lower .95"]
        a_hi   <- rmst_a$adjusted.result[1, "upper .95"]
        a_p    <- rmst_a$adjusted.result[1, "p"]
        """)

        return {
            "n_high":        n_high,
            "n_low":         n_low,
            "strat":         strat_method,
            "none_diff":     float(ro.r("u_diff")[0]),
            "none_lo":       float(ro.r("u_lo")[0]),
            "none_hi":       float(ro.r("u_hi")[0]),
            "none_p":        float(ro.r("u_p")[0]),
            "mgmt_diff":     float(ro.r("m_diff")[0]),
            "mgmt_lo":       float(ro.r("m_lo")[0]),
            "mgmt_hi":       float(ro.r("m_hi")[0]),
            "mgmt_p":        float(ro.r("m_p")[0]),
            "mgmt_age_diff": float(ro.r("ma_diff")[0]),
            "mgmt_age_lo":   float(ro.r("ma_lo")[0]),
            "mgmt_age_hi":   float(ro.r("ma_hi")[0]),
            "mgmt_age_p":    float(ro.r("ma_p")[0]),
            "all_diff":      float(ro.r("a_diff")[0]),
            "all_lo":        float(ro.r("a_lo")[0]),
            "all_hi":        float(ro.r("a_hi")[0]),
            "all_p":         float(ro.r("a_p")[0]),
        }

    except Exception as e:
        print(f"  survRM2 error: {e}")
        return None

# =============================================================================
# RUN RMST PER MODULE — using per-cluster scoring AND stratification
# =============================================================================

print(f"\nRunning RMST (tau={SEL_TAU}y, adjustment={ADJUST_MODE})...")

cluster_results = {}

for cid in sig_cluster_ids:
    members = cluster_dict_fp.get(cid, [])
    cols    = [c for c in members if c in props_fp.columns]
    if not cols:
        print(f"  Module {cid}: no matching columns — skipping")
        continue

    row_opt = rmst_mgmt_age_optimal[rmst_mgmt_age_optimal["cluster"] == cid]
    scoring = row_opt["scoring"].iloc[0]       if len(row_opt) else "mean"
    strat   = row_opt["stratification"].iloc[0] if len(row_opt) else "quartile_vs_rest"

    if scoring == "mean":
        score = props_fp[cols].mean(axis=1)
    elif scoring == "sum":
        score = props_fp[cols].sum(axis=1)
    elif scoring == "max":
        score = props_fp[cols].max(axis=1)
    elif scoring == "pca" and len(cols) >= 2:
        X_s   = StandardScaler().fit_transform(props_fp[cols].values)
        score = pd.Series(
            PCA(n_components=1).fit_transform(X_s).flatten(),
            index=props_fp.index)
    else:
        score = props_fp[cols].mean(axis=1)

    res = run_rmst_module(score, clin_fp, SEL_TAU, strat)

    if res:
        cluster_results[cid] = res
        d = res[f"{ADJUST_MODE}_diff"] * 12
        p = res[f"{ADJUST_MODE}_p"]
        print(f"  Module {cid} ({scoring}, {strat}): "
              f"diff={d:+.2f}mo  p={p:.4f}  "
              f"n_high={res['n_high']}  n_low={res['n_low']}")
    else:
        print(f"  Module {cid}: RMST failed")

print(f"\nResults: {len(cluster_results)}/{len(sig_cluster_ids)} modules")

# =============================================================================
# BUILD ROWS
# =============================================================================

def fmt_p(p):
    if np.isnan(p):
        return "—"
    if p < 0.0001:
        return f"{p:.2e}"
    return f"{p:.4f}"

def get_vals(res, mode):
    return (
        res[f"{mode}_diff"] * 12,
        res[f"{mode}_lo"]   * 12,
        res[f"{mode}_hi"]   * 12,
        res[f"{mode}_p"],
    )

rows = []
for cid, res in cluster_results.items():
    d, lo, hi, p = get_vals(res, ADJUST_MODE)
    sig = p < FP_SIG_ALPHA
    rows.append({
        "label": f"Module {cid}",
        "diff":  d,
        "lo":    lo,
        "hi":    hi,
        "p":     p,
        "sig":   sig,
        "dir":   "pos" if d >= 0 else "neg",
    })

rows_pos = sorted([r for r in rows if r["dir"] == "pos"],
                   key=lambda x: x["diff"], reverse=True)
rows_neg = sorted([r for r in rows if r["dir"] == "neg"],
                   key=lambda x: x["diff"], reverse=True)
fp_rows  = rows_pos + rows_neg
n_rows   = len(fp_rows)
y_pos    = np.arange(n_rows - 1, -1, -1, dtype=float)

# =============================================================================
# PLOT
# =============================================================================

fig, ax = plt.subplots(figsize=(FP_FIG_W, FP_FIG_H))

for i, row in enumerate(fp_rows):
    y   = y_pos[i]
    col = (FP_COL_POS if row["dir"] == "pos" else FP_COL_NEG) if row["sig"] else FP_COL_NS
    ax.errorbar(
        x=row["diff"], y=y,
        xerr=[[row["diff"] - row["lo"]], [row["hi"] - row["diff"]]],
        fmt=FP_MARKER, color=col,
        markersize=FP_MARKER_SZ, capsize=FP_CAP_SZ,
        linewidth=FP_LW, elinewidth=FP_ELW, zorder=3,
    )

ax.axvline(x=0, color="#7F8C8D", linewidth=0.9, linestyle="--", zorder=2)

all_lo = [r["lo"] for r in fp_rows]
all_hi = [r["hi"] for r in fp_rows]
x_pad  = max((max(all_hi) - min(all_lo)) * 0.08, 0.5)
ax.set_xlim(min(all_lo) - x_pad, max(all_hi) + x_pad)
ax.set_ylim(-0.7, n_rows - 0.3)
ax.set_yticks([])
ax.set_xlabel(
    f"RMST difference (months, τ={SEL_TAU}y)  |  high vs low module abundance",
    fontsize=FP_FS_AXIS)
adj_labels = {
    "mgmt":     "MGMT",
    "mgmt_age": "MGMT + Age",
    "all":      "MGMT + Age + Sex + p53",
    "none":     "unadjusted",
}
ax.set_title(
    f"Co-occurrence module survival associations (τ={SEL_TAU} years)\n"
    f"Adjusted for {adj_labels.get(ADJUST_MODE, ADJUST_MODE)}  |  "
    f"blue = protective, red = harmful, gray = n.s.",
    fontsize=FP_FS_TITLE, pad=8)
for sp in ["top", "right", "left"]:
    ax.spines[sp].set_visible(False)
ax.spines["bottom"].set_linewidth(0.6)
ax.tick_params(axis="x", labelsize=FP_FS_AXIS, width=0.6, length=3)
ax.grid(axis="x", linewidth=0.3, alpha=0.4, zorder=0)

trans      = ax.transAxes
y_lo, y_hi = ax.get_ylim()

def y_frac(y_data):
    return (y_data - y_lo) / (y_hi - y_lo)

header_frac = y_frac(n_rows - 0.1)
ax.text(X_LABEL, header_frac, "Module",
        transform=trans, fontsize=FP_FS_HDR, fontweight="bold",
        ha="right", va="bottom")
ax.text(X_EST, header_frac, "RMST diff (months)",
        transform=trans, fontsize=FP_FS_HDR, fontweight="bold",
        ha="left", va="bottom")
ax.text(X_P, header_frac, "adj p",
        transform=trans, fontsize=FP_FS_HDR, fontweight="bold",
        ha="left", va="bottom")

for i, row in enumerate(fp_rows):
    yf  = y_frac(y_pos[i])
    col = (FP_COL_POS if row["dir"] == "pos" else FP_COL_NEG) if row["sig"] else FP_COL_NS
    ax.text(X_LABEL, yf, row["label"],
            transform=trans, fontsize=FP_FS_ROW,
            ha="right", va="center", color="#2C3E50")
    ax.text(X_EST, yf, f"{row['diff']:+.1f}",
            transform=trans, fontsize=FP_FS_ROW,
            ha="left", va="center", color=col)
    ax.text(X_P, yf, fmt_p(row["p"]) + (" *" if row["sig"] else ""),
            transform=trans, fontsize=FP_FS_ROW,
            ha="left", va="center", color=col)

fig.subplots_adjust(left=0.20, right=0.80)
fig.savefig(f"{FP_SAVE_STEM}.png", dpi=FP_DPI, bbox_inches="tight")
fig.savefig(f"{FP_SAVE_STEM}.pdf", bbox_inches="tight")
plt.show()
plt.close(fig)
print(f"Saved {FP_SAVE_STEM}.png/.pdf")
```

```python
# %%
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from matplotlib.collections import PatchCollection
from matplotlib.colors import to_rgb, to_hex
from matplotlib.lines import Line2D
from matplotlib import cm
import s3fs
import zarr
import anndata as ad
from cloudpathlib import S3Path
from gbmcollaboration.definitions import DATA_PATH_CHUV_FILES, DATA_PATH_UKER_FILES, DATA_PATH_CR_FILES

# Curated cell type name -> color (all 101 celltype_level4_curated names)
CELLTYPE_COLORS = {
    # Astrocytes - cyans
    "Astrocyte_IFN_response": "#00CED1",
    "Astrocyte_homeostatic": "#76D7EA",
    "Astrocyte_inflammatory": "#00FFFF",
    "Astrocyte_intermediate_reactive": "#40E0D0",
    "Astrocyte_reactive": "#00E5EE",
    
    # B cells / Plasma - light purples
    "B_cell": "#9370DB",
    "Plasma_cell": "#8A2BE2",
    "Plasmablast": "#9932CC",
    
    # Doublets - grays
    "Doublet_Astro_Oligodendro": "#A9A9A9",
    "Doublet_B_TNK": "#A9A9A9",
    "Doublet_Endothelial_Malignant": "#A9A9A9",
    "Doublet_Endothelial_Myeloid": "#A9A9A9",
    "Doublet_Endothelial_Stromal": "#A9A9A9",
    "Doublet_MalignantGBM_nos": "#A9A9A9",
    "Doublet_Mural_Endo": "#A9A9A9",
    "Doublet_Mural_Immune": "#A9A9A9",
    "Doublet_Mural_Malignant": "#A9A9A9",
    "Doublet_Mural_Myeloid": "#A9A9A9",
    "Doublet_Neuron_Astro": "#A9A9A9",
    "Doublet_Neuron_Malignant": "#A9A9A9",
    "Doublet_Neuron_Microgl": "#A9A9A9",
    "Doublet_Neuron_OligoD": "#A9A9A9",
    "Doublet_Neutro_Endothelial": "#A9A9A9",
    "Doublet_Neutro_Mural": "#A9A9A9",
    "Doublet_Neutro_TNK": "#A9A9A9",
    "Doublet_Neutrophils_Momac": "#A9A9A9",
    "Doublet_Oligo_malignant": "#A9A9A9",
    "Doublet_Oligo_neuron": "#A9A9A9",
    "Doublet_TNK_Malignant": "#A9A9A9",
    "Doublet_TNK_Mural": "#A9A9A9",
    "Doublet_TNK_Myeloid": "#A9A9A9",
    "Doublet_TNK_nos": "#A9A9A9",
    "Doublet_T_Oligodendro": "#A9A9A9",
    "Doublet_momac_Neutro": "#A9A9A9",
    "Doublet_momac_Stroma": "#A9A9A9",
    "Doublet_momac_malignant": "#A9A9A9",
    
    # Vasculature - dark reds to browns
    "Endothelial": "#8B0000",
    "Endothelial_arterial": "#A52A2A",
    "Endothelial_capillary": "#800000",
    "Endothelial_cycling": "#B22222",
    "Endothelial_tip": "#722F37",
    "Endothelial_veinous": "#6B3A3A",
    "Pericyte": "#5C4033",
    "Pericyte_cycling": "#6B4423",
    "Pericyte_fibro-like": "#7B3F00",
    "Perivascular_fibroblast_cycling": "#8B4513",
    "Perivascular_fibroblast_type-I": "#A0522D",
    "Perivascular_fibroblast_type-II": "#8B5A2B",
    "Perivascular_fibroblast_type-III": "#704214",
    "VSMC": "#4A2C2A",
    
    # Myeloid - yellows to oranges
    # Macrophages
    "Macrophage_C1Q_pos": "#E69500",
    "Macrophage_IFN_response": "#FF8C00",
    "Macrophage_LAM": "#FFA500",
    "Macrophage_PVM": "#FFB347",
    "Macrophage_cycling": "#FFAA00",
    # Microglia
    "Microglia_IFN_response": "#FFD700",
    "Microglia_TREM2_pos": "#FFCC00",
    "Microglia_homeostatic": "#F0E68C",
    "Microglia_intermediate_reactive": "#FFE135",
    # DCs
    "cDC_CD207_pos": "#FF7F00",
    "cDC1": "#FF9933",
    "cDC2": "#FFBF00",
    "cDC2_cycling": "#FFD54F",
    "pDC": "#FFC125",
    # Monocytes / Neutrophils / Mast
    "Monocyte_nos": "#CC7000",
    "Neutrophil": "#D2691E",
    "Neutrophil_immature": "#E08000",
    "Mast_cell": "#DAA520",
    
    # Malignant - vibrant greens to blues
    "Malignant_AC-like_IFN_response": "#00BFA5",
    "Malignant_AC1": "#00CC66",
    "Malignant_AC2": "#00E676",
    "Malignant_MES-hypoxia": "#40E0D0",
    "Malignant_MES1": "#00CED1",
    "Malignant_NPC2": "#4169E1",
    "Malignant_OPC-like_invasive": "#1E90FF",
    "Malignant_OPC1": "#00BFFF",
    "Malignant_OPC2": "#6495ED",
    "Malignant_cycling": "#3CB371",
    "Malignant_other_1": "#2E8B57",
    "Malignant_other_2": "#20B2AA",
    "Malignant_other_stressed": "#5F9EA0",
    
    # Neurons - magentas/pinks
    "Neuron_nos": "#FF1493",
    "Neuron_excitatory_CDH24": "#FF69B4",
    "Neuron_excitatory_nos": "#FF85C1",
    "Neuron_inhibitory_LAMP5": "#DB7093",
    "Neuron_inhibitory_NGF": "#FF6EB4",
    "Neuron_inhibitory_PVALB": "#C71585",
    "Neuron_inhibitory_SST": "#E75480",
    "Neuron_inhibitory_VIP": "#DE3163",
    
    # NK cells - purples
    "NK_KLRC1_pos": "#BA55D3",
    "NK_nos": "#DA70D6",
    
    # Oligodendrocytes - teals
    "Oligodendrocyte_nos": "#2F9C95",
    "Oligodendrocyte_reactive": "#008B8B",
    
    # Other
    "Other_LowCounts": "#808080",
    
    # T cells - vibrant reds
    "T_NK_cycling": "#FF6347",
    "T_cell_CD4": "#B22222",
    "T_cell_CD8": "#FF4136",
    "T_cell_IFN_response": "#E31A1C",
    "T_cell_nos": "#DC143C",
    "T_cell_other": "#FF6B6B",
    "T_reg": "#CD5C5C",
}


def plot_visium_celltypes(
    sample_ids,
    celltypes,
    celltype_key=None,
    mode="blend",
    img_key="hires",
    figsize_per_panel=(8, 8),
    spot_alpha=0.85,
    he_alpha=1.0,
    clip_q=(1, 99),
    gamma=1.0,
    show_legend=True,
    legend_fontsize=9,
    title_fontsize=11,
    save_path=None,
):
    """
    Plot cell type abundances on Visium H&E images.
    
    Parameters
    ----------
    sample_ids : str or list of sample IDs (e.g., "CH_G_196a_vis")
    celltypes : str or list of cell type names (celltype_level4_curated format)
    celltype_key : DataFrame mapping scanvi -> curated names (default: from adata.uns)
    mode : 'blend' (weighted color mix) or 'max' (dominant cell type only)
    """
    if isinstance(sample_ids, str):
        sample_ids = [sample_ids]
    if isinstance(celltypes, str):
        celltypes = [celltypes]
    
    # Get celltype mapping if not provided
    if celltype_key is None:
        celltype_key = adata.uns['celltype_annotation_key']
    
    # Build curated -> scanvi mapping
    curated_to_scanvi = dict(zip(celltype_key['celltype_level4_curated'], celltype_key.index))
    
    # Map requested celltypes to scanvi names
    scanvi_map = {}
    for ct in celltypes:
        if ct in curated_to_scanvi:
            scanvi_map[ct] = curated_to_scanvi[ct]
        else:
            print(f"Warning: {ct} not found in celltype mapping, using as-is")
            scanvi_map[ct] = ct
    
    # Build color map
    ct_colors = dict(CELLTYPE_COLORS)
    need_colors = [ct for ct in celltypes if ct not in ct_colors]
    if need_colors:
        cmap = plt.colormaps.get_cmap("tab20")
        for i, ct in enumerate(need_colors):
            ct_colors[ct] = to_hex(cmap(i % 20))
    ct_rgb = {ct: np.array(to_rgb(ct_colors[ct])) for ct in celltypes}
    
    # Center mapping with zarr paths as lists
    center_map = {
        "CH": ("CHUV", DATA_PATH_CHUV_FILES),
        "ER": ("UKER", DATA_PATH_UKER_FILES),
        "CR": ("CR", DATA_PATH_CR_FILES),
    }
    
    # Create figure
    n_samples = len(sample_ids)
    n_cols = min(3, n_samples)
    n_rows = int(np.ceil(n_samples / n_cols))
    fig, axes = plt.subplots(n_rows, n_cols, 
                             figsize=(figsize_per_panel[0]*n_cols, figsize_per_panel[1]*n_rows),
                             squeeze=False)
    
    s3 = s3fs.S3FileSystem(anon=False)
    
    for idx, sample_id in enumerate(sample_ids):
        ax = axes[idx // n_cols, idx % n_cols]
        
        try:
            # Determine center
            prefix = sample_id[:2]
            if prefix not in center_map:
                raise ValueError(f"Unknown center prefix '{prefix}'")
            center_name, data_paths = center_map[prefix]
            
            # Find zarr path from list
            zarr_list = data_paths["Spatial Zarr Files"]
            zarr_path = None
            for p in zarr_list:
                if sample_id in p:
                    zarr_path = p
                    break
            
            if zarr_path is None:
                raise ValueError(f"No zarr found for {sample_id}")
            
            # Load image from Zarr
            store = s3.get_mapper(zarr_path)
            root = zarr.open_group(store=store, mode="r")
            table_keys = list(root["tables"].group_keys())
            table_grp = root["tables"]["table"] if "table" in table_keys else root["tables"][table_keys[0]]
            adata_img = ad.read_zarr(table_grp)
            
            lib = next(iter(adata_img.uns["spatial"].keys()))
            img = adata_img.uns["spatial"][lib]["images"][img_key]
            scalefactors = adata_img.uns["spatial"][lib]["scalefactors"]
            sf_key = "tissue_hires_scalef" if img_key == "hires" else "tissue_lowres_scalef"
            sf = float(scalefactors[sf_key])
            
            # Coordinates - prefer imagecol/imagerow (already in hires space)
            if {"imagecol", "imagerow"}.issubset(adata_img.obs.columns):
                coords_hires = adata_img.obs[["imagecol", "imagerow"]].to_numpy()
                coords_are_hires = True
            elif {"pxl_col_in_fullres", "pxl_row_in_fullres"}.issubset(adata_img.obs.columns):
                coords_hires = adata_img.obs[["pxl_col_in_fullres", "pxl_row_in_fullres"]].to_numpy() * sf
                coords_are_hires = True
            elif "spatial" in adata_img.obsm:
                coords_hires = np.asarray(adata_img.obsm["spatial"]) * sf
                coords_are_hires = True
            else:
                raise ValueError("No coordinates found")
            
            # Image extent in hires pixel space
            h_px, w_px = img.shape[:2]
            img_extent = [0.0, w_px, h_px, 0.0]
            
            # Spot coordinates in hires pixel space
            x_px = coords_hires[:, 0]
            y_px = coords_hires[:, 1]
            
            # Spot radius in hires pixels (55µm diameter, 100µm pitch)
            # Estimate from coordinate spacing
            from sklearn.neighbors import NearestNeighbors
            nn = NearestNeighbors(n_neighbors=2).fit(coords_hires)
            d, _ = nn.kneighbors(coords_hires)
            median_pitch_px = float(np.median(d[:, 1]))
            r_px = median_pitch_px * 0.55 / 2  # 55µm spot diameter / 100µm pitch / 2
            
            pad_frac = 0.05
            span = max(x_px.max() - x_px.min(), y_px.max() - y_px.min())
            pad_px = pad_frac * span
            x_lim = (x_px.min() - pad_px, x_px.max() + pad_px)
            y_lim = (y_px.min() - pad_px, y_px.max() + pad_px)
            
            # Load C2L data
            c2l_base = S3Path(str(data_paths["Cell2location"]))
            c2l_h5ad = c2l_base / sample_id / "cohort_level_4.h5ad"
            h5ad_str = str(c2l_h5ad).replace("s3://", "")
            with s3.open(h5ad_str, "rb") as f:
                adata_c2l = ad.read_h5ad(f)
            
            x_means = adata_c2l.obsm["means_cell_abundance_w_sf"]
            x_means = x_means.toarray() if hasattr(x_means, "toarray") else np.asarray(x_means)
            
            # Factor names
            factor_names = None
            if "mod" in adata_c2l.uns and isinstance(adata_c2l.uns["mod"], dict):
                factor_names = adata_c2l.uns["mod"].get("factor_names")
            if factor_names is None:
                factor_names = adata_c2l.uns.get("factor_names") or adata_c2l.uns.get("cell_type_names")
            if factor_names is None:
                factor_names = [f"celltype_{i}" for i in range(x_means.shape[1])]
            factor_names = [str(x) for x in factor_names]
            
            # Align barcodes
            img_idx = pd.Index(adata_img.obs_names).astype(str)
            c2l_idx = pd.Index(adata_c2l.obs_names).astype(str)
            pos = c2l_idx.get_indexer(img_idx)
            if np.any(pos < 0):
                raise ValueError(f"Barcode mismatch")
            c2l_means = x_means[pos, :]
            c2l_map = {name: c2l_means[:, j] for j, name in enumerate(factor_names)}
            
            n_spots = len(x_px)
            
            # Get abundance for requested cell types (using scanvi names)
            M = np.zeros((n_spots, len(celltypes)))
            for j, ct in enumerate(celltypes):
                scanvi_name = scanvi_map[ct]
                if scanvi_name in c2l_map:
                    M[:, j] = c2l_map[scanvi_name]
                else:
                    print(f"Warning: {scanvi_name} (for {ct}) not in C2L. Available: {factor_names[:10]}")
            
            # Normalize each cell type
            M_norm = np.zeros_like(M)
            for j in range(M.shape[1]):
                v = M[:, j]
                if clip_q:
                    vmin, vmax = np.percentile(v, clip_q[0]), np.percentile(v, clip_q[1])
                else:
                    vmin, vmax = v.min(), v.max()
                if vmax > vmin:
                    M_norm[:, j] = np.clip((v - vmin) / (vmax - vmin), 0, 1)
            
            if gamma != 1.0:
                M_norm = M_norm ** gamma
            
            # Compute colors
            if mode == "max":
                top_idx = np.argmax(M_norm, axis=1)
                top_val = np.max(M_norm, axis=1)
                rgb = np.array([ct_rgb[celltypes[i]] for i in top_idx])
                intensity = top_val
            else:  # blend
                rgb = np.zeros((n_spots, 3))
                total_w = np.zeros(n_spots)
                for j, ct in enumerate(celltypes):
                    w = M_norm[:, j]
                    rgb += w[:, None] * ct_rgb[ct]
                    total_w += w
                total_w = np.maximum(total_w, 1e-10)
                rgb = rgb / total_w[:, None]
                rgb = np.clip(rgb, 0, 1)
                intensity = np.clip(total_w / len(celltypes), 0, 1)
            
            # Opacity based on intensity
            alpha_spots = np.clip(intensity, 0.05, 1.0) * float(spot_alpha)
            rgba = np.column_stack([rgb, alpha_spots])
            
            # Create patches
            patches = [Circle((x_px[i], y_px[i]), r_px) for i in range(n_spots)]
            pc = PatchCollection(patches, linewidth=0.0)
            pc.set_facecolors(rgba)
            
            # Plot
            ax.imshow(img, extent=img_extent, alpha=he_alpha)
            ax.add_collection(pc)
            ax.set_xlim(x_lim)
            ax.set_ylim(y_lim[1], y_lim[0])
            ax.set_aspect("equal")
            ax.axis("off")
            ax.set_title(sample_id, fontsize=title_fontsize)
            
        except Exception as e:
            import traceback
            ax.set_title(f"{sample_id}: ERROR", fontsize=title_fontsize)
            ax.text(0.5, 0.5, str(e)[:80], ha='center', va='center', transform=ax.transAxes, fontsize=8, wrap=True)
            ax.axis("off")
            traceback.print_exc()
    
    # Hide empty panels
    for idx in range(n_samples, n_rows * n_cols):
        axes[idx // n_cols, idx % n_cols].axis("off")
    
    # Legend
    if show_legend:
        handles = [Line2D([0], [0], marker='o', color='w', markerfacecolor=ct_rgb[ct], 
                          markersize=10, label=ct) for ct in celltypes]
        if n_samples == 1:
            axes[0, 0].legend(handles=handles, loc='upper right', fontsize=legend_fontsize, 
                              frameon=True, framealpha=0.9, edgecolor='none')
        else:
            fig.legend(handles=handles, loc='center right', bbox_to_anchor=(1.12, 0.5), 
                       fontsize=legend_fontsize, frameon=False)
    
    plt.tight_layout()
    
    if save_path:
        fig.savefig(save_path, dpi=300, bbox_inches="tight", facecolor="white")
        print(f"Saved: {save_path}")
    
    plt.show()
    return fig, axes
```

### ECM Neighborhood Plot

```python
# %%
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from matplotlib.collections import PatchCollection
from matplotlib.colors import to_rgb, to_hex
from matplotlib.lines import Line2D
import s3fs
import zarr
import anndata as ad
from cloudpathlib import S3Path
from gbmcollaboration.definitions import DATA_PATH_CHUV_FILES, DATA_PATH_UKER_FILES, DATA_PATH_CR_FILES

# Load ECM annotations
ECM_PATH = "/home/sagemaker-user/custom-file-systems/efs/fs-09913c1f7db79b6fd/Analysis_Repo/Notebooks/ecm_neighborhood_annotations.csv"
ECM_DF = pd.read_csv(ECM_PATH)

# ECM neighborhood colors (10 neighborhoods, 0-9) - distinct, vibrant colors
ECM_COLORS = {
    0: "#377EB8",  # blue
    1: "#FF7F00",  # orange
    2: "#4DAF4A",  # green
    3: "#E41A1C",  # red
    4: "#984EA3",  # purple
    5: "#8B4513",  # brown
    6: "#FF1493",  # magenta/pink
    7: "#808080",  # gray
    8: "#9ACD32",  # yellow-green
    9: "#008B8B",  # dark cyan
}


def plot_visium_ecm(
    sample_ids,
    neighborhoods=None,
    img_key="hires",
    figsize_per_panel=(8, 8),
    spot_alpha=0.85,
    he_alpha=1.0,
    show_he_only=False,
    show_legend=True,
    legend_fontsize=9,
    title_fontsize=11,
    save_path=None,
):
    """
    Plot ECM neighborhoods on Visium H&E images.
    
    Parameters
    ----------
    sample_ids : str or list of sample IDs
    neighborhoods : list of int, optional
        Which neighborhoods to show (0-9). If None, shows all.
    show_he_only : bool
        If True, show side-by-side: spots+H&E and H&E only
    """
    if isinstance(sample_ids, str):
        sample_ids = [sample_ids]
    
    if neighborhoods is None:
        neighborhoods = list(range(10))
    
    # Center mapping
    center_map = {
        "CH": ("CHUV", DATA_PATH_CHUV_FILES),
        "ER": ("UKER", DATA_PATH_UKER_FILES),
        "CR": ("CR", DATA_PATH_CR_FILES),
    }
    
    # Figure layout
    n_samples = len(sample_ids)
    if show_he_only:
        n_cols = 2
        n_rows = n_samples
        fig_w = figsize_per_panel[0] * 2
        fig_h = figsize_per_panel[1] * n_samples
    else:
        n_cols = min(3, n_samples)
        n_rows = int(np.ceil(n_samples / n_cols))
        fig_w = figsize_per_panel[0] * n_cols
        fig_h = figsize_per_panel[1] * n_rows
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(fig_w, fig_h), squeeze=False)
    
    s3 = s3fs.S3FileSystem(anon=False)
    
    for idx, sample_id in enumerate(sample_ids):
        if show_he_only:
            ax_spots = axes[idx, 0]
            ax_he = axes[idx, 1]
        else:
            ax_spots = axes[idx // n_cols, idx % n_cols]
            ax_he = None
        
        try:
            # Determine center
            prefix = sample_id[:2]
            if prefix not in center_map:
                raise ValueError(f"Unknown center prefix '{prefix}'")
            center_name, data_paths = center_map[prefix]
            
            # Find zarr path
            zarr_list = data_paths["Spatial Zarr Files"]
            zarr_path = None
            for p in zarr_list:
                if sample_id in p:
                    zarr_path = p
                    break
            
            if zarr_path is None:
                raise ValueError(f"No zarr found for {sample_id}")
            
            # Load image from Zarr
            store = s3.get_mapper(zarr_path)
            root = zarr.open_group(store=store, mode="r")
            table_keys = list(root["tables"].group_keys())
            table_grp = root["tables"]["table"] if "table" in table_keys else root["tables"][table_keys[0]]
            adata_img = ad.read_zarr(table_grp)
            
            lib = next(iter(adata_img.uns["spatial"].keys()))
            img = adata_img.uns["spatial"][lib]["images"][img_key]
            scalefactors = adata_img.uns["spatial"][lib]["scalefactors"]
            sf_key = "tissue_hires_scalef" if img_key == "hires" else "tissue_lowres_scalef"
            sf = float(scalefactors[sf_key])
            
            # Coordinates
            if {"pxl_col_in_fullres", "pxl_row_in_fullres"}.issubset(adata_img.obs.columns):
                coords_base = adata_img.obs[["pxl_col_in_fullres", "pxl_row_in_fullres"]].to_numpy()
            else:
                coords_base = np.asarray(adata_img.obsm["spatial"])
            
            # Compute scale
            from sklearn.neighbors import NearestNeighbors
            nn = NearestNeighbors(n_neighbors=2).fit(coords_base)
            d, _ = nn.kneighbors(coords_base)
            median_pitch_px = float(np.median(d[:, 1]))
            
            pitch_um = 100.0
            spot_diam_um = 55.0
            
            h_px, w_px = img.shape[:2]
            coords_max = coords_base.max(axis=0)
            coords_fullres = (coords_max[0] > 3 * w_px) or (coords_max[1] > 3 * h_px)
            
            if coords_fullres:
                um_per_px_fullres = pitch_um / median_pitch_px
                um_per_px_img = um_per_px_fullres / sf
                x_um = coords_base[:, 0] * um_per_px_fullres
                y_um = coords_base[:, 1] * um_per_px_fullres
            else:
                um_per_px_img = pitch_um / median_pitch_px
                x_um = coords_base[:, 0] * um_per_px_img
                y_um = coords_base[:, 1] * um_per_px_img
            
            w_um = w_px * um_per_px_img
            h_um = h_px * um_per_px_img
            img_extent = [0.0, w_um, h_um, 0.0]
            
            pad_frac = 0.05
            span = max(x_um.max() - x_um.min(), y_um.max() - y_um.min())
            pad_um = pad_frac * span
            x_lim = (x_um.min() - pad_um, x_um.max() + pad_um)
            y_lim = (y_um.min() - pad_um, y_um.max() + pad_um)
            
            r_um = spot_diam_um / 2.0
            
            # Get ECM annotations for this sample
            ecm_sample = ECM_DF[ECM_DF["sample"] == sample_id].copy()
            
            # Build barcode -> ECM neighborhood mapping
            # Extract barcode from spot_id (format: sample_BARCODE-1)
            ecm_sample["barcode"] = ecm_sample["spot_id"].str.replace(f"{sample_id}_", "", regex=False)
            barcode_to_ecm = dict(zip(ecm_sample["barcode"], ecm_sample["ECM_neighborhood"]))
            
            # Match to image barcodes
            img_barcodes = adata_img.obs_names.astype(str)
            ecm_values = np.array([barcode_to_ecm.get(bc, -1) for bc in img_barcodes])
            
            n_spots = len(x_um)
            
            # Create colors and alphas
            rgba = np.zeros((n_spots, 4))
            for i in range(n_spots):
                ecm = ecm_values[i]
                if ecm in neighborhoods and ecm >= 0:
                    rgba[i, :3] = to_rgb(ECM_COLORS[ecm])
                    rgba[i, 3] = spot_alpha
                else:
                    rgba[i, 3] = 0  # transparent if not in selected neighborhoods
            
            # Create patches
            patches = [Circle((x_um[i], y_um[i]), r_um) for i in range(n_spots)]
            pc = PatchCollection(patches, linewidth=0.0)
            pc.set_facecolors(rgba)
            
            # Plot with spots
            ax_spots.imshow(img, extent=img_extent, alpha=he_alpha)
            ax_spots.add_collection(pc)
            ax_spots.set_xlim(x_lim)
            ax_spots.set_ylim(y_lim[1], y_lim[0])
            ax_spots.set_aspect("equal")
            ax_spots.axis("off")
            ax_spots.set_title(f"{sample_id}" + (" - ECM neighborhoods" if show_he_only else ""), 
                              fontsize=title_fontsize)
            
            # Plot H&E only if requested
            if ax_he is not None:
                ax_he.imshow(img, extent=img_extent, alpha=1.0)
                ax_he.set_xlim(x_lim)
                ax_he.set_ylim(y_lim[1], y_lim[0])
                ax_he.set_aspect("equal")
                ax_he.axis("off")
                ax_he.set_title(f"{sample_id} - H&E only", fontsize=title_fontsize)
            
        except Exception as e:
            import traceback
            ax_spots.set_title(f"{sample_id}: ERROR", fontsize=title_fontsize)
            ax_spots.text(0.5, 0.5, str(e)[:80], ha='center', va='center', 
                         transform=ax_spots.transAxes, fontsize=8, wrap=True)
            ax_spots.axis("off")
            if ax_he is not None:
                ax_he.axis("off")
            traceback.print_exc()
    
    # Hide empty panels (only relevant when not show_he_only)
    if not show_he_only:
        for idx in range(n_samples, n_rows * n_cols):
            axes[idx // n_cols, idx % n_cols].axis("off")
    
    # Legend
    if show_legend:
        handles = [Line2D([0], [0], marker='o', color='w', 
                          markerfacecolor=ECM_COLORS[n], markersize=10, 
                          label=f"ECM {n}") for n in sorted(neighborhoods)]
        if n_samples == 1 and not show_he_only:
            axes[0, 0].legend(handles=handles, loc='upper right', fontsize=legend_fontsize,
                              frameon=True, framealpha=0.9, edgecolor='none')
        else:
            fig.legend(handles=handles, loc='center right', bbox_to_anchor=(1.08, 0.5),
                       fontsize=legend_fontsize, frameon=False)
    
    plt.tight_layout()
    
    if save_path:
        fig.savefig(save_path, dpi=300, bbox_inches="tight", facecolor="white")
        print(f"Saved: {save_path}")
    
    plt.show()
    return fig, axes
```

```python
sample_ids = ["CR_G_348a_vis"]
neighborhoods = None  # None = all (0-9), or e.g. [0, 1, 2] for specific ones

fig, axes = plot_visium_ecm(
    sample_ids=sample_ids,
    neighborhoods=neighborhoods,
    figsize_per_panel=(8, 8),
    spot_alpha=0.85,
    he_alpha=1.0,
    show_he_only=True,  # True = side-by-side spots+H&E and H&E only
    show_legend=True,
    # save_path="ecm_neighborhoods.png",
)
```

### Plot TIL Zones

```python
# %%
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from matplotlib.collections import PatchCollection
from matplotlib.colors import to_rgb
from matplotlib.lines import Line2D
import s3fs
import zarr
import anndata as ad
from cloudpathlib import S3Path
from gbmcollaboration.definitions import DATA_PATH_CHUV_FILES, DATA_PATH_UKER_FILES, DATA_PATH_CR_FILES

# TIL zone colors
TIL_COLORS = {
    "TIL+": "#E31A1C",   # red - TIL positive core
    "BZ1": "#FF7F00",    # orange - border zone 1 (closer to TIL+)
    "BZ2": "#FFD700",    # yellow - border zone 2 (further from TIL+)
    "TIL-": "#377EB8",   # blue - TIL negative
}


def plot_visium_til(
    sample_ids,
    adata,
    zones=None,
    exclude_til_negative=True,
    img_key="hires",
    figsize_per_panel=(8, 8),
    spot_alpha=0.85,
    he_alpha=1.0,
    show_he_only=False,
    show_legend=True,
    legend_fontsize=9,
    title_fontsize=11,
    save_path=None,
):
    """
    Plot TIL zones on Visium H&E images.
    
    Parameters
    ----------
    sample_ids : str or list of sample IDs
    adata : AnnData with TIL_zone column in obs
    zones : list of str, optional
        Which zones to show (e.g., ["TIL+", "BZ1", "BZ2", "TIL-"]). If None, shows all.
    exclude_til_negative : bool
        If True (default), don't plot TIL- spots
    show_he_only : bool
        If True, show side-by-side: spots+H&E and H&E only
    """
    if isinstance(sample_ids, str):
        sample_ids = [sample_ids]
    
    if "TIL_zone" not in adata.obs.columns:
        raise ValueError("TIL_zone not found in adata.obs. Run the TIL annotation loading cell first.")
    
    # Determine which zones to plot
    available_zones = adata.obs["TIL_zone"].dropna().unique().tolist()
    if zones is None:
        zones = available_zones
    if exclude_til_negative and "TIL-" in zones:
        zones = [z for z in zones if z != "TIL-"]
    
    # Center mapping
    center_map = {
        "CH": ("CHUV", DATA_PATH_CHUV_FILES),
        "ER": ("UKER", DATA_PATH_UKER_FILES),
        "CR": ("CR", DATA_PATH_CR_FILES),
    }
    
    # Figure layout
    n_samples = len(sample_ids)
    if show_he_only:
        n_cols = 2
        n_rows = n_samples
        fig_w = figsize_per_panel[0] * 2
        fig_h = figsize_per_panel[1] * n_samples
    else:
        n_cols = min(3, n_samples)
        n_rows = int(np.ceil(n_samples / n_cols))
        fig_w = figsize_per_panel[0] * n_cols
        fig_h = figsize_per_panel[1] * n_rows
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(fig_w, fig_h), squeeze=False)
    
    s3 = s3fs.S3FileSystem(anon=False)
    
    for idx, sample_id in enumerate(sample_ids):
        if show_he_only:
            ax_spots = axes[idx, 0]
            ax_he = axes[idx, 1]
        else:
            ax_spots = axes[idx // n_cols, idx % n_cols]
            ax_he = None
        
        try:
            # Determine center
            prefix = sample_id[:2]
            if prefix not in center_map:
                raise ValueError(f"Unknown center prefix '{prefix}'")
            center_name, data_paths = center_map[prefix]
            
            # Find zarr path
            zarr_list = data_paths["Spatial Zarr Files"]
            zarr_path = None
            for p in zarr_list:
                if sample_id in p:
                    zarr_path = p
                    break
            
            if zarr_path is None:
                raise ValueError(f"No zarr found for {sample_id}")
            
            # Load image from Zarr
            store = s3.get_mapper(zarr_path)
            root = zarr.open_group(store=store, mode="r")
            table_keys = list(root["tables"].group_keys())
            table_grp = root["tables"]["table"] if "table" in table_keys else root["tables"][table_keys[0]]
            adata_img = ad.read_zarr(table_grp)
            
            lib = next(iter(adata_img.uns["spatial"].keys()))
            img = adata_img.uns["spatial"][lib]["images"][img_key]
            scalefactors = adata_img.uns["spatial"][lib]["scalefactors"]
            sf_key = "tissue_hires_scalef" if img_key == "hires" else "tissue_lowres_scalef"
            sf = float(scalefactors[sf_key])
            
            # Coordinates
            if {"pxl_col_in_fullres", "pxl_row_in_fullres"}.issubset(adata_img.obs.columns):
                coords_base = adata_img.obs[["pxl_col_in_fullres", "pxl_row_in_fullres"]].to_numpy()
            else:
                coords_base = np.asarray(adata_img.obsm["spatial"])
            
            # Compute scale
            from sklearn.neighbors import NearestNeighbors
            nn = NearestNeighbors(n_neighbors=2).fit(coords_base)
            d, _ = nn.kneighbors(coords_base)
            median_pitch_px = float(np.median(d[:, 1]))
            
            pitch_um = 100.0
            spot_diam_um = 55.0
            
            h_px, w_px = img.shape[:2]
            coords_max = coords_base.max(axis=0)
            coords_fullres = (coords_max[0] > 3 * w_px) or (coords_max[1] > 3 * h_px)
            
            if coords_fullres:
                um_per_px_fullres = pitch_um / median_pitch_px
                um_per_px_img = um_per_px_fullres / sf
                x_um = coords_base[:, 0] * um_per_px_fullres
                y_um = coords_base[:, 1] * um_per_px_fullres
            else:
                um_per_px_img = pitch_um / median_pitch_px
                x_um = coords_base[:, 0] * um_per_px_img
                y_um = coords_base[:, 1] * um_per_px_img
            
            w_um = w_px * um_per_px_img
            h_um = h_px * um_per_px_img
            img_extent = [0.0, w_um, h_um, 0.0]
            
            pad_frac = 0.05
            span = max(x_um.max() - x_um.min(), y_um.max() - y_um.min())
            pad_um = pad_frac * span
            x_lim = (x_um.min() - pad_um, x_um.max() + pad_um)
            y_lim = (y_um.min() - pad_um, y_um.max() + pad_um)
            
            r_um = spot_diam_um / 2.0
            
            # Get TIL zone for this sample from adata
            sample_mask = adata.obs["sample"].astype(str) == sample_id
            adata_sample = adata[sample_mask]
            
            # Build barcode -> TIL zone mapping
            # adata barcodes include sample prefix, need to extract just barcode part
            barcode_to_til = {}
            for bc in adata_sample.obs_names:
                bc_str = str(bc)
                til_zone = adata_sample.obs.loc[bc, "TIL_zone"]
                # Extract barcode suffix (after sample_id_)
                if bc_str.startswith(sample_id + "_"):
                    barcode_only = bc_str.replace(sample_id + "_", "")
                else:
                    barcode_only = bc_str
                barcode_to_til[barcode_only] = til_zone
            
            # Match to image barcodes
            img_barcodes = adata_img.obs_names.astype(str)
            til_values = [barcode_to_til.get(bc, None) for bc in img_barcodes]
            
            n_spots = len(x_um)
            
            # Create colors and alphas
            rgba = np.zeros((n_spots, 4))
            for i in range(n_spots):
                til = til_values[i]
                if til in zones and til is not None:
                    rgba[i, :3] = to_rgb(TIL_COLORS.get(til, "#808080"))
                    rgba[i, 3] = spot_alpha
                else:
                    rgba[i, 3] = 0  # transparent
            
            # Create patches
            patches = [Circle((x_um[i], y_um[i]), r_um) for i in range(n_spots)]
            pc = PatchCollection(patches, linewidth=0.0)
            pc.set_facecolors(rgba)
            
            # Plot with spots
            ax_spots.imshow(img, extent=img_extent, alpha=he_alpha)
            ax_spots.add_collection(pc)
            ax_spots.set_xlim(x_lim)
            ax_spots.set_ylim(y_lim[1], y_lim[0])
            ax_spots.set_aspect("equal")
            ax_spots.axis("off")
            ax_spots.set_title(f"{sample_id}" + (" - TIL zones" if show_he_only else ""),
                              fontsize=title_fontsize)
            
            # Plot H&E only if requested
            if ax_he is not None:
                ax_he.imshow(img, extent=img_extent, alpha=1.0)
                ax_he.set_xlim(x_lim)
                ax_he.set_ylim(y_lim[1], y_lim[0])
                ax_he.set_aspect("equal")
                ax_he.axis("off")
                ax_he.set_title(f"{sample_id} - H&E only", fontsize=title_fontsize)
            
        except Exception as e:
            import traceback
            ax_spots.set_title(f"{sample_id}: ERROR", fontsize=title_fontsize)
            ax_spots.text(0.5, 0.5, str(e)[:80], ha='center', va='center',
                         transform=ax_spots.transAxes, fontsize=8, wrap=True)
            ax_spots.axis("off")
            if ax_he is not None:
                ax_he.axis("off")
            traceback.print_exc()
    
    # Hide empty panels
    if not show_he_only:
        for idx in range(n_samples, n_rows * n_cols):
            axes[idx // n_cols, idx % n_cols].axis("off")
    
    # Legend - order zones logically
    if show_legend:
        zone_order = ["TIL+", "BZ1", "BZ2", "TIL-"]
        legend_zones = [z for z in zone_order if z in zones]
        handles = [Line2D([0], [0], marker='o', color='w',
                          markerfacecolor=TIL_COLORS[z], markersize=10,
                          label=z) for z in legend_zones]
        if n_samples == 1 and not show_he_only:
            axes[0, 0].legend(handles=handles, loc='upper right', fontsize=legend_fontsize,
                              frameon=True, framealpha=0.9, edgecolor='none')
        else:
            fig.legend(handles=handles, loc='center right', bbox_to_anchor=(1.08, 0.5),
                       fontsize=legend_fontsize, frameon=False)
    
    plt.tight_layout()
    
    if save_path:
        fig.savefig(save_path, dpi=300, bbox_inches="tight", facecolor="white")
        print(f"Saved: {save_path}")
    
    plt.show()
    return fig, axes
```

```python
sample_ids = ["ER_G_672a_vis"]

fig, axes = plot_visium_til(
    sample_ids=sample_ids,
    adata=adata,
    zones=None,  # None = all available, or e.g. ["TIL+", "BZ1", "BZ2"]
    exclude_til_negative=True,  # True = don't show TIL- spots
    figsize_per_panel=(8, 8),
    spot_alpha=0.8,
    he_alpha=1.0,
    show_he_only=True,  # True = side-by-side spots+H&E and H&E only
    show_legend=True,
    # save_path="til_zones.png",
)
```

```python
# %%
import tempfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from matplotlib.collections import PatchCollection
from matplotlib.colors import to_rgb, to_hex
from matplotlib.lines import Line2D
from matplotlib import cm
from matplotlib.backends.backend_pdf import PdfPages
import s3fs
import openslide
from pathlib import Path

from gbmcollaboration import CHUVDataCenter, UKERDataCenter, CRDataCenter

# =============================================================================
# Configuration
# =============================================================================

SIMPLIFIED_ANNOTATION_MAPPING = Path(
    "/home/sagemaker-user/custom-file-systems/efs/fs-09913c1f7db79b6fd/resources/SingleCell_Annotation_Key_v3.csv"
)
ECM_ANNOTATIONS_PATH = "/home/sagemaker-user/custom-file-systems/efs/fs-09913c1f7db79b6fd/Analysis_Repo/Notebooks/ecm_neighborhood_annotations.csv"

REFERENCE_COLUMN = "celltype_level4_scanvi"
DECONV_LEVEL_MAP = {
    "deconv_cohort_level_1": "level1",
    "deconv_cohort_level_2": "level2",
    "deconv_cohort_level_3": "level3",
    "deconv_cohort_level_4": "level4",
}

S3_ALIAS = "pa-bi4jtrdsa151p4d3m-jepjxh9qrwgxphe8hd3r6exicahnoeuw1b-s3alias"
VISIUM_PREFIX = {
    "CH": "b5bd8840-cc6e-4de2-a78a-22315e9d1734",
    "ER": "40ce7e61-5ff6-467e-9972-0c85705f086c",
    "CR": "f0415978-2385-4067-b215-13e699a0f7a7",
}
ALIGNED_DIRNAME = "aligned_features_DinoV2ViTBaseDistilled_aquavit"

# Cell type colors
CELLTYPE_COLORS = {
    # T cells - vibrant reds
    "T_cell_IFN_response": "#E31A1C",
    "T_cell_CD8": "#FF4136",
    "T_cell_CD4": "#B22222",
    "T_cell_nos": "#DC143C",
    "T_cell_other": "#FF6B6B",
    "T_reg": "#CD5C5C",
    # Myeloid - yellows to oranges
    "Macrophage_IFN_response": "#FF8C00",
    "Macrophage_LAM": "#FFA500",
    "Macrophage_PVM": "#FFB347",
    "Macrophage_C1Q_pos": "#E69500",
    "Macrophage_cycling": "#FFAA00",
    "Microglia_IFN_response": "#FFD700",
    "Microglia_TREM2_pos": "#FFCC00",
    "Microglia_homeostatic": "#F0E68C",
    "Microglia_intermediate_reactive": "#FFE135",
    "cDC_CD207_pos": "#FF7F00",
    "cDC1": "#FF9933",
    "cDC2": "#FFBF00",
    "cDC2_cycling": "#FFD54F",
    "pDC": "#FFC125",
    "Monocyte_nos": "#CC7000",
    "Neutrophil": "#D2691E",
    "Neutrophil_immature": "#E08000",
    "Mast_cell": "#DAA520",
    # Malignant - greens to blues
    "Malignant_AC-like_IFN_response": "#00BFA5",
    "Malignant_AC1": "#00CC66",
    "Malignant_AC2": "#00E676",
    "Malignant_MES-hypoxia": "#40E0D0",
    "Malignant_MES1": "#00CED1",
    "Malignant_NPC2": "#4169E1",
    "Malignant_OPC-like_invasive": "#1E90FF",
    "Malignant_OPC1": "#00BFFF",
    "Malignant_OPC2": "#6495ED",
    "Malignant_cycling": "#3CB371",
    "Malignant_other_1": "#2E8B57",
    "Malignant_other_2": "#20B2AA",
    "Malignant_other_stressed": "#5F9EA0",
    # Astrocytes - cyans
    "Astrocyte_IFN_response": "#00CED1",
    "Astrocyte_homeostatic": "#76D7EA",
    "Astrocyte_inflammatory": "#00FFFF",
    "Astrocyte_intermediate_reactive": "#40E0D0",
    "Astrocyte_reactive": "#00E5EE",
    # Oligodendrocytes - teals
    "Oligodendrocyte_nos": "#2F9C95",
    "Oligodendrocyte_reactive": "#008B8B",
    # Vasculature - dark reds to browns
    "Endothelial": "#8B0000",
    "Endothelial_arterial": "#A52A2A",
    "Endothelial_capillary": "#800000",
    "Endothelial_cycling": "#B22222",
    "Endothelial_tip": "#722F37",
    "Endothelial_veinous": "#6B3A3A",
    "Pericyte": "#5C4033",
    "Pericyte_cycling": "#6B4423",
    "Pericyte_fibro-like": "#7B3F00",
    "Perivascular_fibroblast_cycling": "#8B4513",
    "Perivascular_fibroblast_type-I": "#A0522D",
    "Perivascular_fibroblast_type-II": "#8B5A2B",
    "Perivascular_fibroblast_type-III": "#704214",
    "VSMC": "#4A2C2A",
    # Neurons - magentas/pinks
    "Neuron_nos": "#FF1493",
    "Neuron_excitatory_CDH24": "#FF69B4",
    "Neuron_excitatory_nos": "#FF85C1",
    "Neuron_inhibitory_LAMP5": "#DB7093",
    "Neuron_inhibitory_NGF": "#FF6EB4",
    "Neuron_inhibitory_PVALB": "#C71585",
    "Neuron_inhibitory_SST": "#E75480",
    "Neuron_inhibitory_VIP": "#DE3163",
    # Lymphoid
    "B_cell": "#9370DB",
    "NK_KLRC1_pos": "#BA55D3",
    "NK_nos": "#DA70D6",
    "Plasma_cell": "#8A2BE2",
    "Plasmablast": "#9932CC",
    "T_NK_cycling": "#FF6347",
}

# ECM neighborhood colors
ECM_COLORS = {
    0: "#377EB8",  # blue
    1: "#FF7F00",  # orange
    2: "#4DAF4A",  # green
    3: "#E41A1C",  # red
    4: "#984EA3",  # purple
    5: "#8B4513",  # brown
    6: "#FF1493",  # magenta/pink
    7: "#808080",  # gray
    8: "#9ACD32",  # yellow-green
    9: "#008B8B",  # dark cyan
}

# TIL zone colors
TIL_COLORS = {
    "TIL+": "#E31A1C",
    "BZ1": "#FF7F00",
    "BZ2": "#FFD700",
    "TIL-": "#377EB8",
}

# =============================================================================
# Caching
# =============================================================================

_SPATIAL_CACHE = {}
_ECM_DF = None

def _get_spatial_object(center_prefix, normalization, deconvolution_slot_name,
                        exclude_low_quality_visium, exclude_idh1_mut):
    key = (center_prefix, normalization, deconvolution_slot_name,
           exclude_low_quality_visium, exclude_idh1_mut)
    if key in _SPATIAL_CACHE:
        return _SPATIAL_CACHE[key]
    
    loader_map = {"CH": CHUVDataCenter, "ER": UKERDataCenter, "CR": CRDataCenter}
    spatial_obj = loader_map[center_prefix].load_spatial(
        normalization=normalization,
        deconvolution_slot_name=deconvolution_slot_name,
        list_sample_id=None,
        exclude_low_quality_visium=exclude_low_quality_visium,
        exclude_idh1_mut=exclude_idh1_mut,
    )
    level = DECONV_LEVEL_MAP.get(deconvolution_slot_name)
    if level is not None:
        spatial_obj.celltype_df = _map_celltype_columns(spatial_obj.celltype_df, level)
    _SPATIAL_CACHE[key] = spatial_obj
    return spatial_obj

def _map_celltype_columns(celltype_df, level):
    map_df = pd.read_csv(SIMPLIFIED_ANNOTATION_MAPPING)
    target_col = f"celltype_{level}_curated"
    ct_mapping = map_df.set_index(REFERENCE_COLUMN)[target_col].dropna().astype(str).to_dict()
    rename_dict = {col: ct_mapping[col] for col in celltype_df.columns if col in ct_mapping}
    return celltype_df.rename(columns=rename_dict)

def _get_ecm_df():
    global _ECM_DF
    if _ECM_DF is None:
        _ECM_DF = pd.read_csv(ECM_ANNOTATIONS_PATH)
    return _ECM_DF

# =============================================================================
# Main plotting function
# =============================================================================

def plot_visium(
    sample_id,
    # What to plot
    plot_type="celltypes",  # "celltypes", "ecm", "til", "he_only"
    celltypes=None,
    ecm_neighborhoods=None,
    til_zones=None,
    exclude_til_negative=True,
    mode="blend",  # for celltypes: "blend" or "max"
    # Data loading
    adata=None,  # required for TIL zones
    deconvolution_slot_name="deconv_cohort_level_4",
    normalization="raw_count",
    exclude_low_quality_visium=True,
    exclude_idh1_mut=True,
    # Display
    figsize=(12, 10),
    spot_alpha=0.9,
    he_alpha=0.8,
    clip_q=(5, 95),
    gamma=0.6,
    margin_px=1000,
    max_view_px=9000,
    spot_diam_um=55.0,
    fallback_mpp=0.5,
    title=None,
    title_fontsize=14,
    # Output
    save_pdf=False,
    save_path=None,
    save_dpi=300,
):
    """
    Plot Visium data on aligned high-resolution H&E.
    
    Parameters
    ----------
    sample_id : str
        Sample ID (e.g., "CH_G_196a_vis")
    plot_type : str
        "celltypes" - cell type abundances
        "ecm" - ECM neighborhoods
        "til" - TIL zones
        "he_only" - just the H&E image
    celltypes : list of str
        Cell types to plot (celltype_level4_curated names)
    ecm_neighborhoods : list of int
        ECM neighborhoods to plot (0-9), None for all
    til_zones : list of str
        TIL zones to plot, None for all (respects exclude_til_negative)
    adata : AnnData
        Required for TIL zones (must have TIL_zone in obs)
    save_pdf : bool
        If True, save as compressed PDF
    save_path : str
        Path to save (auto-generates if None and save_pdf=True)
    
    Returns
    -------
    dict with keys: "fig", "ax", "legend_fig", "legend_ax"
    """
    
    s3 = s3fs.S3FileSystem(anon=False)
    center = sample_id[:2]
    prefix = VISIUM_PREFIX[center]
    
    def s3key(uri):
        return uri.replace("s3://", "")
    
    # Load aligned coordinates
    coords_uri = f"s3://{S3_ALIAS}/{prefix}/v7/image_features/{ALIGNED_DIRNAME}/{sample_id}/filtered_coords.csv"
    with s3.open(s3key(coords_uri), "rb") as f:
        coords_df = pd.read_csv(f)
    coords_df["barcode"] = coords_df["barcode"].astype(str)
    
    # Load H&E image
    he_uri = f"s3://{S3_ALIAS}/{prefix}/v7/converted_he/{sample_id}.tif"
    tmp = tempfile.NamedTemporaryFile(suffix=".tif", delete=False)
    tmp.close()
    s3.get(s3key(he_uri), tmp.name)
    slide = openslide.OpenSlide(tmp.name)
    
    # Get coordinates
    x_full = coords_df["pxl_col_in_fullres"].values
    y_full = coords_df["pxl_row_in_fullres"].values
    barcodes = coords_df["barcode"].values
    
    # Compute view region
    x0 = int(x_full.min() - margin_px)
    y0 = int(y_full.min() - margin_px)
    x1 = int(x_full.max() + margin_px)
    y1 = int(y_full.max() + margin_px)
    w, h = x1 - x0, y1 - y0
    
    # Determine downsample level
    scale = max(w / max_view_px, h / max_view_px, 1.0)
    level = slide.get_best_level_for_downsample(scale)
    ds = slide.level_downsamples[level]
    
    # Read region
    region = slide.read_region((x0, y0), level, (int(w/ds), int(h/ds))).convert("RGB")
    img = np.asarray(region)
    
    # Spot radius
    mpp = float(slide.properties.get("openslide.mpp-x", fallback_mpp))
    r = (spot_diam_um / 2) / mpp / ds
    
    # Transform coordinates to image space
    x = (x_full - x0) / ds
    y = (y_full - y0) / ds
    n_spots = len(x)
    
    # Get spot colors based on plot_type
    if plot_type == "he_only":
        rgba = None
        legend_items = []
        
    elif plot_type == "celltypes":
        if celltypes is None:
            raise ValueError("celltypes required for plot_type='celltypes'")
        
        # Load cell type data
        spatial = _get_spatial_object(center, normalization, deconvolution_slot_name,
                                      exclude_low_quality_visium, exclude_idh1_mut)
        cell_df = spatial.celltype_df.xs(sample_id, level=0).reset_index()
        
        # Merge with coordinates
        merged = coords_df.merge(cell_df, on="barcode", how="inner")
        
        # Build colors
        ct_colors = dict(CELLTYPE_COLORS)
        need = [ct for ct in celltypes if ct not in ct_colors]
        if need:
            cmap = plt.colormaps.get_cmap("tab20")
            for i, ct in enumerate(need):
                ct_colors[ct] = to_hex(cmap(i % 20))
        ct_rgb = {ct: np.array(to_rgb(ct_colors[ct])) for ct in celltypes}
        
        # Get abundance matrix
        M = merged[celltypes].values
        vmin = np.percentile(M, clip_q[0], axis=0)
        vmax = np.percentile(M, clip_q[1], axis=0)
        M_norm = np.clip((M - vmin) / (vmax - vmin + 1e-10), 0, 1) ** gamma
        
        if mode == "max":
            idx = np.argmax(M_norm, axis=1)
            rgb = np.array([ct_rgb[celltypes[i]] for i in idx])
            alpha = np.max(M_norm, axis=1)
        else:  # blend
            rgb = np.zeros((M_norm.shape[0], 3))
            tot = np.zeros(M_norm.shape[0])
            for j, ct in enumerate(celltypes):
                rgb += M_norm[:, j, None] * ct_rgb[ct]
                tot += M_norm[:, j]
            rgb /= np.maximum(tot[:, None], 1e-6)
            alpha = tot / len(celltypes)
        
        rgba = np.column_stack([rgb, alpha * spot_alpha])
        
        # Update coordinates to match merged data
        x = (merged["pxl_col_in_fullres"].values - x0) / ds
        y = (merged["pxl_row_in_fullres"].values - y0) / ds
        n_spots = len(x)
        
        legend_items = [(ct, ct_rgb[ct]) for ct in celltypes]
        
    elif plot_type == "ecm":
        ecm_df = _get_ecm_df()
        ecm_sample = ecm_df[ecm_df["sample"] == sample_id].copy()
        ecm_sample["barcode"] = ecm_sample["spot_id"].str.replace(f"{sample_id}_", "", regex=False)
        
        barcode_to_ecm = dict(zip(ecm_sample["barcode"], ecm_sample["ECM_neighborhood"]))
        
        if ecm_neighborhoods is None:
            ecm_neighborhoods = list(range(10))
        
        rgba = np.zeros((n_spots, 4))
        for i, bc in enumerate(barcodes):
            ecm = barcode_to_ecm.get(bc, -1)
            if ecm in ecm_neighborhoods and ecm >= 0:
                rgba[i, :3] = to_rgb(ECM_COLORS[ecm])
                rgba[i, 3] = spot_alpha
        
        legend_items = [(f"ECM {n}", to_rgb(ECM_COLORS[n])) for n in sorted(ecm_neighborhoods)]
        
    elif plot_type == "til":
        if adata is None:
            raise ValueError("adata required for plot_type='til'")
        if "TIL_zone" not in adata.obs.columns:
            raise ValueError("TIL_zone not in adata.obs")
        
        sample_mask = adata.obs["sample"].astype(str) == sample_id
        adata_sample = adata[sample_mask]
        
        # Build barcode -> TIL mapping
        barcode_to_til = {}
        for bc in adata_sample.obs_names:
            bc_str = str(bc)
            til = adata_sample.obs.loc[bc, "TIL_zone"]
            bc_only = bc_str.replace(f"{sample_id}_", "") if bc_str.startswith(f"{sample_id}_") else bc_str
            barcode_to_til[bc_only] = til
        
        # Determine zones to show
        available = list(adata_sample.obs["TIL_zone"].dropna().unique())
        if til_zones is None:
            til_zones = available
        if exclude_til_negative and "TIL-" in til_zones:
            til_zones = [z for z in til_zones if z != "TIL-"]
        
        rgba = np.zeros((n_spots, 4))
        for i, bc in enumerate(barcodes):
            til = barcode_to_til.get(bc)
            if til in til_zones and til is not None:
                rgba[i, :3] = to_rgb(TIL_COLORS.get(til, "#808080"))
                rgba[i, 3] = spot_alpha
        
        zone_order = ["TIL+", "BZ1", "BZ2", "TIL-"]
        legend_items = [(z, to_rgb(TIL_COLORS[z])) for z in zone_order if z in til_zones]
    
    else:
        raise ValueError(f"Unknown plot_type: {plot_type}")
    
    # Create main figure
    fig, ax = plt.subplots(figsize=figsize)
    ax.imshow(img, alpha=he_alpha)
    
    if rgba is not None:
        patches = [Circle((x[i], y[i]), r) for i in range(n_spots)]
        pc = PatchCollection(patches, linewidth=0)
        pc.set_facecolors(rgba)
        ax.add_collection(pc)
    
    ax.axis("off")
    if title is None:
        title = f"{sample_id} - {plot_type}"
    ax.set_title(title, fontsize=title_fontsize)
    
    plt.tight_layout()
    
    # Create legend figure
    legend_fig = None
    legend_ax = None
    if legend_items:
        legend_fig, legend_ax = plt.subplots(figsize=(4, max(2, len(legend_items) * 0.4)))
        handles = [Line2D([0], [0], marker='o', color='w', markerfacecolor=color, 
                          markersize=12, label=label) for label, color in legend_items]
        legend_ax.legend(handles=handles, loc='center', frameon=False, fontsize=10)
        legend_ax.axis("off")
        legend_fig.tight_layout()
    
    # Save
    if save_pdf:
        if save_path is None:
            save_path = f"{sample_id}_{plot_type}.pdf"
        
        with PdfPages(save_path) as pdf:
            pdf.savefig(fig, dpi=save_dpi, bbox_inches="tight")
            if legend_fig is not None:
                pdf.savefig(legend_fig, dpi=150, bbox_inches="tight")
        print(f"Saved: {save_path}")
    
    plt.show()
    slide.close()
    
    return {"fig": fig, "ax": ax, "legend_fig": legend_fig, "legend_ax": legend_ax}
```

```python
# Cell: Plot cell type abundances on H&E

import tempfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.patches import Circle
from matplotlib.collections import PatchCollection
from matplotlib.colors import to_rgb, to_hex, LinearSegmentedColormap
from PIL import Image
import s3fs
import openslide

from gbmcollaboration import CHUVDataCenter, UKERDataCenter, CRDataCenter

# ===== USER CONFIG =====
sample_id = "CH_G_174a_vis"
celltypes = ["Malignant_MES1",
            "Malignant_AC1",
            "Malignant_AC2",
            "Malignant_NPC2", 
            "Malignant_OPC1", 
            "Malignant_OPC2",]
mode = "max"              # or "blend" or "max"

figsize = (14, 12)
spot_alpha = 0.99
he_alpha = 0.6
margin_px = 800
max_view_px = 6000
spot_diam_um = 55.0
fallback_mpp = 0.5
clip_q = (5, 95)
gamma = 0.6

shared_scale = True              # True = all cell types share global min/max
show_legend = True
show_title = False
title = None
title_fontsize = 14
label_color = "black"             # color for cell type labels in legend

# Data loading
deconvolution_slot_name = "deconv_cohort_level_4"
normalization = "raw_count"
exclude_low_quality_visium = True
exclude_idh1_mut = True

# Output
save_png = True
save_dpi = 200
jpeg_quality = 92
output_format = "png"
# ========================

S3_ALIAS = "pa-bi4jtrdsa151p4d3m-jepjxh9qrwgxphe8hd3r6exicahnoeuw1b-s3alias"
VISIUM_PREFIX = {
    "CH": "b5bd8840-cc6e-4de2-a78a-22315e9d1734",
    "ER": "40ce7e61-5ff6-467e-9972-0c85705f086c",
    "CR": "f0415978-2385-4067-b215-13e699a0f7a7",
}
ALIGNED_DIRNAME = "aligned_features_DinoV2ViTBaseDistilled_aquavit"

SIMPLIFIED_ANNOTATION_MAPPING = "/home/sagemaker-user/custom-file-systems/efs/fs-09913c1f7db79b6fd/resources/SingleCell_Annotation_Key_v3.csv"
REFERENCE_COLUMN = "celltype_level4_scanvi"
DECONV_LEVEL_MAP = {
    "deconv_cohort_level_1": "level1",
    "deconv_cohort_level_2": "level2",
    "deconv_cohort_level_3": "level3",
    "deconv_cohort_level_4": "level4",
}

CELLTYPE_COLORS = {
    "Astrocyte_homeostatic": "#8C0CF2",
    "Neuron_nos": "#0089EB",
    "Neuron_excitatory_nos": "#00B7BE",
    "Neuron_inhibitory_LAMP5": "#00E069",
    "Neuron_inhibitory_PVALB": "#F4D900",

    "Macrophage_LAM": "#CA1607",
    "Malignant_MES-hypoxia": "#138417", 
    "Microglia_TREM2_pos": "#B27A00",

    "T_cell_IFN_response": "#9DAC00", 
    "T_cell_CD8": "#0086AB",
    "T_cell_CD4": "#7087FF",
    "Macrophage_IFN_response": "#CB072E",
    "Macrophage_C1Q_pos": "#D36000",

    "Malignant_MES1": "#CC0B19",
    "Malignant_AC1": "#E69500", 
    "Malignant_AC2": "#3BA500",
    "Malignant_NPC2": "#027B88", 
    "Malignant_OPC1": "#5EADFF", 
    "Malignant_OPC2": "#D300FB",
    
    "Malignant_cycling": "#3CB371",
    "Malignant_AC-like_IFN_response": "#00BFA5",
    "Malignant_OPC-like_invasive": "#1E90FF",
    "T_cell_nos": "#DC143C",
    "T_cell_other": "#FF6B6B", "T_reg": "#CD5C5C",
    "Macrophage_PVM": "#FFB347", 
    "Macrophage_cycling": "#FFAA00",
    "Microglia_IFN_response": "#FFD700", 
    "Microglia_homeostatic": "#F0E68C", "Microglia_intermediate_reactive": "#FFE135",
    "cDC_CD207_pos": "#FF7F00", "cDC1": "#FF9933", "cDC2": "#FFBF00",
    "cDC2_cycling": "#FFD54F", "pDC": "#FFC125",
    "Monocyte_nos": "#CC7000", "Neutrophil": "#D2691E",
    "Neutrophil_immature": "#E08000", "Mast_cell": "#DAA520",
    "Malignant_other_1": "#2E8B57",
    "Malignant_other_2": "#20B2AA", "Malignant_other_stressed": "#5F9EA0",
    "Astrocyte_IFN_response": "#00CED1", 
    "Astrocyte_inflammatory": "#00FFFF", "Astrocyte_intermediate_reactive": "#40E0D0",
    "Astrocyte_reactive": "#00E5EE",
    "Oligodendrocyte_nos": "#2F9C95", "Oligodendrocyte_reactive": "#008B8B",
    "Endothelial": "#8B0000", "Endothelial_arterial": "#A52A2A",
    "Endothelial_capillary": "#800000", "Endothelial_cycling": "#B22222",
    "Endothelial_tip": "#722F37", "Endothelial_veinous": "#6B3A3A",
    "Pericyte": "#5C4033", "Pericyte_cycling": "#6B4423",
    "Pericyte_fibro-like": "#7B3F00",
    "Perivascular_fibroblast_cycling": "#8B4513",
    "Perivascular_fibroblast_type-I": "#A0522D",
    "Perivascular_fibroblast_type-II": "#8B5A2B",
    "Perivascular_fibroblast_type-III": "#704214",
    "VSMC": "#4A2C2A",
    "Neuron_excitatory_CDH24": "#00A969",
    "Neuron_inhibitory_NGF": "#C40025",
    "Neuron_inhibitory_SST": "#E83EFF",
    "Neuron_inhibitory_VIP": "#B7A7FF",
    "B_cell": "#9370DB", "NK_KLRC1_pos": "#BA55D3", "NK_nos": "#DA70D6",
    "Plasma_cell": "#8A2BE2", "Plasmablast": "#9932CC", "T_NK_cycling": "#FF6347",
}

# --- Spatial data loader with caching ---
_SPATIAL_CACHE = {}

def _get_spatial_object(center_prefix):
    key = (center_prefix, normalization, deconvolution_slot_name,
           exclude_low_quality_visium, exclude_idh1_mut)
    if key in _SPATIAL_CACHE:
        return _SPATIAL_CACHE[key]
    
    loader_map = {"CH": CHUVDataCenter, "ER": UKERDataCenter, "CR": CRDataCenter}
    spatial_obj = loader_map[center_prefix].load_spatial(
        normalization=normalization,
        deconvolution_slot_name=deconvolution_slot_name,
        list_sample_id=None,
        exclude_low_quality_visium=exclude_low_quality_visium,
        exclude_idh1_mut=exclude_idh1_mut,
    )
    level = DECONV_LEVEL_MAP.get(deconvolution_slot_name)
    if level is not None:
        map_df = pd.read_csv(SIMPLIFIED_ANNOTATION_MAPPING)
        target_col = f"celltype_{level}_curated"
        ct_mapping = map_df.set_index(REFERENCE_COLUMN)[target_col].dropna().astype(str).to_dict()
        rename_dict = {col: ct_mapping[col] for col in spatial_obj.celltype_df.columns if col in ct_mapping}
        spatial_obj.celltype_df = spatial_obj.celltype_df.rename(columns=rename_dict)
    _SPATIAL_CACHE[key] = spatial_obj
    return spatial_obj

# --- Load data ---
s3 = s3fs.S3FileSystem(anon=False)
center = sample_id[:2]
prefix = VISIUM_PREFIX[center]

def s3key(uri):
    return uri.replace("s3://", "")

coords_uri = f"s3://{S3_ALIAS}/{prefix}/v7/image_features/{ALIGNED_DIRNAME}/{sample_id}/filtered_coords.csv"
with s3.open(s3key(coords_uri), "rb") as f:
    coords_df = pd.read_csv(f)
coords_df["barcode"] = coords_df["barcode"].astype(str)

he_uri = f"s3://{S3_ALIAS}/{prefix}/v7/converted_he/{sample_id}.tif"
tmp = tempfile.NamedTemporaryFile(suffix=".tif", delete=False)
tmp.close()
s3.get(s3key(he_uri), tmp.name)
slide = openslide.OpenSlide(tmp.name)

spatial = _get_spatial_object(center)
cell_df = spatial.celltype_df.xs(sample_id, level=0).reset_index()
merged = coords_df.merge(cell_df, on="barcode", how="inner")

missing_ct = [ct for ct in celltypes if ct not in merged.columns]
if missing_ct:
    available = sorted([c for c in cell_df.columns if c != "barcode"])
    raise ValueError(f"Cell types not found: {missing_ct}\nAvailable: {available[:20]}...")

# --- Compute view region ---
x_full = merged["pxl_col_in_fullres"].values
y_full = merged["pxl_row_in_fullres"].values

x0 = max(0, int(x_full.min() - margin_px))
y0 = max(0, int(y_full.min() - margin_px))
x1 = int(x_full.max() + margin_px)
y1 = int(y_full.max() + margin_px)
w, h = x1 - x0, y1 - y0

scale = max(w / max_view_px, h / max_view_px, 1.0)
level = slide.get_best_level_for_downsample(scale)
ds = slide.level_downsamples[level]

region = slide.read_region((x0, y0), level, (int(w / ds), int(h / ds))).convert("RGB")
img = np.asarray(region)

mpp = float(slide.properties.get("openslide.mpp-x", fallback_mpp))
r = (spot_diam_um / 2) / mpp / ds

x = (x_full - x0) / ds
y = (y_full - y0) / ds
n_spots = len(x)

# --- Build colors ---
ct_colors = dict(CELLTYPE_COLORS)
need = [ct for ct in celltypes if ct not in ct_colors]
if need:
    cmap_fallback = plt.colormaps.get_cmap("tab20")
    for i, ct in enumerate(need):
        ct_colors[ct] = to_hex(cmap_fallback(i % 20))
ct_rgb = {ct: np.array(to_rgb(ct_colors[ct])) for ct in celltypes}

# Abundance matrix + normalization
M = merged[celltypes].values.astype(np.float64)

if shared_scale:
    global_lo = np.percentile(M, clip_q[0])
    global_hi = np.percentile(M, clip_q[1])
    vmin_vals = np.full(len(celltypes), global_lo)
    vmax_vals = np.full(len(celltypes), global_hi)
else:
    vmin_vals = np.percentile(M, clip_q[0], axis=0)
    vmax_vals = np.percentile(M, clip_q[1], axis=0)

M_norm = np.clip((M - vmin_vals) / (vmax_vals - vmin_vals + 1e-10), 0, 1) ** gamma

abundance_ranges = {}
for j, ct in enumerate(celltypes):
    abundance_ranges[ct] = (float(vmin_vals[j]), float(vmax_vals[j]))

if mode == "max":
    idx = np.argmax(M_norm, axis=1)
    rgb = np.array([ct_rgb[celltypes[i]] for i in idx])
    alpha = np.max(M_norm, axis=1)
else:
    rgb = np.zeros((M_norm.shape[0], 3))
    tot = np.zeros(M_norm.shape[0])
    for j, ct in enumerate(celltypes):
        rgb += M_norm[:, j, None] * ct_rgb[ct]
        tot += M_norm[:, j]
    rgb /= np.maximum(tot[:, None], 1e-6)
    alpha = np.clip(tot / len(celltypes), 0, 1)

rgba = np.column_stack([rgb, alpha * spot_alpha])

# --- Plot ---
if show_legend:
    legend_width = 3.5
    fig = plt.figure(figsize=(figsize[0] + legend_width, figsize[1]))
    gs = fig.add_gridspec(1, 2, width_ratios=[figsize[0], legend_width], wspace=0.02)
    ax = fig.add_subplot(gs[0])
    ax_legend = fig.add_subplot(gs[1])
else:
    fig, ax = plt.subplots(figsize=figsize)

ax.imshow(img, alpha=he_alpha)

visible_mask = rgba[:, 3] > 0.01
patches = [Circle((x[i], y[i]), r) for i in range(n_spots) if visible_mask[i]]
rgba_visible = rgba[visible_mask]
if len(patches) > 0:
    pc = PatchCollection(patches, linewidth=0)
    pc.set_facecolors(rgba_visible)
    ax.add_collection(pc)

ax.axis("off")
ax.grid(False)

if show_title:
    if title is None:
        ct_short = ", ".join([ct.replace("Malignant_", "Mal_") for ct in celltypes])
        auto_title = f"{sample_id} — {ct_short}"
    else:
        auto_title = title
    ax.set_title(auto_title, fontsize=title_fontsize, fontweight="bold", pad=10)

# --- Legend with gradient colorbars ---
if show_legend:
    ax_legend.axis("off")
    ax_legend.grid(False)

    n_ct = len(celltypes)
    bar_height = 0.06
    bar_width_frac = 0.7
    spacing = max(0.12, 0.85 / max(n_ct, 1))
    start_y = 0.5 + (n_ct - 1) * spacing / 2

    for j, ct in enumerate(celltypes):
        y_pos = start_y - j * spacing
        color = ct_rgb[ct]
        lo, hi = abundance_ranges[ct]

        gradient = np.linspace(0, 1, 256).reshape(1, -1)
        cmap_ct = LinearSegmentedColormap.from_list(
            f"cmap_{ct}", [(1, 1, 1, 0), (*color, 1)], N=256
        )

        bar_x = 0.15
        inset = ax_legend.inset_axes(
            [bar_x, y_pos - bar_height / 2, bar_width_frac, bar_height],
            transform=ax_legend.transAxes
        )
        inset.imshow(gradient, aspect="auto", cmap=cmap_ct, extent=[0, 1, 0, 1])
        inset.set_xlim(0, 1)
        inset.set_ylim(0, 1)
        inset.set_xticks([0, 1])
        inset.set_xticklabels([f"{lo:.1f}", f"{hi:.1f}"], fontsize=7)
        inset.set_yticks([])
        inset.tick_params(axis="x", length=2, pad=1)
        inset.grid(False)
        for spine in inset.spines.values():
            spine.set_linewidth(0.5)
            spine.set_color("0.5")

        # Cell type label
        display_name = ct.replace("_", " ")
        ax_legend.text(
            bar_x + bar_width_frac / 2, y_pos + bar_height / 2 + 0.015,
            display_name, transform=ax_legend.transAxes,
            ha="center", va="bottom", fontsize=9, fontweight="bold",
            color=label_color
        )

    # Scale annotation
    scale_text = "Shared scale" if shared_scale else "Per-type scale"
    mode_text = f"{scale_text} | {mode} | γ={gamma} | clip={clip_q}"
    ax_legend.text(
        0.5, 0.02, mode_text, transform=ax_legend.transAxes,
        ha="center", va="bottom", fontsize=7, color="0.5", fontstyle="italic"
    )

    ax_legend.text(
        0.5, 0.97, "C2L Abundance", transform=ax_legend.transAxes,
        ha="center", va="top", fontsize=11, fontweight="bold"
    )

plt.tight_layout()

# --- Save ---
if save_png:
    ct_slug = "_".join([ct[:15] for ct in celltypes])
    out_path = f"{sample_id}_celltypes_{ct_slug}"

    if output_format == "jpg":
        out_path += ".jpg"
        fig.savefig(out_path, dpi=save_dpi, bbox_inches="tight", pad_inches=0.1,
                    pil_kwargs={"quality": jpeg_quality, "optimize": True})
    else:
        out_path += ".png"
        fig.savefig(out_path, dpi=save_dpi, bbox_inches="tight", pad_inches=0.1)
        img_pil = Image.open(out_path)
        img_pil.save(out_path, optimize=True)

    import os
    fsize_mb = os.path.getsize(out_path) / (1024 * 1024)
    print(f"Saved: {out_path} ({fsize_mb:.1f} MB)")

plt.show()
slide.close()
```
